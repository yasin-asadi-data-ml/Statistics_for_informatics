```python
# ============================================================
# NUMPY — CONCEPTUAL AND PRACTICAL GUIDE FOR DA / DS / ML
# ============================================================
#
# NumPy = Numerical Python.
#
# NumPy is one of the fundamental libraries of the Python
# Data Science ecosystem.
#
# Its central object is:
#
#                         ndarray
#                           |
#             multidimensional numerical array
#                           |
#          +----------------+----------------+
#          |                |                |
#       shape              dtype           dimensions
#          |                |                |
#      (rows, cols)       data type          axes
#
#
# The most important mental shift is:
#
# Python list:
#
#     [10, 20, 30, 40]
#
# NumPy array:
#
#     np.array([10, 20, 30, 40])
#
# A Python list is primarily a general-purpose container.
#
# A NumPy ndarray is designed for numerical computation.
#
# Therefore NumPy gives you:
#
#     ARRAY
#       ↓
#     numerical operations
#       ↓
#     vectorization
#       ↓
#     statistics
#       ↓
#     linear algebra
#       ↓
#     transformations
#       ↓
#     scientific / DA / DS / ML computation
#
# ============================================================
# 1. WHY NUMPY MATTERS FOR DA / DS / ML
# ============================================================
#
# NumPy is not just "another Python library".
#
# A large part of the Python scientific ecosystem is built around
# the numerical-array model introduced by NumPy.
#
# Conceptually:
#
#                  NumPy ndarray
#                       |
#        +--------------+--------------+
#        |              |              |
#      pandas          SciPy       scikit-learn
#        |              |              |
#      tables        statistics       ML
#      data          science         models
#
#
# You will repeatedly encounter NumPy arrays when working with:
#
#     pandas
#     SciPy
#     scikit-learn
#     matplotlib
#     scientific computing
#     image processing
#     signal processing
#     deep learning ecosystems
#
# ============================================================
# 2. IMPORTING NUMPY
# ============================================================
#
# The standard convention is:
#
# import numpy as np
#
# Then:
#
# np.array(...)
# np.mean(...)
# np.arange(...)
#
# `np` is simply an alias.
#
# It does NOT create a different library.
#
#     import numpy as np
#
# means:
#
#     numpy
#       ↓
#     accessible through the shorter name
#       ↓
#     np
#
# ============================================================
# 3. THE CENTRAL CONCEPT: ndarray
# ============================================================
#
# The most important NumPy object is:
#
#     numpy.ndarray
#
# Usually you create one with:
#
#     np.array(...)
#
# Example:
#
# import numpy as np
#
# x = np.array([10, 20, 30, 40])
#
# Conceptually:
#
# Python data
#     ↓
# [10, 20, 30, 40]
#     ↓
# np.array()
#     ↓
# NumPy ndarray
#     ↓
# numerical operations
#
# ============================================================
# 4. WHY ARRAY INSTEAD OF LIST?
# ============================================================
#
# Consider:
#
# numbers = [10, 20, 30]
#
# With ordinary Python lists:
#
# numbers * 2
#
# means:
#
# [10, 20, 30, 10, 20, 30]
#
# It repeats the list.
#
# But:
#
# numbers = np.array([10, 20, 30])
#
# numbers * 2
#
# produces:
#
# [20, 40, 60]
#
# Why?
#
# NumPy interprets arithmetic operations element-by-element.
#
#     [10, 20, 30]
#          × 2
#            ↓
#     [20, 40, 60]
#
# This is one of the foundations of VECTORISATION.
#
# ============================================================
# 5. VECTORISATION — THE CORE NUMPY IDEA
# ============================================================
#
# Instead of writing:
#
# result = []
#
# for value in values:
#     result.append(value * 2)
#
# NumPy allows:
#
# result = values * 2
#
# Conceptually:
#
#             values
#                |
#      +---------+---------+
#      |         |         |
#     10        20        30
#      |         |         |
#     ×2        ×2        ×2
#      |         |         |
#     20        40        60
#      |         |         |
#      +---------+---------+
#                |
#             result
#
# This is called vectorized computation.
#
# The important idea is not simply "shorter code".
#
# The important idea is:
#
#     describe the operation on the array
#     instead of manually controlling every element.
#
# ============================================================
# 6. CREATING ARRAYS — np.array()
# ============================================================
#
# np.array(object)
#
# This converts suitable Python sequences or other array-like
# objects into an ndarray.
#
# Example:
#
# x = np.array([10, 20, 30])
#
# 1-dimensional:
#
# [10, 20, 30]
#
# Conceptually:
#
# shape = (3,)
#
# There are 3 elements along one axis.
#
# ============================================================
# 7. 2-D ARRAYS
# ============================================================
#
# x = np.array([
#     [10, 20, 30],
#     [40, 50, 60]
# ])
#
# Visual:
#
#             columns
#           0    1    2
#        +----+----+----+
# row 0  | 10 | 20 | 30 |
#        +----+----+----+
# row 1  | 40 | 50 | 60 |
#        +----+----+----+
#
# shape:
#
# (2, 3)
#
# Meaning:
#
#     2 rows
#     3 columns
#
# ============================================================
# 8. 3-D ARRAYS
# ============================================================
#
# A 3-D array can be visualized as multiple 2-D matrices:
#
#              axis 0
#               ↓
#
#        +---------------+
#        | 2-D matrix 0  |
#        +---------------+
#        | 2-D matrix 1  |
#        +---------------+
#        | 2-D matrix 2  |
#        +---------------+
#
# Example shape:
#
# (3, 2, 4)
#
# means:
#
#     3 blocks
#     2 rows per block
#     4 columns per row
#
# ============================================================
# 9. DIMENSION, AXIS, SHAPE — DO NOT CONFUSE THEM
# ============================================================
#
# Suppose:
#
# x = np.array([
#     [10, 20, 30],
#     [40, 50, 60]
# ])
#
# shape:
#
# (2, 3)
#
# number of dimensions:
#
# x.ndim
#
# → 2
#
# number of elements:
#
# x.size
#
# → 6
#
# data type:
#
# x.dtype
#
# → integer dtype
#
# The mental model:
#
#     shape
#       ↓
#     "How is the data arranged?"
#
#     ndim
#       ↓
#     "How many axes are there?"
#
#     size
#       ↓
#     "How many total elements?"
#
#     dtype
#       ↓
#     "What kind of numerical values are stored?"
#
# ============================================================
# 10. ndarray ATTRIBUTES YOU MUST KNOW
# ============================================================
#
# For:
#
# x = np.array([
#     [10, 20, 30],
#     [40, 50, 60]
# ])
#
# Important attributes:
#
# x.ndim
#     → number of dimensions / axes
#
# x.shape
#     → size along each axis
#
# x.size
#     → total number of elements
#
# x.dtype
#     → element data type
#
# x.itemsize
#     → bytes used by each element
#
# x.nbytes
#     → total bytes occupied by array data
#
# x.T
#     → transpose view
#
# ============================================================
# 11. SHAPE IS ONE OF THE MOST IMPORTANT NUMPY CONCEPTS
# ============================================================
#
# Consider:
#
# x = np.array([10, 20, 30])
#
# shape:
#
# (3,)
#
# This is NOT:
#
# (1, 3)
#
# and NOT:
#
# (3, 1)
#
# These are different array shapes.
#
# 1-D:
#
# [10, 20, 30]
#
# shape = (3,)
#
# Row vector:
#
# [[10, 20, 30]]
#
# shape = (1, 3)
#
# Column vector:
#
# [[10],
#  [20],
#  [30]]
#
# shape = (3, 1)
#
# This distinction becomes extremely important in ML.
#
# ============================================================
# 12. AXES — THE CONCEPT THAT CONFUSES BEGINNERS
# ============================================================
#
# For:
#
# x = np.array([
#     [10, 20, 30],
#     [40, 50, 60]
# ])
#
# shape = (2, 3)
#
# axis 0:
#
# moves DOWN the rows.
#
# axis 1:
#
# moves ACROSS the columns.
#
# Visual:
#
#              axis 1 →
#
#             10   20   30
#             40   50   60
#              ↓
#            axis 0
#
# Another useful mental model:
#
# axis = the direction along which NumPy performs an operation.
#
# ============================================================
# 13. np.zeros()
# ============================================================
#
# np.zeros(shape)
#
# creates an array filled with zeros.
#
# Example:
#
# x = np.zeros(5)
#
# result:
#
# [0. 0. 0. 0. 0.]
#
# 2-D:
#
# x = np.zeros((2, 3))
#
# result:
#
# [[0. 0. 0.]
#  [0. 0. 0.]]
#
# Why useful?
#
#     initialize storage
#     ↓
#     later fill/update values
#
# Common in:
#
#     numerical algorithms
#     matrix operations
#     ML initialization
#     simulations
#     image arrays
#
# ============================================================
# 14. np.ones()
# ============================================================
#
# np.ones(shape)
#
# creates an array filled with ones.
#
# Example:
#
# np.ones((2, 3))
#
# →
#
# [[1. 1. 1.]
#  [1. 1. 1.]]
#
# ============================================================
# 15. np.full()
# ============================================================
#
# np.full(shape, fill_value)
#
# creates an array where every element has the same value.
#
# Example:
#
# np.full((2, 3), 7)
#
# →
#
# [[7 7 7]
#  [7 7 7]]
#
# Mental model:
#
# shape + value
#      ↓
# repeated value
#
# ============================================================
# 16. np.empty()
# ============================================================
#
# np.empty(shape)
#
# allocates an array without initializing its elements to a
# meaningful value.
#
# Important:
#
# np.empty()
#
# does NOT mean:
#
# "fill with zeros."
#
# It means:
#
# "allocate the memory; the existing memory contents are not
# initialized."
#
# Therefore:
#
#     np.empty(...)
#
# should only be used when you are going to overwrite all elements
# before reading them.
#
# ============================================================
# 17. np.zeros_like(), np.ones_like(), np.full_like()
# ============================================================
#
# These functions create arrays using another array's shape and,
# depending on the function, its dtype.
#
# Example:
#
# x = np.array([
#     [10, 20],
#     [30, 40]
# ])
#
# np.zeros_like(x)
#
# →
#
# [[0, 0],
#  [0, 0]]
#
# Mental model:
#
# existing array
#       ↓
# copy its structural characteristics
#       ↓
# create a new initialized array
#
# Very useful when building numerical algorithms.
#
# ============================================================
# 18. np.arange()
# ============================================================
#
# np.arange(start, stop, step)
#
# creates regularly spaced values.
#
# Example:
#
# np.arange(0, 10, 2)
#
# →
#
# [0 2 4 6 8]
#
# IMPORTANT:
#
# stop is normally EXCLUDED.
#
# Think:
#
# start
#   ↓
# 0 → 2 → 4 → 6 → 8
#                 ↓
#                stop=10
#                 ↓
#                stop
#
# ============================================================
# 19. np.arange() VS range()
# ============================================================
#
# Python:
#
# range(0, 10, 2)
#
# produces a range object.
#
# NumPy:
#
# np.arange(0, 10, 2)
#
# produces a NumPy ndarray.
#
# This matters because NumPy arrays participate directly in
# vectorized numerical operations.
#
# ============================================================
# 20. FLOATING-POINT WARNING WITH ARANGE()
# ============================================================
#
# Be careful with:
#
# np.arange(0, 1, 0.1)
#
# when floating-point boundaries matter.
#
# Floating-point representation is not exact for many decimal
# fractions.
#
# For a specified number of evenly spaced points, prefer:
#
# np.linspace()
#
# when appropriate.
#
# ============================================================
# 21. np.linspace()
# ============================================================
#
# np.linspace(start, stop, num)
#
# creates `num` evenly spaced values between start and stop.
#
# Unlike arange(), the number of points is specified.
#
# Example:
#
# np.linspace(0, 1, 5)
#
# →
#
# [0.   0.25 0.5  0.75 1.  ]
#
# Visual:
#
# 0 ──── 0.25 ──── 0.5 ──── 0.75 ──── 1
#
# Useful for:
#
#     plotting
#     simulations
#     numerical experiments
#     generating continuous ranges
#
# ============================================================
# 22. ARANGE VS LINSPACE
# ============================================================
#
# arange:
#
# "I want values separated by this STEP."
#
# linspace:
#
# "I want this NUMBER OF evenly spaced values."
#
# Example:
#
# np.arange(0, 10, 2)
#
# → step = 2
#
# np.linspace(0, 10, 6)
#
# → 6 points
#
# Mental model:
#
#     arange  → spacing-oriented
#     linspace → point-count-oriented
#
# ============================================================
# 23. np.eye()
# ============================================================
#
# np.eye(n)
#
# creates an identity matrix.
#
# Example:
#
# np.eye(3)
#
# →
#
# [[1. 0. 0.]
#  [0. 1. 0.]
#  [0. 0. 1.]]
#
# Identity matrix:
#
#             1 0 0
#             0 1 0
#             0 0 1
#
# Important in:
#
#     linear algebra
#     ML
#     regularization
#     transformations
#     numerical algorithms
#
# ============================================================
# 24. np.diag()
# ============================================================
#
# np.diag()
#
# can:
#
#     extract a diagonal
#
# or:
#
#     construct a diagonal matrix
#
# Example:
#
# x = np.array([
#     [1, 2],
#     [3, 4]
# ])
#
# np.diag(x)
#
# →
#
# [1 4]
#
# Conversely:
#
# np.diag([1, 2, 3])
#
# →
#
# [[1 0 0]
#  [0 2 0]
#  [0 0 3]]
#
# ============================================================
# 25. RANDOM NUMBERS — np.random
# ============================================================
#
# NumPy provides random-number generation through:
#
#     np.random
#
# Modern NumPy recommends creating a random Generator:
#
#     rng = np.random.default_rng()
#
# Then:
#
#     rng.random(...)
#     rng.integers(...)
#     rng.normal(...)
#
# Mental model:
#
# random generator
#       ↓
# probability distribution
#       ↓
# generated sample
#
# ============================================================
# 26. np.random.default_rng()
# ============================================================
#
# rng = np.random.default_rng(42)
#
# creates a reproducible random-number generator.
#
# The number:
#
# 42
#
# is a seed.
#
# Same seed + same operations
#       ↓
# same pseudorandom sequence
#
# This is extremely important for:
#
#     experiments
#     debugging
#     ML reproducibility
#     simulations
#
# ============================================================
# 27. rng.random()
# ============================================================
#
# rng.random(5)
#
# generates 5 random floating-point values approximately in:
#
# [0, 1)
#
# Conceptually:
#
# 0 ------------------------ 1
# |                          |
# possible generated values
#
# ============================================================
# 28. rng.integers()
# ============================================================
#
# rng.integers(low, high, size)
#
# generates random integers.
#
# Example:
#
# rng = np.random.default_rng(42)
#
# values = rng.integers(1, 7, size=10)
#
# This simulates 10 dice rolls:
#
# possible values:
#
# 1, 2, 3, 4, 5, 6
#
# IMPORTANT:
#
# `high` is excluded.
#
# Therefore:
#
# rng.integers(1, 7)
#
# can produce 1 through 6.
#
# ============================================================
# 29. RANDOM NORMAL DISTRIBUTION
# ============================================================
#
# rng.normal(loc, scale, size)
#
# generates values from a normal distribution.
#
#     loc
#       → mean
#
#     scale
#       → standard deviation
#
# Example:
#
# rng.normal(loc=100, scale=15, size=1000)
#
# conceptually:
#
#                 /\ 
#                /  \
#               /    \
#              /      \
#             /        \
#        -----100---------
#
# Useful for:
#
#     simulation
#     synthetic data
#     statistical experiments
#     ML testing
#
# ============================================================
# 30. OTHER IMPORTANT RANDOM DISTRIBUTIONS
# ============================================================
#
# NumPy's Generator provides distributions such as:
#
#     rng.uniform()
#     rng.normal()
#     rng.binomial()
#     rng.poisson()
#     rng.exponential()
#     rng.lognormal()
#     rng.choice()
#
# These are important for simulation and probabilistic modeling.
#
# ============================================================
# 31. rng.choice()
# ============================================================
#
# rng.choice(array, size=..., replace=...)
#
# randomly selects elements.
#
# Example:
#
# data = np.array([10, 20, 30, 40])
#
# rng.choice(data, size=3)
#
# selects 3 values.
#
# With:
#
# replace=True
#
# an element can be selected more than once.
#
# With:
#
# replace=False
#
# an element cannot be selected again.
#
# This distinction is fundamental to sampling.
#
# ============================================================
# 32. INDEXING — THE FOUNDATION OF DATA ACCESS
# ============================================================
#
# Given:
#
# x = np.array([10, 20, 30, 40, 50])
#
# Python indexing starts at zero:
#
# index:     0   1   2   3   4
#            ↓   ↓   ↓   ↓   ↓
# value:    10  20  30  40  50
#
# x[0]
#   → 10
#
# x[2]
#   → 30
#
# x[-1]
#   → 50
#
# ============================================================
# 33. SLICING
# ============================================================
#
# x[start:stop:step]
#
# Example:
#
# x = np.array([10, 20, 30, 40, 50])
#
# x[1:4]
#
# →
#
# [20, 30, 40]
#
# Again:
#
# start included
# stop excluded
#
# Visual:
#
# index:
#     0   1   2   3   4
#         └───────┘
#          selected
#
# ============================================================
# 34. 2-D INDEXING
# ============================================================
#
# x = np.array([
#     [10, 20, 30],
#     [40, 50, 60]
# ])
#
# x[row, column]
#
# x[0, 1]
#
# →
# 20
#
# x[1, 2]
#
# →
# 60
#
# Visual:
#
#          col 0  col 1  col 2
# row 0      10     20     30
# row 1      40     50     60
#                        ↑
#                     x[1,2]
#
# ============================================================
# 35. ROW AND COLUMN SELECTION
# ============================================================
#
# x[0, :]
#
# → first row
#
# x[:, 1]
#
# → second column
#
# Visual:
#
# x:
#
#     10 20 30
#     40 50 60
#
# x[0, :]
#
#     10 20 30
#     --------
#
# x[:, 1]
#
#       20
#       50
#
# The colon means:
#
# "take all values along this axis."
#
# ============================================================
# 36. BOOLEAN MASKING — ONE OF THE MOST IMPORTANT NUMPY SKILLS
# ============================================================
#
# Suppose:
#
# x = np.array([10, 25, 40, 15, 60])
#
# condition:
#
# x > 20
#
# produces:
#
# [False, True, True, False, True]
#
# This is a BOOLEAN MASK.
#
# Then:
#
# x[x > 20]
#
# gives:
#
# [25, 40, 60]
#
# Conceptual flow:
#
# data
#   ↓
# condition
#   ↓
# True / False mask
#   ↓
# select True positions
#   ↓
# filtered data
#
# This is fundamental in DA and DS.
#
# ============================================================
# 37. COMBINING BOOLEAN CONDITIONS
# ============================================================
#
# For NumPy arrays, use:
#
# &   → elementwise AND
# |   → elementwise OR
# ~   → elementwise NOT
#
# Example:
#
# x = np.array([10, 25, 40, 15, 60])
#
# mask = (x > 20) & (x < 50)
#
# result:
#
# [25, 40]
#
# IMPORTANT:
#
# Use parentheses around each comparison.
#
# Correct:
#
# (x > 20) & (x < 50)
#
# Do not use Python's scalar:
#
# and
#
# for elementwise NumPy comparisons.
#
# ============================================================
# 38. np.where()
# ============================================================
#
# np.where(condition, value_if_true, value_if_false)
#
# is a vectorized conditional.
#
# Example:
#
# x = np.array([10, 20, 30])
#
# np.where(x >= 20, "high", "low")
#
# →
#
# ["low", "high", "high"]
#
# Mental model:
#
# condition
#     ↓
# +---+---+
# |       |
# T       F
# ↓       ↓
# true    false
# value   value
#
# This is extremely useful for feature creation and data cleaning.
#
# ============================================================
# 39. np.select()
# ============================================================
#
# np.select()
#
# handles multiple conditions.
#
# Conceptually:
#
# condition 1 → result 1
# condition 2 → result 2
# condition 3 → result 3
# otherwise   → default
#
# Useful when creating categories from numerical variables.
#
# Example concept:
#
# score < 50
#     → "Fail"
#
# score < 75
#     → "Pass"
#
# otherwise
#     → "Excellent"
#
# ============================================================
# 40. BASIC ARITHMETIC
# ============================================================
#
# Given:
#
# x = np.array([10, 20, 30])
# y = np.array([1, 2, 3])
#
# x + y
#     → [11, 22, 33]
#
# x - y
#     → [9, 18, 27]
#
# x * y
#     → [10, 40, 90]
#
# x / y
#     → [10., 10., 10.]
#
# x ** 2
#     → [100, 400, 900]
#
# x % y
#     → elementwise remainder
#
# ============================================================
# 41. ELEMENTWISE OPERATION VS MATRIX MULTIPLICATION
# ============================================================
#
# THIS DISTINCTION IS CRITICAL.
#
# For arrays:
#
# x * y
#
# means:
#
# element-by-element multiplication.
#
# Matrix multiplication uses:
#
# x @ y
#
# or:
#
# np.matmul(x, y)
#
# Example:
#
# A = np.array([
#     [1, 2],
#     [3, 4]
# ])
#
# B = np.array([
#     [5, 6],
#     [7, 8]
# ])
#
# A * B
#
# →
#
# [[5, 12],
#  [21, 32]]
#
# But:
#
# A @ B
#
# →
#
# [[19, 22],
#  [43, 50]]
#
# Because:
#
# first output element:
#
# 1×5 + 2×7
# = 5 + 14
# = 19
#
# ============================================================
# 42. np.add(), np.subtract(), np.multiply()
# ============================================================
#
# NumPy also provides explicit universal functions:
#
# np.add(x, y)
# np.subtract(x, y)
# np.multiply(x, y)
# np.divide(x, y)
# np.power(x, y)
#
# These correspond conceptually to:
#
# x + y
# x - y
# x * y
# x / y
# x ** y
#
# Usually the operators are more readable.
#
# The named functions become useful when working with NumPy's
# universal-function machinery.
#
# ============================================================
# 43. UNIVERSAL FUNCTIONS — UFUNCS
# ============================================================
#
# A ufunc performs elementwise operations on arrays.
#
# Example:
#
# x = np.array([1, 4, 9])
#
# np.sqrt(x)
#
# →
#
# [1., 2., 3.]
#
# Conceptually:
#
# [1, 4, 9]
#    ↓
# sqrt each element
#    ↓
# [1, 2, 3]
#
# Important ufuncs include:
#
#     np.sqrt()
#     np.exp()
#     np.log()
#     np.log10()
#     np.abs()
#     np.round()
#     np.floor()
#     np.ceil()
#     np.sin()
#     np.cos()
#     np.tan()
#
# ============================================================
# 44. np.abs()
# ============================================================
#
# np.abs(x)
#
# returns absolute values.
#
# [-5, -2, 3]
#      ↓
# [ 5,  2, 3]
#
# Very useful for:
#
#     errors
#     distances
#     deviations
#     residual analysis
#
# ============================================================
# 45. EXPONENTIAL AND LOGARITHMIC FUNCTIONS
# ============================================================
#
# np.exp(x)
#
# calculates eˣ.
#
# np.log(x)
#
# natural logarithm.
#
# np.log10(x)
#
# base-10 logarithm.
#
# np.log2(x)
#
# base-2 logarithm.
#
# These are heavily used in:
#
#     probability
#     ML
#     optimization
#     transformations
#     scientific computing
#
# ============================================================
# 46. WHY LOG TRANSFORMS MATTER IN DA / DS
# ============================================================
#
# Suppose:
#
# income:
#
# [1000, 2000, 5000, 10000, 100000]
#
# Values may be heavily right-skewed.
#
# log transformation:
#
# x
# ↓
# log(x)
# ↓
# compressed scale
#
# This can make distributions easier to inspect and can sometimes
# improve modeling behavior.
#
# Important:
#
# log transformation is a modeling/analysis choice, not something
# to apply automatically.
#
# ============================================================
# 47. ROUNDING FUNCTIONS
# ============================================================
#
# np.round(x)
# np.floor(x)
# np.ceil(x)
# np.trunc(x)
#
# Example:
#
# x = np.array([1.2, 1.8, -1.2, -1.8])
#
# floor:
#
# [-?]
#
# Conceptually:
#
# floor → greatest integer <= x
#
# ceil:
#
#     smallest integer >= x
#
# trunc:
#
#     remove fractional part toward zero
#
# round:
#
#     round to the requested precision
#
# ============================================================
# 48. COMPARISON FUNCTIONS
# ============================================================
#
# NumPy comparisons are vectorized:
#
# x > 10
# x >= 10
# x < 10
# x <= 10
# x == 10
# x != 10
#
# Example:
#
# x = np.array([5, 10, 15])
#
# x > 10
#
# →
#
# [False, False, True]
#
# This produces a Boolean array.
#
# ============================================================
# 49. np.logical_and(), np.logical_or(), np.logical_not()
# ============================================================
#
# These explicitly perform logical operations elementwise.
#
# np.logical_and(a, b)
# np.logical_or(a, b)
# np.logical_not(a)
#
# They are alternatives to:
#
# &
# |
# ~
#
# when explicit function syntax is preferable.
#
# ============================================================
# 50. AGGREGATION — TURN MANY VALUES INTO ONE
# ============================================================
#
# Aggregation is a major NumPy concept.
#
# Examples:
#
# np.sum()
# np.mean()
# np.median()
# np.min()
# np.max()
# np.std()
# np.var()
#
# Conceptually:
#
# MANY VALUES
#     ↓
#   aggregate
#     ↓
# ONE SUMMARY VALUE
#
# ============================================================
# 51. np.sum()
# ============================================================
#
# np.sum(x)
#
# adds values.
#
# Example:
#
# x = np.array([10, 20, 30])
#
# np.sum(x)
#
# → 60
#
# 2-D:
#
# x = np.array([
#     [1, 2, 3],
#     [4, 5, 6]
# ])
#
# np.sum(x)
#
# → 21
#
# ============================================================
# 52. AXIS WITH SUM
# ============================================================
#
# x =
#
# [[1, 2, 3],
#  [4, 5, 6]]
#
# np.sum(x, axis=0)
#
# → [5, 7, 9]
#
# Why?
#
# column-wise:
#
# [1 + 4, 2 + 5, 3 + 6]
#
# = [5, 7, 9]
#
# np.sum(x, axis=1)
#
# → [6, 15]
#
# row-wise:
#
# [1 + 2 + 3,
#  4 + 5 + 6]
#
# = [6, 15]
#
# IMPORTANT MENTAL MODEL:
#
# axis=0
#     ↓
# collapse rows
#     ↓
# result for each column
#
# axis=1
#     ↓
# collapse columns
#     ↓
# result for each row
#
# ============================================================
# 53. np.mean()
# ============================================================
#
# np.mean(x)
#
# calculates arithmetic mean.
#
# x = [10, 20, 30]
#
# np.mean(x)
#
# → 20
#
# Along an axis:
#
# x =
#
# [[1, 2, 3],
#  [4, 5, 6]]
#
# np.mean(x, axis=0)
#
# → [2.5, 3.5, 4.5]
#
# np.mean(x, axis=1)
#
# → [2., 5.]
#
# This is extremely common in ML preprocessing.
#
# ============================================================
# 54. np.median()
# ============================================================
#
# np.median(x)
#
# calculates the median.
#
# It is useful when the data may contain extreme values.
#
# Example:
#
# x = np.array([10, 10, 10, 10, 1000])
#
# np.mean(x)
#     → strongly affected by 1000
#
# np.median(x)
#     → 10
#
# In practical DA:
#
# mean and median should often be examined together.
#
# ============================================================
# 55. np.min() AND np.max()
# ============================================================
#
# np.min(x)
#     → smallest value
#
# np.max(x)
#     → largest value
#
# Along axes:
#
# np.min(x, axis=0)
# np.max(x, axis=0)
#
# These are basic descriptive statistics.
#
# ============================================================
# 56. np.argmin() AND np.argmax()
# ============================================================
#
# IMPORTANT DISTINCTION:
#
# min()
#     → gives the minimum VALUE
#
# argmin()
#     → gives the POSITION of the minimum
#
# Example:
#
# x = np.array([50, 10, 80, 20])
#
# np.min(x)
#     → 10
#
# np.argmin(x)
#     → 1
#
# Because:
#
# index:  0   1   2   3
# value: 50  10  80  20
#             ↑
#            min
#
# Similarly:
#
# np.argmax(x)
#     → 2
#
# This is extremely useful when finding the location of the best
# or worst observation.
#
# ============================================================
# 57. np.ptp()
# ============================================================
#
# ptp = peak-to-peak range.
#
# np.ptp(x)
#
# = maximum - minimum
#
# Example:
#
# x = [10, 20, 50]
#
# ptp = 50 - 10
#     = 40
#
# It measures the total range of observed values.
#
# ============================================================
# 58. np.std()
# ============================================================
#
# np.std(x)
#
# calculates standard deviation.
#
# IMPORTANT:
#
# NumPy's default standard deviation uses:
#
# ddof = 0
#
# This corresponds to population-style normalization.
#
# Example:
#
# np.std(x)
#
# is conceptually based on:
#
# sqrt(
#     Σ(x - mean)² / N
# )
#
# If you want sample standard deviation:
#
# np.std(x, ddof=1)
#
# This distinction is very important.
#
# ============================================================
# 59. np.var()
# ============================================================
#
# np.var(x)
#
# calculates variance.
#
# Default:
#
# ddof=0
#
# Therefore:
#
# np.var(x)
#
# uses:
#
#     denominator = N
#
# To use sample variance:
#
# np.var(x, ddof=1)
#
# uses:
#
#     denominator = N - 1
#
# This is an important connection between NumPy and the Python
# statistics module.
#
# ============================================================
# 60. NUMPY STATISTICS VS PYTHON STATISTICS
# ============================================================
#
# Python statistics:
#
# statistics.variance(x)
#
# → sample variance by default.
#
# NumPy:
#
# np.var(x)
#
# → population-style denominator N by default.
#
# Therefore:
#
# DO NOT blindly assume:
#
#     np.var(x)
#
# and:
#
#     statistics.variance(x)
#
# are identical.
#
# To match sample variance:
#
#     np.var(x, ddof=1)
#
# This is one of the most important details for statistical
# correctness.
#
# ============================================================
# 61. np.percentile()
# ============================================================
#
# np.percentile(x, q)
#
# finds percentile values.
#
# Example:
#
# np.percentile(x, 50)
#
# → median / 50th percentile
#
# np.percentile(x, 25)
#
# → first quartile
#
# np.percentile(x, 75)
#
# → third quartile
#
# Concept:
#
# 0% ───────── 25% ─────── 50% ─────── 75% ─────── 100%
#              Q1          Q2            Q3
#
# ============================================================
# 62. np.quantile()
# ============================================================
#
# np.quantile(x, q)
#
# is closely related to percentile.
#
# The difference is the scale:
#
# percentile:
#
# q = 0 ... 100
#
# quantile:
#
# q = 0 ... 1
#
# Therefore:
#
# np.percentile(x, 25)
#
# corresponds conceptually to:
#
# np.quantile(x, 0.25)
#
# ============================================================
# 63. np.average()
# ============================================================
#
# np.average(x)
#
# calculates an average.
#
# It can also accept weights:
#
# np.average(
#     values,
#     weights=weights
# )
#
# Conceptually:
#
# weighted average =
#
# Σ(value × weight)
# ──────────────────
#       Σ weight
#
# Useful for:
#
#     weighted KPIs
#     weighted measurements
#     survey data
#     business analytics
#
# ============================================================
# 64. np.cumsum()
# ============================================================
#
# cumsum = cumulative sum.
#
# x = [10, 20, 30]
#
# np.cumsum(x)
#
# →
#
# [10, 30, 60]
#
# Execution:
#
# 10
# ↓
# 10
#
# 10 + 20
# ↓
# 30
#
# 10 + 20 + 30
# ↓
# 60
#
# Useful for:
#
#     cumulative sales
#     cumulative revenue
#     running totals
#     time-series analysis
#
# ============================================================
# 65. np.cumprod()
# ============================================================
#
# cumulative product.
#
# x = [2, 3, 4]
#
# np.cumprod(x)
#
# →
#
# [2, 6, 24]
#
# Useful for:
#
#     compounded growth
#     multiplicative processes
#     simulations
#
# ============================================================
# 66. np.diff()
# ============================================================
#
# np.diff(x)
#
# calculates differences between consecutive values.
#
# x = [10, 15, 21, 30]
#
# np.diff(x)
#
# →
#
# [5, 6, 9]
#
# because:
#
# 15 - 10 = 5
# 21 - 15 = 6
# 30 - 21 = 9
#
# Very useful in:
#
#     time series
#     changes
#     growth analysis
#     numerical methods
#
# ============================================================
# 67. np.gradient()
# ============================================================
#
# np.gradient()
#
# estimates numerical derivatives / gradients.
#
# It is useful when values represent a sampled function or signal.
#
# Conceptually:
#
# values
#   ↓
# change between neighboring points
#   ↓
# approximate rate of change
#
# This is more scientific-computing oriented than ordinary DA,
# but very useful in numerical analysis and ML mathematics.
#
# ============================================================
# 68. np.cov()
# ============================================================
#
# np.cov()
#
# calculates covariance.
#
# It is useful for understanding how variables vary together.
#
# Example concept:
#
# X
# ↓
# deviations from mean
#
# Y
# ↓
# deviations from mean
#
# corresponding deviations
#        ↓
#      multiply
#        ↓
# covariance
#
# IMPORTANT:
#
# NumPy's default normalization is sample covariance
# (ddof=1).
#
# This is another place where understanding defaults matters.
#
# ============================================================
# 69. np.corrcoef()
# ============================================================
#
# np.corrcoef()
#
# calculates a correlation coefficient matrix.
#
# Suppose:
#
# x = [1, 2, 3]
# y = [2, 4, 6]
#
# np.corrcoef([x, y])
#
# conceptually produces:
#
#             x       y
#       x     1       1
#       y     1       1
#
# The diagonal is always:
#
# 1
#
# because a variable is perfectly correlated with itself.
#
# ============================================================
# 70. CORRELATION MATRIX
# ============================================================
#
# Suppose you have:
#
#     age
#     income
#     spending
#
# A correlation matrix might conceptually look like:
#
#             age   income   spending
#
# age          1     0.4       0.2
# income      0.4     1        0.8
# spending    0.2    0.8        1
#
# This is useful in:
#
#     exploratory data analysis
#     feature analysis
#     multicollinearity investigation
#     ML feature selection
#
# IMPORTANT:
#
# correlation does NOT establish causation.
#
# ============================================================
# 71. RESHAPING — np.reshape()
# ============================================================
#
# Reshaping changes the arrangement of an array without changing
# its total number of elements.
#
# Example:
#
# x = np.arange(6)
#
# x:
#
# [0 1 2 3 4 5]
#
# x.reshape(2, 3)
#
# →
#
# [[0 1 2]
#  [3 4 5]]
#
# Before:
#
# shape = (6,)
#
# After:
#
# shape = (2, 3)
#
# Total elements:
#
# 6
#
# still 6.
#
# ============================================================
# 72. RESHAPE RULE
# ============================================================
#
# The total number of elements must remain unchanged.
#
# 6 elements can become:
#
# (1, 6)
# (2, 3)
# (3, 2)
# (6, 1)
#
# but not:
#
# (2, 4)
#
# because:
#
# 2 × 4 = 8
#
# and the original has 6 elements.
#
# ============================================================
# 73. reshape(-1)
# ============================================================
#
# NumPy can infer one dimension:
#
# x.reshape(2, -1)
#
# If x contains 6 elements:
#
# 2 × ? = 6
#
# therefore:
#
# ? = 3
#
# result shape:
#
# (2, 3)
#
# Mental model:
#
#     -1
#      ↓
# "NumPy, calculate this dimension for me."
#
# Only one dimension can normally be inferred this way.
#
# ============================================================
# 74. ndarray.reshape()
# ============================================================
#
# Instead of:
#
# np.reshape(x, (2, 3))
#
# you can use:
#
# x.reshape(2, 3)
#
# This is an ndarray method.
#
# Important distinction:
#
# FUNCTION:
#
#     np.reshape(x, ...)
#
# METHOD:
#
#     x.reshape(...)
#
# Both express the same conceptual operation.
#
# ============================================================
# 75. flatten()
# ============================================================
#
# x.flatten()
#
# creates a flattened COPY.
#
# Example:
#
# x =
#
# [[1, 2],
#  [3, 4]]
#
# x.flatten()
#
# →
#
# [1, 2, 3, 4]
#
# Because it creates a copy, modifying the flattened result does
# not normally modify the original array.
#
# ============================================================
# 76. ravel()
# ============================================================
#
# np.ravel(x)
#
# or:
#
# x.ravel()
#
# flattens an array.
#
# Unlike flatten(), ravel() may return a VIEW when possible.
#
# Therefore:
#
# flatten()
#     → copy
#
# ravel()
#     → view when possible
#
# This distinction matters for memory and performance.
#
# ============================================================
# 77. VIEW VS COPY — CRITICAL NUMPY CONCEPT
# ============================================================
#
# NumPy can sometimes create another array object that refers to
# the SAME underlying data.
#
# This is called a VIEW.
#
# Conceptually:
#
# ORIGINAL ARRAY
#      |
#      +----------------+
#      |                |
#    original          view
#      |                |
#      +------same------+
#             data
#
# If the view modifies the underlying data, the original may also
# appear changed.
#
# A COPY owns separate data.
#
# ORIGINAL
#    ↓
# COPY
#    ↓
# independent data
#
# This is important for:
#
#     debugging
#     memory efficiency
#     data manipulation
#
# ============================================================
# 78. copy()
# ============================================================
#
# x.copy()
#
# explicitly creates a copy of the array data.
#
# Use it when you need an independent array.
#
# Example mental model:
#
# x
# ↓
# x.copy()
# ↓
# independent data
#
# ============================================================
# 79. TRANSPOSE — .T
# ============================================================
#
# For a 2-D array:
#
# x =
#
# [[1, 2, 3],
#  [4, 5, 6]]
#
# x.T
#
# →
#
# [[1, 4],
#  [2, 5],
#  [3, 6]]
#
# Shape changes:
#
# (2, 3)
#     ↓
# (3, 2)
#
# Conceptually:
#
# rows ↔ columns
#
# ============================================================
# 80. np.transpose()
# ============================================================
#
# np.transpose(x)
#
# performs transposition.
#
# For higher-dimensional arrays, axes can also be rearranged.
#
# This becomes important for:
#
#     images
#     tensors
#     deep learning
#     multidimensional scientific data
#
# ============================================================
# 81. np.swapaxes()
# ============================================================
#
# np.swapaxes(x, axis1, axis2)
#
# exchanges two axes.
#
# This is particularly useful with multidimensional data.
#
# Mental model:
#
# axis 0
#     ↕
# swap
#     ↕
# axis 2
#
# This is common when converting between different data layouts.
#
# ============================================================
# 82. np.moveaxis()
# ============================================================
#
# np.moveaxis(x, source, destination)
#
# moves one or more axes to new positions.
#
# This is especially useful for:
#
#     images
#     tensors
#     deep learning data
#
# Example conceptual layouts:
#
# channels × height × width
#
# versus:
#
# height × width × channels
#
# NumPy gives you tools to rearrange these dimensions.
#
# ============================================================
# 83. np.expand_dims()
# ============================================================
#
# np.expand_dims(x, axis)
#
# inserts a new axis.
#
# Example:
#
# x.shape
#     → (3,)
#
# x[:, np.newaxis]
#
# or:
#
# np.expand_dims(x, axis=1)
#
# gives:
#
# shape = (3, 1)
#
# Concept:
#
# (3,)
#  ↓
# add an axis
#  ↓
# (3, 1)
#
# This is extremely important for broadcasting and ML input shapes.
#
# ============================================================
# 84. np.squeeze()
# ============================================================
#
# np.squeeze(x)
#
# removes axes whose size is 1.
#
# Example:
#
# shape:
#
# (3, 1)
#
# squeeze
#     ↓
# (3,)
#
# Important:
#
# squeeze does NOT remove arbitrary dimensions.
#
# It removes dimensions with size exactly 1.
#
# ============================================================
# 85. np.newaxis
# ============================================================
#
# np.newaxis is an alias commonly used to insert a new axis.
#
# Example:
#
# x = np.array([1, 2, 3])
#
# x[:, np.newaxis]
#
# →
#
# [[1],
#  [2],
#  [3]]
#
# shape:
#
# (3, 1)
#
# This is one of the most useful small NumPy techniques.
#
# ============================================================
# 86. CONCATENATION
# ============================================================
#
# np.concatenate()
#
# joins arrays along an existing axis.
#
# Example:
#
# a = np.array([1, 2])
# b = np.array([3, 4])
#
# np.concatenate([a, b])
#
# →
#
# [1, 2, 3, 4]
#
# Concept:
#
# A + B
#  ↓
# joined array
#
# ============================================================
# 87. np.stack()
# ============================================================
#
# stack() is different from concatenate().
#
# concatenate:
#
# joins along an EXISTING axis.
#
# stack:
#
# creates a NEW axis.
#
# Suppose:
#
# a = [1, 2]
# b = [3, 4]
#
# np.concatenate([a, b])
#
# →
#
# [1, 2, 3, 4]
#
# np.stack([a, b])
#
# →
#
# [[1, 2],
#  [3, 4]]
#
# shape:
#
# (2, 2)
#
# This distinction is extremely important.
#
# ============================================================
# 88. np.vstack()
# ============================================================
#
# vstack = vertical stack.
#
# Example:
#
# a = [1, 2]
# b = [3, 4]
#
# np.vstack([a, b])
#
# →
#
# [[1, 2],
#  [3, 4]]
#
# Think:
#
# one row
#      ↓
# another row
#      ↓
# stack vertically
#
# ============================================================
# 89. np.hstack()
# ============================================================
#
# hstack = horizontal stack.
#
# For compatible arrays:
#
# a = [1, 2]
# b = [3, 4]
#
# np.hstack([a, b])
#
# →
#
# [1, 2, 3, 4]
#
# Think:
#
# [A] + [B]
#   → horizontal joining
#
# ============================================================
# 90. np.column_stack()
# ============================================================
#
# column_stack() is useful for turning 1-D arrays into columns.
#
# a = [1, 2, 3]
# b = [4, 5, 6]
#
# np.column_stack([a, b])
#
# →
#
# [[1, 4],
#  [2, 5],
#  [3, 6]]
#
# This is particularly useful when constructing feature matrices.
#
# ============================================================
# 91. np.split()
# ============================================================
#
# np.split(array, sections)
#
# divides an array into multiple pieces.
#
# Example:
#
# x = np.arange(6)
#
# np.split(x, 3)
#
# →
#
# [array([0, 1]),
#  array([2, 3]),
#  array([4, 5])]
#
# Useful for:
#
#     batching
#     partitioning
#     data preparation
#
# ============================================================
# 92. np.array_split()
# ============================================================
#
# array_split() is more flexible than split().
#
# It can split arrays into unequal sections when necessary.
#
# This is useful when the number of elements cannot be divided
# evenly.
#
# ============================================================
# 93. SORTING
# ============================================================
#
# np.sort(x)
#
# returns a sorted copy.
#
# Example:
#
# x = np.array([30, 10, 20])
#
# np.sort(x)
#
# →
#
# [10, 20, 30]
#
# ============================================================
# 94. ndarray.sort()
# ============================================================
#
# x.sort()
#
# sorts the array in place.
#
# This distinction is important:
#
# np.sort(x)
#     → returns sorted result
#
# x.sort()
#     → modifies x
#
# Mental model:
#
# FUNCTION:
#
#     original
#        ↓
#     sorted copy
#
# METHOD:
#
#     original
#        ↓
#     changed in place
#
# ============================================================
# 95. np.argsort()
# ============================================================
#
# np.argsort(x)
#
# returns the indices that would sort the array.
#
# Example:
#
# x = np.array([30, 10, 20])
#
# np.argsort(x)
#
# →
#
# [1, 2, 0]
#
# Why?
#
# x[1] = 10
# x[2] = 20
# x[0] = 30
#
# So:
#
# indices:
#
# 1 → 10
# 2 → 20
# 0 → 30
#
# This is extremely useful for ranking and retrieving associated
# information.
#
# ============================================================
# 96. np.unique()
# ============================================================
#
# np.unique(x)
#
# returns unique values.
#
# Example:
#
# x = [10, 20, 20, 30, 30, 30]
#
# np.unique(x)
#
# →
#
# [10, 20, 30]
#
# This is very useful in:
#
#     categorical analysis
#     data cleaning
#     exploratory analysis
#
# ============================================================
# 97. np.unique(return_counts=True)
# ============================================================
#
# np.unique(
#     x,
#     return_counts=True
# )
#
# can provide:
#
# unique values
# +
# number of occurrences
#
# Conceptually:
#
# value       count
# 10            1
# 20            2
# 30            3
#
# This is useful for quick frequency analysis.
#
# For larger tabular datasets, pandas.value_counts() is often
# more convenient.
#
# ============================================================
# 98. np.where() FOR FINDING POSITIONS
# ============================================================
#
# np.where(condition)
#
# returns positions where the condition is True.
#
# Example:
#
# x = np.array([10, 25, 40, 15])
#
# np.where(x > 20)
#
# identifies the indices:
#
# 1, 2
#
# because:
#
# x[1] = 25
# x[2] = 40
#
# Distinguish:
#
# x[x > 20]
#     → VALUES
#
# np.where(x > 20)
#     → POSITIONS
#
# ============================================================
# 99. np.nonzero()
# ============================================================
#
# np.nonzero(x)
#
# returns indices of non-zero elements.
#
# Useful for:
#
#     sparse-like data
#     masks
#     locating active elements
#     numerical algorithms
#
# ============================================================
# 100. np.count_nonzero()
# ============================================================
#
# np.count_nonzero(x)
#
# counts non-zero elements.
#
# It can also count along an axis.
#
# Useful for:
#
#     counting True conditions
#     binary data
#     missing-data masks
#     sparse patterns
#
# ============================================================
# 101. np.any()
# ============================================================
#
# np.any(condition)
#
# asks:
#
# "Is AT LEAST ONE element True?"
#
# Example:
#
# x = np.array([5, 10, 20])
#
# np.any(x > 15)
#
# → True
#
# because 20 > 15.
#
# ============================================================
# 102. np.all()
# ============================================================
#
# np.all(condition)
#
# asks:
#
# "Are ALL elements True?"
#
# Example:
#
# np.all(x > 0)
#
# → True
#
# because every value is positive.
#
# These are extremely useful for validation.
#
# ============================================================
# 103. np.isfinite()
# ============================================================
#
# np.isfinite(x)
#
# identifies finite values.
#
# Useful for detecting:
#
#     NaN
#     +inf
#     -inf
#
# Conceptual flow:
#
# raw numerical data
#        ↓
# isfinite()
#        ↓
# True = finite
# False = NaN or infinity
#
# ============================================================
# 104. np.isnan()
# ============================================================
#
# np.isnan(x)
#
# identifies NaN values.
#
# Example:
#
# x = np.array([1.0, np.nan, 3.0])
#
# np.isnan(x)
#
# →
#
# [False, True, False]
#
# This is important for data cleaning.
#
# ============================================================
# 105. np.isinf()
# ============================================================
#
# np.isinf(x)
#
# identifies positive and negative infinity.
#
# Useful when numerical operations may produce:
#
#     inf
#     -inf
#
# ============================================================
# 106. np.nanmean()
# ============================================================
#
# np.nanmean(x)
#
# calculates the mean while ignoring NaN values.
#
# Example:
#
# x = np.array([10.0, np.nan, 30.0])
#
# np.nanmean(x)
#
# → 20.0
#
# Compare:
#
# np.mean(x)
#
# may produce NaN.
#
# Therefore:
#
# ordinary function
#     → NaN participates
#
# nan-aware function
#     → NaN is ignored
#
# ============================================================
# 107. NAN-AWARE STATISTICAL FUNCTIONS
# ============================================================
#
# Important functions include:
#
# np.nanmean()
# np.nanmedian()
# np.nanstd()
# np.nanvar()
# np.nanmin()
# np.nanmax()
# np.nansum()
# np.nanprod()
# np.nanpercentile()
# np.nanquantile()
#
# These are useful when numerical arrays contain missing values
# represented as NaN.
#
# ============================================================
# 108. IMPORTANT CAUTION ABOUT NaN
# ============================================================
#
# NaN handling is NOT the same as "fixing missing data".
#
# Example:
#
# np.nanmean()
#
# merely ignores NaN during that calculation.
#
# It does NOT:
#
#     infer the missing value
#     impute the value
#     repair the dataset
#
# In DA/ML, missing-data handling is a modeling/data-quality
# decision.
#
# ============================================================
# 109. np.clip()
# ============================================================
#
# np.clip(x, min, max)
#
# forces values into a specified interval.
#
# Example:
#
# x = [10, 50, 120]
#
# np.clip(x, 0, 100)
#
# →
#
# [10, 50, 100]
#
# Concept:
#
# below minimum
#     ↓
# minimum
#
# inside range
#     ↓
# unchanged
#
# above maximum
#     ↓
# maximum
#
# Useful for:
#
#     data cleaning
#     bounded measurements
#     image values
#     preprocessing
#
# ============================================================
# 110. np.minimum() AND np.maximum()
# ============================================================
#
# These operate elementwise.
#
# np.minimum(a, b)
#
# chooses the smaller value at each position.
#
# np.maximum(a, b)
#
# chooses the larger value at each position.
#
# These are useful for vectorized conditional logic.
#
# ============================================================
# 111. BROADCASTING — ONE OF NUMPY'S MOST IMPORTANT CONCEPTS
# ============================================================
#
# Broadcasting allows NumPy to perform operations between arrays
# with compatible shapes.
#
# Example:
#
# x = np.array([10, 20, 30])
#
# x + 5
#
# conceptually:
#
# [10, 20, 30]
#       +
# [ 5,  5,  5]
#       ↓
# [15, 25, 35]
#
# NumPy does not necessarily physically create that repeated array.
#
# It applies the scalar across the array conceptually.
#
# ============================================================
# 112. BROADCASTING WITH 2-D ARRAYS
# ============================================================
#
# X =
#
# [[10, 20, 30],
#  [40, 50, 60]]
#
# Add:
#
# [1, 2, 3]
#
# NumPy interprets:
#
# [[10, 20, 30],      [[1, 2, 3],
#  [40, 50, 60]]   +   [1, 2, 3]]
#
# result:
#
# [[11, 22, 33],
#  [41, 52, 63]]
#
# The 1-D array is broadcast across the rows.
#
# ============================================================
# 113. BROADCASTING SHAPE RULE
# ============================================================
#
# Shapes are compared from the RIGHTMOST dimension.
#
# Two dimensions are compatible if:
#
#     they are equal
#
# OR:
#
#     one of them is 1
#
# OR:
#
#     one dimension is absent
#
# Example:
#
# (2, 3)
# (3,)
#
# compare from right:
#
#     3 = 3
#
# then:
#
#     2
#
# second shape has no corresponding dimension.
#
# Compatible.
#
# ============================================================
# 114. BROADCASTING VISUAL
# ============================================================
#
# Matrix:
#
# A =
#
# [ 10 20 30 ]
# [ 40 50 60 ]
#
# Vector:
#
# b =
#
# [ 1  2  3 ]
#
# Broadcast:
#
# [ 1 2 3 ]
# [ 1 2 3 ]
#
# Then:
#
# [10 20 30]   [1 2 3]   [11 22 33]
# [40 50 60] + [1 2 3] = [41 52 63]
#
# This is one of the core NumPy mental models.
#
# ============================================================
# 115. WHY BROADCASTING MATTERS IN ML
# ============================================================
#
# Many ML operations involve:
#
# matrix
#     ↓
# subtract feature means
#     ↓
# divide by feature standard deviations
#
# If:
#
# X.shape = (1000, 5)
#
# and:
#
# mean.shape = (5,)
#
# then:
#
# X - mean
#
# broadcasts the 5 feature means across all 1000 observations.
#
# Conceptually:
#
#              5 feature means
#                    ↓
#     +-------------------------------+
#     | repeated across observations |
#     +-------------------------------+
#                    ↓
#               X - mean
#
# This is fundamental to numerical ML.
#
# ============================================================
# 116. np.dot()
# ============================================================
#
# np.dot() performs dot products and, depending on dimensions,
# related multiplication operations.
#
# For 1-D vectors:
#
# a = [1, 2, 3]
# b = [4, 5, 6]
#
# dot product:
#
# 1×4 + 2×5 + 3×6
# = 4 + 10 + 18
# = 32
#
# np.dot(a, b)
#
# → 32
#
# ============================================================
# 117. np.matmul() AND @
# ============================================================
#
# Matrix multiplication:
#
# np.matmul(A, B)
#
# or:
#
# A @ B
#
# The @ operator is usually the clearest notation.
#
# Example:
#
# A @ B
#
# represents mathematical matrix multiplication.
#
# This is fundamental in:
#
#     linear algebra
#     regression
#     neural networks
#     transformations
#
# ============================================================
# 118. np.linalg — LINEAR ALGEBRA
# ============================================================
#
# NumPy provides linear algebra tools through:
#
#     np.linalg
#
# Important functions include:
#
#     np.linalg.inv()
#     np.linalg.solve()
#     np.linalg.det()
#     np.linalg.eig()
#     np.linalg.eigh()
#     np.linalg.svd()
#     np.linalg.norm()
#     np.linalg.matrix_rank()
#     np.linalg.lstsq()
#
# These are especially important for:
#
#     DS
#     ML
#     scientific computing
#
# ============================================================
# 119. np.linalg.norm()
# ============================================================
#
# A norm measures magnitude/length.
#
# For vector:
#
# x = [3, 4]
#
# Euclidean norm:
#
# √(3² + 4²)
# = √25
# = 5
#
# np.linalg.norm(x)
#
# → 5
#
# Visual:
#
#       (3,4)
#         *
#        /|
#       / |
#      /  | 4
#     /   |
#    *----*
#       3
#
# distance = 5
#
# Norms are fundamental in:
#
#     distance calculations
#     optimization
#     regularization
#     ML
#
# ============================================================
# 120. np.linalg.solve()
# ============================================================
#
# Solves a system:
#
# A x = b
#
# Instead of manually computing:
#
# x = inverse(A) × b
#
# prefer:
#
# np.linalg.solve(A, b)
#
# when solving a linear system.
#
# This is generally more appropriate numerically than explicitly
# calculating the inverse just to solve the system.
#
# ============================================================
# 121. np.linalg.inv()
# ============================================================
#
# Calculates a matrix inverse:
#
# A⁻¹
#
# such that:
#
# A × A⁻¹ = I
#
# where I is the identity matrix.
#
# Useful when you genuinely need the inverse.
#
# But for solving:
#
# A x = b
#
# use:
#
# np.linalg.solve(A, b)
#
# rather than:
#
# np.linalg.inv(A) @ b
#
# ============================================================
# 122. np.linalg.det()
# ============================================================
#
# determinant:
#
# np.linalg.det(A)
#
# The determinant provides important information about a square
# matrix.
#
# Particularly:
#
# det(A) = 0
#
# means the matrix is singular and does not have an ordinary
# inverse.
#
# ============================================================
# 123. np.linalg.matrix_rank()
# ============================================================
#
# matrix rank tells you the number of linearly independent
# dimensions represented by a matrix.
#
# This is important for:
#
#     feature redundancy
#     linear algebra
#     regression
#     dimensionality analysis
#
# ============================================================
# 124. np.linalg.eig()
# ============================================================
#
# Computes eigenvalues and eigenvectors.
#
# Conceptually:
#
# A v = λ v
#
# where:
#
# A = matrix
# v = eigenvector
# λ = eigenvalue
#
# This becomes important in:
#
#     PCA
#     dimensionality reduction
#     dynamical systems
#     spectral methods
#
# ============================================================
# 125. np.linalg.svd()
# ============================================================
#
# Singular Value Decomposition:
#
# A = U Σ Vᵀ
#
# SVD decomposes a matrix into:
#
#     U
#     Σ
#     Vᵀ
#
# It is extremely important in:
#
#     dimensionality reduction
#     PCA-related mathematics
#     recommendation systems
#     signal processing
#     matrix approximation
#
# You do not need to memorize the derivation initially.
#
# Understand the mental model:
#
# complicated matrix
#       ↓
# decompose
#       ↓
# simpler mathematical components
#       ↓
# analyze / compress / reconstruct
#
# ============================================================
# 126. np.linalg.lstsq()
# ============================================================
#
# Least-squares solution.
#
# It solves problems related to:
#
#     A x ≈ b
#
# rather than requiring an exact solution.
#
# This connects directly to:
#
#     linear regression
#     numerical optimization
#     statistical modeling
#
# ============================================================
# 127. np.isclose()
# ============================================================
#
# Floating-point calculations are not always exactly equal.
#
# Instead of:
#
# a == b
#
# use:
#
# np.isclose(a, b)
#
# when numerical tolerance matters.
#
# Example:
#
# floating-point calculations
#       ↓
# tiny numerical difference
#       ↓
# exact == may say False
#       ↓
# isclose() can say True
#
# This is extremely important in scientific programming and ML.
#
# ============================================================
# 128. np.allclose()
# ============================================================
#
# np.allclose(a, b)
#
# asks whether all corresponding elements are sufficiently close.
#
# Very useful for testing:
#
#     expected array
#             vs
#     calculated array
#
# Example concept:
#
# expected
#     ↓
# calculated
#     ↓
# numerical tolerance
#     ↓
# allclose
#     ↓
# True / False
#
# ============================================================
# 129. DATA TYPES — dtype
# ============================================================
#
# NumPy arrays have a dtype.
#
# Common types:
#
#     np.int8
#     np.int16
#     np.int32
#     np.int64
#
#     np.float16
#     np.float32
#     np.float64
#
#     np.bool_
#
#     np.complex64
#     np.complex128
#
# Example:
#
# x = np.array([1, 2, 3], dtype=np.float64)
#
# dtype:
#
# float64
#
# ============================================================
# 130. WHY DTYPE MATTERS
# ============================================================
#
# dtype controls:
#
#     what values can be represented
#     memory consumption
#     numerical precision
#     arithmetic behavior
#
# Conceptually:
#
# data
#   ↓
# dtype
#   ↓
# representation in memory
#   ↓
# numerical computation
#
# This becomes important with large datasets.
#
# ============================================================
# 131. np.astype()
# ============================================================
#
# ndarray.astype()
#
# converts an array to another dtype.
#
# Example:
#
# x = np.array([1, 2, 3])
#
# y = x.astype(np.float64)
#
# Concept:
#
# integer array
#      ↓
# type conversion
#      ↓
# floating-point array
#
# Useful in:
#
#     preprocessing
#     ML
#     memory optimization
#     compatibility with algorithms
#
# ============================================================
# 132. INTEGER VS FLOAT
# ============================================================
#
# Consider:
#
# x = np.array([1, 2, 3])
#
# x.dtype
#
# may be an integer dtype.
#
# If you need fractional values:
#
# x.astype(float)
#
# converts to floating-point representation.
#
# This matters because:
#
# integer
#     → whole-number representation
#
# float
#     → fractional numerical representation
#
# ============================================================
# 133. MEMORY AND ITEMSIZE
# ============================================================
#
# x.itemsize
#
# tells you the number of bytes used by each element.
#
# x.nbytes
#
# gives total bytes used by the array data.
#
# Conceptually:
#
# number of elements
#        ×
# bytes per element
#        =
# total data bytes
#
# This matters when working with:
#
#     large datasets
#     images
#     scientific arrays
#     ML datasets
#
# ============================================================
# 134. CONTIGUOUS MEMORY — BASIC IDEA
# ============================================================
#
# NumPy arrays are designed around efficient numerical storage.
#
# You do NOT need to learn memory layout deeply at the beginning.
#
# But understand:
#
# array
#   ↓
# numerical data stored in memory
#   ↓
# efficient bulk operations
#
# As your datasets become large, memory layout and views become
# increasingly relevant.
#
# ============================================================
# 135. np.ascontiguousarray()
# ============================================================
#
# Ensures an array is stored in C-contiguous memory layout when
# appropriate.
#
# This is more advanced.
#
# Learn it when working with:
#
#     low-level numerical code
#     C/C++ interfaces
#     performance-sensitive workflows
#
# It is NOT a first-priority DA function.
#
# ============================================================
# 136. np.asarray()
# ============================================================
#
# np.asarray(x)
#
# converts input into an ndarray when necessary.
#
# Compared conceptually with np.array():
#
# np.asarray()
#     → often avoids copying if the input is already an array
#
# np.array()
#     → generally creates an array object and can be configured
#       to copy
#
# This distinction becomes important for memory-aware code.
#
# ============================================================
# 137. np.asarray_chkfinite()
# ============================================================
#
# More specialized:
#
# np.asarray_chkfinite()
#
# converts to an array while checking for NaN or infinity.
#
# Useful in numerical workflows where invalid finite values must
# be rejected early.
#
# ============================================================
# 138. np.where() — DATA CLEANING PATTERN
# ============================================================
#
# Suppose:
#
# age = np.array([20, 30, -5, 40])
#
# Negative age is invalid.
#
# You could create:
#
# cleaned_age = np.where(age < 0, np.nan, age)
#
# Concept:
#
# invalid?
#   |
#   +-- YES → replace
#   |
#   +-- NO  → preserve
#
# This is a useful pattern for vectorized cleaning.
#
# For full data-cleaning workflows, pandas is usually more convenient.
#
# ============================================================
# 139. np.isnan() + np.where()
# ============================================================
#
# Missing-data workflow:
#
# data
#   ↓
# np.isnan(data)
#   ↓
# Boolean mask
#   ↓
# inspect / replace / calculate
#
# This is a fundamental numerical-array workflow.
#
# ============================================================
# 140. np.nan_to_num()
# ============================================================
#
# np.nan_to_num()
#
# replaces NaN and infinities with finite values according to the
# selected rules.
#
# Example concept:
#
# NaN
#   ↓
# replacement
#
# +inf
#   ↓
# replacement
#
# -inf
#   ↓
# replacement
#
# IMPORTANT:
#
# Replacing missing/infinite values is a data-analysis decision.
#
# Do not automatically replace everything with zero without
# understanding what the missing value means.
#
# ============================================================
# 141. np.searchsorted()
# ============================================================
#
# Given a sorted array, searchsorted() finds where a value could
# be inserted while preserving order.
#
# Example:
#
# x = [10, 20, 30]
#
# search for 25:
#
# position → between 20 and 30
#
# This is useful for:
#
#     binning
#     lookup
#     sorted data
#     numerical algorithms
#
# ============================================================
# 142. np.digitize()
# ============================================================
#
# digitize() assigns values to bins.
#
# Concept:
#
# continuous value
#       ↓
# interval/bin
#       ↓
# category number
#
# Example concept:
#
# age:
#
# 0–17
# 18–64
# 65+
#
# digitize can help convert numerical values into bins.
#
# This is useful in DA and feature engineering.
#
# ============================================================
# 143. np.histogram()
# ============================================================
#
# np.histogram()
#
# counts how many values fall into numerical bins.
#
# Concept:
#
# values
#   ↓
# bins
#   ↓
# counts
#
# Example:
#
# values:
#
# 1, 2, 2, 3, 4, 8
#
# bins:
#
# 0–2
# 2–4
# 4–6
# 6–10
#
# result:
#
# count per interval
#
# This is fundamental to distribution analysis.
#
# ============================================================
# 144. HISTOGRAM MENTAL MODEL
# ============================================================
#
# A histogram answers:
#
# "How many observations fall into each numerical interval?"
#
# DATA
# ↓
# divide number line into bins
# ↓
# count observations
# ↓
# frequency distribution
#
# This is extremely useful in EDA.
#
# ============================================================
# 145. np.bincount()
# ============================================================
#
# bincount() counts occurrences of non-negative integer values.
#
# Example:
#
# x = [0, 1, 1, 2, 2, 2]
#
# np.bincount(x)
#
# →
#
# [1, 2, 3]
#
# Meaning:
#
# value 0 → 1 occurrence
# value 1 → 2 occurrences
# value 2 → 3 occurrences
#
# Useful for efficient categorical/integer counting.
#
# ============================================================
# 146. np.meshgrid()
# ============================================================
#
# meshgrid() creates coordinate grids.
#
# It is important for:
#
#     visualization
#     numerical functions
#     optimization
#     simulations
#
# Concept:
#
# x coordinates
#      ↓
# repeated across grid
#
# y coordinates
#      ↓
# repeated across grid
#
#       ↓
#
# 2-D coordinate system
#
# More advanced for DA, but useful in scientific computing.
#
# ============================================================
# 147. np.fromfunction()
# ============================================================
#
# Creates an array by evaluating a function over index coordinates.
#
# Concept:
#
# indices
#   ↓
# function
#   ↓
# generated array
#
# Useful for mathematical/numerical array construction.
#
# ============================================================
# 148. np.fromiter()
# ============================================================
#
# Creates an array from an iterable.
#
# Useful when values are generated iteratively.
#
# It is more specialized than np.array().
#
# ============================================================
# 149. np.loadtxt()
# ============================================================
#
# Reads numerical text files.
#
# Example conceptual workflow:
#
# CSV/TXT file
#      ↓
# np.loadtxt()
#      ↓
# ndarray
#      ↓
# numerical analysis
#
# It is useful for simple numerical files.
#
# For real-world CSV data containing:
#
#     missing values
#     mixed types
#     categorical variables
#     dates
#     complex columns
#
# pandas.read_csv() is generally more appropriate.
#
# ============================================================
# 150. np.genfromtxt()
# ============================================================
#
# genfromtxt() is more flexible than loadtxt().
#
# It can handle certain missing-data situations and more complex
# structured text input.
#
# Still:
#
# general tabular data
#     → pandas
#
# numerical text arrays
#     → NumPy
#
# ============================================================
# 151. np.save()
# ============================================================
#
# np.save("data.npy", array)
#
# stores a NumPy array in NumPy's binary .npy format.
#
# Concept:
#
# ndarray
#   ↓
# np.save()
#   ↓
# data.npy
#
# ============================================================
# 152. np.load()
# ============================================================
#
# array = np.load("data.npy")
#
# restores the NumPy array.
#
# This is useful when you repeatedly work with numerical arrays and
# want efficient NumPy-native storage.
#
# ============================================================
# 153. np.savez()
# ============================================================
#
# save multiple arrays in one .npz archive.
#
# Concept:
#
# array A ─┐
# array B ─┼→ .npz
# array C ─┘
#
# Useful for storing related numerical arrays.
#
# ============================================================
# 154. np.loadtxt / genfromtxt / load / save — CHOOSE CORRECTLY
# ============================================================
#
# Simple numerical text:
#
#     np.loadtxt()
#
# More flexible numerical text:
#
#     np.genfromtxt()
#
# NumPy-native binary:
#
#     np.save()
#     np.load()
#
# Multiple NumPy arrays:
#
#     np.savez()
#
# General business tables:
#
#     pandas.read_csv()
#
# ============================================================
# 155. np.random.shuffle()
# ============================================================
#
# Modern practice generally uses:
#
# rng.shuffle(x)
#
# It shuffles an array in place.
#
# This is useful for:
#
#     randomizing observations
#     experiments
#     simulations
#
# In ML, remember that randomization must be handled carefully to
# avoid data leakage.
#
# ============================================================
# 156. RANDOM SAMPLING
# ============================================================
#
# A common workflow:
#
# dataset
#    ↓
# random selection
#    ↓
# sample
#    ↓
# analysis / experiment
#
# Functions:
#
#     rng.choice()
#     rng.integers()
#     rng.shuffle()
#     rng.permutation()
#
# ============================================================
# 157. rng.permutation()
# ============================================================
#
# permutation() returns a randomized permutation.
#
# Unlike shuffle(), it returns a result rather than simply
# modifying the original array in place.
#
# Conceptually:
#
# [1, 2, 3, 4]
#      ↓
# permutation
#      ↓
# [3, 1, 4, 2]
#
# Useful for randomized ordering.
#
# ============================================================
# 158. np.random.seed VS default_rng
# ============================================================
#
# Older NumPy code frequently uses:
#
# np.random.seed(42)
#
# Modern NumPy code should generally prefer:
#
# rng = np.random.default_rng(42)
#
# then:
#
# rng.normal(...)
# rng.integers(...)
# rng.choice(...)
#
# The Generator-based approach provides a clearer independent
# random-number-generation object.
#
# ============================================================
# 159. NUMPY FUNCTIONS VS ARRAY METHODS
# ============================================================
#
# NumPy often provides both:
#
# function:
#
#     np.sum(x)
#
# method:
#
#     x.sum()
#
# Similar examples:
#
#     np.mean(x)      / x.mean()
#     np.std(x)       / x.std()
#     np.var(x)       / x.var()
#     np.min(x)       / x.min()
#     np.max(x)       / x.max()
#     np.reshape(x)   / x.reshape()
#     np.flatten(x)   / x.flatten()   # where applicable
#
# Not every NumPy function has an array method.
#
# ============================================================
# 160. FUNCTIONS VS METHODS — MENTAL MODEL
# ============================================================
#
# Function:
#
#     np.mean(x)
#
# Think:
#
#     "Use NumPy's mean operation on x."
#
# Method:
#
#     x.mean()
#
# Think:
#
#     "Ask this array object to calculate its mean."
#
# The operation may be conceptually identical.
#
# ============================================================
# 161. IMPORTANT ndarray METHODS / ATTRIBUTES TO KNOW
# ============================================================
#
# x.shape
# x.ndim
# x.size
# x.dtype
# x.itemsize
# x.nbytes
# x.T
#
# x.reshape()
# x.flatten()
# x.ravel()
# x.copy()
# x.astype()
# x.sort()
# x.sum()
# x.mean()
# x.std()
# x.var()
# x.min()
# x.max()
# x.argmin()
# x.argmax()
# x.cumsum()
# x.cumprod()
#
# These are the main array-level operations worth knowing well.
#
# ============================================================
# 162. REDUCTION VS ELEMENTWISE OPERATION
# ============================================================
#
# This distinction is fundamental.
#
# ELEMENTWISE:
#
# [1, 2, 3]
#    ↓
# ×2
#    ↓
# [2, 4, 6]
#
# OUTPUT HAS SAME BASIC SHAPE.
#
#
# REDUCTION:
#
# [1, 2, 3]
#    ↓
# sum
#    ↓
# 6
#
# MANY VALUES
#     ↓
# ONE VALUE
#
# Examples:
#
# elementwise:
#
# np.sqrt()
# np.exp()
# np.abs()
# x * 2
#
# reductions:
#
# np.sum()
# np.mean()
# np.min()
# np.max()
# np.std()
#
# Understanding this distinction makes NumPy much easier.
#
# ============================================================
# 163. KEEPdims — A SMALL BUT IMPORTANT CONCEPT
# ============================================================
#
# Consider:
#
# X.shape = (3, 4)
#
# np.mean(X, axis=0)
#
# gives shape:
#
# (4,)
#
# But:
#
# np.mean(X, axis=0, keepdims=True)
#
# gives:
#
# (1, 4)
#
# Why?
#
# keepdims preserves the reduced dimension as size 1.
#
# This can make broadcasting and shape management easier.
#
# ============================================================
# 164. EXAMPLE OF keepdims
# ============================================================
#
# X:
#
# shape = (3, 4)
#
# mean across rows:
#
# axis=0
#
# ordinary result:
#
# (4,)
#
# with keepdims:
#
# (1, 4)
#
# Visual:
#
# original:
#
# [ a b c d ]
# [ e f g h ]
# [ i j k l ]
#
# ↓ mean axis=0
#
# [mean_ae_i, mean_bf_j, mean_cg_k, mean_dh_l]
#
# shape = (4,)
#
# keepdims=True:
#
# [[mean_ae_i, mean_bf_j, mean_cg_k, mean_dh_l]]
#
# shape = (1, 4)
#
# ============================================================
# 165. NaN AND INFINITY — PRACTICAL CHECK
# ============================================================
#
# A useful data-quality pattern:
#
# finite_mask = np.isfinite(x)
#
# Then:
#
# np.all(finite_mask)
#
# asks:
#
# "Are ALL values finite?"
#
# Or:
#
# np.any(~finite_mask)
#
# asks:
#
# "Does ANY invalid numerical value exist?"
#
# This is useful before numerical modeling.
#
# ============================================================
# 166. DATA VALIDATION PATTERN
# ============================================================
#
# numerical data
#       ↓
# np.isfinite()
#       ↓
# valid / invalid mask
#       ↓
# inspect
#       ↓
# clean / transform
#       ↓
# statistical calculation
#       ↓
# model
#
# This is a highly reusable DA/ML pattern.
#
# ============================================================
# 167. NUMPY FOR FEATURE ENGINEERING
# ============================================================
#
# Suppose:
#
# price = np.array([100, 200, 300])
# quantity = np.array([2, 3, 4])
#
# revenue = price * quantity
#
# →
#
# [200, 600, 1200]
#
# No explicit loop is needed.
#
# Concept:
#
# feature 1
#    ×
# feature 2
#    ↓
# derived feature
#
# This is fundamental to numerical feature engineering.
#
# ============================================================
# 168. NORMALIZATION WITH NUMPY
# ============================================================
#
# Min-max normalization:
#
# x_scaled =
#
# (x - min(x))
# ───────────────
# max(x) - min(x)
#
# NumPy implementation:
#
# x_scaled = (
#     (x - np.min(x))
#     /
#     (np.max(x) - np.min(x))
# )
#
# Concept:
#
# original range
#     ↓
# subtract minimum
#     ↓
# starts at zero
#     ↓
# divide by range
#     ↓
# [0, 1]
#
# For actual ML pipelines, use a fitted preprocessing transformer
# such as sklearn.preprocessing.MinMaxScaler so the transformation
# is learned only from training data.
#
# ============================================================
# 169. STANDARDIZATION WITH NUMPY
# ============================================================
#
# z = (x - mean) / standard deviation
#
# NumPy:
#
# z = (x - np.mean(x)) / np.std(x)
#
# This is mathematically useful for learning.
#
# IMPORTANT:
#
# In ML:
#
# NEVER calculate scaling independently on the complete dataset
# before train/test splitting if that would allow test information
# to influence the transformation.
#
# Use a fitted preprocessing pipeline.
#
# ============================================================
# 170. ML DATA SHAPE — A CRITICAL MENTAL MODEL
# ============================================================
#
# In many ML problems:
#
# X = feature matrix
#
# shape:
#
# (number_of_samples, number_of_features)
#
# Example:
#
# 1000 samples
# 5 features
#
# X.shape:
#
# (1000, 5)
#
# Visual:
#
#             features
#          f1 f2 f3 f4 f5
# sample 1
# sample 2
# sample 3
#   ...
# sample 1000
#
# This is one of the most important NumPy shape conventions for ML.
#
# ============================================================
# 171. y — TARGET VECTOR
# ============================================================
#
# Often:
#
# X.shape = (1000, 5)
#
# y.shape = (1000,)
#
# Meaning:
#
# X:
#     1000 observations
#     5 features
#
# y:
#     one target per observation
#
# Relationship:
#
# X[0] ↔ y[0]
# X[1] ↔ y[1]
# ...
#
# The rows of X and corresponding entries of y must remain aligned.
#
# ============================================================
# 172. FEATURE COLUMN SELECTION
# ============================================================
#
# Suppose:
#
# X.shape = (100, 5)
#
# X[:, 0]
#
# → first feature column.
#
# X[:, 1]
#
# → second feature.
#
# X[:, 2:4]
#
# → third and fourth features.
#
# This is fundamental for numerical ML data.
#
# ============================================================
# 173. ROW SELECTION
# ============================================================
#
# X[0]
#
# → first observation.
#
# X[0, :]
#
# → first observation / row.
#
# X[:10]
#
# → first ten observations.
#
# X[10:20]
#
# → observations 10 through 19.
#
# ============================================================
# 174. BOOLEAN FEATURE FILTERING
# ============================================================
#
# Suppose:
#
# ages = np.array([15, 20, 30, 12, 40])
#
# adults = ages >= 18
#
# mask:
#
# [False, True, True, False, True]
#
# ages[adults]
#
# →
#
# [20, 30, 40]
#
# This is the NumPy foundation of many filtering operations.
#
# ============================================================
# 175. RANDOM TRAIN/TEST SPLIT — CONCEPT
# ============================================================
#
# NumPy can help with random permutation:
#
# indices = rng.permutation(len(X))
#
# Then:
#
# train_indices
# test_indices
#
# can be selected.
#
# Concept:
#
# dataset
#     ↓
# random permutation of indices
#     ↓
# split indices
#     ↓
# X_train / X_test
#
# In practical ML:
#
# sklearn.model_selection.train_test_split()
#
# is generally preferable because it provides a higher-level,
# clearer interface.
#
# ============================================================
# 176. NUMPY FOR MATRIX REPRESENTATION IN ML
# ============================================================
#
# Many ML algorithms can be understood using:
#
# X
# ↓
# matrix of observations/features
#
# w
# ↓
# vector of model parameters
#
# X @ w
# ↓
# predictions / linear combination
#
# Example:
#
# X.shape = (100, 5)
# w.shape = (5,)
#
# X @ w
#
# result shape:
#
# (100,)
#
# Each observation receives:
#
# x₁w₁ + x₂w₂ + ... + x₅w₅
#
# This is one of the most important mathematical foundations
# underneath linear models and neural networks.
#
# ============================================================
# 177. DOT PRODUCT AS WEIGHTED SUM
# ============================================================
#
# One sample:
#
# x = [10, 2, 5]
#
# weights:
#
# w = [0.5, 3, -1]
#
# x @ w:
#
# 10×0.5 + 2×3 + 5×(-1)
#
# = 5 + 6 - 5
#
# = 6
#
# This is the basic mathematical operation behind a linear model:
#
# weighted features
#       ↓
# weighted sum
#       ↓
# prediction
#
# ============================================================
# 178. NUMPY AND PCA — CONCEPTUAL CONNECTION
# ============================================================
#
# PCA relies heavily on linear algebra.
#
# Conceptually:
#
# feature matrix X
#       ↓
# center features
#       ↓
# covariance structure
#       ↓
# eigenvectors / SVD
#       ↓
# principal components
#       ↓
# lower-dimensional representation
#
# NumPy provides the numerical building blocks:
#
#     mean
#     matrix multiplication
#     covariance
#     eig
#     SVD
#
# In practical DS/ML:
#
# sklearn.decomposition.PCA
#
# is usually the appropriate high-level implementation.
#
# ============================================================
# 179. NUMPY AND DISTANCES
# ============================================================
#
# Euclidean distance between:
#
# a = [1, 2]
# b = [4, 6]
#
# difference:
#
# a - b
#     = [-3, -4]
#
# square:
#
# [9, 16]
#
# sum:
#
# 25
#
# square root:
#
# 5
#
# NumPy:
#
# np.linalg.norm(a - b)
#
# → 5
#
# This is fundamental in:
#
#     KNN
#     clustering
#     similarity analysis
#     scientific computing
#
# ============================================================
# 180. NUMPY AND ERROR CALCULATIONS
# ============================================================
#
# Suppose:
#
# actual = np.array([10, 20, 30])
# predicted = np.array([12, 18, 33])
#
# errors:
#
# predicted - actual
#
# → [2, -2, 3]
#
# absolute errors:
#
# np.abs(predicted - actual)
#
# → [2, 2, 3]
#
# squared errors:
#
# (predicted - actual) ** 2
#
# → [4, 4, 9]
#
# mean squared error:
#
# np.mean((predicted - actual) ** 2)
#
# = 17 / 3
#
# = 5.666666...
#
# This shows how NumPy provides the low-level numerical operations
# behind many ML metrics.
#
# ============================================================
# 181. NUMPY AND LOGICAL CONDITIONS
# ============================================================
#
# NumPy lets you express conditions over entire arrays.
#
# Example:
#
# errors = np.array([-2, 1, -5, 3])
#
# np.abs(errors) > 2
#
# →
#
# [False, False, True, True]
#
# Then:
#
# np.sum(np.abs(errors) > 2)
#
# counts how many errors exceed 2 in magnitude.
#
# This pattern is extremely useful:
#
# condition
#     ↓
# Boolean mask
#     ↓
# aggregate / filter / transform
#
# ============================================================
# 182. np.floor_divide()
# ============================================================
#
# np.floor_divide(a, b)
#
# performs floor division elementwise.
#
# It corresponds conceptually to:
#
# a // b
#
# This is useful for:
#
#     binning
#     integer indexing
#     numerical grouping
#
# ============================================================
# 183. np.mod()
# ============================================================
#
# np.mod(a, b)
#
# gives elementwise remainder.
#
# Equivalent conceptually to:
#
# a % b
#
# Useful for:
#
#     periodic patterns
#     grouping
#     cyclic calculations
#
# ============================================================
# 184. TRIGONOMETRIC FUNCTIONS
# ============================================================
#
# Important:
#
# np.sin()
# np.cos()
# np.tan()
#
# and:
#
# np.arcsin()
# np.arccos()
# np.arctan()
#
# angles are generally expressed in radians.
#
# Degree conversion:
#
# np.deg2rad()
# np.rad2deg()
#
# These are more important in:
#
#     engineering
#     physics
#     signal processing
#     scientific computing
#
# than ordinary business DA.
#
# ============================================================
# 185. np.pi AND np.e
# ============================================================
#
# NumPy provides mathematical constants:
#
# np.pi
#     → π
#
# np.e
#     → Euler's number
#
# Useful for:
#
#     scientific calculations
#     probability
#     trigonometry
#     exponential models
#
# ============================================================
# 186. NUMPY RANDOM DISTRIBUTIONS — WHAT TO KNOW
# ============================================================
#
# Important Generator methods:
#
# rng.uniform()
#     → uniform distribution
#
# rng.normal()
#     → normal distribution
#
# rng.integers()
#     → random integers
#
# rng.binomial()
#     → binomial distribution
#
# rng.poisson()
#     → Poisson distribution
#
# rng.exponential()
#     → exponential distribution
#
# rng.lognormal()
#     → log-normal distribution
#
# rng.choice()
#     → random sampling
#
# rng.shuffle()
#     → in-place random permutation
#
# rng.permutation()
#     → returned permutation
#
# ============================================================
# 187. UNIFORM DISTRIBUTION
# ============================================================
#
# rng.uniform(0, 1, 1000)
#
# generates approximately uniformly distributed values.
#
# Visual concept:
#
# probability
#     |
#     | ─────────────────
#     |
#     +----------------------→ value
#       0                  1
#
# Every region has approximately equal density.
#
# ============================================================
# 188. BINOMIAL DISTRIBUTION
# ============================================================
#
# rng.binomial(n, p, size)
#
# models the number of successes in repeated independent
# Bernoulli trials.
#
# Example:
#
# n = 10 trials
# p = 0.5 probability of success
#
# possible result:
#
# 0, 1, 2, ..., 10 successes
#
# Useful for:
#
#     probability simulation
#     A/B testing simulations
#     reliability
#     discrete probabilistic models
#
# ============================================================
# 189. POISSON DISTRIBUTION
# ============================================================
#
# rng.poisson(lam, size)
#
# models counts occurring over a fixed interval under appropriate
# assumptions.
#
# Examples:
#
#     number of calls per minute
#     arrivals per hour
#     events per interval
#
# Useful in:
#
#     operations analysis
#     queueing
#     reliability
#     simulation
#
# ============================================================
# 190. EXPONENTIAL DISTRIBUTION
# ============================================================
#
# rng.exponential(scale, size)
#
# is commonly used to model waiting times under appropriate
# assumptions.
#
# Examples:
#
#     time between events
#     service intervals
#
# Again:
#
# the distribution is a model assumption, not an automatic truth
# about the data.
#
# ============================================================
# 191. NUMPY AND DESCRIPTIVE DATA ANALYSIS
# ============================================================
#
# For a numerical feature:
#
# x
# ↓
# np.min()
# np.max()
# np.mean()
# np.median()
# np.std()
# np.percentile()
# np.quantile()
#
# This gives a compact numerical summary.
#
# However, for business datasets:
#
# pandas
#
# is generally better for:
#
#     groupby
#     categorical analysis
#     missing-data management
#     joins
#     dates
#     tables
#
# NumPy remains the numerical engine underneath much of the work.
#
# ============================================================
# 192. NUMPY AND DATA ANALYSIS — PRACTICAL PRIORITIES
# ============================================================
#
# MASTER THESE:
#
# np.array()
# np.arange()
# np.linspace()
# np.zeros()
# np.ones()
# np.full()
#
# x.shape
# x.ndim
# x.size
# x.dtype
#
# indexing
# slicing
# boolean masking
#
# np.where()
# np.select()
#
# np.sum()
# np.mean()
# np.median()
# np.min()
# np.max()
# np.std()
# np.var()
# np.percentile()
# np.quantile()
#
# np.unique()
# np.sort()
# np.argsort()
#
# np.isnan()
# np.isfinite()
# np.nanmean()
# np.nanmedian()
#
# np.concatenate()
# np.stack()
# np.vstack()
# np.hstack()
#
# reshape()
# flatten()
# ravel()
# transpose / .T
# expand_dims()
# squeeze()
#
# broadcasting
#
# ============================================================
# 193. NUMPY AND DATA SCIENCE — PRACTICAL PRIORITIES
# ============================================================
#
# In addition to the DA tools, learn:
#
# np.cov()
# np.corrcoef()
# np.histogram()
# np.diff()
# np.cumsum()
# np.gradient()
#
# np.linalg.norm()
# np.linalg.solve()
# np.linalg.lstsq()
# np.linalg.eig()
# np.linalg.svd()
#
# random Generator:
#
# np.random.default_rng()
# rng.normal()
# rng.uniform()
# rng.integers()
# rng.choice()
# rng.permutation()
#
# ============================================================
# 194. NUMPY AND MACHINE LEARNING — PRACTICAL PRIORITIES
# ============================================================
#
# Most important concepts:
#
# 1. ndarray
# 2. shape
# 3. axis
# 4. dtype
# 5. indexing
# 6. slicing
# 7. boolean masking
# 8. broadcasting
# 9. vectorization
# 10. reshape
# 11. transpose
# 12. matrix multiplication
# 13. dot product
# 14. aggregation
# 15. mean/std/variance
# 16. covariance/correlation
# 17. random sampling
# 18. linear algebra
#
# These concepts appear repeatedly in:
#
#     preprocessing
#     feature engineering
#     linear models
#     clustering
#     dimensionality reduction
#     neural-network mathematics
#     optimization
#
# ============================================================
# 195. WHAT NUMPY DOES BETTER THAN PYTHON LISTS
# ============================================================
#
# Python list:
#
#     general-purpose container
#
# NumPy ndarray:
#
#     numerical array
#
# NumPy gives:
#
#     vectorized arithmetic
#     broadcasting
#     multidimensional structure
#     efficient numerical storage
#     mathematical functions
#     statistical operations
#     linear algebra
#     random simulation
#
# Therefore:
#
# Python list
#      ↓
# general Python programming
#
# NumPy ndarray
#      ↓
# numerical computing
#
# ============================================================
# 196. WHAT NUMPY DOES NOT REPLACE
# ============================================================
#
# NumPy is NOT a replacement for:
#
# pandas:
#
#     tabular data analysis
#     joins
#     groupby
#     labels
#     dates
#
# SciPy:
#
#     advanced scientific/statistical functions
#
# statsmodels:
#
#     statistical modeling and inference
#
# scikit-learn:
#
#     machine learning workflows
#
# matplotlib:
#
#     visualization
#
# seaborn:
#
#     statistical visualization
#
# ============================================================
# 197. NUMPY + PANDAS
# ============================================================
#
# Typical workflow:
#
# CSV
#  ↓
# pandas
#  ↓
# DataFrame
#  ↓
# cleaning / grouping / joining
#  ↓
# NumPy numerical operations
#  ↓
# SciPy / statsmodels / sklearn
#  ↓
# results
#
# In practice, you will often move between:
#
# pandas Series/DataFrame
#          ↕
# NumPy ndarray
#
# ============================================================
# 198. NUMPY + SCIKIT-LEARN
# ============================================================
#
# Typical ML workflow:
#
# DataFrame
#     ↓
# cleaning
#     ↓
# feature matrix X
#     ↓
# NumPy-like numerical representation
#     ↓
# preprocessing
#     ↓
# model
#     ↓
# predictions
#     ↓
# metrics
#
# You should understand NumPy arrays because ML libraries frequently
# expect data with:
#
#     samples × features
#
# structure.
#
# ============================================================
# 199. NUMPY + SCIPY
# ============================================================
#
# NumPy:
#
# numerical arrays
# basic linear algebra
# mathematical functions
#
# SciPy:
#
# advanced scientific algorithms
# statistics
# optimization
# signal processing
# numerical integration
#
# Conceptually:
#
# NumPy
#   ↓
# numerical foundation
#   ↓
# SciPy
#   ↓
# higher-level scientific computation
#
# ============================================================
# 200. NUMPY + STATISTICS
# ============================================================
#
# Python `statistics`:
#
#     statistics.mean()
#     statistics.median()
#     statistics.stdev()
#
# NumPy:
#
#     np.mean()
#     np.median()
#     np.std()
#
# NumPy becomes more useful when the data is already represented
# as multidimensional numerical arrays.
#
# ============================================================
# 201. COMMON NUMPY MISTAKE — axis
# ============================================================
#
# Beginners often memorize:
#
# axis=0 = columns
# axis=1 = rows
#
# This can become confusing.
#
# Better mental model:
#
# ASK:
#
# "Which dimension am I COLLAPSING?"
#
# For:
#
# X.shape = (rows, columns)
#
# axis=0:
#
# collapse the row dimension
#     ↓
# one result per column
#
# axis=1:
#
# collapse the column dimension
#     ↓
# one result per row
#
# ============================================================
# 202. COMMON MISTAKE — SHAPE
# ============================================================
#
# These are different:
#
# (3,)
# (1, 3)
# (3, 1)
#
# Never treat them as interchangeable.
#
# In ML:
#
#     (n_features,)
#
# may represent one vector.
#
# while:
#
#     (1, n_features)
#
# represents one row/sample.
#
# and:
#
#     (n_features, 1)
#
# represents a column vector.
#
# ============================================================
# 203. COMMON MISTAKE — * VS @
# ============================================================
#
# x * y
#     → elementwise multiplication
#
# x @ y
#     → matrix multiplication
#
# They are NOT interchangeable.
#
# ============================================================
# 204. COMMON MISTAKE — COPY VS VIEW
# ============================================================
#
# Some slicing operations create views.
#
# Therefore:
#
# subset = x[:3]
#
# does not necessarily mean:
#
# "create an independent copy."
#
# If independent data is required:
#
# subset = x[:3].copy()
#
# This is an important practical habit.
#
# ============================================================
# 205. COMMON MISTAKE — INTEGER DIVISION / DTYPE
# ============================================================
#
# Be aware of dtype when performing operations.
#
# Numerical operations may produce a dtype different from the
# original input.
#
# When fractional values are required, use floating-point arrays.
#
# Example:
#
# x = np.array([1, 2, 3], dtype=float)
#
# ============================================================
# 206. COMMON MISTAKE — FLOATING-POINT EQUALITY
# ============================================================
#
# Avoid assuming:
#
# calculated_value == expected_value
#
# is always reliable for floating-point calculations.
#
# Prefer:
#
# np.isclose()
#
# or:
#
# np.allclose()
#
# when a tolerance is appropriate.
#
# ============================================================
# 207. COMMON MISTAKE — NAN
# ============================================================
#
# NaN is not an ordinary number.
#
# Do not test:
#
# x == np.nan
#
# to find NaN.
#
# Use:
#
# np.isnan(x)
#
# or:
#
# np.isfinite(x)
#
# ============================================================
# 208. COMMON MISTAKE — MODIFYING DATA UNINTENTIONALLY
# ============================================================
#
# Methods such as:
#
# x.sort()
#
# modify the array in place.
#
# While:
#
# np.sort(x)
#
# returns a sorted result.
#
# Always know whether an operation:
#
#     returns a new result
#
# or:
#
#     changes the existing array.
#
# ============================================================
# 209. COMMON MISTAKE — np.empty()
# ============================================================
#
# Do NOT assume:
#
# np.empty(5)
#
# gives:
#
# [0, 0, 0, 0, 0]
#
# It does not initialize the values.
#
# Use:
#
# np.zeros(5)
#
# if zeros are required.
#
# ============================================================
# 210. COMMON MISTAKE — RANDOMNESS
# ============================================================
#
# If you need reproducible experiments:
#
# rng = np.random.default_rng(42)
#
# Then use that generator.
#
# Otherwise:
#
# random experiments
#     ↓
# different results
#     ↓
# difficult debugging / comparison
#
# Reproducibility matters in ML.
#
# ============================================================
# 211. PERFORMANCE — WHY VECTORISATION MATTERS
# ============================================================
#
# Python loop:
#
# for x in data:
#     result.append(x * 2)
#
# NumPy:
#
# result = data * 2
#
# The NumPy version expresses the numerical operation at the array
# level.
#
# NumPy can then use optimized low-level implementations.
#
# The lesson is NOT:
#
# "Never use Python loops."
#
# The lesson is:
#
# "When operating on numerical arrays, think in terms of whole-array
# operations first."
#
# ============================================================
# 212. VECTORIZATION MENTAL MODEL
# ============================================================
#
# Traditional thinking:
#
#     item 1 → operation
#     item 2 → operation
#     item 3 → operation
#     ...
#
# NumPy thinking:
#
#     whole array
#          ↓
#     vectorized operation
#          ↓
#     whole result
#
# This change in thinking is one of the most important skills to
# develop.
#
# ============================================================
# 213. A COMPLETE MINI DATA-ANALYSIS EXAMPLE
# ============================================================
#
# import numpy as np
#
# sales = np.array([
#     120,
#     150,
#     180,
#     200,
#     250,
#     300
# ])
#
# mean_sales = np.mean(sales)
# median_sales = np.median(sales)
# std_sales = np.std(sales)
# minimum = np.min(sales)
# maximum = np.max(sales)
#
# q1 = np.percentile(sales, 25)
# q3 = np.percentile(sales, 75)
#
# print("Mean:", mean_sales)
# print("Median:", median_sales)
# print("Standard deviation:", std_sales)
# print("Minimum:", minimum)
# print("Maximum:", maximum)
# print("Q1:", q1)
# print("Q3:", q3)
#
# ============================================================
# HOW TO THINK ABOUT IT
# ============================================================
#
# RAW SALES
#     ↓
#     ├── mean       → center
#     ├── median     → robust center
#     ├── std        → spread
#     ├── min/max    → range
#     └── quartiles  → distribution position
#
# This is the basic descriptive-statistics pattern.
#
# ============================================================
# 214. A COMPLETE VECTORISED FEATURE EXAMPLE
# ============================================================
#
# import numpy as np
#
# price = np.array([100, 200, 150, 300])
# quantity = np.array([2, 1, 4, 3])
#
# revenue = price * quantity
#
# Conceptual flow:
#
# price:
#
# [100, 200, 150, 300]
#
# quantity:
#
# [  2,   1,   4,   3]
#
# elementwise multiplication:
#
# [100×2, 200×1, 150×4, 300×3]
#
# ↓
#
# [200, 200, 600, 900]
#
# No explicit loop is required.
#
# ============================================================
# 215. A COMPLETE FILTERING EXAMPLE
# ============================================================
#
# import numpy as np
#
# sales = np.array([
#     100,
#     250,
#     80,
#     400,
#     150
# ])
#
# high_sales = sales[sales >= 200]
#
# Conceptual execution:
#
# sales
#     ↓
# compare each value >= 200
#     ↓
# [False, True, False, True, False]
#     ↓
# keep True positions
#     ↓
# [250, 400]
#
# This is one of the most important NumPy patterns for DA.
#
# ============================================================
# 216. A COMPLETE TRANSFORMATION EXAMPLE
# ============================================================
#
# import numpy as np
#
# sales = np.array([
#     100,
#     250,
#     80,
#     400,
#     150
# ])
#
# sales_k = sales / 1000
#
# Concept:
#
# original units
#     ↓
# vectorized division
#     ↓
# transformed units
#
# Result:
#
# [0.10, 0.25, 0.08, 0.40, 0.15]
#
# ============================================================
# 217. A COMPLETE NORMALIZATION EXAMPLE
# ============================================================
#
# import numpy as np
#
# x = np.array([10, 20, 30, 40])
#
# x_min = np.min(x)
# x_max = np.max(x)
#
# normalized = (
#     (x - x_min)
#     /
#     (x_max - x_min)
# )
#
# Concept:
#
# [10, 20, 30, 40]
#       ↓
# subtract 10
#       ↓
# [0, 10, 20, 30]
#       ↓
# divide by 30
#       ↓
# [0, 0.333..., 0.666..., 1]
#
# This demonstrates:
#
#     broadcasting
#     vectorization
#     aggregation
#     transformation
#
# all in one example.
#
# ============================================================
# 218. A COMPLETE ML MATRIX EXAMPLE
# ============================================================
#
# import numpy as np
#
# X = np.array([
#     [2, 3],
#     [4, 5],
#     [6, 7]
# ])
#
# w = np.array([0.5, 2.0])
#
# predictions = X @ w
#
# For first row:
#
# [2, 3] @ [0.5, 2]
#
# = 2×0.5 + 3×2
# = 1 + 6
# = 7
#
# Second:
#
# 4×0.5 + 5×2
# = 2 + 10
# = 12
#
# Third:
#
# 6×0.5 + 7×2
# = 3 + 14
# = 17
#
# Therefore:
#
# predictions:
#
# [7, 12, 17]
#
# This is the basic numerical structure of a linear model.
#
# ============================================================
# 219. NUMPY FOR DA / DS / ML — PRIORITY TABLE
# ============================================================
#
# +----------------------+-----+-----+-----+
# | NumPy concept        | DA  | DS  | ML  |
# +----------------------+-----+-----+-----+
# | ndarray              | ★★★ | ★★★★★|★★★★★|
# | shape                | ★★★★|★★★★★|★★★★★|
# | axis                 | ★★★★|★★★★★|★★★★★|
# | dtype                | ★★★ |★★★★ |★★★★ |
# | indexing             |★★★★★|★★★★★|★★★★★|
# | slicing              |★★★★★|★★★★★|★★★★★|
# | boolean masking      |★★★★★|★★★★★|★★★★★|
# | vectorization        |★★★★ |★★★★★|★★★★★|
# | broadcasting         |★★★ |★★★★★|★★★★★|
# | reshape              |★★★★ |★★★★★|★★★★★|
# | statistics           |★★★★★|★★★★★|★★★★ |
# | random sampling      |★★★ |★★★★ |★★★★ |
# | covariance           |★★★★ |★★★★★|★★★★ |
# | linear algebra       |★★  |★★★★ |★★★★★|
# | matrix multiplication|★★  |★★★★ |★★★★★|
# | SVD                   |★   |★★★★ |★★★★★|
# | file I/O              |★★★ |★★★ |★★★ |
# +----------------------+-----+-----+-----+
#
# ============================================================
# 220. THE NUMPY FUNCTIONS YOU SHOULD MASTER
# ============================================================
#
# ARRAY CREATION:
#
# np.array()
# np.asarray()
# np.zeros()
# np.ones()
# np.full()
# np.empty()
# np.zeros_like()
# np.ones_like()
# np.full_like()
# np.arange()
# np.linspace()
# np.eye()
# np.diag()
#
# SHAPE / STRUCTURE:
#
# x.shape
# x.ndim
# x.size
# x.reshape()
# x.flatten()
# x.ravel()
# x.T
# np.transpose()
# np.expand_dims()
# np.squeeze()
# np.swapaxes()
# np.moveaxis()
#
# INDEXING / SELECTION:
#
# indexing
# slicing
# boolean masks
# np.where()
# np.select()
# np.nonzero()
# np.count_nonzero()
#
# BASIC NUMERICAL OPERATIONS:
#
# +
# -
# *
# /
# **
# @
#
# np.add()
# np.subtract()
# np.multiply()
# np.divide()
# np.power()
#
# MATHEMATICAL UFUNCS:
#
# np.abs()
# np.sqrt()
# np.exp()
# np.log()
# np.log10()
# np.log2()
# np.floor()
# np.ceil()
# np.round()
# np.sin()
# np.cos()
# np.tan()
#
# AGGREGATION:
#
# np.sum()
# np.mean()
# np.median()
# np.min()
# np.max()
# np.std()
# np.var()
# np.ptp()
# np.percentile()
# np.quantile()
# np.average()
#
# POSITION / ORDER:
#
# np.argmin()
# np.argmax()
# np.argsort()
# np.sort()
# np.unique()
# np.searchsorted()
#
# DATA QUALITY:
#
# np.isnan()
# np.isinf()
# np.isfinite()
# np.nanmean()
# np.nanmedian()
# np.nanstd()
# np.nanvar()
# np.nanmin()
# np.nanmax()
# np.nansum()
# np.nanpercentile()
# np.nanquantile()
# np.nan_to_num()
#
# LOGICAL:
#
# np.any()
# np.all()
# np.logical_and()
# np.logical_or()
# np.logical_not()
#
# TRANSFORMATION:
#
# np.clip()
# np.diff()
# np.gradient()
# np.cumsum()
# np.cumprod()
#
# COMBINING:
#
# np.concatenate()
# np.stack()
# np.vstack()
# np.hstack()
# np.column_stack()
# np.split()
# np.array_split()
#
# RANDOM:
#
# np.random.default_rng()
# rng.random()
# rng.integers()
# rng.normal()
# rng.uniform()
# rng.binomial()
# rng.poisson()
# rng.exponential()
# rng.lognormal()
# rng.choice()
# rng.shuffle()
# rng.permutation()
#
# STATISTICAL RELATIONSHIPS:
#
# np.cov()
# np.corrcoef()
# np.histogram()
# np.bincount()
#
# LINEAR ALGEBRA:
#
# np.dot()
# np.matmul()
# np.linalg.norm()
# np.linalg.solve()
# np.linalg.inv()
# np.linalg.det()
# np.linalg.matrix_rank()
# np.linalg.eig()
# np.linalg.eigh()
# np.linalg.svd()
# np.linalg.lstsq()
#
# NUMERICAL COMPARISON:
#
# np.isclose()
# np.allclose()
#
# DATA I/O:
#
# np.loadtxt()
# np.genfromtxt()
# np.save()
# np.load()
# np.savez()
#
# ============================================================
# 221. FUNCTIONS YOU DO NOT NEED TO MEMORIZE YET
# ============================================================
#
# NumPy is enormous.
#
# You do NOT need to memorize every function in the library.
#
# Specialized functions exist for:
#
#     Fourier transforms
#     polynomial operations
#     financial calculations
#     floating-point manipulation
#     bitwise operations
#     structured arrays
#     masked arrays
#     datetime arrays
#     random distributions
#     advanced indexing
#     memory layout
#     C interoperability
#     FFT
#     testing
#
# Learn them when your work requires them.
#
# The goal is:
#
#     KNOW THE CORE
#          ↓
#     RECOGNIZE THE PROBLEM
#          ↓
#     KNOW WHICH NUMPY FAMILY TO SEARCH
#          ↓
#     LOOK UP SPECIALIZED FUNCTIONS
#
# This is much better than memorizing hundreds of names.
#
# ============================================================
# 222. NUMPY LEARNING HIERARCHY
# ============================================================
#
# LEVEL 1 — ABSOLUTE FOUNDATION
#
# ndarray
# shape
# ndim
# size
# dtype
# indexing
# slicing
#
#                         ↓
#
# LEVEL 2 — ARRAY COMPUTATION
#
# vectorization
# elementwise operations
# Boolean masks
# broadcasting
#
#                         ↓
#
# LEVEL 3 — DATA ANALYSIS
#
# sum
# mean
# median
# std
# var
# min
# max
# percentiles
# unique
# sorting
#
#                         ↓
#
# LEVEL 4 — ARRAY MANIPULATION
#
# reshape
# transpose
# concatenate
# stack
# split
# squeeze
# expand_dims
#
#                         ↓
#
# LEVEL 5 — NUMERICAL SCIENCE
#
# random
# covariance
# correlation
# histogram
# diff
# cumulative operations
#
#                         ↓
#
# LEVEL 6 — LINEAR ALGEBRA
#
# dot
# @
# norm
# solve
# eig
# SVD
# least squares
#
#                         ↓
#
# LEVEL 7 — ML APPLICATION
#
# X
# y
# feature matrices
# standardization
# matrix multiplication
# distances
# transformations
# numerical model operations
#
# ============================================================
# 223. THE MOST IMPORTANT NUMPY MENTAL MODEL
# ============================================================
#
# Do not think:
#
#     "NumPy is a collection of functions."
#
# Think:
#
#                    ARRAY
#                      |
#          +-----------+-----------+
#          |           |           |
#        SHAPE        TYPE        DATA
#          |           |           |
#          +-----------+-----------+
#                      |
#                ARRAY OPERATION
#                      |
#        +-------------+-------------+
#        |             |             |
#   elementwise     reduction     reshape
#        |             |             |
#        +-------------+-------------+
#                      |
#                 NEW ARRAY / VALUE
#
# NumPy is fundamentally about:
#
#     representing numerical data
#     +
#     transforming numerical data
#     +
#     performing operations efficiently over arrays.
#
# ============================================================
# 224. THE MOST IMPORTANT EXECUTION PATTERN
# ============================================================
#
# For almost every NumPy operation, mentally ask:
#
#     1. What is the input array?
#
#     2. What is its shape?
#
#     3. What is its dtype?
#
#     4. What operation is being performed?
#
#     5. Is it elementwise or a reduction?
#
#     6. If reduction:
#            which axis?
#
#     7. Is broadcasting involved?
#
#     8. What will the output shape be?
#
#     9. Is the original array modified?
#
#    10. Is the result a view or a copy?
#
# This checklist gives you a much stronger mental model than
# memorizing isolated syntax.
#
# ============================================================
# 225. THE NUMPY VALUE-FLOW MODEL
# ============================================================
#
# Example:
#
# X.shape = (1000, 5)
#
# mean = X.mean(axis=0)
#
# Execution:
#
# X
# |
# | 1000 observations
# | 5 features
# ↓
# mean(axis=0)
# ↓
# collapse observation axis
# ↓
# one mean per feature
# ↓
# shape = (5,)
#
# Then:
#
# X - mean
#
# ↓
# broadcasting
# ↓
# subtract each feature's mean
# ↓
# centered X
#
# This tiny example contains:
#
#     shape
#     axis
#     reduction
#     broadcasting
#     vectorization
#
# These are some of the most important NumPy concepts for ML.
#
# ============================================================
# 226. THE NUMPY DATA-SCIENCE PIPELINE
# ============================================================
#
# REAL DATA
#     ↓
# pandas DataFrame
#     ↓
# select numerical data
#     ↓
# NumPy ndarray
#     ↓
# inspect shape / dtype
#     ↓
# clean invalid numerical values
#     ↓
# vectorized transformations
#     ↓
# descriptive statistics
#     ↓
# feature engineering
#     ↓
# linear algebra / numerical operations
#     ↓
# SciPy / statsmodels / scikit-learn
#
# NumPy is therefore a bridge between:
#
# Python
#   ↓
# numerical computation
#   ↓
# Data Science
#   ↓
# Machine Learning
#
# ============================================================
# 227. FINAL DA / DS / ML CHEAT SHEET
# ============================================================
#
# DATA:
#
# np.array()
#
# STRUCTURE:
#
# shape
# ndim
# size
# dtype
#
# SELECT:
#
# indexing
# slicing
# boolean masks
# where()
#
# TRANSFORM:
#
# reshape()
# astype()
# clip()
# abs()
# log()
# sqrt()
#
# SUMMARIZE:
#
# sum()
# mean()
# median()
# std()
# var()
# min()
# max()
# percentile()
# quantile()
#
# RELATIONSHIPS:
#
# cov()
# corrcoef()
#
# DISTRIBUTION:
#
# histogram()
#
# RANDOM:
#
# default_rng()
# normal()
# uniform()
# integers()
# choice()
# permutation()
#
# COMBINE:
#
# concatenate()
# stack()
#
# LINEAR ALGEBRA:
#
# @
# dot()
# norm()
# solve()
# eig()
# svd()
# lstsq()
#
# DATA QUALITY:
#
# isnan()
# isfinite()
# nanmean()
# nanstd()
# nan_to_num()
#
# ============================================================
# 228. CORE IDEA
# ============================================================
#
# NumPy is fundamentally about one idea:
#
#     NUMERICAL DATA
#          ↓
#     REPRESENT AS ARRAYS
#          ↓
#     DESCRIBE THEIR SHAPE
#          ↓
#     OPERATE ON WHOLE ARRAYS
#          ↓
#     USE VECTORISATION + BROADCASTING
#          ↓
#     TRANSFORM / SUMMARIZE / ANALYZE
#          ↓
#     USE LINEAR ALGEBRA WHEN NEEDED
#          ↓
#     FEED NUMERICAL DATA INTO DS / ML WORKFLOWS
#
# If you truly understand:
#
#     ndarray
#     shape
#     axis
#     dtype
#     indexing
#     slicing
#     Boolean masking
#     vectorization
#     broadcasting
#     reshape
#     aggregation
#     matrix multiplication
#     linear algebra
#
# then you have learned the conceptual core of NumPy.
#
# ============================================================
# 229. FINAL LEARNING PRIORITY FOR YOU
# ============================================================
#
# For a DA → DS → ML path, I would prioritize NumPy in this order:
#
# ------------------------------------------------------------
# PRIORITY 1 — MASTER
# ------------------------------------------------------------
#
# np.array()
# shape
# ndim
# size
# dtype
# indexing
# slicing
# Boolean masking
# np.where()
# vectorized arithmetic
# np.sum()
# np.mean()
# np.min()
# np.max()
# np.std()
# np.var()
# np.median()
# np.percentile()
# np.unique()
#
# ------------------------------------------------------------
# PRIORITY 2 — MASTER FOR DS / ML
# ------------------------------------------------------------
#
# reshape()
# transpose / .T
# concatenate()
# stack()
# broadcasting
# np.expand_dims()
# np.squeeze()
# np.isnan()
# np.isfinite()
# np.nanmean()
# np.cov()
# np.corrcoef()
# np.diff()
# np.cumsum()
#
# ------------------------------------------------------------
# PRIORITY 3 — MASTER FOR ML MATHEMATICS
# ------------------------------------------------------------
#
# @
# np.dot()
# np.matmul()
# np.linalg.norm()
# np.linalg.solve()
# np.linalg.lstsq()
# np.linalg.eig()
# np.linalg.svd()
#
# ------------------------------------------------------------
# PRIORITY 4 — KNOW WELL
# ------------------------------------------------------------
#
# np.random.default_rng()
# rng.normal()
# rng.uniform()
# rng.integers()
# rng.choice()
# rng.permutation()
# np.histogram()
# np.clip()
# np.argsort()
# np.all()
# np.any()
# np.isclose()
# np.allclose()
#
# ------------------------------------------------------------
# PRIORITY 5 — LEARN WHEN NEEDED
# ------------------------------------------------------------
#
# advanced axis manipulation
# structured arrays
# FFT
# polynomial tools
# advanced memory-layout tools
# specialized random distributions
# advanced indexing machinery
# low-level interoperability
#
# ============================================================
# FINAL CORE MODEL
# ============================================================
#
#                   NUMPY
#                     |
#                     ↓
#                  ndarray
#                     |
#          +----------+----------+
#          |          |          |
#        shape       dtype      data
#          |          |          |
#          +----------+----------+
#                     ↓
#              ARRAY OPERATIONS
#                     |
#       +-------------+-------------+
#       |             |             |
#   elementwise     reduction     reshape
#       |             |             |
#       +-------------+-------------+
#                     ↓
#                broadcasting
#                     ↓
#              numerical result
#                     |
#          +----------+----------+
#          |          |          |
#          ↓          ↓          ↓
#         DA         DS         ML
#          |          |          |
#     summaries   analysis   feature/model
#                             mathematics
#
# The key habit to develop is:
#
#     "Before I perform a NumPy operation,
#      what are the SHAPE, AXES, and DATA TYPE?"
#
# Then ask:
#
#     "Is this operation elementwise,
#      a reduction,
#      a transformation,
#      or matrix/linear-algebra computation?"
#
# Once that becomes intuitive, NumPy stops looking like a huge
# collection of unrelated functions and becomes a coherent system
# for numerical array computation.
#
# ============================================================
# END OF NUMPY NOTES
# ============================================================
```
