# PYTHON STATISTICS LIBRARY


```python
# =================================================================================================
# PYTHON STATISTICS LIBRARY
# =================================================================================================
#
# A CONCEPTUAL GUIDE FOR DATA ANALYSIS (DA), DATA SCIENCE (DS), AND MACHINE LEARNING (ML)
#
# Python's built-in `statistics` module provides a collection of functions for
# descriptive statistics, basic relationships between variables, simple
# regression, and normal-distribution calculations.
#
# It is especially useful for:
#
#     DA  → quick descriptive statistics, summaries, spread, percentiles
#     DS  → exploratory analysis, statistical reasoning, data understanding
#     ML  → preprocessing ideas, feature analysis, correlations, normalization concepts
#
# It is NOT a replacement for:
#
#     NumPy       → vectorized numerical computation
#     SciPy       → advanced statistical tests and probability distributions
#     pandas      → tabular data analysis
#     statsmodels → statistical modeling and inference
#     scikit-learn → machine-learning workflows
#
# Think of `statistics` as:
#
#     "A small, readable, Python-native toolbox for fundamental statistics."
#
#
# =================================================================================================
# 01. THE BIG PICTURE
# =================================================================================================
#
# Most practical statistics questions can be organized like this:
#
#
#                         DATA
#                          │
#                          ▼
#              ┌─────────────────────────┐
#              │ What is a typical value?│
#              └─────────────────────────┘
#                          │
#                  mean / median / mode
#                          │
#                          ▼
#              ┌─────────────────────────┐
#              │ How spread out is data? │
#              └─────────────────────────┘
#                          │
#             variance / stdev / quantiles
#                          │
#                          ▼
#              ┌─────────────────────────┐
#              │ How do variables relate?│
#              └─────────────────────────┘
#                          │
#             covariance / correlation
#                          │
#                          ▼
#              ┌─────────────────────────┐
#              │ Can we model a pattern? │
#              └─────────────────────────┘
#                          │
#                    regression
#                          │
#                          ▼
#              ┌─────────────────────────┐
#              │ What probability model? │
#              └─────────────────────────┘
#                          │
#                       NormalDist
#
#
# The most important mental model is:
#
#     CENTER
#       ↓
#     SPREAD
#       ↓
#     POSITION
#       ↓
#     RELATIONSHIP
#       ↓
#     MODEL
#
#
# =================================================================================================
# 02. IMPORTING THE LIBRARY
# =================================================================================================
#
# The module is part of Python's standard library.
#
# No installation is required.
#
#     import statistics
#
# Then functions can be accessed as:
#
#     statistics.mean(data)
#     statistics.median(data)
#     statistics.stdev(data)
#
# Or individual functions can be imported:
#
#     from statistics import mean, median, stdev
#
# For learning and clarity, this guide generally uses:
#
#     import statistics
#
#
# =================================================================================================
# 03. THE CORE STATISTICAL QUESTIONS
# =================================================================================================
#
# Imagine:
#
#     exam_scores = [55, 60, 65, 70, 75, 80, 95]
#
#
# We immediately want to know:
#
#     "What is typical?"
#              ↓
#           CENTER
#
#     "How much do values vary?"
#              ↓
#           SPREAD
#
#     "Where does a particular value sit?"
#              ↓
#           POSITION
#
#     "Do two variables move together?"
#              ↓
#         RELATIONSHIP
#
#
# These four ideas form much of practical descriptive statistics.
#
#
# =================================================================================================
# 04. MEAN — THE ARITHMETIC AVERAGE
# =================================================================================================
#
# FUNCTION:
#
#     statistics.mean(data)
#
# MEANING:
#
#     mean = sum of values / number of values
#
#
# Example:
#
#     data = [10, 20, 30]
#
#     mean = (10 + 20 + 30) / 3
#          = 60 / 3
#          = 20
#
#
# Visual:
#
#     10 -------- 20 -------- 30
#                    ▲
#                  mean
#
#
# The mean is a BALANCE POINT.
#
# Imagine placing the observations on a number line.
# The mean is the point where the data would balance if every observation
# had equal weight.
#
#
# -------------------------------------------------------------------------------------------------
# PRACTICAL EXAMPLE
# -------------------------------------------------------------------------------------------------
#
import statistics

scores = [10, 20, 30, 40, 50]

mean_score = statistics.mean(scores)

print(mean_score)
#
# Output:
#
#     30
#
#
# WHY?
#
#     10 + 20 + 30 + 40 + 50 = 150
#
#     150 / 5 = 30
#
#
# VALUE FLOW:
#
#     [10, 20, 30, 40, 50]
#                ↓
#             sum = 150
#                ↓
#           count = 5
#                ↓
#           150 / 5
#                ↓
#              30
#
#
# DA:
#     Very useful for numerical summaries and KPI analysis.
#
# DS:
#     Fundamental descriptive statistic.
#
# ML:
#     Important for understanding feature distributions and preprocessing.
#
#
# IMPORTANT:
#
# The mean is sensitive to extreme values.
#
#     [10, 20, 30, 40, 1000]
#
# The 1000 pulls the mean strongly upward.
#
#
# =================================================================================================
# 05. FMEAN — FAST FLOATING-POINT MEAN
# =================================================================================================
#
# FUNCTION:
#
#     statistics.fmean(data)
#
# fmean means "floating-point mean".
#
# It is designed specifically for numerical data and returns a float.
#
#     statistics.fmean([10, 20, 30])
#
#     → 20.0
#
#
# CONCEPT:
#
#     mean()
#       ↓
#     general arithmetic mean
#
#     fmean()
#       ↓
#     floating-point numerical mean
#
#
# `fmean()` also supports weights:
#
#     statistics.fmean(data, weights=weights)
#
#
# Example:
#
#     values = [10, 20, 30]
#     weights = [1, 2, 3]
#
# Weighted mean concept:
#
#     10×1 + 20×2 + 30×3
#     -------------------
#          1 + 2 + 3
#
#     = 140 / 6
#     = 23.333...
#
#
# VALUE FLOW:
#
#     value
#       ×
#     importance
#       ↓
#     weighted contribution
#       ↓
#     total weighted contribution
#       ↓
#     divide by total weight
#       ↓
#     weighted mean
#
#
# DA:
#     Useful when observations have different importance.
#
# DS:
#     Useful for weighted descriptive statistics.
#
# ML:
#     Useful conceptually when dealing with sample weights.
#
#
# =================================================================================================
# 06. WEIGHTED MEAN — THE CONCEPT
# =================================================================================================
#
# A normal mean treats every observation equally.
#
#     [10, 20, 30]
#
#     10 → equal importance
#     20 → equal importance
#     30 → equal importance
#
#
# A weighted mean says:
#
#     "Some observations count more than others."
#
#
# Example:
#
#     Product A: rating = 4, reviews = 1000
#     Product B: rating = 5, reviews = 10
#
# Treating both equally may not represent the overall rating well.
#
# Weighting allows:
#
#     observation
#          ↓
#     importance
#          ↓
#     contribution to average
#
#
# =================================================================================================
# 07. MEDIAN — THE MIDDLE VALUE
# =================================================================================================
#
# FUNCTION:
#
#     statistics.median(data)
#
# The median is the middle value after sorting the data.
#
#
# Example:
#
#     [10, 20, 30, 40, 50]
#
#                 30
#                  ▲
#                median
#
#
# For an odd number of observations:
#
#     [10, 20, 30, 40, 50]
#             ↓
#            30
#
#
# For an even number:
#
#     [10, 20, 30, 40]
#
# The two middle values are:
#
#     20 and 30
#
# Therefore:
#
#     median = (20 + 30) / 2
#            = 25
#
#
# VISUAL:
#
#     10 ---- 20 ---- 30 ---- 40
#              ↑       ↑
#           middle  middle
#                 ↓
#                25
#
#
# WHY MEDIAN MATTERS:
#
# The median is much less sensitive to extreme values than the mean.
#
#
# Example:
#
#     salaries = [30, 32, 35, 38, 500]
#
#     mean   = strongly pulled upward
#     median = 35
#
#
# For income, house prices, response times, etc.,
# median can often describe the "typical" observation better than mean.
#
#
# DA:
#     Extremely useful.
#
# DS:
#     Fundamental robust measure of center.
#
# ML:
#     Useful when inspecting skewed feature distributions.
#
#
# =================================================================================================
# 08. MEDIAN_LOW
# =================================================================================================
#
# FUNCTION:
#
#     statistics.median_low(data)
#
# For an even-sized dataset, it returns the LOWER middle value.
#
#
#     [10, 20, 30, 40]
#
#             ↑    ↑
#            20   30
#
#     median_low → 20
#
#
# This is different from:
#
#     median → 25
#
#
# WHY IT EXISTS:
#
# Sometimes you need an actual observed value rather than the average
# of the two middle values.
#
#
# =================================================================================================
# 09. MEDIAN_HIGH
# =================================================================================================
#
# FUNCTION:
#
#     statistics.median_high(data)
#
# For an even-sized dataset, it returns the HIGHER middle value.
#
#
#     [10, 20, 30, 40]
#
#             ↑    ↑
#            20   30
#
#     median_high → 30
#
#
# Therefore:
#
#     median_low  → 20
#     median      → 25
#     median_high → 30
#
#
# Think:
#
#                 middle pair
#                    ↓
#              [20] [30]
#                ↑     ↑
#              LOW   HIGH
#
#
# =================================================================================================
# 10. MEDIAN_GROUPED
# =================================================================================================
#
# FUNCTION:
#
#     statistics.median_grouped(data, interval=1)
#
# This is used when observations represent grouped or continuous values
# and an interpolated median is desired.
#
# It is more specialized than ordinary median().
#
#
# Conceptually:
#
#     ordinary median
#          ↓
#     choose middle observation
#
#     grouped median
#          ↓
#     estimate where the middle lies
#     inside a group/interval
#
#
# Example:
#
#     statistics.median_grouped(data, interval=10)
#
#
# The `interval` describes the width of the grouping interval.
#
# This function is useful for understanding grouped data,
# but it is much less frequently needed in everyday DA/DS/ML work
# than `median()`.
#
#
# =================================================================================================
# 11. MEAN VS MEDIAN
# =================================================================================================
#
# The most important distinction:
#
#
#     MEAN
#       ↓
#     uses every value
#       ↓
#     sensitive to extreme values
#
#
#     MEDIAN
#       ↓
#     uses position after sorting
#       ↓
#     resistant to extreme values
#
#
# Example:
#
#     data = [10, 10, 10, 10, 1000]
#
#
#     mean:
#
#     (10 + 10 + 10 + 10 + 1000) / 5
#                         ↓
#                        208
#
#
#     median:
#
#     [10, 10, 10, 10, 1000]
#             ↓
#             10
#
#
# The extreme value affects:
#
#     mean → heavily
#     median → very little
#
#
# GENERAL PATTERN:
#
#     Symmetric data
#          ↓
#       mean often useful
#
#     Strongly skewed data
#          ↓
#       median often useful
#
#
# =================================================================================================
# 12. MODE — MOST FREQUENT VALUE
# =================================================================================================
#
# FUNCTION:
#
#     statistics.mode(data)
#
# Mode answers:
#
#     "Which value occurs most frequently?"
#
#
# Example:
#
#     data = [10, 20, 20, 30, 20, 40]
#
#     20 occurs three times.
#
#     mode = 20
#
#
# VISUAL:
#
#     10 → █
#     20 → ███
#     30 → █
#     40 → █
#
#              ↑
#            mode
#
#
# Unlike mean and median, mode works naturally for categorical data
# when the category values can be compared for frequency.
#
# Example:
#
#     colors = ["red", "blue", "blue", "green", "blue"]
#
#     mode → "blue"
#
#
# DA:
#     Useful for categorical frequency summaries.
#
# DS:
#     Useful for exploratory analysis.
#
# ML:
#     Can be useful for understanding categorical variables,
#     although pandas/scikit-learn are usually used for actual preprocessing.
#
#
# =================================================================================================
# 13. MULTIMODE
# =================================================================================================
#
# FUNCTION:
#
#     statistics.multimode(data)
#
# Sometimes several values are tied for highest frequency.
#
#
#     [10, 10, 20, 20, 30]
#
#     10 → 2 occurrences
#     20 → 2 occurrences
#     30 → 1 occurrence
#
#
# Therefore:
#
#     multimode(data)
#
#     → [10, 20]
#
#
# CONCEPT:
#
#     one most frequent value
#             ↓
#           mode()
#
#     multiple equally frequent values
#             ↓
#         multimode()
#
#
# =================================================================================================
# 14. QUANTILES — POSITION INSIDE A DISTRIBUTION
# =================================================================================================
#
# FUNCTION:
#
#     statistics.quantiles(data, n=4, method='exclusive')
#
# Quantiles divide ordered data into sections.
#
#
# Think:
#
#     smallest
#        │
#        ▼
#     ─────────────────────────────────────────
#        │          │          │          │
#        Q1         Q2         Q3       maximum
#
#
# For quartiles:
#
#     n = 4
#
# gives three cut points:
#
#     Q1 = 25th percentile
#     Q2 = 50th percentile
#     Q3 = 75th percentile
#
#
# VISUAL:
#
#     0%          25%          50%          75%         100%
#     │------------│------------│------------│------------│
#                  Q1           Q2           Q3
#
#
# Example:
#
data = [10, 20, 30, 40, 50, 60, 70, 80]
#
quartiles = statistics.quantiles(data, n=4)
#
print(quartiles)
#
#
# IMPORTANT:
#
# `quantiles()` returns the internal cut points,
# not the minimum and maximum.
#
#
# WHY QUANTILES MATTER:
#
# They tell us WHERE observations sit in a distribution.
#
# This is extremely important in:
#
#     DA
#     ↓
#     customer segmentation
#
#     DS
#     ↓
#     distribution analysis
#
#     ML
#     ↓
#     feature distribution inspection
#
#
# =================================================================================================
# 15. PERCENTILES AND QUANTILES
# =================================================================================================
#
# A percentile answers:
#
#     "What value separates a certain percentage of observations?"
#
#
# Examples:
#
#     50th percentile → median
#     25th percentile → Q1
#     75th percentile → Q3
#
#
# GENERAL IDEA:
#
#     25% of observations
#     │
#     ▼
#     Q1
#
#     50%
#     │
#     ▼
#     median
#
#     75%
#     │
#     ▼
#     Q3
#
#
# =================================================================================================
# 16. QUANTILE METHODS
# =================================================================================================
#
# `statistics.quantiles()` supports different definitions of how quantiles
# are interpolated.
#
# Common methods include:
#
#     method='exclusive'
#     method='inclusive'
#
#
# The important conceptual point is:
#
# Different statistical conventions can produce slightly different
# percentile boundaries.
#
# Therefore, when reporting quantiles in research or analytics,
# the chosen convention should be known and documented when precision matters.
#
#
# =================================================================================================
# 17. RANGE — IMPORTANT CONCEPT, BUT NOT A statistics FUNCTION
# =================================================================================================
#
# The `statistics` module does NOT provide a dedicated `range()` statistical
# function.
#
# But the range is fundamental:
#
#     range = maximum - minimum
#
#
# Example:
#
#     [10, 20, 30, 50]
#
#     max = 50
#     min = 10
#
#     range = 50 - 10
#           = 40
#
#
# Python:
#
data = [10, 20, 30, 50]
range_value = max(data) - min(data)
#
#
# RANGE tells us the total width of the observed values.
#
# It is simple but very sensitive to extreme values.
#
#
# =================================================================================================
# 18. VARIANCE — HOW MUCH VALUES VARY
# =================================================================================================
#
# FUNCTIONS:
#
#     statistics.variance(data)
#     statistics.pvariance(data)
#
#
# Variance measures average SQUARED deviation from the mean.
#
#
# Start with:
#
#     data
#       ↓
#     mean
#       ↓
#     distance of each value from mean
#       ↓
#     square the distances
#       ↓
#     average them
#       ↓
#     variance
#
#
# Example:
#
#     data = [10, 20, 30]
#
#     mean = 20
#
#
# Deviations:
#
#     10 - 20 = -10
#     20 - 20 =   0
#     30 - 20 = +10
#
#
# Squared deviations:
#
#     (-10)² = 100
#     (  0)² =   0
#     (+10)² = 100
#
#
# The square removes negative signs:
#
#     100 + 0 + 100
#     = 200
#
#
# Population variance:
#
#     200 / 3
#     = 66.666...
#
#
# Sample variance:
#
#     200 / (3 - 1)
#     = 100
#
#
# This distinction is CRITICAL.
#
#
# =================================================================================================
# 19. SAMPLE VS POPULATION
# =================================================================================================
#
# This is one of the most important statistical concepts in Python.
#
#
# POPULATION:
#
#     You have ALL observations you care about.
#
#     Example:
#
#     every machine produced by one factory this week
#
#
# SAMPLE:
#
#     You have a subset and want to estimate the population.
#
#     Example:
#
#     100 machines selected from 100,000 machines
#
#
# Therefore:
#
#     pvariance()
#         ↓
#     population variance
#
#     variance()
#         ↓
#     sample variance
#
#
# Mental model:
#
#     POPULATION
#     ┌─────────────────────────────┐
#     │ ● ● ● ● ● ● ● ● ● ● ● ● ● │
#     └─────────────────────────────┘
#                  ↑
#               all data
#
#
#     SAMPLE
#     ┌─────────────────────────────┐
#     │ ● ● ● ● ● ● ● ● ● ● ● ● ● │
#     └─────────────────────────────┘
#             ↑   ↑   ↑
#          selected observations
#
#
# Sample variance uses n - 1 in the denominator.
#
# This is associated with Bessel's correction.
#
#
# =================================================================================================
# 20. PVARIANCE
# =================================================================================================
#
# FUNCTION:
#
#     statistics.pvariance(data)
#
# Use it when your data represents the complete population of interest.
#
#
# Example:
#
data = [10, 20, 30]
population_variance = statistics.pvariance(data)
#
#
# CONCEPT:
#
#     deviations
#          ↓
#     squared deviations
#          ↓
#     divide by N
#          ↓
#     population variance
#
#
# =================================================================================================
# 21. VARIANCE
# =================================================================================================
#
# FUNCTION:
#
#     statistics.variance(data)
#
# Use it for SAMPLE variance.
#
#
# CONCEPT:
#
#     deviations
#          ↓
#     squared deviations
#          ↓
#     divide by n - 1
#          ↓
#     sample variance
#
#
# GENERAL RULE:
#
#     pvariance → population
#     variance  → sample
#
#
# =================================================================================================
# 22. STANDARD DEVIATION
# =================================================================================================
#
# FUNCTIONS:
#
#     statistics.stdev(data)
#     statistics.pstdev(data)
#
#
# Standard deviation is:
#
#     square root of variance
#
#
# Why do we need it?
#
# Variance uses squared units.
#
# If the original data is measured in:
#
#     meters
#
# variance is measured in:
#
#     meters²
#
# Standard deviation returns to:
#
#     meters
#
#
# VISUAL:
#
#     variance
#         ↓
#     square root
#         ↓
#     standard deviation
#
#
# =================================================================================================
# 23. STDEV
# =================================================================================================
#
# FUNCTION:
#
#     statistics.stdev(data)
#
# Calculates SAMPLE standard deviation.
#
#
# Concept:
#
#     sample data
#          ↓
#     sample variance
#          ↓
#     √
#          ↓
#     sample standard deviation
#
#
# Example:
#
data = [10, 20, 30]
sample_sd = statistics.stdev(data)
#
#
# =================================================================================================
# 24. PSTDEV
# =================================================================================================
#
# FUNCTION:
#
#     statistics.pstdev(data)
#
# Calculates POPULATION standard deviation.
#
#
# Therefore:
#
#     pstdev → population
#     stdev  → sample
#
#
# =================================================================================================
# 25. VARIANCE VS STANDARD DEVIATION
# =================================================================================================
#
# Suppose:
#
#     mean = 50
#     standard deviation = 5
#
#
# A useful mental model is:
#
#     mean
#       │
#       │
#     45 ───────── approximately 1 SD below
#       │
#     50 ───────── mean
#       │
#     55 ───────── approximately 1 SD above
#
#
# Standard deviation gives a sense of the typical scale of deviations
# around the mean.
#
# It does NOT mean every observation lies within one standard deviation.
#
#
# =================================================================================================
# 26. DESCRIPTIVE STATISTICS AS A SYSTEM
# =================================================================================================
#
# These functions should NOT be memorized as unrelated functions.
#
#
#                         DATA
#                          │
#             ┌────────────┼────────────┐
#             ↓            ↓            ↓
#           CENTER       SPREAD       POSITION
#             │            │            │
#      mean/median/mode  variance     quantiles
#                       stdev
#
#
# CENTER:
#
#     "What is typical?"
#
# SPREAD:
#
#     "How much variation exists?"
#
# POSITION:
#
#     "Where is a value relative to the distribution?"
#
#
# This is the core of descriptive analysis.
#
#
# =================================================================================================
# 27. GEOMETRIC MEAN
# =================================================================================================
#
# FUNCTION:
#
#     statistics.geometric_mean(data)
#
# The geometric mean is based on MULTIPLICATION rather than addition.
#
#
# For positive values:
#
#     geometric mean
#
#     = (x1 × x2 × ... × xn)^(1/n)
#
#
# Why is this useful?
#
# It is especially meaningful for:
#
#     growth rates
#     multiplicative changes
#     ratios
#     investment returns
#
#
# Example:
#
# Suppose something grows:
#
#     +10%
#     +20%
#
# You should not simply average the percentages in every context,
# because growth compounds multiplicatively.
#
#
# Concept:
#
#     value
#       ↓
#     multiplicative effect
#       ↓
#     geometric mean
#
#
# IMPORTANT:
#
# Geometric mean is intended for appropriate positive numeric data.
#
#
# DA:
#     Useful for growth-rate analysis.
#
# DS:
#     Useful when data has multiplicative structure.
#
# ML:
#     Usually less central than arithmetic mean, but conceptually useful
#     for multiplicative feature behavior.
#
#
# =================================================================================================
# 28. HARMONIC MEAN
# =================================================================================================
#
# FUNCTION:
#
#     statistics.harmonic_mean(data)
#
# Harmonic mean is particularly useful for averaging RATES.
#
#
# Example:
#
#     speed = distance / time
#
# When the same distance is traveled at different speeds,
# the ordinary arithmetic mean of speeds can be misleading.
#
#
# Harmonic mean:
#
#     n
#     ---------------------------
#     1/x1 + 1/x2 + ... + 1/xn
#
#
# Mental model:
#
#     arithmetic mean
#         ↓
#     "average values"
#
#     harmonic mean
#         ↓
#     "average rates/ratios in appropriate situations"
#
#
# A famous relationship:
#
#     arithmetic mean ≥ geometric mean ≥ harmonic mean
#
# for appropriate positive data.
#
#
# =================================================================================================
# 29. MEAN TYPES — ONE CONCEPTUAL MAP
# =================================================================================================
#
#
#                 AVERAGE
#                    │
#       ┌────────────┼────────────┐
#       │            │            │
#       ▼            ▼            ▼
#     MEAN       GEOMETRIC     HARMONIC
#                  MEAN          MEAN
#
#
# MEAN:
#     additive quantities
#
# GEOMETRIC MEAN:
#     multiplicative changes / growth
#
# HARMONIC MEAN:
#     rates / ratios in appropriate contexts
#
#
# `fmean()`:
#     numerical floating-point mean, also supports weights
#
#
# =================================================================================================
# 30. COVARIANCE — HOW TWO VARIABLES MOVE TOGETHER
# =================================================================================================
#
# FUNCTION:
#
#     statistics.covariance(x, y)
#
# Covariance asks:
#
#     "When X changes, does Y tend to change in the same direction
#      or the opposite direction?"
#
#
# Imagine:
#
#     X ↑
#     Y ↑
#
#     → positive covariance
#
#
#     X ↑
#     Y ↓
#
#     → negative covariance
#
#
#     X changes
#     Y has no consistent relationship
#
#     → covariance near zero
#
#
# IMPORTANT:
#
# Covariance's numerical magnitude depends on the units of the variables.
#
# Therefore covariance is useful for direction,
# but correlation is usually easier to interpret for strength.
#
#
# =================================================================================================
# 31. COVARIANCE VISUAL
# =================================================================================================
#
# Positive relationship:
#
#     Y
#     │             ●
#     │          ●
#     │       ●
#     │    ●
#     │ ●
#     └──────────────── X
#
#
# Negative relationship:
#
#     Y
#     │ ●
#     │    ●
#     │       ●
#     │          ●
#     │             ●
#     └──────────────── X
#
#
# No obvious linear relationship:
#
#     Y
#     │    ●      ●
#     │       ●
#     │ ●        ●
#     │      ●
#     │         ●
#     └──────────────── X
#
#
# =================================================================================================
# 32. CORRELATION
# =================================================================================================
#
# FUNCTION:
#
#     statistics.correlation(x, y)
#
# Correlation measures the strength and direction of association.
#
#
# For Pearson correlation:
#
#     -1 ≤ r ≤ +1
#
#
#     r = +1
#         ↓
#     perfect positive linear relationship
#
#
#     r = 0
#         ↓
#     no linear correlation
#
#
#     r = -1
#         ↓
#     perfect negative linear relationship
#
#
# Visual:
#
#     -1             0             +1
#      │-------------│-------------│
#      ↓             ↓             ↓
#    negative      none        positive
#
#
# =================================================================================================
# 33. CORRELATION IS NOT CAUSATION
# =================================================================================================
#
# This is one of the most important rules in Data Science.
#
#
#     X
#      \
#       → correlation ←
#      /
#     Y
#
#
# A high correlation does NOT prove:
#
#     X causes Y
#
# It only indicates that the variables have a statistical relationship
# according to the correlation measure.
#
#
# Possible explanation:
#
#     X ← Z → Y
#
# A third variable Z may influence both.
#
#
# DA:
#     Extremely useful for exploratory analysis.
#
# DS:
#     Essential for feature understanding.
#
# ML:
#     Useful for feature analysis and detecting strongly related variables.
#
# BUT:
#
# Correlation analysis alone should not be used as a feature-selection
# strategy without understanding the modeling problem.
#
#
# =================================================================================================
# 34. PEARSON CORRELATION
# =================================================================================================
#
# `statistics.correlation(x, y)` uses linear correlation by default.
#
# Concept:
#
#     How closely do the points follow a straight-line relationship?
#
#
# Strong positive:
#
#     ●
#       ●
#         ●
#           ●
#             ●
#
# Strong negative:
#
#             ●
#           ●
#         ●
#       ●
#     ●
#
#
# =================================================================================================
# 35. RANK CORRELATION
# =================================================================================================
#
# `statistics.correlation()` can also support rank-based correlation
# in Python versions that provide the rank method.
#
# Conceptually:
#
#     original values
#          ↓
#     replace values with ranks
#          ↓
#     compare rank movement
#
#
# This is useful when the relationship is monotonic but not necessarily
# linear.
#
#
# Example concept:
#
#     X increases
#     Y generally increases
#
# but the relationship may curve rather than form a straight line.
#
#
# Always check the Python version's documentation for the exact supported
# `method` options when writing version-specific code.
#
#
# =================================================================================================
# 36. COVARIANCE → CORRELATION
# =================================================================================================
#
# These two concepts are closely related:
#
#
#     covariance
#          ↓
#     direction of joint variation
#          ↓
#     normalize by individual variability
#          ↓
#     correlation
#          ↓
#     standardized relationship from -1 to +1
#
#
# Therefore:
#
#     covariance → raw joint variation
#     correlation → standardized association
#
#
# =================================================================================================
# 37. LINEAR REGRESSION
# =================================================================================================
#
# FUNCTION:
#
#     statistics.linear_regression(x, y)
#
# This calculates a simple linear regression relationship:
#
#     y = slope × x + intercept
#
#
# The model tries to find a straight line that describes the relationship
# between X and Y.
#
#
# VISUAL:
#
#     Y
#     │                 ●
#     │             ●
#     │          ●
#     │       ●
#     │    ●
#     │  ───────────────── regression line
#     │
#     └──────────────────────── X
#
#
# CONCEPT:
#
#     X
#     ↓
#     linear model
#     ↓
#     predicted Y
#
#
# =================================================================================================
# 38. LINEAR REGRESSION OUTPUT
# =================================================================================================
#
# Example:
#
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
#
regression = statistics.linear_regression(x, y)
#
print(regression)
#
#
# The result provides:
#
#     slope
#     intercept
#
#
# Here the relationship is:
#
#     y = 2x + 0
#
#
# VALUE FLOW:
#
#     x
#     ↓
#     regression analysis
#     ↓
#     slope + intercept
#     ↓
#     equation
#     ↓
#     predicted y
#
#
# =================================================================================================
# 39. SLOPE — WHAT DOES IT MEAN?
# =================================================================================================
#
# Suppose:
#
#     y = 2x + 5
#
#
# The slope is:
#
#     2
#
#
# Meaning:
#
#     when x increases by 1,
#     the predicted y increases by 2.
#
#
# VISUAL:
#
#     Δy
#     ↑
#     │       /
#     │      /
#     │     /
#     │    /
#     │___/________→ Δx
#
#
# slope = Δy / Δx
#
#
# =================================================================================================
# 40. INTERCEPT
# =================================================================================================
#
# In:
#
#     y = mx + b
#
#     m = slope
#     b = intercept
#
#
# The intercept is the predicted y value when:
#
#     x = 0
#
#
# Example:
#
#     y = 2x + 5
#
#     x = 0
#
#     y = 5
#
#
# Therefore:
#
#     intercept = 5
#
#
# =================================================================================================
# 41. PROPORTIONAL LINEAR REGRESSION
# =================================================================================================
#
# `linear_regression()` supports:
#
#     proportional=True
#
# Conceptually, this forces:
#
#     intercept = 0
#
#
# Therefore the model becomes:
#
#     y = slope × x
#
#
# This is appropriate only when theory or the problem requires the line
# to pass through the origin.
#
#
# IMPORTANT:
#
# Do NOT force an intercept of zero merely because it looks simpler.
#
# It changes the model.
#
#
# =================================================================================================
# 42. LINEAR REGRESSION VS CORRELATION
# =================================================================================================
#
#
# CORRELATION:
#
#     "How strongly and in what direction are X and Y linearly associated?"
#
#
# REGRESSION:
#
#     "What line can we use to describe/predict Y from X?"
#
#
# Relationship:
#
#     correlation
#          ↓
#     association strength
#
#     regression
#          ↓
#     predictive equation
#
#
# Example:
#
#     correlation → r = 0.9
#
#     regression → y = 3.2x + 4.1
#
#
# They answer related but different questions.
#
#
# =================================================================================================
# 43. NORMAL DISTRIBUTION
# =================================================================================================
#
# The `statistics` module contains a particularly useful class:
#
#     statistics.NormalDist
#
#
# It represents a normal (Gaussian) probability distribution.
#
#
# VISUAL:
#
#                    /\
#                  /    \
#                /        \
#              /            \
#            /                \
#     ______/__________________\______
#
#                   mean
#
#
# A normal distribution is described by:
#
#     mean
#     standard deviation
#
#
#     NormalDist(mean, stdev)
#
#
# =================================================================================================
# 44. CREATING A NORMAL DISTRIBUTION
# =================================================================================================
#
# Example:
#
from statistics import NormalDist
#
dist = NormalDist(mu=100, sigma=15)
#
#
# Mental model:
#
#     mean = 100
#     standard deviation = 15
#
#
#     70    85    100    115    130
#      │     │      │      │      │
#      -2σ   -1σ   mean   +1σ   +2σ
#
#
# `mu` represents the mean.
#
# `sigma` represents the standard deviation.
#
#
# =================================================================================================
# 45. NORMAL DISTRIBUTION PDF
# =================================================================================================
#
# METHOD:
#
#     dist.pdf(x)
#
# PDF = probability density function.
#
#
# IMPORTANT:
#
# For a continuous distribution, `pdf(x)` is NOT the probability that
# X equals exactly x.
#
# It describes the density around x.
#
#
# Visual:
#
#                    /\
#                   /  \
#                  /    \
#                 /      \
#     ___________/________\___________
#                     ↑
#                    x
#
#
# Higher density:
#
#     observations are more concentrated around that region.
#
#
# =================================================================================================
# 46. NORMAL DISTRIBUTION CDF
# =================================================================================================
#
# METHOD:
#
#     dist.cdf(x)
#
# CDF = cumulative distribution function.
#
#
# It answers:
#
#     "What proportion/probability is at or below x?"
#
#
# Visual:
#
#     probability accumulated
#
#     0% ─────────────────────────────── 100%
#                    ↑
#                   x
#
#
# Example:
#
#     dist.cdf(100)
#
# For a normal distribution centered at 100,
# approximately half the distribution lies below 100.
#
#
# Therefore:
#
#     CDF = cumulative probability
#
#
# =================================================================================================
# 47. PDF VS CDF
# =================================================================================================
#
#
# PDF:
#
#     "How dense is the distribution around x?"
#
#
# CDF:
#
#     "How much probability has accumulated up to x?"
#
#
# Think:
#
#     PDF
#       ↓
#     local shape/density
#
#
#     CDF
#       ↓
#     accumulated probability
#
#
# This distinction is extremely important in probability.
#
#
# =================================================================================================
# 48. INVERSE CDF
# =================================================================================================
#
# METHOD:
#
#     dist.inv_cdf(p)
#
# It performs the reverse operation of `cdf()`.
#
#
# CDF:
#
#     x
#     ↓
#     probability
#
#
# INV_CDF:
#
#     probability
#     ↓
#     x
#
#
# Example concept:
#
#     What value marks the 95th percentile?
#
#     p = 0.95
#          ↓
#     inv_cdf(0.95)
#          ↓
#     threshold value
#
#
# This is useful for:
#
#     percentiles
#     thresholds
#     confidence calculations
#     probability-based decisions
#
#
# =================================================================================================
# 49. QUANTILES OF NORMAL DISTRIBUTION
# =================================================================================================
#
# METHOD:
#
#     dist.quantiles(n=4)
#
# This produces cut points dividing the normal distribution into
# equal-probability regions.
#
#
# For quartiles:
#
#     n = 4
#
# gives:
#
#     25%
#     50%
#     75%
#
#
# This connects:
#
#     NormalDist
#          ↓
#     probability distribution
#          ↓
#     quantiles
#
#
# =================================================================================================
# 50. Z-SCORE
# =================================================================================================
#
# METHOD:
#
#     dist.zscore(x)
#
# A z-score tells us how many standard deviations a value is from
# the distribution mean.
#
#
# Formula:
#
#     z = (x - μ) / σ
#
#
# Example:
#
#     mean = 100
#     stdev = 15
#     x = 130
#
#
#     z = (130 - 100) / 15
#
#       = 30 / 15
#
#       = 2
#
#
# Therefore:
#
#     130 is 2 standard deviations above the mean.
#
#
# VISUAL:
#
#     70       85       100       115       130
#      │        │         │         │         │
#     -2σ      -1σ       μ        +1σ       +2σ
#                                           ↑
#                                           x
#
#
# Z-score is one of the most useful conceptual bridges between
# statistics and machine learning.
#
#
# =================================================================================================
# 51. Z-SCORE AS STANDARDIZATION
# =================================================================================================
#
# The general transformation is:
#
#     original value
#          ↓
#     subtract mean
#          ↓
#     divide by standard deviation
#          ↓
#     standardized value
#
#
# This changes the scale to:
#
#     mean ≈ 0
#     standard deviation ≈ 1
#
#
# Example:
#
#     x = 130
#     mean = 100
#     sd = 15
#
#     z = 2
#
#
# ML CONNECTION:
#
# Standardization is important for many algorithms, including methods
# sensitive to feature scale.
#
# Examples include:
#
#     logistic regression
#     linear regression in many workflows
#     SVM
#     k-nearest neighbors
#     PCA
#     neural networks
#
# The actual preprocessing should normally be performed with tools such as
# scikit-learn's StandardScaler rather than manually for production ML.
#
#
# =================================================================================================
# 52. NORMAL DISTRIBUTION OVERLAP
# =================================================================================================
#
# METHOD:
#
#     dist.overlap(other)
#
# This calculates the overlap between two normal distributions.
#
#
# Concept:
#
#     Distribution A
#            /\
#           /  \
#          /    \
#         /      \
#
#             /\       Distribution B
#            /  \
#           /    \
#
#
# The more their probability mass overlaps,
# the greater the overlap measure.
#
#
# This is a specialized method and is much less frequently needed
# than mean, median, standard deviation, correlation, and regression.
#
#
# =================================================================================================
# 53. NORMAL DISTRIBUTION SAMPLING
# =================================================================================================
#
# METHOD:
#
#     dist.samples(n)
#
# Generates random samples from the normal distribution.
#
#
# Concept:
#
#     NormalDist(mean, stdev)
#                ↓
#             sampling
#                ↓
#     random observations
#
#
# Example:
#
dist = NormalDist(100, 15)
samples = dist.samples(10)
#
#
# The generated values will vary because sampling is random.
#
# Therefore, do NOT expect a fixed list of values.
#
#
# ML / DS:
#
# Useful for:
#
#     simulation
#     demonstrations
#     probabilistic experiments
#     teaching
#
# For serious random-number generation and simulation workflows,
# NumPy is usually more appropriate.
#
#
# =================================================================================================
# 54. NORMALDIST.FROM_SAMPLES
# =================================================================================================
#
# CLASS METHOD:
#
#     NormalDist.from_samples(data)
#
# It estimates a normal distribution from sample data.
#
#
# Data:
#
#     [95, 102, 98, 110, 105, ...]
#              ↓
#       estimate mean
#              ↓
#       estimate standard deviation
#              ↓
#          NormalDist
#
#
# Conceptually:
#
#     observed data
#          ↓
#     estimate parameters
#          ↓
#     fitted normal distribution
#
#
# This is useful for learning how data can be summarized by a probability
# distribution.
#
#
# =================================================================================================
# 55. NORMALDIST.FROM_MEAN_SD
# =================================================================================================
#
# Depending on Python version, `NormalDist` provides construction helpers
# for creating a distribution from a mean and standard deviation.
#
# The fundamental concept remains:
#
#     mean
#       +
#     standard deviation
#       ↓
#     NormalDist
#
#
# For version-specific code, check the Python documentation for the exact
# available constructor/class-method names in your installed version.
#
#
# =================================================================================================
# 56. NORMALDIST MEAN
# =================================================================================================
#
# ATTRIBUTE:
#
#     dist.mean
#
# Gives the mean of the normal distribution.
#
#
# Example:
#
dist = NormalDist(100, 15)
print(dist.mean)
#
#     → 100.0
#
#
# =================================================================================================
# 57. NORMALDIST STDEV
# =================================================================================================
#
# ATTRIBUTE:
#
#     dist.stdev
#
# Gives the standard deviation.
#
#
# Example:
#
print(dist.stdev)
#
#     → 15.0
#
#
# =================================================================================================
# 58. NORMALDIST VARIANCE
# =================================================================================================
#
# ATTRIBUTE:
#
#     dist.variance
#
# Variance is:
#
#     stdev²
#
#
# Example:
#
#     stdev = 15
#
#     variance = 15²
#              = 225
#
#
# Relationship:
#
#     stdev
#       ↓ square
#     variance
#
#     variance
#       ↓ square root
#     stdev
#
#
# =================================================================================================
# 59. NORMALDIST OBJECT MODEL
# =================================================================================================
#
# A NormalDist object combines:
#
#     PARAMETERS
#         │
#         ├── mean
#         └── stdev
#
#     with OPERATIONS
#         │
#         ├── pdf()
#         ├── cdf()
#         ├── inv_cdf()
#         ├── quantiles()
#         ├── zscore()
#         ├── samples()
#         └── overlap()
#
#
# This is an important Python idea:
#
#     object
#       ↓
#     stores related state
#       +
#     provides related behavior
#
#
# =================================================================================================
# 60. NORMAL DISTRIBUTION — THE 68–95–99.7 RULE
# =================================================================================================
#
# For an approximately normal distribution:
#
#
#             about 68%
#         ┌───────────────┐
#         │               │
#     ────┤────── μ ──────├────
#        -1σ             +1σ
#
#
#             about 95%
#      ┌───────────────────────┐
#      │                       │
#   ───┤────────── μ ──────────├───
#     -2σ                    +2σ
#
#
#             about 99.7%
#   ┌───────────────────────────────┐
#   │                               │
# ──┤────────────── μ ──────────────├──
#  -3σ                            +3σ
#
#
# This is a conceptual rule, not a universal law for arbitrary data.
#
#
# =================================================================================================
# 61. STATISTICS LIBRARY — COMPLETE FUNCTION MAP
# =================================================================================================
#
# A practical map of the major functionality:
#
#
#     CENTER
#     ├── mean()
#     ├── fmean()
#     ├── median()
#     ├── median_low()
#     ├── median_high()
#     ├── median_grouped()
#     ├── mode()
#     └── multimode()
#
#
#     GEOMETRIC / RATE AVERAGES
#     ├── geometric_mean()
#     └── harmonic_mean()
#
#
#     POSITION
#     └── quantiles()
#
#
#     SPREAD
#     ├── variance()
#     ├── pvariance()
#     ├── stdev()
#     └── pstdev()
#
#
#     RELATIONSHIPS
#     ├── covariance()
#     └── correlation()
#
#
#     MODELING
#     └── linear_regression()
#
#
#     PROBABILITY DISTRIBUTION
#     └── NormalDist
#           ├── pdf()
#           ├── cdf()
#           ├── inv_cdf()
#           ├── quantiles()
#           ├── zscore()
#           ├── samples()
#           └── overlap()
#
#
# =================================================================================================
# 62. STATISTICS LIBRARY — WHAT YOU REALLY NEED TO MASTER
# =================================================================================================
#
# You do NOT need to give every function equal learning priority.
#
#
# TIER 1 — MUST KNOW FOR DA / DS / ML
# ─────────────────────────────────────
#
#     mean()
#     fmean()
#     median()
#     mode()
#     multimode()
#     quantiles()
#     variance()
#     pvariance()
#     stdev()
#     pstdev()
#     covariance()
#     correlation()
#
#
# These form the practical descriptive-statistics foundation.
#
#
# TIER 2 — IMPORTANT CONCEPTUALLY
# ─────────────────────────────────
#
#     geometric_mean()
#     harmonic_mean()
#     linear_regression()
#     NormalDist
#     NormalDist.cdf()
#     NormalDist.inv_cdf()
#     NormalDist.zscore()
#
#
# TIER 3 — SPECIALIZED
# ─────────────────────
#
#     median_low()
#     median_high()
#     median_grouped()
#     NormalDist.pdf()
#     NormalDist.quantiles()
#     NormalDist.overlap()
#     NormalDist.samples()
#
#
# Learn what they mean and recognize when they are useful.
# They do not deserve the same practice time as mean, median,
# standard deviation, quantiles, and correlation.
#
#
# =================================================================================================
# 63. DATA ANALYSIS (DA) — MOST USEFUL FUNCTIONS
# =================================================================================================
#
# For Data Analysis, prioritize:
#
#
#     1. mean()
#     2. median()
#     3. mode()
#     4. multimode()
#     5. quantiles()
#     6. stdev()
#     7. variance()
#     8. correlation()
#     9. covariance()
#    10. fmean()
#
#
# Typical DA workflow:
#
#
#     DATA
#       ↓
#     inspect center
#       ↓
#     mean / median / mode
#       ↓
#     inspect spread
#       ↓
#     stdev / variance
#       ↓
#     inspect distribution position
#       ↓
#     quantiles
#       ↓
#     inspect relationships
#       ↓
#     correlation / covariance
#       ↓
#     BUSINESS INTERPRETATION
#
#
# Example business question:
#
#     "What is the typical customer purchase?"
#
#         ↓
#     mean / median
#
#     "How variable are purchases?"
#
#         ↓
#     stdev / quantiles
#
#     "Which variables are related to sales?"
#
#         ↓
#     correlation
#
#
# =================================================================================================
# 64. DATA SCIENCE (DS) — MOST USEFUL FUNCTIONS
# =================================================================================================
#
# Data Science requires a broader statistical mindset.
#
# Prioritize:
#
#     mean
#     median
#     quantiles
#     variance
#     stdev
#     covariance
#     correlation
#     linear_regression
#     NormalDist
#
#
# DS workflow:
#
#
#     raw data
#        ↓
#     distribution
#        ↓
#     center + spread
#        ↓
#     relationships
#        ↓
#     assumptions
#        ↓
#     statistical model
#        ↓
#     prediction / inference
#
#
# The statistics module gives you the fundamental building blocks,
# but advanced DS requires SciPy, statsmodels, NumPy, pandas,
# scikit-learn, and often specialized libraries.
#
#
# =================================================================================================
# 65. MACHINE LEARNING (ML) — WHERE statistics HELPS
# =================================================================================================
#
# The statistics module is NOT an ML library.
#
# However, it teaches concepts that are fundamental to ML.
#
#
# IMPORTANT CONNECTIONS:
#
#
#     mean
#       ↓
#     feature center
#
#     stdev
#       ↓
#     feature spread
#
#     quantiles
#       ↓
#     feature distribution / outliers
#
#     correlation
#       ↓
#     relationship between features
#
#     covariance
#       ↓
#     joint variation
#
#     z-score
#       ↓
#     standardization concept
#
#     linear_regression
#       ↓
#     basic predictive modeling
#
#     NormalDist
#       ↓
#     probability/distribution reasoning
#
#
# =================================================================================================
# 66. CORRELATION AND FEATURE ENGINEERING
# =================================================================================================
#
# Suppose:
#
#     X1 = house size
#     X2 = number of rooms
#     X3 = price
#
#
# You may calculate:
#
#     correlation(X1, X3)
#     correlation(X2, X3)
#
#
# Conceptual matrix:
#
#
#                 size    rooms    price
#
#     size         1.00    0.85     0.78
#
#     rooms        0.85    1.00     0.70
#
#     price        0.78    0.70     1.00
#
#
# This can reveal:
#
#     strong relationship between size and rooms
#     strong relationship between size and price
#
#
# But:
#
#     correlation ≠ causation
#     correlation ≠ guaranteed predictive importance
#
#
# A feature can have low marginal correlation with a target
# and still be useful to a nonlinear ML model.
#
#
# =================================================================================================
# 67. OUTLIERS — WHY MEAN AND STDEV CAN CHANGE
# =================================================================================================
#
# Consider:
#
#     data A:
#
#     [10, 11, 12, 13, 14]
#
#
# versus:
#
#     data B:
#
#     [10, 11, 12, 13, 1000]
#
#
# The extreme observation:
#
#     ↑
#     pulls mean
#     ↑
#     increases variance
#     ↑
#     increases standard deviation
#
#
# Therefore:
#
#     mean + stdev
#
# should often be interpreted together with:
#
#     median + quantiles
#
#
# This is a very important DA/DS habit.
#
#
# =================================================================================================
# 68. A PRACTICAL DISTRIBUTION PROFILE
# =================================================================================================
#
# Instead of asking only:
#
#     "What is the average?"
#
# Ask:
#
#
#     CENTER
#     ├── mean
#     ├── median
#     └── mode
#
#     SPREAD
#     ├── range
#     ├── variance
#     └── standard deviation
#
#     POSITION
#     └── quantiles
#
#     RELATIONSHIP
#     ├── covariance
#     └── correlation
#
#
# This gives a much more complete picture.
#
#
# =================================================================================================
# 69. A SMALL PRACTICAL EXPLORATION
# =================================================================================================
#
# Let's combine several concepts.
#
sales = [100, 120, 130, 150, 160, 170, 500]
#
mean_sales = statistics.mean(sales)
median_sales = statistics.median(sales)
sd_sales = statistics.stdev(sales)
quartiles = statistics.quantiles(sales, n=4)
#
print("Mean:", mean_sales)
print("Median:", median_sales)
print("Standard deviation:", sd_sales)
print("Quartiles:", quartiles)
#
#
# Think:
#
#     sales
#       ↓
#     ┌─────────────────────────────┐
#     │ center                      │
#     │ mean / median               │
#     ├─────────────────────────────┤
#     │ spread                      │
#     │ standard deviation          │
#     ├─────────────────────────────┤
#     │ position                    │
#     │ quartiles                   │
#     └─────────────────────────────┘
#
#
# Notice the 500.
#
# It will strongly influence the mean and standard deviation.
#
# Therefore comparing:
#
#     mean
#     median
#
# immediately gives useful information about skew/outliers.
#
#
# =================================================================================================
# 70. A PRACTICAL CORRELATION EXAMPLE
# =================================================================================================
#
# Suppose:
#
#     study_hours = [1, 2, 3, 4, 5]
#     exam_scores = [50, 55, 65, 75, 85]
#
#
# Conceptually:
#
#     study_hours ↑
#            │
#            │
#            ▼
#     exam_scores ↑
#
#
# Calculate:
#
study_hours = [1, 2, 3, 4, 5]
exam_scores = [50, 55, 65, 75, 85]
#
r = statistics.correlation(study_hours, exam_scores)
#
print(r)
#
#
# A value close to +1 means strong positive linear association.
#
#
# But do NOT conclude:
#
#     "Studying causes exactly this score increase."
#
# without an appropriate causal design.
#
#
# =================================================================================================
# 71. A PRACTICAL REGRESSION EXAMPLE
# =================================================================================================
#
regression = statistics.linear_regression(study_hours, exam_scores)
#
print("Slope:", regression.slope)
print("Intercept:", regression.intercept)
#
#
# If the fitted relationship is approximately:
#
#     score = 8x + 42
#
#
# Then:
#
#     x = 1
#     ↓
#     predicted score ≈ 50
#
#
#     x = 5
#     ↓
#     predicted score ≈ 82
#
#
# This is a model.
#
# It is not a statement that every future observation must exactly
# follow the line.
#
#
# =================================================================================================
# 72. THE STATISTICAL PIPELINE
# =================================================================================================
#
#
#                    RAW DATA
#                       │
#                       ▼
#               ┌───────────────┐
#               │ CENTER        │
#               │ mean/median   │
#               └───────┬───────┘
#                       │
#                       ▼
#               ┌───────────────┐
#               │ SPREAD        │
#               │ stdev/variance│
#               └───────┬───────┘
#                       │
#                       ▼
#               ┌───────────────┐
#               │ POSITION      │
#               │ quantiles     │
#               └───────┬───────┘
#                       │
#                       ▼
#               ┌───────────────┐
#               │ RELATIONSHIP  │
#               │ correlation   │
#               │ covariance    │
#               └───────┬───────┘
#                       │
#                       ▼
#               ┌───────────────┐
#               │ MODEL         │
#               │ regression    │
#               └───────┬───────┘
#                       │
#                       ▼
#               ┌───────────────┐
#               │ PROBABILITY   │
#               │ NormalDist    │
#               └───────────────┘
#
#
# This is a much better mental model than memorizing a list of functions.
#
#
# =================================================================================================
# 73. COMMON MISTAKE — MEAN IS NOT ALWAYS "THE AVERAGE"
# =================================================================================================
#
# "Average" is an informal word.
#
# Mathematically, several different averages exist:
#
#     arithmetic mean
#     geometric mean
#     harmonic mean
#     median
#     mode
#
#
# Therefore:
#
#     "average"
#
# does NOT automatically mean:
#
#     mean()
#
#
# Ask:
#
#     What kind of quantity am I summarizing?
#
#
# =================================================================================================
# 74. COMMON MISTAKE — SAMPLE VS POPULATION
# =================================================================================================
#
# Do not automatically use:
#
#     variance()
#
# without asking:
#
#     "Is this a sample or the complete population?"
#
#
# Remember:
#
#     variance()
#         → sample variance
#
#     pvariance()
#         → population variance
#
#
# and:
#
#     stdev()
#         → sample standard deviation
#
#     pstdev()
#         → population standard deviation
#
#
# =================================================================================================
# 75. COMMON MISTAKE — CORRELATION DOES NOT MEAN CAUSATION
# =================================================================================================
#
# A correlation of:
#
#     +0.95
#
# does NOT mean:
#
#     X causes Y.
#
#
# It means:
#
#     X and Y have a very strong positive linear association
#     according to that correlation measure.
#
#
# =================================================================================================
# 76. COMMON MISTAKE — STANDARD DEVIATION IS NOT "ERROR"
# =================================================================================================
#
# Standard deviation measures the spread of observations.
#
# It does not automatically mean:
#
#     measurement error
#
# or:
#
#     prediction error
#
#
# Those are different concepts.
#
#
# Standard deviation:
#
#     data variability
#
#
# Prediction error:
#
#     difference between prediction and observed outcome
#
#
# =================================================================================================
# 77. COMMON MISTAKE — PDF IS NOT PROBABILITY AT A POINT
# =================================================================================================
#
# For continuous distributions:
#
#     pdf(x)
#
# is a density.
#
# Probability comes from AREA under the curve over an interval.
#
#
#     density
#        ↓
#     integrate over interval
#        ↓
#     probability
#
#
# For example:
#
#     P(a ≤ X ≤ b)
#
# corresponds to the area under the PDF between a and b.
#
#
# =================================================================================================
# 78. EDGE CASES
# =================================================================================================
#
# Empty data:
#
#     many statistics functions raise StatisticsError.
#
#
# One observation:
#
#     mean() can be calculated.
#
# But sample variance/stdev cannot meaningfully estimate variability
# from only one observation.
#
#
# Constant data:
#
#     [5, 5, 5, 5]
#
# has:
#
#     variance = 0
#     standard deviation = 0
#
#
# This matters for standardization:
#
#     (x - mean) / stdev
#
# cannot be performed when:
#
#     stdev = 0
#
#
# =================================================================================================
# 79. ERROR HANDLING
# =================================================================================================
#
# The statistics module provides:
#
#     statistics.StatisticsError
#
#
# Example:
#
try:
#     result = statistics.mean([])
    pass
except statistics.StatisticsError:
    print("The dataset is not suitable for this calculation.")
#
#
# The exact handling should depend on the application.
#
# In Data Science, missing/empty data often needs explicit preprocessing
# rather than silently replacing it with a value.
#
#
# =================================================================================================
# 80. STATISTICS LIBRARY VS NUMPY
# =================================================================================================
#
#
# statistics
#     ↓
#     simple
#     readable
#     standard-library
#     educational
#     basic statistical calculations
#
#
# NumPy
#     ↓
#     arrays
#     vectorization
#     numerical computing
#     large-scale numerical operations
#
#
# Think:
#
#     statistics
#          ↓
#     "I want a statistical answer."
#
#     NumPy
#          ↓
#     "I want to perform numerical computation efficiently on arrays."
#
#
# For professional DA/DS:
#
#     NumPy + pandas
#
# will normally become much more important than the standalone
# statistics module.
#
#
# =================================================================================================
# 81. STATISTICS LIBRARY VS PANDAS
# =================================================================================================
#
# `statistics` works primarily with Python data collections.
#
# pandas is designed around tabular data:
#
#     rows
#     columns
#     Series
#     DataFrame
#
#
# Example conceptual difference:
#
#     statistics.mean([10, 20, 30])
#
# versus:
#
#     dataframe["sales"].mean()
#
#
# For real-world DA:
#
#     CSV / Excel / SQL
#           ↓
#         pandas
#           ↓
#     filtering / grouping / joining
#           ↓
#     statistical summaries
#
#
# `statistics` is excellent for learning the underlying concepts,
# but pandas is usually the practical workhorse for tabular DA.
#
#
# =================================================================================================
# 82. STATISTICS LIBRARY VS SCIPY
# =================================================================================================
#
# statistics:
#
#     basic descriptive statistics
#     basic correlation/covariance
#     simple regression
#     normal distribution
#
#
# SciPy:
#
#     probability distributions
#     hypothesis tests
#     statistical tests
#     optimization
#     advanced scientific statistics
#
#
# Think:
#
#     statistics
#          ↓
#     FUNDAMENTALS
#
#     SciPy
#          ↓
#     ADVANCED SCIENTIFIC STATISTICS
#
#
# =================================================================================================
# 83. STATISTICS LIBRARY VS STATSMODELS
# =================================================================================================
#
# statistics:
#
#     basic calculations
#
#
# statsmodels:
#
#     statistical modeling
#     regression inference
#     confidence intervals
#     p-values
#     statistical tests
#     model diagnostics
#     econometrics/time-series tools
#
#
# Think:
#
#     statistics
#          ↓
#     "calculate the statistic"
#
#     statsmodels
#          ↓
#     "fit and statistically analyze a model"
#
#
# =================================================================================================
# 84. STATISTICS LIBRARY VS SCIKIT-LEARN
# =================================================================================================
#
# statistics:
#
#     understand fundamental statistics
#
#
# scikit-learn:
#
#     machine-learning workflow
#     preprocessing
#     train/test splitting
#     models
#     pipelines
#     cross-validation
#     metrics
#     feature selection
#
#
# Think:
#
#     statistics
#          ↓
#     understand the statistical building blocks
#
#     scikit-learn
#          ↓
#     build machine-learning systems
#
#
# =================================================================================================
# 85. WHAT TO LEARN DEEPLY FOR YOUR DA/DS/ML PATH
# =================================================================================================
#
# If your goal is practical Data Analysis → Data Science → ML,
# learn these concepts VERY well:
#
#
#     1. mean
#     2. median
#     3. mode
#     4. weighted mean
#     5. quantiles
#     6. variance
#     7. standard deviation
#     8. sample vs population
#     9. covariance
#    10. correlation
#    11. z-score
#    12. normal distribution
#    13. linear regression
#
#
# The function names are secondary.
#
# The concepts are primary.
#
#
# =================================================================================================
# 86. THE MOST IMPORTANT STATISTICAL MENTAL MODEL
# =================================================================================================
#
#
#                         DATA
#                          │
#                          ▼
#                 "What is typical?"
#                          │
#                  mean / median
#                          │
#                          ▼
#                 "How variable?"
#                          │
#                  variance / stdev
#                          │
#                          ▼
#                 "Where am I?"
#                          │
#                      quantiles
#                          │
#                          ▼
#                 "How related?"
#                          │
#               covariance / correlation
#                          │
#                          ▼
#                 "Can I model it?"
#                          │
#                    regression
#                          │
#                          ▼
#                 "What probability?"
#                          │
#                      NormalDist
#
#
# This sequence is far more important than memorizing isolated functions.
#
#
# =================================================================================================
# 87. FUNCTION → QUESTION MAP
# =================================================================================================
#
# When you face a dataset, ask:
#
#
# QUESTION                                      FUNCTION
# ─────────────────────────────────────────────────────────────
#
# What is the arithmetic center?               mean()
#
# What is the numerical mean as float?         fmean()
#
# What is the middle value?                    median()
#
# What is the lower middle?                    median_low()
#
# What is the higher middle?                   median_high()
#
# What is the grouped/interpolated median?     median_grouped()
#
# What occurs most often?                      mode()
#
# Are there several modes?                     multimode()
#
# What are the quartiles/quantiles?            quantiles()
#
# What is the multiplicative average?          geometric_mean()
#
# What is the rate-based average?              harmonic_mean()
#
# How variable is the population?              pvariance()
#
# How variable is the sample?                  variance()
#
# Population standard deviation?               pstdev()
#
# Sample standard deviation?                   stdev()
#
# How do X and Y vary together?                covariance()
#
# How strongly are X and Y associated?         correlation()
#
# What linear equation describes X → Y?        linear_regression()
#
# What is the normal-distribution density?     NormalDist.pdf()
#
# What probability is ≤ x?                     NormalDist.cdf()
#
# What x corresponds to probability p?         NormalDist.inv_cdf()
#
# What is the standardized value?              NormalDist.zscore()
#
# What are normal-distribution cut points?     NormalDist.quantiles()
#
# Generate normal random observations?          NormalDist.samples()
#
# How much do two normal distributions overlap?
#                                               NormalDist.overlap()
#
#
# =================================================================================================
# 88. A PRACTICAL DA CHECKLIST
# =================================================================================================
#
# Given a numerical column:
#
#
#     1. Look at the center
#
#         mean
#         median
#
#              ↓
#
#         Are they similar?
#
#
#     2. Look at spread
#
#         stdev
#         quantiles
#
#              ↓
#
#         Is variability large?
#
#
#     3. Look for skew/outliers
#
#         mean vs median
#         quartiles
#
#              ↓
#
#         Are extreme values affecting the mean?
#
#
#     4. Look at relationships
#
#         correlation
#         covariance
#
#              ↓
#
#         Which variables move together?
#
#
#     5. If appropriate, model
#
#         linear_regression
#
#
# =================================================================================================
# 89. A PRACTICAL DS CHECKLIST
# =================================================================================================
#
# For Data Science:
#
#     DATA
#       ↓
#     distribution
#       ↓
#     missingness / quality
#       ↓
#     center
#       ↓
#     spread
#       ↓
#     skew / outliers
#       ↓
#     relationships
#       ↓
#     assumptions
#       ↓
#     statistical modeling
#       ↓
#     ML modeling
#
#
# The statistics module supports the middle part of this pipeline.
#
# It does NOT perform the entire DS workflow.
#
#
# =================================================================================================
# 90. A PRACTICAL ML CONNECTION
# =================================================================================================
#
# Suppose a feature is:
#
#     age = [18, 20, 22, 25, 80]
#
#
# First:
#
#     mean
#     median
#     quantiles
#     stdev
#
#
# Then ask:
#
#     Is 80 an outlier?
#     Is the distribution skewed?
#     Should the feature be transformed?
#     Does the model require scaling?
#
#
# Then:
#
#     correlation(feature, target)
#
# may provide an initial exploratory relationship.
#
#
# Finally:
#
#     scikit-learn
#
# would normally perform the actual preprocessing/modeling pipeline.
#
#
# =================================================================================================
# 91. IMPORTANT LIMITATION OF THE statistics MODULE
# =================================================================================================
#
# Do NOT think:
#
#     "If I know the statistics module, I know statistics."
#
#
# The module gives you COMPUTATIONAL TOOLS.
#
# Statistics as a discipline also requires understanding:
#
#     probability
#     sampling
#     distributions
#     estimation
#     uncertainty
#     confidence intervals
#     hypothesis testing
#     statistical power
#     experimental design
#     regression
#     assumptions
#     causal inference
#     Bayesian reasoning
#
#
# The module is a TOOLBOX.
#
# Statistics is the DISCIPLINE.
#
#
# =================================================================================================
# 92. CORE IDEA
# =================================================================================================
#
# The most important principle to remember is:
#
#
#     DATA
#       ↓
#     describe its CENTER
#       ↓
#     describe its SPREAD
#       ↓
#     understand its POSITION
#       ↓
#     examine RELATIONSHIPS
#       ↓
#     build MODELS when appropriate
#
#
# In Python:
#
#     mean / median / mode
#             ↓
#     variance / stdev
#             ↓
#     quantiles
#             ↓
#     covariance / correlation
#             ↓
#     linear_regression
#             ↓
#     NormalDist
#
#
# The goal is NOT to memorize:
#
#     "mean() does X"
#
# The stronger mental model is:
#
#     "I have data.
#
#      First I ask what is typical.
#      Then I ask how variable it is.
#      Then I ask where observations sit.
#      Then I ask how variables relate.
#      Then, if appropriate, I construct a model."
#
#
# =================================================================================================
# 93. FINAL PRACTICAL PRIORITY FOR DA → DS → ML
# =================================================================================================
#
#                         LEARNING PRIORITY
#
#     ┌───────────────────────────────────────────────┐
#     │ VERY HIGH                                     │
#     │                                               │
#     │ mean()                                        │
#     │ median()                                      │
#     │ quantiles()                                   │
#     │ variance() / pvariance()                     │
#     │ stdev() / pstdev()                            │
#     │ correlation()                                 │
#     │ covariance()                                  │
#     └───────────────────────────────────────────────┘
#
#     ┌───────────────────────────────────────────────┐
#     │ HIGH                                          │
#     │                                               │
#     │ fmean()                                       │
#     │ mode() / multimode()                          │
#     │ geometric_mean()                              │
#     │ harmonic_mean()                               │
#     │ linear_regression()                            │
#     │ NormalDist                                   │
#     │ zscore() / cdf() / inv_cdf()                  │
#     └───────────────────────────────────────────────┘
#
#     ┌───────────────────────────────────────────────┐
#     │ SPECIALIZED                                   │
#     │                                               │
#     │ median_low()                                  │
#     │ median_high()                                 │
#     │ median_grouped()                              │
#     │ NormalDist.overlap()                          │
#     │ NormalDist.samples()                          │
#     └───────────────────────────────────────────────┘
#
#
# For your DA/DS/ML path, the most important outcome is not mastery of
# every method in isolation.
#
# It is being able to look at a dataset and naturally think:
#
#
#     "CENTER?"
#          ↓
#     mean / median
#
#     "SPREAD?"
#          ↓
#     variance / stdev
#
#     "POSITION?"
#          ↓
#     quantiles
#
#     "RELATIONSHIP?"
#          ↓
#     covariance / correlation
#
#     "PREDICTIVE PATTERN?"
#          ↓
#     regression
#
#     "PROBABILITY / STANDARDIZATION?"
#          ↓
#     NormalDist / z-score
#
#
# That mental model is the real foundation.
#
# =================================================================================================
```
