```python
# ============================================================
# SCIpy FOR STATISTICS, DATA ANALYSIS, DATA SCIENCE & ML
# ============================================================
#
# Goal:
# Learn SciPy conceptually, especially the parts that matter for:
#
#     Data Analysis (DA)
#     Data Science (DS)
#     Machine Learning (ML)
#     Statistical analysis
#     Scientific computing
#
# IMPORTANT:
#
# SciPy is a large scientific-computing ecosystem.
#
# You do NOT need to memorize every SciPy function.
#
# For statistics and DS/DA/ML, the most important part is:
#
#     scipy.stats
#
# But several other SciPy modules are extremely useful:
#
#     scipy.linalg       → linear algebra
#     scipy.optimize     → optimization
#     scipy.spatial      → distances / geometry
#     scipy.special      → mathematical special functions
#     scipy.interpolate  → interpolation
#
# The central mental model is:
#
#     NumPy
#       ↓
#     numerical arrays
#       ↓
#     SciPy
#       ↓
#     scientific/statistical algorithms
#       ↓
#     pandas / sklearn / statsmodels
#       ↓
#     real DA / DS / ML workflows
#
#
# ============================================================
# 01. WHAT IS SciPy?
# ============================================================
#
# SciPy = Scientific Python
#
# It builds on NumPy.
#
# NumPy gives you:
#
#     arrays
#     vectorized arithmetic
#     linear algebra basics
#     random-number generation
#
# SciPy gives you:
#
#     probability distributions
#     statistical tests
#     optimization
#     numerical integration
#     interpolation
#     advanced linear algebra
#     distance calculations
#     special mathematical functions
#     signal processing
#     scientific algorithms
#
#
# Think of the relationship like this:
#
#     NumPy
#       │
#       ├── "I have numbers and arrays."
#       │
#       ↓
#     SciPy
#       │
#       ├── "I need mathematical/statistical algorithms."
#       │
#       ↓
#     pandas / sklearn / statsmodels
#       │
#       └── "I want to analyze data / build models."
#
#
# SciPy is NOT a replacement for:
#
#     pandas
#     NumPy
#     scikit-learn
#     statsmodels
#
# It complements them.
#
#
# ============================================================
# 02. INSTALLATION AND IMPORTING
# ============================================================
#
# Usually:
#
#     pip install scipy
#
# Common imports:
#
#     import scipy
#     import scipy.stats as stats
#     import scipy.linalg as linalg
#     import scipy.optimize as optimize
#     import scipy.spatial as spatial
#     import scipy.special as special
#
#
# The most important one for statistics is:
#
#     from scipy import stats
#
#
# Then:
#
#     stats.mean(...)             # actually prefer NumPy for this
#     stats.norm(...)
#     stats.ttest_ind(...)
#     stats.pearsonr(...)
#
#
# ============================================================
# 03. THE MOST IMPORTANT IDEA:
#     SciPy.stats IS NOT JUST "MORE STATISTICS"
# ============================================================
#
# Think about statistics as several layers:
#
#
#                  STATISTICS
#                       │
#        ┌──────────────┼──────────────┐
#        ↓              ↓              ↓
#   DESCRIPTIVE      PROBABILITY     INFERENCE
#        │              │              │
#        ↓              ↓              ↓
#   mean/std/etc.   distributions   hypothesis tests
#                                  confidence intervals
#                                  p-values
#
#                       ↓
#
#                    SciPy
#                  scipy.stats
#
#
# scipy.stats is especially strong in:
#
#     probability distributions
#     hypothesis testing
#     correlation
#     non-parametric statistics
#     statistical transformations
#     resampling
#     distribution fitting
#     contingency tables
#
#
# ============================================================
# 04. THE SCIPY.STATS MENTAL MODEL
# ============================================================
#
# Almost everything in scipy.stats can be understood through:
#
#
#       DATA
#        │
#        ↓
#   ┌───────────────┐
#   │ Mathematical  │
#   │ model         │
#   └───────────────┘
#        │
#        ↓
#   probability distribution
#        │
#        ├── PDF
#        ├── CDF
#        ├── PPF
#        ├── random samples
#        └── statistics
#
#
# OR:
#
#
#       DATA
#         ↓
#   ┌─────────────┐
#   │ Statistical │
#   │ question    │
#   └─────────────┘
#         ↓
#   statistical test
#         ↓
#   statistic + p-value
#         ↓
#   conclusion
#
#
# ============================================================
# 05. DESCRIPTIVE STATISTICS
# ============================================================
#
# SciPy has descriptive-statistics functionality, but remember:
#
#     NumPy
#     pandas
#     statistics
#
# are often more convenient for basic descriptive statistics.
#
# SciPy becomes particularly valuable when you move from:
#
#     "What does my data look like?"
#
# to:
#
#     "What statistical process generated this data?"
#
# and:
#
#     "Is this difference/correlation statistically significant?"
#
#
# Important functions include:
#
#     stats.describe()
#     stats.gmean()
#     stats.hmean()
#     stats.skew()
#     stats.kurtosis()
#     stats.variation()
#
#
# Example:
#
#     import numpy as np
#     from scipy import stats
#
#     x = np.array([10, 12, 13, 15, 20])
#
#     stats.describe(x)
#
#
# Conceptually:
#
#     DATA
#       │
#       ├── minimum
#       ├── maximum
#       ├── mean
#       ├── variance
#       ├── skewness
#       └── kurtosis
#
#
# ============================================================
# 06. stats.describe()
# ============================================================
#
#     stats.describe(x)
#
# gives a compact statistical description.
#
# Conceptually:
#
#     x
#     │
#     ├── n observations
#     ├── min / max
#     ├── mean
#     ├── variance
#     ├── skewness
#     └── kurtosis
#
#
# Example:
#
#     x = np.array([1, 2, 3, 4, 5])
#     result = stats.describe(x)
#
#
# It is useful when you want a quick statistical profile.
#
# In real DA:
#
#     raw dataset
#          ↓
#     data-quality inspection
#          ↓
#     descriptive statistics
#          ↓
#     visualization
#          ↓
#     statistical modeling
#
#
# ============================================================
# 07. GEOMETRIC MEAN
# ============================================================
#
#     stats.gmean(x)
#
# Geometric mean is appropriate for multiplicative quantities.
#
# Arithmetic mean:
#
#     (x1 + x2 + ... + xn) / n
#
# Geometric mean:
#
#     (x1 × x2 × ... × xn) ** (1/n)
#
#
# Example:
#
#     growth rates:
#
#         +10%
#         +20%
#         +30%
#
# Growth is multiplicative, so geometric averaging can make
# more sense than ordinary averaging.
#
#
# Mental model:
#
#     additive process  → arithmetic mean
#
#     multiplicative process → geometric mean
#
#
# ============================================================
# 08. HARMONIC MEAN
# ============================================================
#
#     stats.hmean(x)
#
# Useful for quantities such as:
#
#     rates
#     speeds
#     ratios
#
#
# Classic example:
#
# Same distance:
#
#     60 km/h going out
#     40 km/h returning
#
# The average speed is NOT:
#
#     (60 + 40) / 2 = 50
#
# because speed is a rate.
#
# Harmonic mean:
#
#     2 / (1/60 + 1/40)
#
#     = 48 km/h
#
#
# Mental model:
#
#     ordinary values → arithmetic mean
#     multiplicative values → geometric mean
#     rates → harmonic mean
#
#
# ============================================================
# 09. SKEWNESS
# ============================================================
#
#     stats.skew(x)
#
# Skewness describes asymmetry.
#
#
# Symmetric:
#
#             /\ 
#            /  \
#           /    \
#     ______/      \______
#
#
# Positive/right skew:
#
#             /\
#            /  \
#           /    \__________
#     ______/
#
#
# Negative/left skew:
#
#       __________
#             /\
#            /  \
#     ______/    \
#
#
# Positive skew:
#
#     long tail → right
#
# Negative skew:
#
#     long tail → left
#
#
# Useful for:
#
#     income
#     transaction values
#     response times
#     insurance losses
#     financial variables
#
#
# ============================================================
# 10. KURTOSIS
# ============================================================
#
#     stats.kurtosis(x)
#
# Kurtosis is related to tail behavior and peakedness.
#
# Important:
#
# Do not reduce kurtosis to simply:
#
#     "how pointy is the distribution?"
#
# In statistics, kurtosis is strongly related to tail weight.
#
# SciPy's default uses Fisher's definition:
#
#     normal distribution → kurtosis ≈ 0
#
# while Pearson kurtosis would make normal kurtosis = 3.
#
# This distinction matters.
#
#
# ============================================================
# 11. VARIATION
# ============================================================
#
#     stats.variation(x)
#
# Coefficient of variation:
#
#     CV = standard deviation / mean
#
#
# It measures relative variability.
#
#
# Example:
#
# Dataset A:
#
#     mean = 100
#     std  = 10
#
#     CV = 0.10
#
#
# Dataset B:
#
#     mean = 10
#     std  = 5
#
#     CV = 0.50
#
#
# Dataset B is much more variable relative to its mean.
#
#
# Useful when comparing variability across different scales.
#
#
# ============================================================
# 12. PROBABILITY DISTRIBUTIONS
# ============================================================
#
# THIS is one of the most important parts of scipy.stats.
#
#
# A probability distribution answers:
#
#     "How likely are different values?"
#
#
# Example:
#
#     Height
#
#     probability
#          │
#          │           /\
#          │         /    \
#          │       /        \
#          │______/____________\____
#                height
#
#
# SciPy provides many distributions.
#
#
# Major examples:
#
#     stats.norm
#     stats.binom
#     stats.poisson
#     stats.uniform
#     stats.expon
#     stats.t
#     stats.chi2
#     stats.f
#     stats.gamma
#     stats.beta
#     stats.lognorm
#     stats.weibull_min
#     stats.bernoulli
#     stats.multivariate_normal
#
#
# ============================================================
# 13. DISTRIBUTION OBJECTS
# ============================================================
#
# A very important SciPy idea:
#
#     stats.norm
#
# represents the normal distribution family.
#
# You can think of it as a mathematical machine:
#
#
#                NORMAL DISTRIBUTION
#                       │
#          ┌────────────┼────────────┐
#          ↓            ↓            ↓
#         PDF          CDF          PPF
#          │            │            │
#      density       probability    quantile
#
#
# And you can specify parameters:
#
#     mean = μ
#     standard deviation = σ
#
#
# Example:
#
#     distribution = stats.norm(loc=100, scale=15)
#
#
# Here:
#
#     loc   → mean
#     scale → standard deviation
#
#
# ============================================================
# 14. PDF — PROBABILITY DENSITY FUNCTION
# ============================================================
#
#     stats.norm.pdf(x)
#
# PDF tells you the HEIGHT of a continuous probability density.
#
#
# IMPORTANT:
#
# PDF value is NOT generally:
#
#     "probability that X equals x"
#
# For a continuous variable:
#
#     P(X = exact value) = 0
#
# Instead, probability is represented by AREA.
#
#
# Visual:
#
#              density
#                 │
#              ___│___
#            /   │   \
#          /     │     \
#        /_______│_______\
#                 x
#
# Area under curve
#       ↓
# probability
#
#
# Example:
#
#     stats.norm.pdf(0)
#
# asks:
#
#     "What is the standard normal density at x = 0?"
#
#
# ============================================================
# 15. CDF — CUMULATIVE DISTRIBUTION FUNCTION
# ============================================================
#
#     stats.norm.cdf(x)
#
# CDF answers:
#
#     P(X ≤ x)
#
#
# Visual:
#
#     probability
#         1.0 ───────────────────
#                          ______
#                       __/
#                    __/
#                 __/
#              __/
#         0 ───/──────────────────
#             x
#
#
# Example:
#
#     stats.norm.cdf(1.96)
#
# approximately:
#
#     0.975
#
#
# Meaning:
#
#     P(Z ≤ 1.96) ≈ 0.975
#
#
# This is extremely important in:
#
#     confidence intervals
#     hypothesis testing
#     z-scores
#     probability calculations
#
#
# ============================================================
# 16. PPF — PERCENT POINT FUNCTION
# ============================================================
#
#     stats.norm.ppf(p)
#
# PPF is the inverse of the CDF.
#
#
# CDF:
#
#     x → probability
#
# PPF:
#
#     probability → x
#
#
# Diagram:
#
#     x ───────────────→ CDF ───────────────→ probability
#     x ←────────────── PPF ←─────────────── probability
#
#
# Example:
#
#     stats.norm.ppf(0.975)
#
# gives approximately:
#
#     1.96
#
#
# This is extremely useful for:
#
#     critical values
#     confidence intervals
#     quantiles
#     statistical tests
#
#
# ============================================================
# 17. SF — SURVIVAL FUNCTION
# ============================================================
#
#     stats.norm.sf(x)
#
# SF means:
#
#     P(X > x)
#
#
# Since:
#
#     P(X > x) = 1 - P(X ≤ x)
#
# you could write:
#
#     1 - stats.norm.cdf(x)
#
# but:
#
#     stats.norm.sf(x)
#
# is often numerically more reliable in extreme tails.
#
#
# Mental model:
#
#     CDF:
#
#         area ←──────── x
#
#     SF:
#
#         x ─────────→ area
#
#
# Useful in:
#
#     reliability
#     survival analysis
#     extreme probabilities
#     risk analysis
#
#
# ============================================================
# 18. ISF — INVERSE SURVIVAL FUNCTION
# ============================================================
#
#     stats.norm.isf(p)
#
# ISF is the inverse of SF.
#
#
# Useful when working with upper-tail probabilities.
#
#
# CDF / PPF:
#
#     lower tail
#
# SF / ISF:
#
#     upper tail
#
#
# ============================================================
# 19. RVS — RANDOM VARIATES
# ============================================================
#
#     stats.norm.rvs(...)
#
# generates random samples from a distribution.
#
# Example:
#
#     samples = stats.norm.rvs(
#         loc=100,
#         scale=15,
#         size=1000,
#         random_state=42
#     )
#
#
# Concept:
#
#     probability distribution
#              ↓
#        random sampling
#              ↓
#          simulated data
#
#
# This is extremely useful for:
#
#     simulation
#     Monte Carlo methods
#     testing algorithms
#     synthetic data
#     statistical experiments
#
#
# ============================================================
# 20. PMF — PROBABILITY MASS FUNCTION
# ============================================================
#
# PMF is for DISCRETE distributions.
#
# Example:
#
#     stats.binom.pmf(k, n, p)
#
#
# Continuous:
#
#     PDF → density
#
# Discrete:
#
#     PMF → probability
#
#
# Visual:
#
#     continuous:
#
#          smooth curve
#
#     discrete:
#
#          │       │
#          │   │   │
#          │ │ │ │ │
#          └────────────
#             values
#
#
# ============================================================
# 21. COMMON DISCRETE DISTRIBUTIONS
# ============================================================
#
# ------------------------------------------------------------
# Bernoulli
# ------------------------------------------------------------
#
# One trial:
#
#     success / failure
#     1 / 0
#
#     stats.bernoulli
#
#
# Example:
#
#     probability of success = 0.7
#
#
# ------------------------------------------------------------
# Binomial
# ------------------------------------------------------------
#
# Number of successes in n independent trials.
#
#     stats.binom
#
#
# Example:
#
#     10 attempts
#     probability of success = 0.7
#
#     stats.binom.pmf(7, 10, 0.7)
#
#
# Mental model:
#
#     Bernoulli × Bernoulli × ... × Bernoulli
#                       ↓
#                   Binomial
#
#
# ------------------------------------------------------------
# Poisson
# ------------------------------------------------------------
#
# Number of events in a fixed interval.
#
#     stats.poisson
#
# Examples:
#
#     calls per minute
#     failures per month
#     customers per hour
#     accidents per year
#
#
# ============================================================
# 22. COMMON CONTINUOUS DISTRIBUTIONS
# ============================================================
#
# ------------------------------------------------------------
# Normal
# ------------------------------------------------------------
#
#     stats.norm
#
# Used everywhere:
#
#     measurement error
#     standardized values
#     approximate distributions
#     statistical inference
#
#
# ------------------------------------------------------------
# Uniform
# ------------------------------------------------------------
#
#     stats.uniform
#
# Every value in an interval has equal density.
#
#
# ------------------------------------------------------------
# Exponential
# ------------------------------------------------------------
#
#     stats.expon
#
# Often used for waiting times.
#
#
# ------------------------------------------------------------
# Gamma
# ------------------------------------------------------------
#
#     stats.gamma
#
# Useful for:
#
#     waiting times
#     positive continuous variables
#     Bayesian/statistical modeling
#
#
# ------------------------------------------------------------
# Beta
# ------------------------------------------------------------
#
#     stats.beta
#
# Values are restricted to:
#
#     0 ≤ x ≤ 1
#
# Particularly useful for:
#
#     probabilities
#     proportions
#     Bayesian modeling
#
#
# ------------------------------------------------------------
# Lognormal
# ------------------------------------------------------------
#
#     stats.lognorm
#
# Useful when:
#
#     log(X)
#
# is approximately normally distributed.
#
# Common in:
#
#     income
#     biological measurements
#     multiplicative processes
#
#
# ------------------------------------------------------------
# Weibull
# ------------------------------------------------------------
#
#     stats.weibull_min
#
# Important for:
#
#     reliability
#     lifetime analysis
#     failure-time modeling
#
#
# ============================================================
# 23. T DISTRIBUTION
# ============================================================
#
#     stats.t
#
# Extremely important in classical statistics.
#
# It appears when estimating a mean with unknown population
# standard deviation, especially with smaller samples.
#
#
# Visual:
#
# Normal:
#
#                 /\
#                /  \
#               /    \
#
# t distribution:
#
#               /\
#              /  \
#             /    \
#     _______/      \_______
#
# t has heavier tails.
#
#
# As degrees of freedom increase:
#
#     t distribution → normal distribution
#
#
# ============================================================
# 24. CHI-SQUARE DISTRIBUTION
# ============================================================
#
#     stats.chi2
#
# Important for:
#
#     variance-related inference
#     goodness-of-fit
#     contingency-table tests
#     categorical-data analysis
#
#
# It is asymmetric and only takes non-negative values.
#
#
# ============================================================
# 25. F DISTRIBUTION
# ============================================================
#
#     stats.f
#
# Important for:
#
#     comparing variances
#     ANOVA
#     regression inference
#
#
# Conceptually:
#
#     variance estimate
#           ÷
#     another variance estimate
#           ↓
#       F statistic
#
#
# ============================================================
# 26. DISTRIBUTION METHODS YOU SHOULD KNOW
# ============================================================
#
# For many continuous distributions:
#
#     pdf()
#     logpdf()
#     cdf()
#     logcdf()
#     sf()
#     logsf()
#     ppf()
#     isf()
#     rvs()
#     stats()
#     entropy()
#     expect()
#     fit()
#
#
# Discrete distributions additionally use:
#
#     pmf()
#     logpmf()
#
#
# You do NOT need to memorize every method separately.
#
# Learn the pattern:
#
#
#     VALUE
#       │
#       ├── pdf / pmf
#       │
#       ├── cdf
#       │
#       ├── ppf
#       │
#       └── rvs
#
#
# This pattern repeats across distributions.
#
#
# ============================================================
# 27. logpdf / logpmf
# ============================================================
#
# Instead of:
#
#     pdf(x)
#
# you can calculate:
#
#     logpdf(x)
#
# Why?
#
# Probability products can become extremely small.
#
#
# Suppose:
#
#     p1 = 0.001
#     p2 = 0.001
#     p3 = 0.001
#
# Product:
#
#     p1 × p2 × p3 = 0.000000001
#
#
# For hundreds/thousands of observations:
#
#     p1 × p2 × ... × pn
#
# can underflow toward zero.
#
#
# Logs transform:
#
#     multiplication
#          ↓
#       addition
#
#     log(p1 × p2)
#          =
#     log(p1) + log(p2)
#
#
# This is fundamental in:
#
#     maximum likelihood
#     probabilistic ML
#     Bayesian methods
#     Naive Bayes
#     statistical modeling
#
#
# ============================================================
# 28. DISTRIBUTION PARAMETERS
# ============================================================
#
# A distribution is not just a name.
#
# It has PARAMETERS.
#
#
# Normal:
#
#     X ~ Normal(μ, σ)
#
# SciPy:
#
#     stats.norm(loc=μ, scale=σ)
#
#
# Example:
#
#     dist = stats.norm(loc=100, scale=15)
#
#
# Mental model:
#
#     distribution family
#             +
#          parameters
#             ↓
#       specific distribution
#
#
# ============================================================
# 29. DISTRIBUTION STATISTICS
# ============================================================
#
#     dist.stats()
#
# can return quantities such as:
#
#     mean
#     variance
#     skewness
#     kurtosis
#
#
# Example:
#
#     stats.norm.stats(moments="mvsk")
#
#
# Meaning:
#
#     m → mean
#     v → variance
#     s → skewness
#     k → kurtosis
#
#
# ============================================================
# 30. DISTRIBUTION ENTROPY
# ============================================================
#
#     dist.entropy()
#
# Entropy measures uncertainty.
#
#
# Intuition:
#
#     highly predictable
#           ↓
#        low entropy
#
#     highly uncertain
#           ↓
#        high entropy
#
#
# Entropy is important in:
#
#     information theory
#     decision trees
#     ML
#     probabilistic modeling
#
#
# Note:
#
# For ML decision trees, scikit-learn usually performs the
# practical computation for you.
#
#
# ============================================================
# 31. FITTING A DISTRIBUTION TO DATA
# ============================================================
#
#     distribution.fit(data)
#
# attempts to estimate distribution parameters from observed data.
#
#
# Concept:
#
#     observed data
#           ↓
#     assume distribution family
#           ↓
#     estimate parameters
#           ↓
#     fitted distribution
#
#
# Example:
#
#     data = ...
#
#     parameters = stats.norm.fit(data)
#
#
# For normal:
#
#     estimated mean
#     estimated standard deviation
#
#
# This is an important bridge between:
#
#     raw data
#          ↓
#     probability model
#
#
# ============================================================
# 32. HYPOTHESIS TESTING
# ============================================================
#
# One of the MOST IMPORTANT uses of scipy.stats.
#
#
# A hypothesis test asks:
#
#     "Is the observed evidence sufficiently inconsistent
#      with the null hypothesis?"
#
#
# General structure:
#
#
#     H0
#     │
#     ↓
#   collect data
#     │
#     ↓
#   calculate test statistic
#     │
#     ↓
#   calculate p-value
#     │
#     ↓
#   decision / interpretation
#
#
# ============================================================
# 33. T-TESTS
# ============================================================
#
# Important functions:
#
#     stats.ttest_1samp()
#     stats.ttest_ind()
#     stats.ttest_rel()
#
#
# ------------------------------------------------------------
# One-sample t-test
# ------------------------------------------------------------
#
#     stats.ttest_1samp(x, popmean)
#
# Question:
#
#     "Is the sample mean different from a specified population
#      mean?"
#
#
# ------------------------------------------------------------
# Independent two-sample t-test
# ------------------------------------------------------------
#
#     stats.ttest_ind(x, y)
#
# Question:
#
#     "Do two independent groups have different means?"
#
#
# Example:
#
#     treatment group
#     control group
#
#
# ------------------------------------------------------------
# Paired t-test
# ------------------------------------------------------------
#
#     stats.ttest_rel(before, after)
#
# Question:
#
#     "Did the same subjects change?"
#
#
# Structure:
#
#     person 1: before → after
#     person 2: before → after
#     person 3: before → after
#
#
# ============================================================
# 34. WELCH'S T-TEST
# ============================================================
#
# When two groups may have unequal variances:
#
#     stats.ttest_ind(
#         x,
#         y,
#         equal_var=False
#     )
#
#
# This is Welch's t-test.
#
# Practical rule:
#
#     If equal variance is questionable,
#     Welch's test is often preferable.
#
#
# ============================================================
# 35. P-VALUE
# ============================================================
#
# A p-value is NOT:
#
#     "probability that H0 is true"
#
# Instead, conceptually:
#
#     Assuming H0 is true,
#     how unusual is evidence at least this extreme?
#
#
# Small p-value:
#
#     observed data are difficult to reconcile with H0
#
#
# But:
#
#     small p-value ≠ large practical effect
#
#
# Always separate:
#
#     statistical significance
#
# from:
#
#     practical significance
#
#
# ============================================================
# 36. NON-PARAMETRIC TESTS
# ============================================================
#
# Important when normality/parametric assumptions are unsuitable.
#
# Major functions:
#
#     stats.mannwhitneyu()
#     stats.wilcoxon()
#     stats.kruskal()
#     stats.friedmanchisquare()
#
#
# ------------------------------------------------------------
# Mann-Whitney U
# ------------------------------------------------------------
#
# Compare two independent groups using ranks.
#
#
# ------------------------------------------------------------
# Wilcoxon signed-rank
# ------------------------------------------------------------
#
# Paired observations.
#
#
# ------------------------------------------------------------
# Kruskal-Wallis
# ------------------------------------------------------------
#
# Compare more than two independent groups.
#
#
# ------------------------------------------------------------
# Friedman
# ------------------------------------------------------------
#
# Repeated-measures / related groups.
#
#
# Mental model:
#
#     t-test
#       ↓
#     means / parametric assumptions
#
#     rank-based tests
#       ↓
#     relative ordering / fewer distribution assumptions
#
#
# ============================================================
# 37. NORMALITY TESTS
# ============================================================
#
# Important functions:
#
#     stats.shapiro()
#     stats.normaltest()
#     stats.anderson()
#
#
# They ask whether data are consistent with particular
# distributional assumptions.
#
#
# But:
#
# DO NOT blindly use a normality test as the only way to decide
# whether a method is appropriate.
#
# Also inspect:
#
#     histogram
#     Q-Q plot
#     skewness
#     sample size
#     domain knowledge
#
#
# Conceptually:
#
#     numerical test
#          +
#     visual inspection
#          +
#     domain knowledge
#          ↓
#     assumption assessment
#
#
# ============================================================
# 38. Q-Q PLOTS
# ============================================================
#
# SciPy provides tools such as:
#
#     stats.probplot()
#
#
# Q-Q plot compares:
#
#     observed quantiles
#          vs.
#     theoretical quantiles
#
#
# If points approximately follow a straight line:
#
#     observed distribution
#            ≈
#     theoretical distribution
#
#
# Visual:
#
#     observed
#        │             *
#        │          *
#        │       *
#        │    *
#        │ *
#        └──────────────── theoretical
#
#
# Curvature indicates distributional differences.
#
#
# ============================================================
# 39. CORRELATION
# ============================================================
#
# Important functions:
#
#     stats.pearsonr()
#     stats.spearmanr()
#     stats.kendalltau()
#
#
# ------------------------------------------------------------
# Pearson
# ------------------------------------------------------------
#
# Measures linear association.
#
#
#     r = +1
#         perfect positive linear relationship
#
#     r =  0
#         no linear relationship
#
#     r = -1
#         perfect negative linear relationship
#
#
# IMPORTANT:
#
# r = 0 does NOT necessarily mean:
#
#     "no relationship"
#
# It can mean:
#
#     "no linear relationship"
#
#
# ------------------------------------------------------------
# Spearman
# ------------------------------------------------------------
#
# Rank-based correlation.
#
# Useful for monotonic relationships.
#
#
# ------------------------------------------------------------
# Kendall
# ------------------------------------------------------------
#
# Rank/concordance-based association.
#
# Useful particularly for ordinal/rank relationships.
#
#
# ============================================================
# 40. PEARSON vs SPEARMAN
# ============================================================
#
#
# Linear:
#
#     y
#     │       *
#     │     *
#     │   *
#     │ *
#     └──────── x
#
# Pearson → excellent candidate
#
#
# Monotonic but nonlinear:
#
#     y
#     │          *
#     │       *
#     │     *
#     │   *
#     │ *
#     └──────── x
#
# Spearman can capture the monotonic relationship better.
#
#
# Mental model:
#
#     Pearson  → actual values + linear relationship
#
#     Spearman → ranks + monotonic relationship
#
#
# ============================================================
# 41. CONTINGENCY TABLES
# ============================================================
#
# Important:
#
#     stats.contingency.crosstab()
#     stats.chi2_contingency()
#     stats.fisher_exact()
#
#
# Used with categorical variables.
#
#
# Example:
#
#                  Purchased
#                Yes     No
#
# Male            80     20
# Female          60     40
#
#
# Question:
#
#     "Is gender associated with purchase?"
#
#
# This is a categorical association problem.
#
#
# ============================================================
# 42. CHI-SQUARE TEST OF INDEPENDENCE
# ============================================================
#
#     stats.chi2_contingency(table)
#
#
# Tests whether two categorical variables are independent.
#
#
# Concept:
#
#     observed frequencies
#             vs.
#     expected frequencies under independence
#
#
# If observed counts differ substantially from expected:
#
#     evidence of association
#
#
# ============================================================
# 43. FISHER'S EXACT TEST
# ============================================================
#
#     stats.fisher_exact(table)
#
# Useful especially for small sample contingency tables.
#
#
# Mental model:
#
#     Chi-square
#         ↓
#     often larger samples
#
#     Fisher exact
#         ↓
#     small 2×2 tables
#
#
# ============================================================
# 44. ANOVA
# ============================================================
#
# Important:
#
#     stats.f_oneway()
#
#
# ANOVA asks:
#
#     "Are there differences among several group means?"
#
#
# Example:
#
#     Group A
#     Group B
#     Group C
#
#
# Instead of doing:
#
#     A vs B
#     A vs C
#     B vs C
#
# ANOVA evaluates the overall evidence.
#
#
# Concept:
#
#     variation BETWEEN groups
#                 ÷
#     variation WITHIN groups
#                 ↓
#             F statistic
#
#
# ============================================================
# 45. LEVENE'S TEST
# ============================================================
#
#     stats.levene()
#
# Tests equality of variances.
#
#
# Important when considering methods that assume similar
# variances.
#
#
# ============================================================
# 46. KS TEST
# ============================================================
#
#     stats.kstest()
#
# Kolmogorov-Smirnov test.
#
# Can compare:
#
#     empirical distribution
#          vs.
#     theoretical distribution
#
# or compare two empirical distributions in relevant cases.
#
#
# Concept:
#
#     empirical CDF
#           vs.
#     theoretical CDF
#
#
# Maximum vertical difference:
#
#
#       CDF
#        │
#        │       theoretical
#        │      /
#        │   __/
#        │  /
#        │_/ empirical
#        └────────────
#
#
# ============================================================
# 47. EFFECT SIZE vs P-VALUE
# ============================================================
#
# SciPy tests often return:
#
#     statistic
#     pvalue
#
#
# But you should ask THREE different questions:
#
#     1. Is there evidence of a difference?
#     2. How large is the difference?
#     3. Is the difference practically important?
#
#
# p-value:
#
#     evidence
#
# effect size:
#
#     magnitude
#
# confidence interval:
#
#     uncertainty around estimate
#
#
# Do not use p-value alone.
#
#
# ============================================================
# 48. CONFIDENCE INTERVALS
# ============================================================
#
# SciPy distributions and statistical-result objects can help
# construct confidence intervals.
#
#
# General concept:
#
#     sample
#       ↓
#     estimate
#       ↓
#     uncertainty
#       ↓
#     confidence interval
#
#
# Example idea:
#
#     estimate = 52
#
#     95% CI:
#
#         [48, 56]
#
#
# This does NOT mean:
#
#     "There is a 95% probability the fixed parameter is inside
#      this particular interval."
#
# Confidence intervals have a repeated-sampling interpretation.
#
#
# ============================================================
# 49. RESAMPLING
# ============================================================
#
# Modern SciPy includes resampling tools such as:
#
#     stats.bootstrap()
#     stats.permutation_test()
#
#
# These are VERY useful for data science.
#
#
# ------------------------------------------------------------
# Bootstrap
# ------------------------------------------------------------
#
# Original data:
#
#     [10, 12, 13, 15, 20]
#
# Resample WITH replacement:
#
#     [10, 10, 13, 20, 20]
#
# another:
#
#     [12, 13, 13, 15, 20]
#
# another:
#
#     [10, 12, 15, 15, 20]
#
#
# Repeat many times.
#
#
#     original sample
#           │
#           ↓
#     resample with replacement
#           │
#           ↓
#       calculate statistic
#           │
#           ↓
#        repeat many times
#           │
#           ↓
#     bootstrap distribution
#
#
# Useful for:
#
#     confidence intervals
#     uncertainty estimation
#     medians
#     custom statistics
#     ML evaluation metrics
#
#
# ============================================================
# 50. PERMUTATION TEST
# ============================================================
#
#     stats.permutation_test()
#
# Core idea:
#
# If the null hypothesis says group labels are irrelevant,
# shuffle the labels.
#
#
# Original:
#
#     Group A → A A A A
#     Group B → B B B B
#
# Shuffle:
#
#     A B A B
#     B A B A
#
# Recalculate the statistic many times.
#
#
# This creates a null distribution.
#
#
# Visual:
#
#     observed statistic
#              │
#              ↓
#     ────────████───────
#          null distribution
#
# If observed value lies far in the tail:
#
#     evidence against H0
#
#
# ============================================================
# 51. BOOTSTRAP vs PERMUTATION
# ============================================================
#
# BOOTSTRAP:
#
#     "How uncertain is my estimate?"
#
#
# PERMUTATION:
#
#     "Would this difference occur if the labels did not matter?"
#
#
# Simple mental model:
#
#     Bootstrap
#         ↓
#     uncertainty of statistic
#
#     Permutation
#         ↓
#     null hypothesis / significance
#
#
# ============================================================
# 52. RANKING FUNCTIONS
# ============================================================
#
#     stats.rankdata()
#
# Converts observations to ranks.
#
#
# Example:
#
#     values:
#
#     40   10   30   20
#
# ranks:
#
#      4    1    3    2
#
#
# Ranking is fundamental to:
#
#     Spearman correlation
#     Mann-Whitney
#     Wilcoxon
#     non-parametric statistics
#
#
# ============================================================
# 53. Z-SCORES
# ============================================================
#
#     stats.zscore(x)
#
#
# Standardization:
#
#     z = (x - mean) / standard deviation
#
#
# Example:
#
#     mean = 100
#     std = 10
#     x = 120
#
#     z = (120 - 100) / 10
#       = 2
#
#
# Meaning:
#
#     x is 2 standard deviations above the mean.
#
#
# Visual:
#
#     -3σ   -2σ   -1σ    μ    +1σ   +2σ   +3σ
#      │     │     │     │     │     │     │
#                        ↑
#                       120
#
#
# ============================================================
# 54. WHY Z-SCORE MATTERS FOR ML
# ============================================================
#
# Standardization is important for algorithms sensitive to scale.
#
# Examples:
#
#     KNN
#     K-means
#     SVM
#     PCA
#     many gradient-based models
#
#
# However:
#
# In practical ML pipelines, usually use:
#
#     sklearn.preprocessing.StandardScaler
#
# rather than manually using scipy.stats.zscore()
#
# Why?
#
# Because sklearn's transformer integrates correctly into:
#
#     train/test split
#     pipelines
#     cross-validation
#
#
# So:
#
#     scipy.stats.zscore()
#
# is excellent for:
#
#     analysis
#     understanding
#     quick transformations
#
# while:
#
#     StandardScaler
#
# is generally better for production ML pipelines.
#
#
# ============================================================
# 55. QUANTILES AND PERCENTILES
# ============================================================
#
# SciPy distributions heavily use:
#
#     CDF
#     PPF
#
# for quantiles.
#
#
# Quantile:
#
#     "What value corresponds to this cumulative probability?"
#
#
# Example:
#
#     50th percentile → median
#     90th percentile → value below which 90% lie
#     95th percentile → value below which 95% lie
#
#
# PPF:
#
#     probability → value
#
#
# ============================================================
# 56. TRIMMING AND WINSORIZATION
# ============================================================
#
# Useful functions include:
#
#     stats.trim_mean()
#     stats.mstats.winsorize()
#
#
# ------------------------------------------------------------
# Trimmed mean
# ------------------------------------------------------------
#
# Remove a percentage of extreme observations.
#
#
# Example:
#
#     1 2 3 4 5 6 7 1000
#
# Ordinary mean is strongly affected by 1000.
#
# Trimmed mean reduces the influence of extreme observations.
#
#
# ------------------------------------------------------------
# Winsorization
# ------------------------------------------------------------
#
# Extreme values are replaced by less extreme boundary values
# rather than completely removed.
#
#
# Important:
#
# Outlier treatment should be based on domain understanding,
# not an automatic "delete everything extreme" rule.
#
#
# ============================================================
# 57. IQR
# ============================================================
#
# Interquartile range:
#
#     IQR = Q3 - Q1
#
#
# Concept:
#
#     Q1
#      │
#      ├────────────── middle 50% ──────────────┤
#      │                                        │
#     Q1                                       Q3
#
#
# SciPy can support quantile-related calculations, but for
# everyday data analysis NumPy/pandas are generally more direct.
#
#
# ============================================================
# 58. OUTLIERS AND SCIPY
# ============================================================
#
# SciPy can help calculate:
#
#     z-scores
#     quantiles
#     robust statistics
#     distribution probabilities
#
#
# But remember:
#
#     OUTLIER ≠ ERROR
#
#
# An observation may be:
#
#     valid but unusual
#
# or:
#
#     invalid measurement
#
#
# Statistical detection:
#
#     "unusual"
#
# Data-quality investigation:
#
#     "wrong?"
#
#
# These are different questions.
#
#
# ============================================================
# 59. SPECIAL STATISTICAL FUNCTIONS
# ============================================================
#
# SciPy has:
#
#     scipy.special
#
# This module provides mathematical functions that appear in
# advanced statistics and ML.
#
#
# Important examples:
#
#     special.gamma()
#     special.gammaln()
#     special.erf()
#     special.erfc()
#     special.expit()
#     special.logit()
#     special.softmax()
#
#
# ============================================================
# 60. GAMMA FUNCTION
# ============================================================
#
#     special.gamma(x)
#
# Gamma generalizes factorial.
#
#
# For positive integers:
#
#     Γ(n) = (n - 1)!
#
#
# Therefore:
#
#     Γ(5) = 4! = 24
#
#
# Gamma appears in:
#
#     Gamma distribution
#     Beta distribution
#     Bayesian statistics
#     probability theory
#
#
# ============================================================
# 61. GAMMALN
# ============================================================
#
#     special.gammaln(x)
#
# Gives:
#
#     log(Gamma(x))
#
#
# This is important numerically because gamma values can become
# enormous.
#
#
# Instead of:
#
#     gamma(x)
#
# calculate:
#
#     log(gamma(x))
#
# directly when appropriate.
#
#
# This connects to a major numerical-computing principle:
#
#
#     very large products
#            ↓
#        logarithms
#            ↓
#       manageable sums
#
#
# ============================================================
# 62. ERROR FUNCTION
# ============================================================
#
#     special.erf(x)
#     special.erfc(x)
#
#
# These are strongly connected to the normal distribution.
#
#
# Standard normal CDF:
#
#     Φ(x) = 0.5 × [1 + erf(x / sqrt(2))]
#
#
# Therefore:
#
#     erf
#       ↓
#     normal probability
#
#
# This is one reason special functions are important in statistics.
#
#
# ============================================================
# 63. LOGIT
# ============================================================
#
#     special.logit(p)
#
# Definition:
#
#     logit(p) = log(p / (1 - p))
#
#
# It transforms:
#
#     probability
#          ↓
#     real number
#
#
# Example:
#
#     p = 0.5
#
#     logit(0.5) = 0
#
#
# p close to 0:
#
#     logit(p) → -∞
#
# p close to 1:
#
#     logit(p) → +∞
#
#
# This is fundamental to:
#
#     logistic regression
#     probability modeling
#
#
# ============================================================
# 64. EXPIT / SIGMOID
# ============================================================
#
#     special.expit(x)
#
# Definition:
#
#     expit(x) = 1 / (1 + exp(-x))
#
#
# This is the sigmoid function.
#
#
# Visual:
#
#     1.0 ───────────────
#                    ___
#                 __/
#              __/
#            _/
#          _/
#     0.0 ─────────────────
#             x
#
#
# It transforms:
#
#     real number
#          ↓
#     probability-like value [0, 1]
#
#
# Central to:
#
#     logistic regression
#     neural networks
#     binary classification
#
#
# ============================================================
# 65. SOFTMAX
# ============================================================
#
# SciPy also provides:
#
#     special.softmax()
#
#
# Softmax converts scores into probabilities:
#
#
#     raw scores
#
#     [2.0, 1.0, 0.1]
#          ↓
#       softmax
#          ↓
#     [0.66, 0.24, 0.10]   approximately
#
#
# Properties:
#
#     each probability ≥ 0
#
#     probabilities sum to 1
#
#
# Used in:
#
#     multiclass classification
#     neural networks
#
#
# In practical ML:
#
#     sklearn
#     PyTorch
#     TensorFlow
#
# usually handle this at the model level.
#
# SciPy is valuable for understanding and numerical utilities.
#
#
# ============================================================
# 66. SPATIAL DISTANCES
# ============================================================
#
# Module:
#
#     scipy.spatial
#
#
# This is VERY relevant to ML.
#
# Important:
#
#     spatial.distance
#
#
# Examples:
#
#     euclidean
#     cityblock
#     cosine
#     minkowski
#     chebyshev
#     jaccard
#
#
# ============================================================
# 67. EUCLIDEAN DISTANCE
# ============================================================
#
# Euclidean distance:
#
#     d = sqrt(
#             (x1-y1)^2 +
#             (x2-y2)^2 +
#             ...
#         )
#
#
# Visual:
#
#       B
#       ●
#       |\
#       | \
#       |  \  distance
#       |   \
#       |____\
#       ●     ●
#       A
#
#
# This is ordinary geometric distance.
#
#
# Important for:
#
#     KNN
#     K-means
#     clustering
#     similarity
#     anomaly detection
#
#
# ============================================================
# 68. MANHATTAN DISTANCE
# ============================================================
#
#     cityblock
#
#
# Distance:
#
#     |x1-y1| + |x2-y2| + ...
#
#
# Visual:
#
#     ●───────┐
#             │
#             │
#             ●
#
# You cannot travel diagonally.
#
#
# Useful when movement is grid-like or when L1 geometry is
# desirable.
#
#
# ============================================================
# 69. COSINE DISTANCE
# ============================================================
#
# Cosine similarity measures angle rather than magnitude.
#
#
# Visual:
#
#          B
#         /
#        /
#       / θ
#      /
#     ●──────── A
#
#
# Two vectors pointing in the same direction:
#
#     cosine similarity ≈ 1
#
#
# Opposite directions:
#
#     ≈ -1
#
#
# Common in:
#
#     text analysis
#     embeddings
#     recommendation
#     high-dimensional vectors
#
#
# ============================================================
# 70. LINEAR ALGEBRA
# ============================================================
#
# Module:
#
#     scipy.linalg
#
#
# NumPy already provides many linear algebra operations.
#
# SciPy provides more advanced routines.
#
#
# Important areas:
#
#     matrix decompositions
#     solving systems
#     eigenvalues/eigenvectors
#     matrix inversion
#     least squares
#
#
# Important functions:
#
#     linalg.solve()
#     linalg.lstsq()
#     linalg.inv()
#     linalg.det()
#     linalg.eig()
#     linalg.svd()
#     linalg.norm()
#
#
# ============================================================
# 71. SOLVING LINEAR SYSTEMS
# ============================================================
#
# Instead of:
#
#     A x = b
#
# you want:
#
#     x
#
#
# Use:
#
#     linalg.solve(A, b)
#
#
# This is much better than manually computing:
#
#     inv(A) @ b
#
# when the goal is simply to solve the system.
#
#
# Concept:
#
#     equations
#        ↓
#     matrix representation
#        ↓
#     numerical solver
#        ↓
#     solution
#
#
# ============================================================
# 72. LEAST SQUARES
# ============================================================
#
#     linalg.lstsq()
#
#
# This is closely connected to regression.
#
#
# Data:
#
#     x → y
#
# We want:
#
#     predicted y
#
# as close as possible to:
#
#     observed y
#
#
# Least squares minimizes:
#
#     Σ(y - ŷ)^2
#
#
# Concept:
#
#     observations
#          ↓
#     model predictions
#          ↓
#     residuals
#          ↓
#     squared residuals
#          ↓
#     minimize
#
#
# ============================================================
# 73. SVD
# ============================================================
#
#     linalg.svd()
#
# Singular Value Decomposition:
#
#     A = U Σ Vᵀ
#
#
# This is fundamental in:
#
#     PCA
#     dimensionality reduction
#     low-rank approximation
#     signal processing
#     recommender systems
#
#
# Mental model:
#
#     complicated matrix
#            ↓
#          SVD
#            ↓
#     rotations + scaling + rotations
#
#
# You do not need to derive SVD every time,
# but you should understand what it accomplishes.
#
#
# ============================================================
# 74. EIGENVALUES / EIGENVECTORS
# ============================================================
#
#     linalg.eig()
#
#
# Eigenvector:
#
#     direction that remains fundamentally aligned
#     under a linear transformation.
#
#
#     A v = λ v
#
#
# Important in:
#
#     PCA
#     dynamical systems
#     physics
#     spectral methods
#
#
# ============================================================
# 75. OPTIMIZATION
# ============================================================
#
# Module:
#
#     scipy.optimize
#
#
# This is extremely important for ML.
#
#
# Optimization means:
#
#     find parameters that minimize or maximize an objective.
#
#
# General ML structure:
#
#
#     parameters θ
#          ↓
#     model
#          ↓
#     predictions
#          ↓
#     loss function
#          ↓
#     optimization
#          ↓
#     better θ
#
#
# ============================================================
# 76. MINIMIZATION
# ============================================================
#
# Important:
#
#     optimize.minimize()
#
#
# Example concept:
#
#     f(x) = (x - 3)^2
#
#
# Minimum:
#
#     x = 3
#
#
# SciPy:
#
#     optimize.minimize(f, x0=0)
#
#
# Conceptual process:
#
#     x = 0
#       ↓
#     evaluate loss
#       ↓
#     search
#       ↓
#     evaluate
#       ↓
#     improve
#       ↓
#     x ≈ 3
#
#
# ============================================================
# 77. ROOT FINDING
# ============================================================
#
# Important:
#
#     optimize.root()
#     optimize.brentq()
#     optimize.newton()
#
#
# Root finding asks:
#
#     "Where does f(x) = 0?"
#
#
# Visual:
#
#     y
#     │       /
#     │      /
#     │─────●──────── x
#     │    /
#     │   /
#
# The root is where the curve crosses zero.
#
#
# Useful in:
#
#     scientific computing
#     numerical equations
#     engineering
#     statistical calculations
#
#
# ============================================================
# 78. CURVE FITTING
# ============================================================
#
#     optimize.curve_fit()
#
#
# Concept:
#
#     observed data
#          ↓
#     choose mathematical model
#          ↓
#     estimate parameters
#          ↓
#     fitted curve
#
#
# Example:
#
#     y = a*x + b
#
# Find:
#
#     a
#     b
#
# that best fit observed data.
#
#
# ============================================================
# 79. INTERPOLATION
# ============================================================
#
# Module:
#
#     scipy.interpolate
#
#
# Suppose:
#
#     x = 0 → y = 0
#     x = 1 → y = 10
#
# What about:
#
#     x = 0.5 ?
#
#
# Interpolation estimates values between known points.
#
#
# Visual:
#
#     ●
#       \
#        ● ← estimated
#       /
#     ●
#
#
# Important functions/classes include:
#
#     interp1d
#     CubicSpline
#     PchipInterpolator
#     UnivariateSpline
#
#
# Useful in:
#
#     missing-value estimation
#     scientific measurements
#     time-series processing
#     engineering data
#
#
# ============================================================
# 80. NUMERICAL INTEGRATION
# ============================================================
#
# Module:
#
#     scipy.integrate
#
#
# Important:
#
#     integrate.quad()
#     integrate.simpson()
#     integrate.trapezoid()
#
#
# Integration means:
#
#     area under a curve
#
#
# Visual:
#
#     y
#     │       /\
#     │      /██\
#     │     /████\
#     │____/██████\____
#     └──────────────── x
#
# shaded area
#      ↓
#    integral
#
#
# In probability:
#
#     area under PDF
#           ↓
#       probability
#
#
# ============================================================
# 81. NUMERICAL DIFFERENTIATION
# ============================================================
#
# SciPy also has tools for numerical derivatives and related
# scientific calculations, although symbolic differentiation is
# generally handled by SymPy and automatic differentiation by
# ML frameworks.
#
#
# Concept:
#
#     function
#        ↓
#     slope
#        ↓
#     derivative
#
#
# ML connection:
#
#     optimization
#          ↓
#     gradient
#          ↓
#     parameter update
#
#
# But for neural networks:
#
#     PyTorch / TensorFlow / JAX
#
# are normally used rather than SciPy.
#
#
# ============================================================
# 82. NUMERICAL INTEGRATION + PROBABILITY
# ============================================================
#
# A probability density must satisfy:
#
#     total area = 1
#
#
# Therefore:
#
#     integral(PDF over all x) = 1
#
#
# Probability between a and b:
#
#     integral from a to b of PDF(x)
#
#
# This explains why:
#
#     CDF
#
# is fundamentally connected to integration.
#
#
# ============================================================
# 83. SCIPY FOR DATA ANALYSIS
# ============================================================
#
# DA usefulness:
#
#     ████████████████████  scipy.stats
#     █████████████████     distributions
#     ███████████████       hypothesis testing
#     ███████████████       correlation
#     █████████████         resampling
#     ███████████           spatial distances
#     █████████             optimization
#     ███████               interpolation
#
#
# Typical DA workflow:
#
#
#     pandas
#       ↓
#     clean data
#       ↓
#     descriptive analysis
#       ↓
#     SciPy statistical tests
#       ↓
#     visualization
#       ↓
#     business conclusion
#
#
# ============================================================
# 84. SCIPY FOR DATA SCIENCE
# ============================================================
#
# DS usefulness:
#
#     VERY HIGH
#
# Especially:
#
#     scipy.stats
#     scipy.optimize
#     scipy.linalg
#     scipy.spatial
#     scipy.special
#
#
# Data Science mental map:
#
#
#       DATA
#        │
#        ↓
#     NumPy / pandas
#        │
#        ↓
#     SciPy
#        │
#     ┌──┼─────────────┐
#     ↓  ↓             ↓
#   stats optimize   spatial
#     │    │            │
#     ↓    ↓            ↓
# inference fitting   distances
#     │    │            │
#     └────┼────────────┘
#          ↓
#       analysis
#
#
# ============================================================
# 85. SCIPY FOR MACHINE LEARNING
# ============================================================
#
# SciPy is important, but it is NOT the primary ML library.
#
#
# Typical hierarchy:
#
#
#     NumPy
#       ↓
#     numerical foundation
#
#     SciPy
#       ↓
#     scientific/statistical algorithms
#
#     scikit-learn
#       ↓
#     classical machine learning
#
#     PyTorch / TensorFlow / JAX
#       ↓
#     deep learning / automatic differentiation
#
#
# SciPy is especially useful for:
#
#     optimization
#     distributions
#     statistical tests
#     distances
#     linear algebra
#     scientific preprocessing
#     special functions
#
#
# ============================================================
# 86. SCIPY + SCIKIT-LEARN
# ============================================================
#
# They are complementary.
#
#
# SciPy:
#
#     "I need a scientific/statistical algorithm."
#
# scikit-learn:
#
#     "I need a complete classical ML workflow."
#
#
# Example:
#
#     SciPy:
#
#         stats.pearsonr()
#
#     scikit-learn:
#
#         LinearRegression()
#
#
# SciPy:
#
#     statistical/scientific primitive
#
# sklearn:
#
#     ML estimator/pipeline ecosystem
#
#
# ============================================================
# 87. SCIPY + PANDAS
# ============================================================
#
# pandas:
#
#     table/dataframe manipulation
#
# SciPy:
#
#     mathematical/statistical computation
#
#
# Typical:
#
#     df
#      ↓
#     pandas
#      ↓
#     extract columns
#      ↓
#     NumPy / SciPy
#      ↓
#     statistical calculation
#
#
# Example:
#
#     group1 = df.loc[df["group"] == "A", "value"]
#     group2 = df.loc[df["group"] == "B", "value"]
#
#     stats.ttest_ind(group1, group2)
#
#
# ============================================================
# 88. SCIPY + STATISTICS LIBRARY
# ============================================================
#
# Python's:
#
#     statistics
#
# is excellent for:
#
#     mean
#     median
#     mode
#     variance
#     standard deviation
#     quantiles
#     correlation
#     simple descriptive statistics
#
#
# SciPy is much broader:
#
#     probability distributions
#     statistical tests
#     resampling
#     distribution fitting
#     advanced numerical statistics
#     optimization
#     scientific computing
#
#
# Mental model:
#
#     statistics
#          ↓
#     basic everyday statistics
#
#     scipy.stats
#          ↓
#     statistical analysis ecosystem
#
#
# ============================================================
# 89. SCIPY + NUMPY
# ============================================================
#
# NumPy:
#
#     arrays
#     vectorized operations
#     numerical computation
#
# SciPy:
#
#     specialized scientific algorithms
#
#
# Example:
#
#     np.mean(x)
#     np.std(x)
#
# vs.
#
#     stats.ttest_ind(x, y)
#     stats.norm.cdf(x)
#     stats.bootstrap(...)
#
#
# A useful rule:
#
#     NumPy = numerical foundation
#
#     SciPy = scientific algorithms
#
#
# ============================================================
# 90. THE MOST IMPORTANT scipy.stats FUNCTIONS FOR YOU
# ============================================================
#
# HIGH PRIORITY:
#
#     stats.describe()
#     stats.skew()
#     stats.kurtosis()
#     stats.zscore()
#
# Distributions:
#
#     stats.norm
#     stats.binom
#     stats.poisson
#     stats.t
#     stats.chi2
#     stats.f
#     stats.expon
#     stats.gamma
#     stats.beta
#     stats.lognorm
#     stats.weibull_min
#
# Distribution methods:
#
#     pdf()
#     pmf()
#     cdf()
#     ppf()
#     sf()
#     isf()
#     rvs()
#     fit()
#
# Tests:
#
#     ttest_1samp()
#     ttest_ind()
#     ttest_rel()
#     mannwhitneyu()
#     wilcoxon()
#     kruskal()
#     f_oneway()
#     chi2_contingency()
#     fisher_exact()
#     shapiro()
#     normaltest()
#     anderson()
#     kstest()
#
# Correlation:
#
#     pearsonr()
#     spearmanr()
#     kendalltau()
#
# Resampling:
#
#     bootstrap()
#     permutation_test()
#
# Data transformations:
#
#     zscore()
#     rankdata()
#     trim_mean()
#
#
# ============================================================
# 91. SECOND-TIER FUNCTIONS FOR DA/DS/ML
# ============================================================
#
# Very useful when the problem requires them:
#
#     stats.gmean()
#     stats.hmean()
#     stats.variation()
#     stats.probplot()
#     stats.levene()
#     stats.trim_mean()
#     stats.mstats.winsorize()
#
# Distribution families:
#
#     gamma
#     beta
#     lognorm
#     weibull_min
#     uniform
#     expon
#     multivariate_normal
#
#
# ============================================================
# 92. SPECIALIZED SCIENTIFIC FUNCTIONS
# ============================================================
#
# Learn when your work requires them:
#
#     scipy.special
#     scipy.linalg
#     scipy.optimize
#     scipy.interpolate
#     scipy.integrate
#     scipy.spatial
#
#
# These are not "extra useless functions."
#
# They become extremely important in specialized DS/ML/research.
#
#
# ============================================================
# 93. A VERY IMPORTANT DISTINCTION:
#     FUNCTION vs DISTRIBUTION OBJECT
# ============================================================
#
# Example:
#
#     stats.norm
#
# is a distribution object/family.
#
#
# Then:
#
#     stats.norm.pdf(...)
#     stats.norm.cdf(...)
#     stats.norm.ppf(...)
#
# operate on that distribution.
#
#
# Think:
#
#     norm
#      │
#      ├── pdf
#      ├── cdf
#      ├── ppf
#      ├── sf
#      ├── isf
#      └── rvs
#
#
# This pattern makes SciPy much easier to learn.
#
#
# ============================================================
# 94. A COMPLETE NORMAL-DISTRIBUTION EXAMPLE
# ============================================================
#
#     from scipy import stats
#
#     dist = stats.norm(loc=100, scale=15)
#
#     # Density at 120
#     density = dist.pdf(120)
#
#     # Probability X <= 120
#     probability = dist.cdf(120)
#
#     # Value corresponding to 95th percentile
#     value = dist.ppf(0.95)
#
#     # Generate 100 observations
#     sample = dist.rvs(size=100, random_state=42)
#
#
# Conceptual flow:
#
#
#     Normal(100, 15)
#          │
#          ├── pdf(120)
#          │       ↓
#          │     density
#          │
#          ├── cdf(120)
#          │       ↓
#          │     P(X ≤ 120)
#          │
#          ├── ppf(.95)
#          │       ↓
#          │     95th percentile
#          │
#          └── rvs()
#                  ↓
#              synthetic data
#
#
# ============================================================
# 95. A STATISTICAL TEST EXAMPLE
# ============================================================
#
#     from scipy import stats
#
#     before = [70, 72, 68, 75, 71]
#     after  = [74, 75, 70, 78, 76]
#
#     result = stats.ttest_rel(before, after)
#
#
# Execution:
#
#     paired observations
#             ↓
#     calculate differences
#             ↓
#     test mean difference
#             ↓
#     t statistic
#             ↓
#     p-value
#
#
# The result typically contains:
#
#     statistic
#     pvalue
#
#
# Do not stop at:
#
#     "p < 0.05"
#
# Also examine:
#
#     direction
#     effect size
#     confidence interval
#     practical meaning
#
#
# ============================================================
# 96. CORRELATION EXAMPLE
# ============================================================
#
#     x = [1, 2, 3, 4, 5]
#     y = [2, 4, 5, 8, 10]
#
#     result = stats.pearsonr(x, y)
#
#
# Concept:
#
#     x
#     │        *
#     │      *
#     │    *
#     │  *
#     │ *
#     └──────────── y
#
#
# Output conceptually:
#
#     correlation coefficient
#     p-value
#
#
# ============================================================
# 97. DISTRIBUTION FITTING EXAMPLE
# ============================================================
#
#     data = ...
#
#     mu, sigma = stats.norm.fit(data)
#
#
# Concept:
#
#     real observations
#          ↓
#     assume normal distribution
#          ↓
#     estimate μ and σ
#          ↓
#     fitted model
#
#
# This is a simple example of:
#
#     parameter estimation
#
#
# ============================================================
# 98. SCIPY AND MAXIMUM LIKELIHOOD
# ============================================================
#
# Distribution fitting is closely related to likelihood.
#
#
# We want parameters θ that make the observed data most plausible:
#
#
#     data
#       ↓
#     likelihood(θ | data)
#       ↓
#     choose θ maximizing likelihood
#
#
# Usually it is easier numerically to maximize:
#
#     log-likelihood
#
# rather than:
#
#     likelihood
#
#
# because:
#
#     product of probabilities
#          ↓
#     log
#          ↓
#     sum of log-probabilities
#
#
# This idea appears throughout:
#
#     statistics
#     ML
#     Bayesian inference
#
#
# ============================================================
# 99. SCIPY OPTIMIZATION + ML
# ============================================================
#
# A machine-learning model often has:
#
#     parameters θ
#
# and:
#
#     loss L(θ)
#
#
# Goal:
#
#     minimize L(θ)
#
#
# SciPy:
#
#     optimize.minimize()
#
# can solve many general optimization problems.
#
#
# However:
#
# Deep-learning frameworks generally use:
#
#     gradient descent
#     Adam
#     SGD
#     automatic differentiation
#
# through:
#
#     PyTorch
#     TensorFlow
#     JAX
#
#
# So SciPy optimization is highly useful for classical/scientific
# optimization, but is not the standard deep-learning optimizer.
#
#
# ============================================================
# 100. SCIPY SPATIAL + ML
# ============================================================
#
# Many ML algorithms depend on distances.
#
#
#     observations
#          ↓
#     represent as vectors
#          ↓
#     calculate distance
#          ↓
#     determine similarity
#
#
# Examples:
#
#     KNN
#     K-means
#     clustering
#     nearest-neighbor search
#     anomaly detection
#
#
# Important:
#
#     scipy.spatial.distance
#
#
# ============================================================
# 101. KD-TREE / SPATIAL SEARCH
# ============================================================
#
# scipy.spatial also provides structures such as:
#
#     KDTree
#     cKDTree
#
#
# Concept:
#
#     many points
#          ↓
#     organize spatially
#          ↓
#     efficiently find nearby points
#
#
# Useful for:
#
#     nearest-neighbor searches
#     spatial data
#     geometric problems
#     scientific computing
#
#
# ============================================================
# 102. SCIPY AND MULTIVARIATE NORMAL DISTRIBUTION
# ============================================================
#
#     stats.multivariate_normal
#
#
# Normal distribution can extend from:
#
#     one variable
#
# to:
#
#     many variables
#
#
# One-dimensional:
#
#     X
#
# Two-dimensional:
#
#     [X1, X2]
#
# Three-dimensional:
#
#     [X1, X2, X3]
#
#
# Multivariate normal requires:
#
#     mean vector
#     covariance matrix
#
#
# This is fundamental for:
#
#     multivariate statistics
#     Gaussian models
#     covariance analysis
#     probabilistic ML
#
#
# ============================================================
# 103. COVARIANCE AS GEOMETRY
# ============================================================
#
# Suppose:
#
#     X increases
#     Y increases
#
# then:
#
#     positive covariance
#
#
# If:
#
#     X increases
#     Y decreases
#
# then:
#
#     negative covariance
#
#
# Covariance matrix:
#
#
#             X       Y
#
#     X      VarX    CovXY
#     Y      CovXY   VarY
#
#
# This matrix becomes fundamental in:
#
#     multivariate normal
#     PCA
#     Gaussian models
#     statistical learning
#
#
# ============================================================
# 104. SCIPY AND PCA
# ============================================================
#
# PCA can be understood through:
#
#     covariance matrix
#          ↓
#     eigenvectors
#          ↓
#     principal directions
#
#
# or through:
#
#     SVD
#
#
# SciPy can provide the underlying linear algebra.
#
# But in practical ML:
#
#     sklearn.decomposition.PCA
#
# is usually the better high-level tool.
#
#
# ============================================================
# 105. SCIPY AND SCIENTIFIC DATA
# ============================================================
#
# SciPy becomes particularly valuable when your data are:
#
#     physical
#     engineering
#     biological
#     financial
#     experimental
#     sensor-based
#     time-dependent
#
#
# Example:
#
#     sensor measurements
#          ↓
#     NumPy
#          ↓
#     SciPy filtering / interpolation / statistics
#          ↓
#     ML model
#
#
# This is one reason SciPy is highly relevant to engineering
# and research-oriented data science.
#
#
# ============================================================
# 106. WHAT YOU SHOULD MEMORIZE
# ============================================================
#
# Do NOT memorize 100+ functions.
#
# Memorize the CONCEPTUAL FAMILIES.
#
#
# ------------------------------------------------------------
# Statistics
# ------------------------------------------------------------
#
#     describe
#     skew
#     kurtosis
#     zscore
#
#
# ------------------------------------------------------------
# Probability
# ------------------------------------------------------------
#
#     distribution
#     pdf / pmf
#     cdf
#     ppf
#     sf
#     rvs
#     fit
#
#
# ------------------------------------------------------------
# Tests
# ------------------------------------------------------------
#
#     ttest
#     Mann-Whitney
#     Wilcoxon
#     ANOVA
#     chi-square
#     Fisher
#     normality tests
#
#
# ------------------------------------------------------------
# Correlation
# ------------------------------------------------------------
#
#     Pearson
#     Spearman
#     Kendall
#
#
# ------------------------------------------------------------
# Resampling
# ------------------------------------------------------------
#
#     bootstrap
#     permutation_test
#
#
# ------------------------------------------------------------
# Scientific computing
# ------------------------------------------------------------
#
#     linalg
#     optimize
#     spatial
#     interpolate
#     integrate
#     special
#
#
# ============================================================
# 107. PRIORITY FOR DATA ANALYSIS
# ============================================================
#
#                 DA PRIORITY
#
#     ████████████████████████
#     scipy.stats
#
#     █████████████████
#     distributions
#
#     ███████████████
#     correlation
#
#     ███████████████
#     hypothesis tests
#
#     ████████████
#     resampling
#
#     █████████
#     spatial
#
#     ███████
#     optimize
#
#
# Most important:
#
#     scipy.stats
#
#
# ============================================================
# 108. PRIORITY FOR DATA SCIENCE
# ============================================================
#
#                 DS PRIORITY
#
#     ████████████████████████
#     scipy.stats
#
#     ███████████████████
#     scipy.linalg
#
#     █████████████████
#     scipy.optimize
#
#     ███████████████
#     scipy.spatial
#
#     █████████████
#     scipy.special
#
#     █████████
#     scipy.interpolate
#
#     ███████
#     scipy.integrate
#
#
# ============================================================
# 109. PRIORITY FOR CLASSICAL MACHINE LEARNING
# ============================================================
#
#                 ML PRIORITY
#
#     ████████████████████
#     scipy.optimize
#
#     ██████████████████
#     scipy.linalg
#
#     █████████████████
#     scipy.spatial
#
#     ████████████████
#     scipy.stats
#
#     █████████████
#     scipy.special
#
#
# BUT:
#
#     scikit-learn
#
# remains your primary classical ML library.
#
#
# ============================================================
# 110. PRIORITY FOR DEEP LEARNING
# ============================================================
#
# SciPy is useful, but it is NOT your main DL library.
#
#
# Priority:
#
#     PyTorch / TensorFlow / JAX
#          ↓
#     NumPy
#          ↓
#     SciPy
#
#
# SciPy still helps with:
#
#     statistics
#     preprocessing
#     scientific calculations
#     evaluation
#     optimization outside the neural network
#     numerical utilities
#
#
# ============================================================
# 111. YOUR PRACTICAL SCIPY LEARNING ORDER
# ============================================================
#
# Since your goal is:
#
#     DA → DS → ML
#
# use this order:
#
#
# STEP 1
#     scipy.stats basics
#
# STEP 2
#     probability distributions
#
# STEP 3
#     PDF / PMF / CDF / PPF / SF / RVS
#
# STEP 4
#     t-tests
#
# STEP 5
#     correlation
#
# STEP 6
#     chi-square / categorical analysis
#
# STEP 7
#     ANOVA
#
# STEP 8
#     non-parametric tests
#
# STEP 9
#     bootstrap / permutation
#
# STEP 10
#     distribution fitting
#
# STEP 11
#     scipy.spatial.distance
#
# STEP 12
#     scipy.linalg
#
# STEP 13
#     scipy.optimize
#
# STEP 14
#     scipy.special
#
# STEP 15
#     interpolate / integrate when your projects need them
#
#
# ============================================================
# 112. THE BIGGER PYTHON STATISTICS ECOSYSTEM
# ============================================================
#
#
#                  STATISTICAL PYTHON
#                         │
#       ┌─────────────────┼─────────────────┐
#       ↓                 ↓                 ↓
#    statistics          NumPy            SciPy
#       │                 │                 │
# basic statistics    arrays/math      statistics +
#       │                 │             scientific algorithms
#       │                 │                 │
#       └─────────────────┼─────────────────┘
#                         ↓
#                      pandas
#                         │
#                    data analysis
#                         ↓
#                    statsmodels
#                         │
#              statistical modeling /
#                 econometrics
#                         ↓
#                  scikit-learn
#                         │
#                 machine learning
#
#
# This distinction is extremely important.
#
#
# ============================================================
# 113. SCIPY vs STATISTICS vs NUMPY
# ============================================================
#
# ------------------------------------------------------------
# statistics
# ------------------------------------------------------------
#
# Best for:
#
#     basic descriptive statistics
#     small/simple Python collections
#     learning statistical concepts
#
#
# ------------------------------------------------------------
# NumPy
# ------------------------------------------------------------
#
# Best for:
#
#     arrays
#     vectorized numerical computation
#     numerical statistics
#     matrix-like operations
#
#
# ------------------------------------------------------------
# SciPy
# ------------------------------------------------------------
#
# Best for:
#
#     statistical tests
#     probability distributions
#     scientific algorithms
#     optimization
#     advanced linear algebra
#     distances
#     special functions
#
#
# ============================================================
# 114. SCIPY vs STATSMODELS
# ============================================================
#
# SciPy:
#
#     "Give me statistical/numerical tools."
#
#
# statsmodels:
#
#     "Build and interpret statistical models."
#
#
# Example:
#
# SciPy:
#
#     t-test
#     correlation
#     distribution
#
#
# statsmodels:
#
#     OLS regression
#     regression coefficients
#     standard errors
#     confidence intervals
#     model summaries
#     statistical inference
#
#
# Both are important.
#
#
# ============================================================
# 115. SCIPY vs SCIKIT-LEARN
# ============================================================
#
# SciPy:
#
#     mathematical/statistical machinery
#
#
# scikit-learn:
#
#     ML workflow
#
#
# Example:
#
#     SciPy
#       ↓
#     distance
#
#     sklearn
#       ↓
#     KNN classifier
#
#
# SciPy:
#
#     optimization algorithm
#
# sklearn:
#
#     model training API
#
#
# ============================================================
# 116. THE MOST IMPORTANT SCIPY CONCEPTUAL PATTERN
# ============================================================
#
#
#              REAL DATA
#                  │
#                  ↓
#             mathematical
#              representation
#                  │
#          ┌───────┼────────┐
#          ↓       ↓        ↓
#       stats    linalg   spatial
#          │       │        │
#          ↓       ↓        ↓
#      inference  models  distance
#          │       │        │
#          └───────┼────────┘
#                  ↓
#             interpretation
#
#
# SciPy is a TOOLBOX.
#
# It does not tell you automatically:
#
#     "which statistical method should I use?"
#
# You must understand:
#
#     data type
#     research question
#     assumptions
#     scale
#     distribution
#     sample size
#     dependence structure
#
#
# ============================================================
# 117. COMMON MISTAKE:
#     USING EVERY TEST JUST BECAUSE IT EXISTS
# ============================================================
#
# Bad workflow:
#
#     data
#       ↓
#     run 20 tests
#       ↓
#     find p < 0.05
#       ↓
#     report it
#
#
# Better:
#
#     research question
#          ↓
#     variable types
#          ↓
#     study design
#          ↓
#     assumptions
#          ↓
#     appropriate method
#          ↓
#     statistic
#          ↓
#     uncertainty / p-value
#          ↓
#     effect size
#          ↓
#     practical interpretation
#
#
# ============================================================
# 118. COMMON MISTAKE:
#     THINKING PDF IS PROBABILITY
# ============================================================
#
# Wrong:
#
#     pdf(x) = probability of x
#
# Better:
#
#     PDF = density
#
#     probability = area under PDF
#
#
#     P(a ≤ X ≤ b)
#
#             ↓
#
#     area from a to b
#
#
# CDF gives that accumulated area:
#
#     CDF(b) - CDF(a)
#
#
# ============================================================
# 119. COMMON MISTAKE:
#     CONFUSING CDF AND PPF
# ============================================================
#
# CDF:
#
#     x → probability
#
#
# PPF:
#
#     probability → x
#
#
# Example:
#
#     CDF(1.96) ≈ 0.975
#
#     PPF(0.975) ≈ 1.96
#
#
# Think:
#
#     CDF = forward direction
#
#     PPF = reverse direction
#
#
# ============================================================
# 120. COMMON MISTAKE:
#     CONFUSING CORRELATION WITH CAUSATION
# ============================================================
#
# scipy.stats.pearsonr(x, y)
#
# can tell you about association.
#
# It does NOT prove:
#
#     X causes Y
#
#
# Possible explanations:
#
#     X → Y
#     Y → X
#     Z → X and Y
#     coincidence
#
#
# Statistics provides evidence.
#
# Study design determines causal strength.
#
#
# ============================================================
# 121. COMMON MISTAKE:
#     IGNORING DATA SCALE
# ============================================================
#
# Before choosing a statistical method ask:
#
#     Is the variable:
#
#         nominal?
#         ordinal?
#         interval?
#         ratio?
#
#
# And:
#
#     numerical?
#     categorical?
#     binary?
#     count?
#     proportion?
#     time-to-event?
#
#
# Distribution choice depends strongly on data structure.
#
#
# ============================================================
# 122. COMMON MISTAKE:
#     BLINDLY TRUSTING NORMALITY TESTS
# ============================================================
#
# A normality test can be statistically significant because
# even a small deviation becomes detectable with a large sample.
#
#
# Therefore use:
#
#     Q-Q plot
#     histogram
#     skewness
#     domain knowledge
#     sample size
#     robustness of intended method
#
#
# Statistics is not:
#
#     "run test → obey p-value"
#
#
# ============================================================
# 123. THE CORE SCIPY.STATS MAP
# ============================================================
#
#
#                     scipy.stats
#                          │
#        ┌─────────────────┼──────────────────┐
#        ↓                 ↓                  ↓
#   DESCRIPTIVE       DISTRIBUTIONS        INFERENCE
#        │                 │                  │
#     skew             normal             t-test
#     kurtosis         binomial           ANOVA
#     zscore            Poisson            chi-square
#     describe          gamma              non-parametric
#                       beta
#                       Weibull
#                          │
#                          ↓
#                  PDF / CDF / PPF
#                  SF / RVS / FIT
#
#        ┌─────────────────┼──────────────────┐
#        ↓                 ↓                  ↓
#   CORRELATION       RESAMPLING        CATEGORICAL
#        │                 │                  │
#     Pearson          bootstrap          chi-square
#     Spearman         permutation         Fisher
#     Kendall
#
#
# ============================================================
# 124. FINAL "WHAT DO I ACTUALLY NEED?" LIST
# ============================================================
#
# For YOUR DA → DS → ML path:
#
#
# █████████████████████████████████████████████
# MUST KNOW
# █████████████████████████████████████████████
#
#     scipy.stats
#
#     stats.norm
#     stats.binom
#     stats.poisson
#     stats.t
#     stats.chi2
#     stats.f
#
#     pdf
#     pmf
#     cdf
#     ppf
#     sf
#     rvs
#
#     ttest_1samp
#     ttest_ind
#     ttest_rel
#
#     pearsonr
#     spearmanr
#
#     chi2_contingency
#     fisher_exact
#
#     f_oneway
#
#     mannwhitneyu
#     wilcoxon
#     kruskal
#
#     shapiro
#     normaltest
#     probplot
#
#     bootstrap
#     permutation_test
#
#     zscore
#     skew
#     kurtosis
#     describe
#
#
# █████████████████████████████████████████████
# SHOULD KNOW
# █████████████████████████████████████████████
#
#     scipy.spatial.distance
#
#     euclidean
#     cityblock
#     cosine
#     minkowski
#
#     scipy.linalg
#
#     solve
#     lstsq
#     eig
#     svd
#     norm
#
#     scipy.optimize
#
#     minimize
#     root
#     curve_fit
#
#     scipy.special
#
#     expit
#     logit
#     softmax
#     gamma
#     gammaln
#     erf
#     erfc
#
#
# █████████████████████████████████████████████
# LEARN WHEN NEEDED
# █████████████████████████████████████████████
#
#     scipy.interpolate
#     scipy.integrate
#     advanced scipy.linalg
#     advanced scipy.optimize
#     advanced distributions
#     signal processing
#     sparse matrices
#     ODE/PDE-related scientific computing
#
#
# ============================================================
# 125. THE BIG PICTURE
# ============================================================
#
#
#                 PYTHON DATA SCIENCE
#
#                        │
#       ┌────────────────┼────────────────┐
#       ↓                ↓                ↓
#     pandas           NumPy           SciPy
#       │                │                │
#    tables          arrays/math      algorithms
#       │                │                │
#       └────────────────┼────────────────┘
#                        ↓
#              statistical analysis
#                        │
#             ┌──────────┴──────────┐
#             ↓                     ↓
#        statsmodels          scikit-learn
#             │                     │
#      statistical models       ML models
#             │                     │
#             └──────────┬──────────┘
#                        ↓
#                     ML / DS
#
#
# ============================================================
# 126. CORE IDEA TO REMEMBER
# ============================================================
#
# Do not think:
#
#     "SciPy = a huge list of functions I must memorize."
#
#
# Think:
#
#
#     SciPy
#       │
#       ├── stats
#       │     ├── distributions
#       │     ├── tests
#       │     ├── correlation
#       │     ├── resampling
#       │     └── statistical transformations
#       │
#       ├── linalg
#       │     └── matrix mathematics
#       │
#       ├── optimize
#       │     └── find best parameters / solutions
#       │
#       ├── spatial
#       │     └── distance / geometry / neighbors
#       │
#       ├── special
#       │     └── advanced mathematical functions
#       │
#       ├── interpolate
#       │     └── estimate between known values
#       │
#       └── integrate
#             └── calculate accumulated area / integrals
#
#
# The MOST IMPORTANT branch for your Statistics → DA → DS → ML
# learning path is:
#
#
#                    scipy.stats
#                         │
#        ┌────────────────┼─────────────────┐
#        ↓                ↓                 ↓
#   distributions       tests          correlation
#        │                │                 │
#   PDF/CDF/PPF        t-tests          Pearson
#   PMF/RVS            ANOVA            Spearman
#   fitting            chi-square       Kendall
#        │                │                 │
#        └────────────────┼─────────────────┘
#                         ↓
#                    inference
#                         │
#                    bootstrap
#                    permutation
#                         ↓
#                  DATA SCIENCE
#
#
# ============================================================
# 127. FINAL PRACTICAL RULE
# ============================================================
#
# For your learning:
#
#     statistics library
#         ↓
#     learn basic statistical concepts
#
#     NumPy
#         ↓
#     learn numerical arrays and vectorized mathematics
#
#     SciPy
#         ↓
#     learn statistical/scientific algorithms
#
#     pandas
#         ↓
#     learn real-world tabular data analysis
#
#     statsmodels
#         ↓
#     learn statistical modeling and inference
#
#     scikit-learn
#         ↓
#     learn classical machine learning
#
#
# You therefore do NOT need to learn every SciPy submodule
# before moving to DA/DS.
#
# The high-value core is:
#
#     scipy.stats
#     scipy.spatial
#     scipy.linalg
#     scipy.optimize
#     scipy.special
#
# and then learn the remaining SciPy modules
# project-by-project.
#
#
# ============================================================
# END
# ============================================================
```
