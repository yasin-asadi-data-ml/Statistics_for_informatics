2. Basic Statistical Concepts
2.1 Population vs. Sample
Population: All cars ever made (or all cars of a certain type).

Sample: Our dataset, which is a subset of all cars. We'll use sample statistics to infer population parameters.

2.2 Types of Variables
Quantitative (numerical):

Continuous: e.g., mpg, weight, acceleration.

Discrete: e.g., cylinders, model_year.

Categorical (qualitative): e.g., origin, name (though name is unique).

2.3 Descriptive vs. Inferential Statistics
Descriptive: Summarize and describe the data (mean, median, variance, plots).

Inferential: Make predictions or generalizations about a population based on sample data (hypothesis tests, confidence intervals).

3. Descriptive Statistics
3.1 Measures of Central Tendency
Mean: The average value.
Median: The middle value when sorted.
Mode: The most frequent value.

Let's compute these for mpg (fuel efficiency).

```python
# Central tendency for mpg
mean_mpg = np.mean(mpg_clean['mpg'])
median_mpg = np.median(mpg_clean['mpg'])
mode_mpg = stats.mode(mpg_clean['mpg'])[0][0]  # scipy mode returns array

print(f"Mean MPG: {mean_mpg:.2f}")
print(f"Median MPG: {median_mpg:.2f}")
print(f"Mode MPG: {mode_mpg}")
```
