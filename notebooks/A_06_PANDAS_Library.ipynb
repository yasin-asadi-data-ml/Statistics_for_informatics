# PANDAS — CONCEPTUAL AND PRACTICAL GUIDE FOR DA / DS / ML


```python
# ============================================================
# PANDAS — CONCEPTUAL AND PRACTICAL GUIDE FOR DA / DS / ML
# ============================================================
#
# Pandas is one of the central Python libraries for working with
# structured data.
#
# The easiest mental model is:
#
#                  PANDAS
#                    │
#          ┌─────────┴─────────┐
#          ↓                   ↓
#       Series             DataFrame
#      1-dimensional       2-dimensional
#       labeled data        labeled table
#
#
# A DataFrame can be mentally viewed as:
#
#                    COLUMNS
#             ┌────────┬────────┬────────┐
#             │ age    │ salary │ city   │
#       ┌─────┼────────┼────────┼────────┤
# ROWS  │  0  │  25     │ 3000   │ Tehran │
#       │  1  │  31     │ 4500   │ Karaj  │
#       │  2  │ 28      │ 3800   │ Tabriz │
#       └─────┴────────┴────────┴────────┘
#
# Pandas gives you tools to:
#
#     LOAD
#       ↓
#     INSPECT
#       ↓
#     CLEAN
#       ↓
#     SELECT
#       ↓
#     TRANSFORM
#       ↓
#     FILTER
#       ↓
#     GROUP
#       ↓
#     COMBINE
#       ↓
#     RESHAPE
#       ↓
#     ANALYZE
#       ↓
#     EXPORT
#
#
# For Data Analysis (DA):
#     Pandas is one of the most important libraries.
#
# For Data Science (DS):
#     Pandas is fundamental for data preparation and exploration.
#
# For Machine Learning (ML):
#     Pandas is extremely useful before the data enters NumPy,
#     scikit-learn, XGBoost, PyTorch, etc.
#
# A very important practical boundary:
#
#     Pandas = labeled/tabular data manipulation
#     NumPy  = numerical array computation
#
# They work together constantly.
#
#
# ============================================================
# 01. THE TWO CORE PANDAS OBJECTS
# ============================================================
#
# Pandas primarily revolves around:
#
#     Series
#     DataFrame
#
# These are not just "Python lists and tables".
#
# They are DATA STRUCTURES WITH LABELS.
#
#
# SERIES
# ------
#
# A Series is approximately:
#
#     index        values
#       ↓             ↓
#       0            85
#       1            91
#       2            76
#
# Think:
#
#              INDEX
#                │
#                ↓
#        ┌───────┬───────┐
#        │   0   │  85   │
#        │   1   │  91   │
#        │   2   │  76   │
#        └───────┴───────┘
#                         ↑
#                       values
#
#
# DATAFRAME
# ---------
#
# A DataFrame is a collection of aligned Series.
#
#             DataFrame
#                 │
#       ┌─────────┼─────────┐
#       ↓         ↓         ↓
#      age      salary     city
#    Series    Series     Series
#
# Each column is a Series sharing the same row index.
#
#
# ============================================================
# 02. IMPORTING PANDAS
# ============================================================
#
import pandas as pd
#
# "pd" is simply the conventional alias.
#
# It allows:
#
#     pd.DataFrame()
#     pd.Series()
#     pd.read_csv()
#
# instead of:
#
#     pandas.DataFrame()
#
#
# ============================================================
# 03. CREATING A SERIES
# ============================================================
#
ages = pd.Series([25, 31, 28, 42])
#
# Conceptually:
#
#     Python list
#          ↓
#     pd.Series()
#          ↓
#     labeled one-dimensional structure
#
# Result:
#
#     0    25
#     1    31
#     2    28
#     3    42
#
# The left side is the INDEX.
# The right side contains the VALUES.
#
#
# ============================================================
# 04. CREATING A SERIES WITH A CUSTOM INDEX
# ============================================================
#
ages = pd.Series(
#     [25, 31, 28],
#     index=["Ali", "Sara", "Reza"]
# )
#
# Now:
#
#     Ali     → 25
#     Sara    → 31
#     Reza    → 28
#
# The index provides labels.
#
# IMPORTANT:
#
# Index labels are not necessarily numerical positions.
#
# This distinction becomes extremely important later.
#
#
# ============================================================
# 05. CREATING A DATAFRAME
# ============================================================
#
df = pd.DataFrame({
#     "name": ["Ali", "Sara", "Reza", "Mina"],
#     "age": [25, 31, 28, 42],
#     "salary": [3000, 4500, 3800, 6000]
# })
#
# Think:
#
#     dictionary
#         ↓
#     DataFrame
#         ↓
#
#     name     age     salary
#     -----------------------
#     Ali      25      3000
#     Sara     31      4500
#     Reza     28      3800
#     Mina     42      6000
#
#
# Each dictionary key becomes a COLUMN.
#
# Each list becomes the values of that column.
#
#
# ============================================================
# 06. DATAFRAME MENTAL MODEL
# ============================================================
#
# A DataFrame is best understood as:
#
#                    DATAFRAME
#                        │
#       ┌────────────────┼────────────────┐
#       ↓                ↓                ↓
#    columns            index            values
#       │                │                │
#       ↓                ↓                ↓
#     age             0,1,2,3         actual data
#    salary
#     name
#
#
# More technically:
#
#     DataFrame
#        ↓
#     labeled 2D data
#        ↓
#     rows + columns
#
#
# ============================================================
# 07. BASIC DATAFRAME PROPERTIES
# ============================================================
#
# Commonly used attributes:
#
#     df.shape
#     df.ndim
#     df.size
#     df.columns
#     df.index
#     df.dtypes
#     df.values
#     df.empty
#
#
# IMPORTANT:
#
# These are mostly ATTRIBUTES, not methods.
#
# Therefore:
#
#     df.shape       # correct
#     df.shape()     # incorrect
#
#
# ============================================================
# 08. SHAPE
# ============================================================
#
print(df.shape)
#
# Example:
#
#     (4, 3)
#
# means:
#
#     4 rows
#     3 columns
#
# Visual:
#
#              3 columns
#        ┌──────────────────┐
#  4     │                  │
#  rows  │     DataFrame    │
#        │                  │
#        └──────────────────┘
#
#
# GENERAL PATTERN:
#
#     df.shape
#
#     → (number_of_rows, number_of_columns)
#
# Very important in DA / DS / ML.
#
#
# ============================================================
# 09. NDIM AND SIZE
# ============================================================
#
print(df.ndim)
print(df.size)
#
# For a normal DataFrame:
#
#     ndim = 2
#
#     size = rows × columns
#
# Example:
#
#     4 × 3 = 12
#
#
# ============================================================
# 10. COLUMNS
# ============================================================
#
print(df.columns)
#
# Gives the column labels.
#
# Conceptually:
#
#     DataFrame
#          │
#          ├── name
#          ├── age
#          └── salary
#
#
# ============================================================
# 11. INDEX
# ============================================================
#
print(df.index)
#
# The index identifies rows.
#
# Default:
#
#     0
#     1
#     2
#     3
#
#
# IMPORTANT MENTAL MODEL:
#
#     column labels → identify columns
#     index labels  → identify rows
#
#
# ============================================================
# 12. DTYPES
# ============================================================
#
print(df.dtypes)
#
# Example conceptually:
#
#     name       object/string
#     age        int64
#     salary     int64
#
# Dtype tells Pandas how the values are represented.
#
# This matters because operations depend on data type.
#
#
# ============================================================
# 13. VALUES
# ============================================================
#
print(df.values)
#
# This exposes the underlying values as a NumPy-style array.
#
# IMPORTANT:
#
# In modern Pandas workflows, normally prefer:
#
#     df
#
# or:
#
#     df.to_numpy()
#
# when you specifically need a NumPy array.
#
#
# ============================================================
# 14. EMPTY
# ============================================================
#
print(df.empty)
#
# Returns:
#
#     True
#     or
#     False
#
# It answers:
#
#     "Does this DataFrame contain zero rows or columns?"
#
#
# ============================================================
# 15. INSPECTING DATA
# ============================================================
#
# One of the first jobs in data analysis is:
#
#     DO NOT ANALYZE THE DATA YET.
#
#     FIRST UNDERSTAND THE DATA.
#
# Typical workflow:
#
#     df.head()
#        ↓
#     df.tail()
#        ↓
#     df.shape
#        ↓
#     df.info()
#        ↓
#     df.describe()
#        ↓
#     df.dtypes
#        ↓
#     missing-value checks
#
#
# ============================================================
# 16. head()
# ============================================================
#
print(df.head())
#
# Shows the first 5 rows by default.
#
#     df.head(10)
#
# shows the first 10 rows.
#
# This is useful because datasets can contain millions of rows.
#
# You usually do NOT want to print everything.
#
#
# ============================================================
# 17. tail()
# ============================================================
#
print(df.tail())
#
# Shows the last 5 rows by default.
#
# Useful for checking:
#
#     final records
#     imported files
#     chronological data
#     whether the end of the dataset looks correct
#
#
# ============================================================
# 18. sample()
# ============================================================
#
print(df.sample(2))
#
# Randomly selects rows.
#
# Mental model:
#
#     entire dataset
#            ↓
#        random selection
#            ↓
#       sample rows
#
# Useful for quickly seeing whether the dataset contains
# unexpected patterns or formatting.
#
# Important:
#
#     sample()
#
# is especially useful for large datasets because head()
# only shows the beginning.
#
#
# ============================================================
# 19. info()
# ============================================================
#
df.info()
#
# This is one of the most important inspection methods.
#
# It provides information about:
#
#     index
#     number of rows
#     columns
#     non-null counts
#     data types
#     memory usage
#
#
# Mental model:
#
#     df
#      ↓
#     structural inspection
#      ↓
#     "What is this dataset?"
#
#
# In Data Analysis:
#
#     df.info()
#
# should become almost automatic.
#
#
# ============================================================
# 20. describe()
# ============================================================
#
print(df.describe())
#
# For numerical columns, this typically provides:
#
#     count
#     mean
#     std
#     min
#     25%
#     50%
#     75%
#     max
#
#
# Conceptual flow:
#
#     raw numerical data
#            ↓
#     descriptive statistics
#            ↓
#     quick statistical profile
#
#
# For example:
#
#     count → number of non-missing observations
#     mean  → arithmetic average
#     std   → standard deviation
#     min   → minimum
#     25%   → first quartile
#     50%   → median
#     75%   → third quartile
#     max   → maximum
#
#
# ============================================================
# 21. describe(include="all")
# ============================================================
#
print(df.describe(include="all"))
#
# Useful when you want information about both numerical
# and non-numerical columns.
#
#
# ============================================================
# 22. SELECTING ONE COLUMN
# ============================================================
#
ages = df["age"]
#
# Result:
#
#     Series
#
# This is a very important concept.
#
#     df["age"]
#
# means:
#
#     "Select the column whose label is 'age'."
#
#
# DataFrame
#     │
#     ├── name
#     ├── age ───────→ selected Series
#     └── salary
#
#
# ============================================================
# 23. SELECTING MULTIPLE COLUMNS
# ============================================================
#
selected = df[["name", "salary"]]
#
# Notice the DOUBLE brackets.
#
#     df["salary"]
#
# returns a Series.
#
#     df[["name", "salary"]]
#
# returns a DataFrame.
#
#
# Mental model:
#
#     ["salary"]
#          ↓
#       one label
#          ↓
#       Series
#
#     [["name", "salary"]]
#          ↓
#       list of labels
#          ↓
#       DataFrame
#
#
# ============================================================
# 24. WHY DOUBLE BRACKETS?
# ============================================================
#
# The outer brackets perform DataFrame indexing.
#
# The inner brackets contain the list of requested columns.
#
# Conceptually:
#
#     df[
#         ["name", "salary"]
#     ]
#
#     DataFrame
#        ↓
#     receive list of column labels
#        ↓
#     return DataFrame containing those columns
#
#
# ============================================================
# 25. iloc — POSITION-BASED SELECTION
# ============================================================
#
print(df.iloc[0])
#
# iloc means:
#
#     integer-location based indexing
#
# It thinks in POSITIONS.
#
#     0 → first row
#     1 → second row
#     2 → third row
#
#
# Example:
#
print(df.iloc[0, 1])
#
# means:
#
#     row position 0
#     column position 1
#
#
# Visual:
#
#             0        1        2
#          name      age    salary
#     0     Ali       25      3000
#     1     Sara      31      4500
#     2     Reza      28      3800
#
#     df.iloc[1, 2]
#             │  │
#             │  └── column position 2
#             └───── row position 1
#
# Result:
#
#     4500
#
#
# ============================================================
# 26. iloc SLICING
# ============================================================
#
print(df.iloc[0:2, 0:2])
#
# Concept:
#
#     rows 0 through 1
#     columns 0 through 1
#
# Python slicing excludes the stop position.
#
#
# ============================================================
# 27. loc — LABEL-BASED SELECTION
# ============================================================
#
# loc works primarily with LABELS.
#
# Example:
#
# df.loc[0, "age"]
#
# means:
#
#     row whose index label is 0
#     column whose label is "age"
#
#
# IMPORTANT:
#
# iloc → position
# loc  → label
#
#
# ============================================================
# 28. iloc vs loc — CRITICAL DISTINCTION
# ============================================================
#
# Think:
#
#     iloc
#       ↓
#     "WHERE IS IT?"
#       ↓
#     position
#
#     loc
#       ↓
#     "WHAT IS ITS LABEL?"
#       ↓
#     label
#
#
# Example:
#
#     df.iloc[0]
#
# means:
#
#     first row
#
# Whereas:
#
#     df.loc[0]
#
# means:
#
#     row whose INDEX LABEL is 0
#
# These happen to be the same when the default index is used.
#
# They can become different after changing the index.
#
#
# ============================================================
# 29. BOOLEAN FILTERING — CORE PANDAS SKILL
# ============================================================
#
# Suppose:
#
#     age = [25, 31, 28, 42]
#
# We want:
#
#     age > 30
#
# Pandas evaluates the comparison for EVERY row.
#
# Conceptually:
#
#     age
#      ↓
#     25   31   28   42
#      ↓
#     > 30
#      ↓
#     F    T    F    T
#
# This Boolean Series becomes a FILTER MASK.
#
#
# ============================================================
# 30. BOOLEAN MASK
# ============================================================
#
# mask = df["age"] > 30
#
# Conceptually:
#
#     row      age      condition
#
#      0        25      False
#      1        31      True
#      2        28      False
#      3        42      True
#
#
# Then:
#
# filtered = df[mask]
#
# means:
#
#     keep rows where mask == True
#
#
# ============================================================
# 31. FILTERING DIRECTLY
# ============================================================
#
# filtered = df[df["age"] > 30]
#
# Data flow:
#
#     df["age"]
#         ↓
#     compare each value with 30
#         ↓
#     Boolean Series
#         ↓
#     use Boolean Series as mask
#         ↓
#     keep matching rows
#
#
# This is one of the most important Pandas patterns for DA.
#
#
# ============================================================
# 32. MULTIPLE CONDITIONS
# ============================================================
#
# Use:
#
#     &
#     |
#     ~
#
# for element-wise Boolean logic.
#
# Example:
#
# filtered = df[
#     (df["age"] > 25) &
#     (df["salary"] > 3500)
# ]
#
# Conceptually:
#
#     age > 25
#         ↓
#     Boolean mask A
#
#     salary > 3500
#         ↓
#     Boolean mask B
#
#     A & B
#         ↓
#     combined mask
#         ↓
#     filtered DataFrame
#
#
# IMPORTANT:
#
# Put each comparison in parentheses.
#
# Do not write:
#
#     df["age"] > 25 and df["salary"] > 3500
#
# because Python's "and" is not the normal element-wise
# operator for Pandas Series.
#
#
# ============================================================
# 33. OR CONDITION
# ============================================================
#
# filtered = df[
#     (df["age"] < 25) |
#     (df["salary"] > 5000)
# ]
#
# "|" means element-wise OR.
#
#
# ============================================================
# 34. NOT CONDITION
# ============================================================
#
# filtered = df[~(df["age"] > 30)]
#
# "~" reverses the Boolean mask.
#
#
# ============================================================
# 35. isin()
# ============================================================
#
# selected = df[df["city"].isin(["Tehran", "Karaj"])]
#
# isin() asks:
#
#     "Is each value contained in this collection?"
#
# Example:
#
#     Tehran → True
#     Karaj  → True
#     Shiraz → False
#
#
# General pattern:
#
#     Series
#       ↓
#     isin(values)
#       ↓
#     Boolean mask
#       ↓
#     filter
#
#
# ============================================================
# 36. BETWEEN()
# ============================================================
#
# selected = df[df["age"].between(25, 35)]
#
# This is useful for numerical ranges.
#
# Conceptually:
#
#     lower bound
#         ↓
#     value
#         ↓
#     upper bound
#
#
# ============================================================
# 37. STRING FILTERING
# ============================================================
#
# Pandas provides the .str accessor for vectorized string operations.
#
# Examples:
#
#     df["name"].str.lower()
#     df["name"].str.upper()
#     df["name"].str.strip()
#     df["name"].str.len()
#     df["name"].str.contains("a")
#     df["name"].str.startswith("A")
#     df["name"].str.endswith("a")
#     df["name"].str.replace("old", "new")
#
#
# Mental model:
#
#     Series of strings
#             ↓
#          .str
#             ↓
#     string operation applied element-by-element
#
#
# ============================================================
# 38. .str.contains()
# ============================================================
#
# mask = df["name"].str.contains("a", case=False, na=False)
#
# This creates a Boolean mask.
#
# Then:
#
#     df[mask]
#
# keeps matching rows.
#
# This is extremely useful in practical Data Analysis.
#
#
# ============================================================
# 39. ASSIGNING A NEW COLUMN
# ============================================================
#
# df["annual_salary"] = df["salary"] * 12
#
# Data flow:
#
#     salary
#        ↓
#     multiply by 12
#        ↓
#     new values
#        ↓
#     annual_salary column
#
#
# This is a fundamental data-transformation pattern.
#
#
# ============================================================
# 40. CREATING COLUMNS FROM MULTIPLE COLUMNS
# ============================================================
#
# df["salary_per_year"] = df["salary"] * 12
#
# More generally:
#
#     new_column = transformation(old_columns)
#
#
# Example:
#
#     df["income_per_age"] = df["salary"] / df["age"]
#
#
# IMPORTANT:
#
# Pandas normally performs these operations element-by-element
# across aligned Series.
#
#
# ============================================================
# 41. VECTORISATION
# ============================================================
#
# This:
#
#     df["salary"] * 12
#
# applies the operation to the entire Series.
#
# Conceptually:
#
#     [3000, 4500, 3800]
#             × 12
#               ↓
#     [36000, 54000, 45600]
#
# You generally do NOT need:
#
#     for each row:
#         multiply salary by 12
#
# Pandas/NumPy can perform the operation in a vectorized manner.
#
#
# ============================================================
# 42. ASSIGN()
# ============================================================
#
# df = df.assign(
#     annual_salary=df["salary"] * 12
# )
#
# assign() creates columns while returning a DataFrame.
#
# It is useful for method chaining.
#
#
# ============================================================
# 43. RENAME COLUMNS
# ============================================================
#
# df = df.rename(
#     columns={"salary": "monthly_salary"}
# )
#
# This changes the column label.
#
# It does NOT automatically change the data itself.
#
#
# ============================================================
# 44. RENAME INDEX
# ============================================================
#
# df = df.rename(index={0: "A", 1: "B"})
#
# This changes index labels.
#
#
# ============================================================
# 45. DROP COLUMNS
# ============================================================
#
# df = df.drop(columns=["age"])
#
# Conceptually:
#
#     DataFrame
#         ↓
#     remove selected column
#         ↓
#     new DataFrame
#
# By default, many Pandas operations return a new object rather
# than modifying the original.
#
#
# ============================================================
# 46. DROP ROWS
# ============================================================
#
# df = df.drop(index=[0, 2])
#
# Removes rows with those index labels.
#
#
# ============================================================
# 47. dropna()
# ============================================================
#
# missing_data_removed = df.dropna()
#
# Removes rows containing missing values by default.
#
# Mental model:
#
#     DataFrame
#         ↓
#     locate missing values
#         ↓
#     remove rows containing them
#
# IMPORTANT:
#
# This can remove a large amount of data.
#
# Never use dropna() blindly.
#
#
# ============================================================
# 48. dropna(subset=...)
# ============================================================
#
# df = df.dropna(subset=["age"])
#
# This means:
#
#     "Remove rows where age is missing."
#
# Other columns can still contain missing values.
#
#
# ============================================================
# 49. fillna()
# ============================================================
#
# df["age"] = df["age"].fillna(df["age"].median())
#
# This is a very important Data Science pattern.
#
# Conceptually:
#
#     missing age
#          ↓
#     calculate median age
#          ↓
#     replace missing value
#
# Why median?
#
# Median is often more robust to extreme values than mean.
#
# The appropriate strategy depends on the dataset.
#
#
# ============================================================
# 50. isna()
# ============================================================
#
# missing_mask = df.isna()
#
# Returns a Boolean structure:
#
#     missing → True
#     present → False
#
#
# ============================================================
# 51. isnull()
# ============================================================
#
# df.isnull()
#
# isnull() and isna() are equivalent interfaces for detecting
# missing values.
#
# Common modern style:
#
#     isna()
#
#
# ============================================================
# 52. NOTNA()
# ============================================================
#
# df.notna()
#
# Gives:
#
#     missing → False
#     present → True
#
#
# ============================================================
# 53. COUNTING MISSING VALUES
# ============================================================
#
# df.isna().sum()
#
# This is a very useful pattern.
#
# First:
#
#     df.isna()
#
# produces True/False.
#
# Then:
#
#     .sum()
#
# counts True values because:
#
#     True  → 1
#     False → 0
#
# Therefore:
#
#     missing count per column
#
#
# ============================================================
# 54. MISSING-VALUE PROFILE
# ============================================================
#
# missing_count = df.isna().sum()
#
# missing_percent = df.isna().mean() * 100
#
# Conceptually:
#
#     missing values
#          ÷
#     total values
#          × 100
#          ↓
#     missing percentage
#
# This is very useful for Data Quality analysis.
#
#
# ============================================================
# 55. DUPLICATES
# ============================================================
#
# df.duplicated()
#
# Returns a Boolean Series indicating duplicate rows.
#
#
# ============================================================
# 56. REMOVE DUPLICATES
# ============================================================
#
# df = df.drop_duplicates()
#
# Removes duplicate rows.
#
# IMPORTANT:
#
# "Duplicate" depends on what columns define identity.
#
# Therefore you can specify:
#
#     df.drop_duplicates(subset=["customer_id"])
#
#
# ============================================================
# 57. UNIQUE VALUES
# ============================================================
#
# df["city"].unique()
#
# Returns the distinct values.
#
# Useful for:
#
#     categories
#     labels
#     data validation
#     exploratory analysis
#
#
# ============================================================
# 58. NUMBER OF UNIQUE VALUES
# ============================================================
#
# df["city"].nunique()
#
# Answers:
#
#     "How many distinct values exist?"
#
# Example:
#
#     Tehran
#     Karaj
#     Tehran
#     Tabriz
#
#     nunique() → 3
#
#
# ============================================================
# 59. VALUE COUNTS
# ============================================================
#
# df["city"].value_counts()
#
# One of the most useful exploratory methods.
#
# It counts occurrences of each distinct value.
#
# Example:
#
#     Tehran    120
#     Karaj      80
#     Tabriz     40
#
# This is useful for:
#
#     categorical distributions
#     class imbalance
#     frequency analysis
#     data quality
#
#
# ============================================================
# 60. SORTING
# ============================================================
#
# df = df.sort_values("salary")
#
# Sorts rows according to salary.
#
# Descending:
#
#     df.sort_values("salary", ascending=False)
#
#
# Multiple columns:
#
#     df.sort_values(
#         ["city", "salary"],
#         ascending=[True, False]
#     )
#
#
# Conceptually:
#
#     grouping key 1
#          ↓
#     grouping/order key 2
#          ↓
#     final row order
#
#
# ============================================================
# 61. sort_index()
# ============================================================
#
# df = df.sort_index()
#
# Sorts according to the index labels.
#
#
# ============================================================
# 62. RESETTING THE INDEX
# ============================================================
#
# df = df.reset_index()
#
# This turns the current index into a column and creates a new
# default integer index.
#
# Common variant:
#
#     df = df.reset_index(drop=True)
#
# This discards the old index instead of making it a column.
#
#
# IMPORTANT:
#
# After filtering:
#
#     df = df[df["age"] > 30]
#
# the original index labels may remain:
#
#     1
#     3
#     7
#     12
#
# reset_index(drop=True) can create:
#
#     0
#     1
#     2
#     3
#
#
# ============================================================
# 63. SETTING AN INDEX
# ============================================================
#
# df = df.set_index("customer_id")
#
# This makes customer_id the row index.
#
# Mental model:
#
#     normal column
#          ↓
#     set_index()
#          ↓
#     row identifier
#
#
# ============================================================
# 64. COPY()
# ============================================================
#
# df2 = df.copy()
#
# This creates an independent DataFrame object.
#
# Very important when you want to work on a separate version.
#
#
# Mental model:
#
#     df
#      ↓
#     copy()
#      ↓
#     separate DataFrame
#
#
# ============================================================
# 65. BASIC MATHEMATICAL AGGREGATIONS
# ============================================================
#
# Important Series/DataFrame methods include:
#
#     sum()
#     mean()
#     median()
#     min()
#     max()
#     std()
#     var()
#     count()
#     prod()
#     quantile()
#
#
# Example:
#
#     df["salary"].mean()
#
# means:
#
#     salary values
#          ↓
#     arithmetic mean
#          ↓
#     one numerical result
#
#
# ============================================================
# 66. SUM()
# ============================================================
#
# total = df["salary"].sum()
#
# Pattern:
#
#     many values
#         ↓
#        sum
#         ↓
#     one value
#
#
# ============================================================
# 67. MEAN()
# ============================================================
#
# average_salary = df["salary"].mean()
#
# Conceptually:
#
#     x₁ + x₂ + ... + xₙ
#     -------------------
#             n
#
#
# ============================================================
# 68. MEDIAN()
# ============================================================
#
# median_salary = df["salary"].median()
#
# Median is the middle value after sorting.
#
# It is often less sensitive to extreme observations than mean.
#
#
# ============================================================
# 69. STD()
# ============================================================
#
# spread = df["salary"].std()
#
# Standard deviation measures dispersion around the mean.
#
# Important Pandas detail:
#
# Pandas Series.std() uses sample standard deviation by default:
#
#     ddof = 1
#
# This differs from NumPy's default:
#
#     np.std(..., ddof=0)
#
# Understanding library defaults is important in statistics.
#
#
# ============================================================
# 70. VAR()
# ============================================================
#
# variance = df["salary"].var()
#
# Variance measures squared dispersion.
#
# Standard deviation:
#
#     std = sqrt(variance)
#
#
# ============================================================
# 71. QUANTILE()
# ============================================================
#
# q50 = df["salary"].quantile(0.50)
# q25 = df["salary"].quantile(0.25)
# q75 = df["salary"].quantile(0.75)
#
# Quantiles divide ordered data according to cumulative proportion.
#
# Important:
#
#     0.50 → median
#     0.25 → first quartile
#     0.75 → third quartile
#
#
# ============================================================
# 72. AGG()
# ============================================================
#
# summary = df["salary"].agg(
#     ["mean", "median", "min", "max", "std"]
# )
#
# agg() allows multiple aggregation operations.
#
# Mental model:
#
#     one Series
#          ↓
#     multiple summary functions
#          ↓
#     statistical summary
#
#
# ============================================================
# 73. AGG() WITH MULTIPLE COLUMNS
# ============================================================
#
# summary = df.agg({
#     "age": ["mean", "min", "max"],
#     "salary": ["mean", "median", "max"]
# })
#
# This is very useful for analytical reports.
#
#
# ============================================================
# 74. GROUPBY — ONE OF THE MOST IMPORTANT PANDAS CONCEPTS
# ============================================================
#
# Suppose we have:
#
#     city      salary
#     Tehran    3000
#     Tehran    4000
#     Karaj     3500
#     Karaj     4500
#
# We want:
#
#     average salary by city
#
# Use:
#
#     df.groupby("city")["salary"].mean()
#
#
# Mental model:
#
#     DATA
#       ↓
#     GROUP BY city
#       ↓
#     ┌───────────────┐
#     │ Tehran rows   │
#     │ Karaj rows    │
#     └───────────────┘
#       ↓
#     aggregate each group
#       ↓
#     mean salary
#
#
# ============================================================
# 75. GROUPBY IS SPLIT → APPLY → COMBINE
# ============================================================
#
# This is a fundamental Data Analysis concept.
#
#                 DATA
#                  ↓
#                SPLIT
#                  ↓
#       ┌──────────┼──────────┐
#       ↓          ↓          ↓
#    Group A    Group B    Group C
#       ↓          ↓          ↓
#      APPLY      APPLY      APPLY
#       ↓          ↓          ↓
#    result A   result B   result C
#       └──────────┼──────────┘
#                  ↓
#                COMBINE
#
# This is the conceptual foundation of groupby().
#
#
# ============================================================
# 76. GROUPBY WITH MULTIPLE AGGREGATIONS
# ============================================================
#
# summary = df.groupby("city")["salary"].agg(
#     ["count", "mean", "median", "min", "max"]
# )
#
# This produces a group-level statistical profile.
#
#
# ============================================================
# 77. GROUPBY MULTIPLE COLUMNS
# ============================================================
#
# summary = df.groupby(
#     ["city", "department"]
# )["salary"].mean()
#
# This creates groups based on combinations:
#
#     city + department
#
#
# ============================================================
# 78. GROUPBY WITH MULTIPLE OUTPUT COLUMNS
# ============================================================
#
# summary = df.groupby("city").agg(
#     average_salary=("salary", "mean"),
#     median_age=("age", "median"),
#     employee_count=("name", "count")
# )
#
# This is an extremely useful practical pattern.
#
#
# ============================================================
# 79. TRANSFORM()
# ============================================================
#
# transform() differs from aggregate().
#
# Aggregation:
#
#     many rows
#        ↓
#     one result per group
#
# Transform:
#
#     many rows
#        ↓
#     result aligned back to original rows
#
#
# Example:
#
# df["city_average_salary"] = (
#     df.groupby("city")["salary"]
#       .transform("mean")
# )
#
#
# Visual:
#
#     salary
#       ↓
#     group by city
#       ↓
#     city mean
#       ↓
#     broadcast mean back to each row
#
#
# This is very useful for:
#
#     normalization
#     feature engineering
#     comparisons against group averages
#     anomaly detection
#
#
# ============================================================
# 80. APPLY()
# ============================================================
#
# apply() allows you to apply a function across a Series or
# DataFrame axis.
#
# Example:
#
# df["salary"] = df["salary"].apply(lambda x: x * 1.10)
#
# Conceptually:
#
#     each salary
#          ↓
#     function
#          ↓
#     transformed salary
#
# Important:
#
# apply() is flexible, but vectorized Pandas/NumPy operations
# are often preferable when available.
#
#
# ============================================================
# 81. MAP()
# ============================================================
#
# map() is especially useful for transforming values in a Series.
#
# Example:
#
# city_code = {
#     "Tehran": 1,
#     "Karaj": 2,
#     "Tabriz": 3
# }
#
# df["city_code"] = df["city"].map(city_code)
#
# Mental model:
#
#     original value
#          ↓
#     lookup mapping
#          ↓
#     transformed value
#
#
# ============================================================
# 82. REPLACE()
# ============================================================
#
# df["city"] = df["city"].replace(
#     {"Tehran": "TEH", "Karaj": "KAR"}
# )
#
# replace() is useful for systematic value replacement.
#
#
# ============================================================
# 83. WHERE()
# ============================================================
#
# df["salary"] = df["salary"].where(
#     df["salary"] > 3000,
#     3000
# )
#
# Concept:
#
#     condition TRUE
#         ↓
#     keep original value
#
#     condition FALSE
#         ↓
#     use replacement value
#
#
# ============================================================
# 84. MASK()
# ============================================================
#
# mask() is conceptually the opposite selection behavior:
#
#     condition TRUE
#         ↓
#     replace
#
#     condition FALSE
#         ↓
#     keep
#
#
# ============================================================
# 85. CLIPPING VALUES
# ============================================================
#
# df["salary"] = df["salary"].clip(
#     lower=3000,
#     upper=6000
# )
#
# Values below 3000 become 3000.
#
# Values above 6000 become 6000.
#
# Values inside the range remain unchanged.
#
#
# Visual:
#
#     1000  → 3000
#     3000  → 3000
#     4500  → 4500
#     6000  → 6000
#     8000  → 6000
#
# Useful in some data-cleaning and feature-engineering tasks.
#
#
# ============================================================
# 86. READING CSV FILES
# ============================================================
#
# df = pd.read_csv("data.csv")
#
# This is one of the most important Pandas operations in real DA.
#
# Data flow:
#
#     CSV file
#         ↓
#     pd.read_csv()
#         ↓
#     DataFrame
#         ↓
#     inspect
#         ↓
#     clean
#         ↓
#     analyze
#
#
# ============================================================
# 87. IMPORTANT read_csv() PARAMETERS
# ============================================================
#
# Common practical parameters include:
#
#     sep
#     header
#     names
#     index_col
#     usecols
#     dtype
#     parse_dates
#     na_values
#     encoding
#     skiprows
#     nrows
#
# Example:
#
# df = pd.read_csv(
#     "data.csv",
#     usecols=["age", "salary", "city"]
# )
#
#
# ============================================================
# 88. READING EXCEL
# ============================================================
#
# df = pd.read_excel("data.xlsx")
#
# Common practical options:
#
#     sheet_name
#     usecols
#     skiprows
#     nrows
#     dtype
#
# Example:
#
# df = pd.read_excel(
#     "data.xlsx",
#     sheet_name="Sales"
# )
#
#
# ============================================================
# 89. READING JSON
# ============================================================
#
# df = pd.read_json("data.json")
#
# Useful when data comes from APIs or JSON-based storage.
#
#
# ============================================================
# 90. READING SQL DATA
# ============================================================
#
# In real Data Analysis, Pandas often receives data from SQL.
#
# Conceptually:
#
#     SQL database
#          ↓
#     query
#          ↓
#     DataFrame
#          ↓
#     Pandas analysis
#
# Typical interface:
#
#     pd.read_sql(...)
#
# The actual database connection is usually provided by a
# database/connection library.
#
#
# ============================================================
# 91. WRITING CSV
# ============================================================
#
# df.to_csv(
#     "cleaned_data.csv",
#     index=False
# )
#
# index=False means:
#
#     "Do not write the DataFrame index as a CSV column."
#
#
# ============================================================
# 92. WRITING EXCEL
# ============================================================
#
# df.to_excel(
#     "cleaned_data.xlsx",
#     index=False
# )
#
#
# ============================================================
# 93. BASIC FILE FORMAT WORKFLOW
# ============================================================
#
#     CSV / Excel / SQL / JSON
#                 ↓
#              Pandas
#                 ↓
#             DataFrame
#                 ↓
#     ┌───────────┼────────────┐
#     ↓           ↓            ↓
#   inspect     clean       transform
#     └───────────┼────────────┘
#                 ↓
#              analyze
#                 ↓
#              export
#
#
# ============================================================
# 94. CONCAT()
# ============================================================
#
# concat() combines objects along an axis.
#
# Example:
#
#     df_all = pd.concat(
#         [df_2024, df_2025],
#         ignore_index=True
#     )
#
#
# Concept:
#
#     DATASET A
#        ↓
#     DATASET B
#        ↓
#     concatenate
#        ↓
#     combined dataset
#
#
# ============================================================
# 95. CONCAT — ROW-WISE
# ============================================================
#
#     pd.concat([df1, df2], axis=0)
#
# means:
#
#     append rows vertically
#
#
# Visual:
#
#     df1
#     ┌───────────┐
#     │ rows      │
#     └───────────┘
#          +
#     df2
#     ┌───────────┐
#     │ rows      │
#     └───────────┘
#          ↓
#     combined
#     ┌───────────┐
#     │ df1 rows  │
#     │ df2 rows  │
#     └───────────┘
#
#
# ============================================================
# 96. CONCAT — COLUMN-WISE
# ============================================================
#
#     pd.concat([df1, df2], axis=1)
#
# combines columns side-by-side based on index alignment.
#
#
# ============================================================
# 97. MERGE()
# ============================================================
#
# merge() is conceptually similar to a SQL JOIN.
#
# Suppose:
#
# customers:
#
#     customer_id   name
#
# orders:
#
#     customer_id   amount
#
# We can combine them:
#
#     pd.merge(
#         customers,
#         orders,
#         on="customer_id"
#     )
#
#
# Concept:
#
#     TABLE A
#         │
#         │ common key
#         ↓
#     JOIN
#         ↑
#         │ common key
#     TABLE B
#         ↓
#     COMBINED TABLE
#
#
# ============================================================
# 98. MERGE TYPES
# ============================================================
#
# Important:
#
#     how="inner"
#     how="left"
#     how="right"
#     how="outer"
#
#
# INNER:
#
#     keep matching keys in both datasets
#
# LEFT:
#
#     keep every row from the left DataFrame
#     and matching rows from the right
#
# RIGHT:
#
#     keep every row from the right DataFrame
#
# OUTER:
#
#     keep keys from both datasets
#
#
# SQL mental model:
#
#     inner → INNER JOIN
#     left  → LEFT JOIN
#     right → RIGHT JOIN
#     outer → FULL OUTER JOIN
#
#
# ============================================================
# 99. MERGE WITH DIFFERENT KEY NAMES
# ============================================================
#
# pd.merge(
#     df1,
#     df2,
#     left_on="customer_id",
#     right_on="client_id"
# )
#
# Useful when the join keys have different column names.
#
#
# ============================================================
# 100. JOIN()
# ============================================================
#
# DataFrame.join() is another way to combine data, especially
# when the relationship is naturally index-based.
#
#     df1.join(df2)
#
# Think:
#
#     index
#       ↓
#     align rows
#       ↓
#     combine columns
#
#
# ============================================================
# 101. PIVOT TABLE
# ============================================================
#
# pivot_table() is extremely useful in Data Analysis.
#
# Suppose:
#
#     city
#     product
#     sales
#
# We want:
#
#                  Product A   Product B
#     Tehran          500         700
#     Karaj           300         450
#
# We can use:
#
#     pd.pivot_table(
#         df,
#         values="sales",
#         index="city",
#         columns="product",
#         aggfunc="sum"
#     )
#
#
# Conceptually:
#
#     long/raw data
#           ↓
#       group + aggregate
#           ↓
#       two-dimensional summary
#
#
# ============================================================
# 102. PIVOT()
# ============================================================
#
# pivot() reshapes data from long to wide form without performing
# an aggregation.
#
# Important distinction:
#
#     pivot()
#         → reshape
#
#     pivot_table()
#         → reshape + aggregate
#
#
# ============================================================
# 103. MELT()
# ============================================================
#
# melt() transforms wide data into long format.
#
# Example wide:
#
#     name   math   physics   chemistry
#
#     Ali     90       80         85
#
# Long:
#
#     name   subject      score
#     Ali    math          90
#     Ali    physics       80
#     Ali    chemistry     85
#
#
# Visual:
#
#     WIDE
#
#     ┌──────┬──────┬─────────┬───────────┐
#     │ name │ math │ physics │ chemistry │
#     └──────┴──────┴─────────┴───────────┘
#
#                  ↓ melt()
#
#     LONG
#
#     ┌──────┬───────────┬───────┐
#     │ name │ subject   │ score │
#     └──────┴───────────┴───────┘
#
#
# Long format is often very useful for plotting and grouped analysis.
#
#
# ============================================================
# 104. STACK() AND UNSTACK()
# ============================================================
#
# These reshape hierarchical/indexed data.
#
# Rough mental model:
#
#     unstack()
#         ↓
#     index levels → columns
#
#     stack()
#         ↓
#     columns → index level
#
# They become especially useful when working with MultiIndex data.
#
#
# ============================================================
# 105. MULTIINDEX
# ============================================================
#
# A MultiIndex allows multiple levels of labels.
#
# Example:
#
#     city      department
#     Tehran    AI
#     Tehran    Network
#     Karaj     AI
#     Karaj     Network
#
# Conceptually:
#
#     level 1 → city
#     level 2 → department
#
# This is powerful but more advanced.
#
# For beginner DA:
#
#     understand groupby()
#     understand pivot_table()
#     understand normal indexes
#
# before deeply using MultiIndex.
#
#
# ============================================================
# 106. DATETIME — PANDAS TIME SERIES
# ============================================================
#
# Dates are extremely important in real Data Analysis.
#
# Example:
#
#     df["date"] = pd.to_datetime(df["date"])
#
# This converts date-like values into Pandas datetime objects.
#
#
# Conceptual flow:
#
#     text
#     "2026-09-13"
#          ↓
#     pd.to_datetime()
#          ↓
#     datetime representation
#          ↓
#     date-aware operations
#
#
# ============================================================
# 107. DATETIME COMPONENTS
# ============================================================
#
# Once a Series contains datetimes:
#
#     df["date"].dt.year
#     df["date"].dt.month
#     df["date"].dt.day
#     df["date"].dt.dayofweek
#     df["date"].dt.hour
#     df["date"].dt.minute
#
# The .dt accessor exposes datetime components.
#
#
# ============================================================
# 108. DATETIME FILTERING
# ============================================================
#
# df[df["date"] >= "2026-01-01"]
#
# This can be used to select records after a particular date.
#
#
# ============================================================
# 109. RESAMPLING
# ============================================================
#
# For time-series data, resample() groups observations into
# time intervals.
#
# Example:
#
#     monthly_sales = (
#         df.set_index("date")["sales"]
#           .resample("ME")
#           .sum()
#     )
#
# Concept:
#
#     daily observations
#            ↓
#        monthly groups
#            ↓
#          sum()
#            ↓
#     monthly totals
#
# This is extremely useful in business analytics.
#
#
# ============================================================
# 110. SHIFT()
# ============================================================
#
# shift() moves values relative to their positions.
#
# Example:
#
#     df["previous_sales"] = df["sales"].shift(1)
#
# Concept:
#
#     current row
#          ↓
#     retrieve previous row's value
#
# This is useful for:
#
#     time-series analysis
#     lag features
#     growth calculations
#     ML feature engineering
#
#
# ============================================================
# 111. DIFF()
# ============================================================
#
# df["sales_change"] = df["sales"].diff()
#
# Calculates the difference between consecutive observations.
#
# Concept:
#
#     current value
#          -
#     previous value
#          ↓
#     change
#
# Example:
#
#     100
#     120
#     135
#
# diff():
#
#     NaN
#     20
#     15
#
#
# ============================================================
# 112. PCT_CHANGE()
# ============================================================
#
# df["growth"] = df["sales"].pct_change()
#
# Measures proportional change between consecutive values.
#
# Conceptually:
#
#     current - previous
#     ------------------
#          previous
#
# Useful for:
#
#     growth rates
#     financial analysis
#     business KPIs
#     time series
#
#
# ============================================================
# 113. ROLLING()
# ============================================================
#
# rolling() creates a moving window.
#
# Example:
#
#     df["moving_average"] = (
#         df["sales"]
#         .rolling(7)
#         .mean()
#     )
#
#
# Mental model:
#
#     [day1]
#     [day1 day2]
#     [day1 day2 day3]
#             ...
#
# With a 7-observation window:
#
#     current observation
#           +
#     previous 6 observations
#           ↓
#     7-value window
#           ↓
#     mean
#
# Useful for:
#
#     smoothing
#     trend analysis
#     time series
#     feature engineering
#
#
# ============================================================
# 114. EXPANDING()
# ============================================================
#
# expanding() creates a window that grows from the beginning.
#
# Example:
#
#     df["running_mean"] = (
#         df["sales"]
#         .expanding()
#         .mean()
#     )
#
# Difference:
#
#     rolling(3)
#         → fixed-size moving window
#
#     expanding()
#         → growing window from the beginning
#
#
# ============================================================
# 115. CATEGORICAL DATA
# ============================================================
#
# Repeated categorical values can be represented using Pandas
# categorical dtype.
#
# Example:
#
#     df["city"] = df["city"].astype("category")
#
# Useful for:
#
#     memory optimization
#     categorical semantics
#     some analytical workflows
#
#
# ============================================================
# 116. ASTYPE()
# ============================================================
#
# astype() changes a column's data type.
#
# Example:
#
#     df["age"] = df["age"].astype("int64")
#
# Or:
#
#     df["category"] = df["category"].astype("category")
#
# Mental model:
#
#     existing representation
#            ↓
#          astype()
#            ↓
#     new dtype representation
#
# Be careful:
#
# conversion can fail if values cannot be converted.
#
#
# ============================================================
# 117. TO_NUMERIC()
# ============================================================
#
# pd.to_numeric(
#     df["salary"],
#     errors="coerce"
# )
#
# Useful when numbers have been imported as text.
#
# Example:
#
#     "3500"
#     "4200"
#     "unknown"
#
# with:
#
#     errors="coerce"
#
# invalid values can become NaN.
#
#
# ============================================================
# 118. TO_DATETIME()
# ============================================================
#
# pd.to_datetime(...)
#
# converts date-like values into datetime representations.
#
#
# ============================================================
# 119. TO_TIMEDELTA()
# ============================================================
#
# pd.to_timedelta(...)
#
# converts values into durations.
#
# Useful for:
#
#     response times
#     durations
#     time differences
#
#
# ============================================================
# 120. MEMORY USAGE
# ============================================================
#
# df.memory_usage()
#
# provides memory consumption information.
#
# For larger datasets:
#
#     df.info(memory_usage="deep")
#
# can provide more detailed memory information.
#
#
# ============================================================
# 121. QUERY()
# ============================================================
#
# query() provides a readable way to filter rows.
#
# Example:
#
#     filtered = df.query("age > 30")
#
# Multiple conditions:
#
#     filtered = df.query(
#         "age > 30 and salary > 4000"
#     )
#
# It can be convenient for readable analytical code.
#
#
# ============================================================
# 122. QUERY() MENTAL MODEL
# ============================================================
#
#     DataFrame
#         ↓
#     textual condition
#         ↓
#     Boolean filtering
#         ↓
#     selected rows
#
# Compare:
#
#     df[df["age"] > 30]
#
# with:
#
#     df.query("age > 30")
#
# Both express the same basic idea.
#
#
# ============================================================
# 123. WHERE TO USE apply(), map(), transform(), agg()
# ============================================================
#
# These four are easy to confuse.
#
# ------------------------------------------------------------
#
# map()
#     → transform individual Series values
#
#     value
#       ↓
#     mapping/function
#       ↓
#     value
#
# ------------------------------------------------------------
#
# apply()
#     → apply a function along a Series/DataFrame axis
#
# ------------------------------------------------------------
#
# agg()
#     → reduce data into summary statistics
#
#     many values
#         ↓
#     one/few summary values
#
# ------------------------------------------------------------
#
# transform()
#     → produce results aligned with original data
#
#     many rows
#         ↓
#     group calculation
#         ↓
#     same number of rows
#
#
# ============================================================
# 124. RESETTING YOUR MENTAL MODEL FOR PANDAS
# ============================================================
#
# Do not think:
#
#     "Pandas is just a big Excel replacement."
#
# Better:
#
#     Pandas is a labeled data-manipulation system.
#
# It combines:
#
#     labels
#       +
#     tabular structures
#       +
#     vectorized operations
#       +
#     grouping
#       +
#     alignment
#       +
#     missing-value handling
#       +
#     reshaping
#       +
#     file/database interfaces
#
#
# ============================================================
# 125. ALIGNMENT — ONE OF THE MOST IMPORTANT PANDAS IDEAS
# ============================================================
#
# Pandas frequently aligns data according to labels.
#
# Suppose:
#
#     A:
#     index  value
#       0      10
#       1      20
#       2      30
#
#     B:
#     index  value
#       1      100
#       2      200
#       3      300
#
# Adding them:
#
#     A + B
#
# aligns by INDEX:
#
#     index  A     B
#       0    10    NaN
#       1    20    100
#       2    30    200
#       3   NaN    300
#
# Result:
#
#     index  result
#       0     NaN
#       1     120
#       2     230
#       3     NaN
#
# This is fundamentally different from blindly adding by position.
#
#
# ============================================================
# 126. ALIGNMENT — CORE IDEA
# ============================================================
#
# Pandas often asks:
#
#     "Which labels correspond?"
#
# rather than simply:
#
#     "Which positions correspond?"
#
# This is one reason Pandas is so useful for labeled data.
#
#
# ============================================================
# 127. COPY VS VIEW — PRACTICAL WARNING
# ============================================================
#
# Pandas indexing can sometimes produce objects whose relationship
# to the original data is not what a beginner expects.
#
# Therefore, when you intentionally want an independent object:
#
#     subset = df.loc[df["age"] > 30].copy()
#
# Then modifications to subset are conceptually separated from df.
#
# This also helps avoid chained-assignment problems.
#
#
# ============================================================
# 128. SETTING VALUES WITH LOC
# ============================================================
#
# Example:
#
#     df.loc[df["age"] > 40, "salary"] = 7000
#
# Read it conceptually as:
#
#     find rows where age > 40
#             ↓
#     select salary column
#             ↓
#     assign 7000
#
# This is one of the cleanest patterns for conditional assignment.
#
#
# ============================================================
# 129. CONDITIONAL COLUMN CREATION
# ============================================================
#
# Example:
#
#     df["senior"] = df["age"] >= 40
#
# Result:
#
#     age
#      ↓
#     comparison
#      ↓
#     True / False
#      ↓
#     new column
#
# This is extremely useful for feature engineering.
#
#
# ============================================================
# 130. MULTI-CONDITION FEATURE
# ============================================================
#
# Example:
#
#     df["high_income"] = (
#         df["salary"] > 5000
#     )
#
# Or:
#
#     df["target_group"] = (
#         (df["age"] > 30) &
#         (df["salary"] > 4000)
#     )
#
# This is a basic example of creating ML-ready features.
#
#
# ============================================================
# 131. BINNING WITH cut()
# ============================================================
#
# pd.cut() converts continuous values into intervals/categories.
#
# Example:
#
#     df["age_group"] = pd.cut(
#         df["age"],
#         bins=[0, 18, 30, 50, 100],
#         labels=[
#             "child",
#             "young",
#             "adult",
#             "senior"
#         ]
#     )
#
# Concept:
#
#     continuous age
#          ↓
#     intervals
#          ↓
#     categorical variable
#
# Useful for:
#
#     reporting
#     segmentation
#     exploratory analysis
#     feature engineering
#
#
# ============================================================
# 132. qcut()
# ============================================================
#
# pd.qcut() creates bins based on quantiles.
#
# Unlike cut():
#
#     cut()
#         → fixed value boundaries
#
#     qcut()
#         → approximately equal-frequency groups
#
# This is useful for:
#
#     quartiles
#     customer segmentation
#     ranking-based analysis
#
#
# ============================================================
# 133. RANK()
# ============================================================
#
# df["salary_rank"] = df["salary"].rank()
#
# rank() assigns relative positions.
#
# Useful for:
#
#     ranking customers
#     ranking products
#     leaderboard-style analysis
#     feature engineering
#
#
# ============================================================
# 134. CORRELATION
# ============================================================
#
# df.corr(numeric_only=True)
#
# Calculates pairwise correlations among numerical columns.
#
# Concept:
#
#     variable A
#          ↘
#           relationship
#          ↗
#     variable B
#
# Correlation is NOT proof of causation.
#
# It measures association according to the selected correlation
# method.
#
#
# ============================================================
# 135. COVARIANCE
# ============================================================
#
# df.cov(numeric_only=True)
#
# calculates pairwise covariance.
#
#
# ============================================================
# 136. CORRELATION WITH ONE COLUMN
# ============================================================
#
# df["age"].corr(df["salary"])
#
# This gives the correlation between two Series.
#
#
# ============================================================
# 137. DESCRIPTIVE ANALYSIS WORKFLOW
# ============================================================
#
# A practical first-pass Data Analysis workflow:
#
#     df = pd.read_csv(...)
#              ↓
#     df.head()
#              ↓
#     df.shape
#              ↓
#     df.info()
#              ↓
#     df.describe()
#              ↓
#     df.isna().sum()
#              ↓
#     df.nunique()
#              ↓
#     df.select_dtypes(...)
#              ↓
#     inspect categories
#              ↓
#     clean
#              ↓
#     analyze
#
#
# ============================================================
# 138. select_dtypes()
# ============================================================
#
# numeric_df = df.select_dtypes(
#     include="number"
# )
#
# This selects columns according to dtype.
#
# Useful when you want:
#
#     numerical features
#     categorical features
#     datetime columns
#
#
# Examples:
#
#     include="number"
#     include="object"
#     include="category"
#     include="datetime"
#
#
# ============================================================
# 139. NUMERICAL VS CATEGORICAL DATA
# ============================================================
#
# A very important Data Science distinction:
#
#     DATA
#       │
#       ├───────────────┐
#       ↓               ↓
#   numerical       categorical
#       │               │
#       ↓               ↓
#   mean/std/etc.    counts/groups
#
# Pandas helps you identify and manipulate both.
#
#
# ============================================================
# 140. VALUE COUNTS FOR ML CLASS DISTRIBUTION
# ============================================================
#
# Suppose:
#
#     df["target"].value_counts()
#
# gives:
#
#     0    950
#     1     50
#
# Then:
#
#     target class 0 → 95%
#     target class 1 → 5%
#
# This immediately warns you about class imbalance.
#
# In ML this matters because accuracy alone may become misleading.
#
#
# ============================================================
# 141. NORMALIZATION / STANDARDIZATION — PANDAS ROLE
# ============================================================
#
# Pandas can perform simple transformations:
#
#     df["x_standardized"] = (
#         df["x"] - df["x"].mean()
#     ) / df["x"].std()
#
# Concept:
#
#     x
#      ↓
#     subtract mean
#      ↓
#     divide by standard deviation
#      ↓
#     standardized value
#
# However:
#
# In production ML pipelines, preprocessing is often better handled
# with tools such as scikit-learn's StandardScaler, especially to
# prevent train/test leakage.
#
#
# ============================================================
# 142. DATA LEAKAGE WARNING
# ============================================================
#
# Suppose you have:
#
#     training data
#     test data
#
# You should NOT calculate preprocessing statistics from the combined
# train + test data when those statistics will influence the model.
#
# Correct conceptual flow:
#
#     TRAIN
#       ↓
#     calculate preprocessing parameters
#       ↓
#     fit transformation
#       ↓
#     transform TRAIN
#
#     TEST
#       ↓
#     use parameters learned from TRAIN
#       ↓
#     transform TEST
#
# Pandas is useful for preparation, but ML pipelines should carefully
# control this process.
#
#
# ============================================================
# 143. RANDOM SAMPLING
# ============================================================
#
# df.sample(
#     frac=0.8,
#     random_state=42
# )
#
# frac=0.8 means:
#
#     select approximately 80% of rows.
#
# random_state makes the random selection reproducible.
#
#
# ============================================================
# 144. TRAIN / TEST SPLIT — PANDAS CAN HELP, BUT...
# ============================================================
#
# You can manually split data using Pandas:
#
#     train = df.sample(frac=0.8, random_state=42)
#     test = df.drop(train.index)
#
# But for ML, normally prefer:
#
#     sklearn.model_selection.train_test_split
#
# because it provides more ML-specific controls such as stratification.
#
#
# ============================================================
# 145. SAMPLE WITH REPLACEMENT
# ============================================================
#
# df.sample(
#     n=100,
#     replace=True,
#     random_state=42
# )
#
# replace=True allows the same row to be selected more than once.
#
# This can be useful for resampling techniques.
#
#
# ============================================================
# 146. SHUFFLING
# ============================================================
#
# shuffled = df.sample(
#     frac=1,
#     random_state=42
# )
#
# frac=1 means:
#
#     select 100% of the rows
#
# but in randomized order.
#
#
# ============================================================
# 147. INDEXING AS A DATA MODEL
# ============================================================
#
# Do not think of the index as merely:
#
#     "the row number"
#
# Better:
#
#     index = labels used to identify/alignment rows
#
# The default integer index LOOKS like row numbers.
#
# But the index can contain:
#
#     IDs
#     dates
#     strings
#     hierarchical labels
#
#
# ============================================================
# 148. INDEX ALIGNMENT IN PRACTICE
# ============================================================
#
# Suppose:
#
#     sales.index
#         = customer IDs
#
#     costs.index
#         = customer IDs
#
# Then:
#
#     sales - costs
#
# can align customers by their labels.
#
# This is one of Pandas' most powerful conceptual features.
#
#
# ============================================================
# 149. METHODS VS ATTRIBUTES
# ============================================================
#
# Attribute:
#
#     df.shape
#     df.columns
#     df.index
#     df.dtypes
#
# Method:
#
#     df.head()
#     df.describe()
#     df.info()
#     df.sort_values()
#     df.drop()
#
# Mental rule:
#
#     attribute → information/property
#
#     method → action/operation
#
#
# ============================================================
# 150. METHOD CHAINING
# ============================================================
#
# Pandas operations can often be chained:
#
#     result = (
#         df
#         .dropna()
#         .query("age > 30")
#         .sort_values("salary", ascending=False)
#     )
#
# Read it as:
#
#     original data
#         ↓
#     remove missing rows
#         ↓
#     keep age > 30
#         ↓
#     sort by salary
#         ↓
#     final DataFrame
#
# This is essentially a data-processing pipeline.
#
#
# ============================================================
# 151. METHOD CHAINING MENTAL MODEL
# ============================================================
#
#     DATA
#       ↓
#     operation 1
#       ↓
#     intermediate DataFrame
#       ↓
#     operation 2
#       ↓
#     intermediate DataFrame
#       ↓
#     operation 3
#       ↓
#     RESULT
#
# This is a powerful way to think about Pandas.
#
#
# ============================================================
# 152. PIPE()
# ============================================================
#
# pipe() can pass a DataFrame through a function.
#
# Conceptually:
#
#     df
#      ↓
#     function
#      ↓
#     transformed df
#
# This is useful for organizing reusable data-processing pipelines.
#
#
# ============================================================
# 153. MEMORY-EFFICIENT COLUMN SELECTION
# ============================================================
#
# When reading large datasets, avoid loading unnecessary columns:
#
#     df = pd.read_csv(
#         "large.csv",
#         usecols=["id", "age", "salary"]
#     )
#
# Concept:
#
#     entire file
#          ↓
#     select required columns during loading
#          ↓
#     smaller DataFrame
#
# This can reduce memory usage and processing work.
#
#
# ============================================================
# 154. CHUNKED READING
# ============================================================
#
# Very large CSV files can be processed in chunks:
#
#     for chunk in pd.read_csv(
#         "large.csv",
#         chunksize=10000
#     ):
#         # process chunk
#         pass
#
# Mental model:
#
#     huge file
#         ↓
#     chunk 1
#         ↓
#     process
#         ↓
#     chunk 2
#         ↓
#     process
#         ↓
#     ...
#
# This prevents the entire dataset from needing to exist in memory
# at once.
#
#
# ============================================================
# 155. DATA QUALITY WORKFLOW
# ============================================================
#
# A practical Pandas data-quality investigation:
#
#     df.info()
#          ↓
#     check dtypes
#          ↓
#     df.isna().sum()
#          ↓
#     missingness
#          ↓
#     df.duplicated().sum()
#          ↓
#     duplicates
#          ↓
#     df.nunique()
#          ↓
#     cardinality
#          ↓
#     value_counts()
#          ↓
#     suspicious categories
#          ↓
#     describe()
#          ↓
#     numerical anomalies
#          ↓
#     clean
#
#
# ============================================================
# 156. COMMON DATA-CLEANING PATTERN
# ============================================================
#
#     RAW DATA
#         ↓
#     inspect structure
#         ↓
#     standardize column names
#         ↓
#     correct data types
#         ↓
#     detect missing values
#         ↓
#     detect duplicates
#         ↓
#     validate categories
#         ↓
#     investigate outliers
#         ↓
#     transform variables
#         ↓
#     validated dataset
#
#
# ============================================================
# 157. COLUMN-NAME CLEANING
# ============================================================
#
# A common pattern:
#
#     df.columns = (
#         df.columns
#         .str.strip()
#         .str.lower()
#         .str.replace(" ", "_")
#     )
#
# Example:
#
#     " Customer Name "
#
# becomes:
#
#     "customer_name"
#
# This makes downstream code more consistent.
#
#
# ============================================================
# 158. DUPLICATE CATEGORY CLEANING
# ============================================================
#
# Real data may contain:
#
#     "Tehran"
#     "tehran"
#     " Tehran "
#
# These may represent the same category.
#
# A typical normalization:
#
#     df["city"] = (
#         df["city"]
#         .str.strip()
#         .str.lower()
#     )
#
# Then:
#
#     df["city"].value_counts()
#
# can reveal the cleaned distribution.
#
#
# ============================================================
# 159. MISSING DATA STRATEGY
# ============================================================
#
# Never start with:
#
#     df.fillna(0)
#
# automatically.
#
# First ask:
#
#     WHY is the value missing?
#
# Then:
#
#     identify missingness
#          ↓
#     understand meaning
#          ↓
#     choose strategy
#          ↓
#     remove / impute / preserve
#
# Possible strategies:
#
#     drop rows
#     drop columns
#     mean
#     median
#     mode
#     forward fill
#     backward fill
#     domain-specific value
#     model-based imputation
#
#
# ============================================================
# 160. FORWARD FILL
# ============================================================
#
# df["value"] = df["value"].ffill()
#
# A missing value receives the most recent previous non-missing value.
#
# Concept:
#
#     10
#     NaN
#     NaN
#     15
#
# becomes:
#
#     10
#     10
#     10
#     15
#
# This can be appropriate for certain time-series data,
# but inappropriate for many ordinary datasets.
#
#
# ============================================================
# 161. BACKWARD FILL
# ============================================================
#
# df["value"] = df["value"].bfill()
#
# Missing values can be filled from the next available value.
#
#
# ============================================================
# 162. EXPLODING LIST-LIKE DATA
# ============================================================
#
# explode() turns list-like values into multiple rows.
#
# Example concept:
#
#     customer    products
#     Ali         [A, B, C]
#
# after:
#
#     customer    product
#     Ali         A
#     Ali         B
#     Ali         C
#
# Useful when data contains nested list-like fields.
#
#
# ============================================================
# 163. STRIPING JSON-LIKE / NESTED DATA
# ============================================================
#
# pd.json_normalize(...)
#
# can flatten nested JSON structures into tabular form.
#
# This is particularly useful when collecting data from APIs.
#
#
# ============================================================
# 164. APPLY VS VECTORISATION
# ============================================================
#
# Prefer:
#
#     df["salary"] * 12
#
# over:
#
#     df["salary"].apply(lambda x: x * 12)
#
# when the operation can naturally be expressed as a vectorized
# Pandas/NumPy operation.
#
# Why?
#
#     vectorized operation
#           ↓
#     generally clearer
#           +
#     often faster
#
# apply() is still useful when the transformation genuinely requires
# a custom Python function.
#
#
# ============================================================
# 165. PANDAS + NUMPY
# ============================================================
#
# Pandas is deeply connected to NumPy.
#
# Typical relationship:
#
#     Pandas DataFrame
#            ↓
#     labeled tabular structure
#            ↓
#     NumPy arrays underneath many numerical operations
#
#
# Typical workflow:
#
#     Pandas
#       ↓
#     clean/select/transform
#       ↓
#     NumPy
#       ↓
#     numerical computation
#       ↓
#     scikit-learn / SciPy / statsmodels
#
#
# ============================================================
# 166. PANDAS + SCIKIT-LEARN
# ============================================================
#
# A very common ML pipeline:
#
#     CSV
#      ↓
#     Pandas
#      ↓
#     cleaning
#      ↓
#     feature selection
#      ↓
#     X and y
#      ↓
#     scikit-learn
#      ↓
#     model
#      ↓
#     evaluation
#
#
# Example:
#
#     X = df[["age", "salary"]]
#     y = df["target"]
#
# Pandas prepares the data.
#
# Scikit-learn performs ML operations.
#
#
# ============================================================
# 167. X AND y MENTAL MODEL
# ============================================================
#
# In supervised learning:
#
#     X
#       ↓
#     input features
#       ↓
#     model
#       ↓
#     prediction
#
#     y
#       ↓
#     target / label
#
# Pandas is often used to construct:
#
#     X = df[feature_columns]
#     y = df[target_column]
#
#
# ============================================================
# 168. SELECTING FEATURES
# ============================================================
#
# feature_columns = [
#     "age",
#     "salary",
#     "experience"
# ]
#
# X = df[feature_columns]
#
# This is one of the most common transitions from Pandas to ML.
#
#
# ============================================================
# 169. TARGET SELECTION
# ============================================================
#
# y = df["failure"]
#
# Now:
#
#     X → predictors
#     y → target
#
#
# ============================================================
# 170. CATEGORICAL VARIABLES BEFORE ML
# ============================================================
#
# Suppose:
#
#     city = Tehran / Karaj / Tabriz
#
# Many ML algorithms need numerical representation.
#
# Pandas can help inspect and prepare categories, but for ML,
# scikit-learn encoders such as OneHotEncoder are often preferable.
#
# Pandas:
#
#     understand
#     clean
#     validate
#     inspect
#
# scikit-learn:
#
#     formal preprocessing pipeline
#     encoding
#     scaling
#     train/test-safe transformations
#
#
# ============================================================
# 171. CORRELATION MATRIX FOR EDA
# ============================================================
#
# corr = df.select_dtypes(
#     include="number"
# ).corr()
#
# This creates a numerical relationship matrix.
#
# Example conceptual output:
#
#              age   salary
#     age      1.00   0.72
#     salary   0.72   1.00
#
# This can help during exploratory analysis.
#
# But:
#
#     correlation ≠ causation
#
#
# ============================================================
# 172. DESCRIPTIVE STATISTICS VS MODELING
# ============================================================
#
# Pandas is excellent for:
#
#     descriptive statistics
#     grouping
#     filtering
#     aggregation
#     cleaning
#     transformation
#
# It is NOT intended to replace:
#
#     scipy
#     statsmodels
#     scikit-learn
#     XGBoost
#     PyTorch
#     TensorFlow
#
# Each library has a different expertise.
#
#
# ============================================================
# 173. PANDAS FOR DATA ANALYSIS (DA)
# ============================================================
#
# For DA, Pandas is HIGH PRIORITY.
#
# Learn very well:
#
#     read_csv()
#     read_excel()
#     read_sql()
#     head()
#     tail()
#     sample()
#     info()
#     describe()
#     shape
#     columns
#     index
#     dtypes
#     loc
#     iloc
#     boolean filtering
#     isin()
#     between()
#     sort_values()
#     value_counts()
#     unique()
#     nunique()
#     groupby()
#     agg()
#     pivot_table()
#     merge()
#     concat()
#     drop()
#     rename()
#     dropna()
#     fillna()
#     duplicated()
#     drop_duplicates()
#     assign()
#     map()
#     apply()
#     to_datetime()
#     dt
#     shift()
#     diff()
#     pct_change()
#     rolling()
#     resample()
#     to_csv()
#     to_excel()
#
#
# ============================================================
# 174. PANDAS FOR DATA SCIENCE (DS)
# ============================================================
#
# For DS, Pandas is HIGH PRIORITY for:
#
#     data ingestion
#     cleaning
#     EDA
#     missing values
#     feature engineering
#     grouping
#     categorical analysis
#     time-series preparation
#     correlation analysis
#     dataset preparation
#     train/test preparation
#
# Especially important:
#
#     loc / iloc
#     boolean masks
#     groupby
#     merge
#     concat
#     missing values
#     dtype management
#     datetime
#     vectorized transformations
#     categorical handling
#
#
# ============================================================
# 175. PANDAS FOR MACHINE LEARNING (ML)
# ============================================================
#
# Pandas is primarily a PREPROCESSING and DATA-UNDERSTANDING tool
# in ML.
#
# Typical role:
#
#     raw dataset
#          ↓
#     Pandas
#          ↓
#     inspect
#          ↓
#     clean
#          ↓
#     validate
#          ↓
#     engineer features
#          ↓
#     create X / y
#          ↓
#     scikit-learn / XGBoost / etc.
#
#
# ============================================================
# 176. PANDAS FOR ML — HIGH-VALUE OPERATIONS
# ============================================================
#
# Particularly useful:
#
#     read_csv()
#     read_excel()
#     info()
#     describe()
#     isna()
#     fillna()
#     dropna()
#     drop_duplicates()
#     astype()
#     to_numeric()
#     to_datetime()
#     loc
#     iloc
#     query()
#     isin()
#     groupby()
#     transform()
#     merge()
#     concat()
#     map()
#     assign()
#     shift()
#     diff()
#     rolling()
#     cut()
#     qcut()
#     get_dummies()
#
#
# ============================================================
# 177. get_dummies()
# ============================================================
#
# pd.get_dummies(df, columns=["city"])
#
# can convert categorical variables into indicator columns.
#
# Concept:
#
#     city
#     Tehran
#     Karaj
#     Tabriz
#
#        ↓
#
#     city_Karaj
#     city_Tehran
#     city_Tabriz
#
# This is called one-hot encoding.
#
# IMPORTANT:
#
# For serious ML pipelines, sklearn's OneHotEncoder is often
# preferable because it integrates with train/test preprocessing
# and pipelines more safely.
#
#
# ============================================================
# 178. FEATURE ENGINEERING WITH PANDAS
# ============================================================
#
# Feature engineering means:
#
#     raw variables
#          ↓
#     meaningful transformation
#          ↓
#     model features
#
# Examples:
#
#     df["age_squared"] = df["age"] ** 2
#
#     df["income_per_age"] = (
#         df["salary"] / df["age"]
#     )
#
#     df["is_senior"] = df["age"] >= 40
#
#     df["year"] = df["date"].dt.year
#
#     df["previous_sales"] = df["sales"].shift(1)
#
# These are classic Pandas-based feature-engineering operations.
#
#
# ============================================================
# 179. GROUP-BASED FEATURE ENGINEERING
# ============================================================
#
# Example:
#
#     df["customer_avg"] = (
#         df.groupby("customer_id")["amount"]
#           .transform("mean")
#     )
#
# Every transaction now receives the average transaction amount
# associated with its customer.
#
# This is powerful, but:
#
# IMPORTANT ML WARNING:
#
# You must check whether the feature uses information that would
# be unavailable at prediction time.
#
# Otherwise, you can create target leakage or temporal leakage.
#
#
# ============================================================
# 180. DATA LEAKAGE AND GROUPBY
# ============================================================
#
# A feature can look mathematically valid but still be invalid
# for ML.
#
# Example:
#
#     target-related future information
#              ↓
#          groupby()
#              ↓
#       engineered feature
#              ↓
#          ML model
#
# If the feature contains information from the future or target,
# model performance may become artificially high.
#
# Therefore:
#
#     mathematically correct
#         ≠
#     ML-valid
#
#
# ============================================================
# 181. PANDAS + POWER BI / BUSINESS ANALYTICS
# ============================================================
#
# For practical Data Analyst work, Pandas is useful for:
#
#     Excel cleanup
#     CSV cleanup
#     joining datasets
#     KPI calculation
#     customer segmentation
#     sales analysis
#     marketing analysis
#     missing-value handling
#     duplicate detection
#     time-series preparation
#     preparing data for Power BI
#
#
# Typical workflow:
#
#     Excel / CSV / SQL
#             ↓
#          Pandas
#             ↓
#       clean / transform
#             ↓
#       validated dataset
#             ↓
#           Power BI
#             ↓
#        dashboard / KPI
#
#
# ============================================================
# 182. PANDAS + SQL
# ============================================================
#
# The two are complementary.
#
# SQL:
#
#     retrieve and manipulate data in databases
#
# Pandas:
#
#     analyze and transform data in Python
#
#
# Common workflow:
#
#     database
#         ↓
#     SQL query
#         ↓
#     DataFrame
#         ↓
#     Pandas analysis
#
#
# ============================================================
# 183. PANDAS + STATISTICS
# ============================================================
#
# Pandas provides convenient descriptive statistics:
#
#     mean
#     median
#     std
#     var
#     quantile
#     covariance
#     correlation
#
# But Pandas is NOT a complete statistics library.
#
# For deeper statistical inference:
#
#     Pandas
#       ↓
#     prepare data
#       ↓
#     SciPy / statsmodels
#       ↓
#     hypothesis tests
#     probability distributions
#     regression inference
#     confidence intervals
#     statistical models
#
#
# ============================================================
# 184. PANDAS + VISUALIZATION
# ============================================================
#
# Pandas can produce basic plots through:
#
#     df.plot()
#
# Examples:
#
#     df["sales"].plot()
#
#     df.plot(
#         x="age",
#         y="salary",
#         kind="scatter"
#     )
#
# However:
#
#     Pandas plotting interface
#          ↓
#     convenient interface
#          ↓
#     Matplotlib underneath
#
# For advanced visualization:
#
#     Matplotlib
#     Seaborn
#     Plotly
#
# are more appropriate.
#
#
# ============================================================
# 185. PANDAS — WHAT YOU SHOULD MASTER
# ============================================================
#
# If your goal is DA + DS + ML, divide learning into levels.
#
# ------------------------------------------------------------
# LEVEL 1 — MUST MASTER
# ------------------------------------------------------------
#
#     Series
#     DataFrame
#     read_csv()
#     head()
#     info()
#     describe()
#     shape
#     columns
#     dtypes
#     loc
#     iloc
#     boolean filtering
#     column creation
#     sort_values()
#     value_counts()
#     unique()
#     nunique()
#     isna()
#     fillna()
#     dropna()
#     drop_duplicates()
#
#
# ------------------------------------------------------------
# LEVEL 2 — MUST BE COMFORTABLE WITH
# ------------------------------------------------------------
#
#     groupby()
#     agg()
#     transform()
#     merge()
#     concat()
#     pivot_table()
#     melt()
#     map()
#     apply()
#     assign()
#     rename()
#     set_index()
#     reset_index()
#     astype()
#     to_numeric()
#     to_datetime()
#     select_dtypes()
#
#
# ------------------------------------------------------------
# LEVEL 3 — VERY USEFUL
# ------------------------------------------------------------
#
#     shift()
#     diff()
#     pct_change()
#     rolling()
#     expanding()
#     resample()
#     cut()
#     qcut()
#     rank()
#     corr()
#     cov()
#     query()
#     explode()
#     json_normalize()
#     get_dummies()
#
#
# ------------------------------------------------------------
# LEVEL 4 — LEARN WHEN NEEDED
# ------------------------------------------------------------
#
#     MultiIndex
#     stack()
#     unstack()
#     pipe()
#     advanced reshaping
#     advanced time-series operations
#     chunked processing
#     performance optimization
#
#
# ============================================================
# 186. THE PANDAS SKILL MAP
# ============================================================
#
#                         PANDAS
#                            │
#       ┌────────────────────┼─────────────────────┐
#       ↓                    ↓                     ↓
#    INPUT                MANIPULATION          OUTPUT
#       │                    │                     │
#       ↓                    ↓                     ↓
# read_csv()             loc / iloc           to_csv()
# read_excel()           filter              to_excel()
# read_sql()             assign              to_sql()
# read_json()            groupby
#                        merge
#                        concat
#                        reshape
#                            │
#                            ↓
#                         ANALYSIS
#                            │
#                  ┌─────────┼─────────┐
#                  ↓         ↓         ↓
#               describe   agg       corr
#                  ↓         ↓         ↓
#               summary   groups   relationships
#
#
# ============================================================
# 187. THE COMPLETE DATA ANALYST PANDAS PIPELINE
# ============================================================
#
#     1. LOAD
#        ↓
#     pd.read_csv()
#
#     2. INSPECT
#        ↓
#     head()
#     info()
#     describe()
#
#     3. UNDERSTAND
#        ↓
#     dtypes
#     unique()
#     value_counts()
#
#     4. CLEAN
#        ↓
#     missing values
#     duplicates
#     invalid values
#     incorrect types
#
#     5. TRANSFORM
#        ↓
#     assign()
#     map()
#     apply()
#     astype()
#
#     6. FILTER
#        ↓
#     loc
#     boolean masks
#     query()
#
#     7. COMBINE
#        ↓
#     merge()
#     concat()
#
#     8. SUMMARIZE
#        ↓
#     groupby()
#     agg()
#     pivot_table()
#
#     9. TIME ANALYSIS
#        ↓
#     to_datetime()
#     resample()
#     rolling()
#     shift()
#     diff()
#
#     10. EXPORT
#         ↓
#     to_csv()
#     to_excel()
#
#
# ============================================================
# 188. THE DATA SCIENCE PANDAS PIPELINE
# ============================================================
#
#     RAW DATA
#         ↓
#     Pandas DataFrame
#         ↓
#     DATA QUALITY
#         ↓
#     missing values
#     duplicates
#     types
#     invalid categories
#         ↓
#     EDA
#         ↓
#     distributions
#     relationships
#     groups
#     correlations
#         ↓
#     FEATURE ENGINEERING
#         ↓
#     transformations
#     lags
#     aggregates
#     categorical preparation
#         ↓
#     X / y
#         ↓
#     ML library
#
#
# ============================================================
# 189. THE MOST IMPORTANT PANDAS PATTERNS
# ============================================================
#
# Pattern 1:
#
#     df["column"]
#
#     → select one column
#
#
# Pattern 2:
#
#     df[["a", "b"]]
#
#     → select multiple columns
#
#
# Pattern 3:
#
#     df[df["x"] > value]
#
#     → filter rows
#
#
# Pattern 4:
#
#     df.loc[condition, "column"] = value
#
#     → conditional assignment
#
#
# Pattern 5:
#
#     df["new"] = transformation(df["old"])
#
#     → create a feature/column
#
#
# Pattern 6:
#
#     df.groupby("category")["value"].mean()
#
#     → group + aggregate
#
#
# Pattern 7:
#
#     df.groupby("category")["value"].transform("mean")
#
#     → group statistic aligned back to rows
#
#
# Pattern 8:
#
#     pd.merge(df1, df2, on="id")
#
#     → relational join
#
#
# Pattern 9:
#
#     pd.concat([df1, df2])
#
#     → combine datasets
#
#
# Pattern 10:
#
#     df.isna().sum()
#
#     → missing-value counts
#
#
# Pattern 11:
#
#     df["category"].value_counts()
#
#     → category frequencies
#
#
# Pattern 12:
#
#     df.sort_values("column")
#
#     → order observations
#
#
# ============================================================
# 190. PANDAS FUNCTIONS / METHODS YOU SHOULD KNOW BY PURPOSE
# ============================================================
#
# ------------------------------------------------------------
# CREATE
# ------------------------------------------------------------
#
# pd.Series()
# pd.DataFrame()
#
# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------
#
# pd.read_csv()
# pd.read_excel()
# pd.read_json()
# pd.read_sql()
#
# ------------------------------------------------------------
# INSPECT
# ------------------------------------------------------------
#
# head()
# tail()
# sample()
# info()
# describe()
# shape
# ndim
# size
# columns
# index
# dtypes
#
# ------------------------------------------------------------
# SELECT
# ------------------------------------------------------------
#
# []
# loc
# iloc
# query()
# select_dtypes()
#
# ------------------------------------------------------------
# FILTER
# ------------------------------------------------------------
#
# boolean masks
# isin()
# between()
# str.contains()
#
# ------------------------------------------------------------
# CLEAN
# ------------------------------------------------------------
#
# isna()
# notna()
# dropna()
# fillna()
# ffill()
# bfill()
# duplicated()
# drop_duplicates()
#
# ------------------------------------------------------------
# TRANSFORM
# ------------------------------------------------------------
#
# assign()
# map()
# apply()
# replace()
# where()
# mask()
# clip()
# astype()
# cut()
# qcut()
#
# ------------------------------------------------------------
# SUMMARIZE
# ------------------------------------------------------------
#
# sum()
# mean()
# median()
# min()
# max()
# std()
# var()
# count()
# quantile()
# agg()
#
# ------------------------------------------------------------
# GROUP
# ------------------------------------------------------------
#
# groupby()
# transform()
#
# ------------------------------------------------------------
# COMBINE
# ------------------------------------------------------------
#
# merge()
# join()
# concat()
#
# ------------------------------------------------------------
# RESHAPE
# ------------------------------------------------------------
#
# pivot()
# pivot_table()
# melt()
# stack()
# unstack()
# explode()
#
# ------------------------------------------------------------
# SORT / INDEX
# ------------------------------------------------------------
#
# sort_values()
# sort_index()
# set_index()
# reset_index()
#
# ------------------------------------------------------------
# TIME SERIES
# ------------------------------------------------------------
#
# pd.to_datetime()
# dt
# resample()
# shift()
# diff()
# pct_change()
# rolling()
# expanding()
#
# ------------------------------------------------------------
# RELATIONSHIPS
# ------------------------------------------------------------
#
# corr()
# cov()
#
# ------------------------------------------------------------
# EXPORT
# ------------------------------------------------------------
#
# to_csv()
# to_excel()
# to_json()
# to_sql()
#
#
# ============================================================
# 191. WHAT IS MOST IMPORTANT FOR YOUR DA / DS / ML PATH?
# ============================================================
#
# If you are learning Pandas specifically for:
#
#     Data Analysis
#     Data Science
#     Machine Learning
#
# do NOT try to memorize every Pandas method.
#
# Master the underlying operations:
#
#     SELECT
#       ↓
#     FILTER
#       ↓
#     TRANSFORM
#       ↓
#     GROUP
#       ↓
#     AGGREGATE
#       ↓
#     COMBINE
#       ↓
#     RESHAPE
#       ↓
#     TIME-SERIES PROCESS
#       ↓
#     ANALYZE
#
#
# Once these concepts are intuitive, individual methods become much
# easier to remember.
#
#
# ============================================================
# 192. THE DEEPEST PANDAS MENTAL MODEL
# ============================================================
#
# Think of almost every Pandas operation as answering one of these
# questions:
#
#     WHERE IS THE DATA?
#         → loc / iloc
#
#     WHICH ROWS DO I WANT?
#         → Boolean filtering / query()
#
#     WHICH COLUMNS DO I WANT?
#         → column selection
#
#     HOW DO I CHANGE THE DATA?
#         → assign / map / apply / replace
#
#     HOW DO I HANDLE MISSING DATA?
#         → isna / fillna / dropna
#
#     HOW DO I SUMMARIZE DATA?
#         → agg / describe
#
#     HOW DO I SPLIT DATA INTO GROUPS?
#         → groupby
#
#     HOW DO I COMBINE DATASETS?
#         → merge / concat / join
#
#     HOW DO I CHANGE THE SHAPE?
#         → pivot / melt / stack / unstack
#
#     HOW DO I HANDLE TIME?
#         → datetime / resample / rolling / shift
#
#     HOW DO I PREPARE DATA FOR ML?
#         → select / clean / transform / engineer features
#
#
# ============================================================
# 193. PANDAS CORE IDEA
# ============================================================
#
# Pandas is fundamentally about turning:
#
#     RAW, MESSY, STRUCTURED DATA
#
#                  ↓
#
#             REPRESENT
#
#                  ↓
#
#              INSPECT
#
#                  ↓
#
#               CLEAN
#
#                  ↓
#
#              TRANSFORM
#
#                  ↓
#
#               FILTER
#
#                  ↓
#
#                GROUP
#
#                  ↓
#
#               COMBINE
#
#                  ↓
#
#               RESHAPE
#
#                  ↓
#
#               ANALYZE
#
#                  ↓
#
#          MODEL-READY / REPORT-READY DATA
#
#
# ============================================================
# 194. CORE IDEA
# ============================================================
#
# The most important principle to remember is:
#
#     PANDAS = LABELED DATA MANIPULATION
#
# A DataFrame is not merely a table.
#
# It is a labeled collection of columns and rows whose labels allow
# Pandas to select, align, combine, transform, and summarize data.
#
# The most reusable mental pattern is:
#
#     SELECT
#       ↓
#     FILTER
#       ↓
#     TRANSFORM
#       ↓
#     GROUP / AGGREGATE
#       ↓
#     COMBINE
#       ↓
#     RESHAPE
#       ↓
#     ANALYZE
#
# For your DA / DS / ML work, the highest-value Pandas skill is NOT
# memorizing hundreds of methods.
#
# It is being able to look at a dataset and think:
#
#     "What is the structure?"
#             ↓
#     "What does each row represent?"
#             ↓
#     "What does each column represent?"
#             ↓
#     "What is missing or invalid?"
#             ↓
#     "Which rows/columns do I need?"
#             ↓
#     "What transformation represents the question?"
#             ↓
#     "Should I filter, group, aggregate, join, or reshape?"
#             ↓
#     "What should the resulting data structure look like?"
#
# Once this mental model becomes intuitive, Pandas stops being a
# collection of commands and becomes a language for thinking about
# structured data.
#
# ============================================================
```
