# STATSMODELS

```python
# ============================================================
# STATSMODELS — STATISTICS, INFERENCE, REGRESSION & TIME SERIES
# ============================================================
#
# PURPOSE OF THIS NOTE
# ------------------------------------------------------------
# Learn statsmodels conceptually, not as a list of functions.
#
# Main mental model:
#
#     DATA
#       ↓
#     MODEL
#       ↓
#     ESTIMATION
#       ↓
#     COEFFICIENTS
#       ↓
#     UNCERTAINTY
#       ↓
#     HYPOTHESIS TESTS
#       ↓
#     DIAGNOSTICS
#       ↓
#     INTERPRETATION
#
#
# statsmodels is especially valuable when you want to answer:
#
#     "What is the relationship?"
#     "How large is the effect?"
#     "Is the effect statistically significant?"
#     "How uncertain is the estimate?"
#     "Does the model satisfy its assumptions?"
#     "What happens after controlling for other variables?"
#
# This makes statsmodels extremely useful for:
#
#     DA  → VERY USEFUL
#     DS  → VERY USEFUL
#     ML  → USEFUL, especially for statistical modeling/inference
#
# ------------------------------------------------------------
# IMPORTANT:
#
# statsmodels is NOT primarily a replacement for:
#
#     pandas       → data manipulation
#     NumPy        → numerical arrays
#     SciPy        → scientific/statistical algorithms
#     scikit-learn → predictive machine learning
#
# Think:
#
#     NumPy       = numerical foundation
#     pandas      = data manipulation
#     SciPy       = scientific/statistical algorithms
#     statsmodels = statistical modeling + inference
#     sklearn     = predictive ML
#
#
# A useful distinction:
#
#                 INFERENCE                PREDICTION
#                    │                         │
#                    ↓                         ↓
#              statsmodels                sklearn
#                    │                         │
#         "What does X do to Y?"       "Can I predict Y?"
#         "Is X significant?"          "How accurate?"
#         "How uncertain?"             "How well generalizes?"
#
# ============================================================
# 01. INSTALLATION AND IMPORT
# ============================================================
#
# Install:
#
#     pip install statsmodels
#
# Import:
#
#     import statsmodels.api as sm
#
# Other important modules:
#
#     import statsmodels.formula.api as smf
#     import statsmodels.api as sm
#
# Time series:
#
#     from statsmodels.tsa.arima.model import ARIMA
#
# Statistical tests:
#
#     from statsmodels.stats import ...
#
# Robust regression:
#
#     from statsmodels.robust.robust_linear_model import RLM
#
#
# The two most important starting points are:
#
#     statsmodels.api
#     statsmodels.formula.api
#
#
# ============================================================
# 02. THE CORE IDEA OF STATSMODELS
# ============================================================
#
# Suppose we have:
#
#     advertising_cost → sales
#
# Data:
#
#     Advertising     Sales
#     ---------------------
#       10              120
#       20              145
#       30              170
#       40              195
#
#
# A statistical model asks:
#
#     Sales = β₀ + β₁ × Advertising + error
#
#
# Visual idea:
#
#     Sales
#       ↑
#       │                         ●
#       │                    ●
#       │               ●
#       │          ●
#       │
#       └────────────────────────────→ Advertising
#
#
# statsmodels estimates:
#
#     β₀       intercept
#     β₁       slope/effect
#
# and also gives:
#
#     standard error
#     confidence interval
#     t-statistic
#     p-value
#     R²
#     adjusted R²
#     residual diagnostics
#     F-test
#     AIC
#     BIC
#     etc.
#
#
# Therefore:
#
#     sklearn:
#
#         X → model → prediction
#
#     statsmodels:
#
#         X → model → coefficient
#                       ↓
#                   uncertainty
#                       ↓
#                    inference
#
#
# ============================================================
# 03. THE MOST IMPORTANT OBJECT IN STATSMODELS:
#     THE RESULTS OBJECT
# ============================================================
#
# A typical workflow:
#
#     model = sm.OLS(y, X)
#     results = model.fit()
#
#
# Notice the two stages:
#
#     MODEL SPECIFICATION
#             ↓
#          fit()
#             ↓
#     ESTIMATED MODEL
#
#
# `results` contains much more than predictions.
#
# Important attributes/methods include:
#
#     results.params
#     results.bse
#     results.tvalues
#     results.pvalues
#     results.conf_int()
#     results.rsquared
#     results.rsquared_adj
#     results.fvalue
#     results.f_pvalue
#     results.resid
#     results.fittedvalues
#     results.predict()
#     results.aic
#     results.bic
#     results.summary()
#
#
# Mental model:
#
#     results
#        │
#        ├── params          → estimated effects
#        ├── bse             → uncertainty
#        ├── tvalues         → standardized effects
#        ├── pvalues         → hypothesis evidence
#        ├── conf_int()      → plausible effect range
#        ├── fittedvalues    → model predictions
#        ├── resid           → errors
#        ├── rsquared        → explanatory fit
#        ├── aic/bic         → model comparison
#        └── summary()       → complete statistical report
#
#
# ============================================================
# 04. OLS — ORDINARY LEAST SQUARES
# ============================================================
#
# OLS is one of the MOST IMPORTANT statsmodels models.
#
#     Ordinary Least Squares
#
# Model:
#
#     y = β₀ + β₁x₁ + β₂x₂ + ... + ε
#
#
# Example:
#
#     house_price =
#         β₀
#         + β₁ × area
#         + β₂ × bedrooms
#         + β₃ × age
#         + error
#
#
# ------------------------------------------------------------
# BASIC EXAMPLE
# ------------------------------------------------------------
#
#     import numpy as np
#     import statsmodels.api as sm
#
#     X = np.array([
#         [1000],
#         [1500],
#         [2000],
#         [2500],
#         [3000]
#     ])
#
#     y = np.array([
#         200,
#         280,
#         350,
#         430,
#         500
#     ])
#
#     X = sm.add_constant(X)
#
#     model = sm.OLS(y, X)
#     results = model.fit()
#
#     print(results.summary())
#
#
# ------------------------------------------------------------
# WHY add_constant?
# ------------------------------------------------------------
#
# Mathematically:
#
#     y = β₀ + β₁x
#
# The β₀ term is the intercept.
#
# The design matrix needs a column of 1s:
#
#             x
#     ┌────────────┐
#     │ 1    1000  │
#     │ 1    1500  │
#     │ 1    2000  │
#     │ 1    2500  │
#     │ 1    3000  │
#     └────────────┘
#
# `sm.add_constant(X)` adds that column.
#
#
# ============================================================
# 05. OLS — WHAT DOES THE MODEL ACTUALLY DO?
# ============================================================
#
# Suppose:
#
#     actual y = 350
#     predicted y = 330
#
# Then:
#
#     residual = actual - predicted
#              = 350 - 330
#              = 20
#
#
# Visual:
#
#     y
#     ↑
#     │              ● actual
#     │              │
#     │              │  residual
#     │              │
#     │              × predicted
#     │             /
#     │            /
#     └────────────────────→ x
#
#
# OLS chooses coefficients that minimize:
#
#     Σ(actual - predicted)²
#
#
# Why square?
#
#     positive error → positive
#     negative error → positive
#
# and large errors receive greater penalty.
#
#
# ============================================================
# 06. OLS RESULTS — THE MOST IMPORTANT OUTPUT
# ============================================================
#
#     results.params
#
# Gives:
#
#     intercept
#     coefficient(s)
#
#
# Example:
#
#     results.params
#
# might conceptually give:
#
#     intercept = 50
#     area      = 0.15
#
#
# Interpretation:
#
#     predicted_price =
#         50 + 0.15 × area
#
#
# If area increases by 100:
#
#     predicted increase = 0.15 × 100
#                        = 15
#
#
# ------------------------------------------------------------
# Standard error
# ------------------------------------------------------------
#
#     results.bse
#
# Measures uncertainty in estimated coefficients.
#
#
# Concept:
#
#     coefficient
#          │
#          ±
#     uncertainty
#          │
#          ↓
#     confidence interval
#
#
# ------------------------------------------------------------
# t-statistic
# ------------------------------------------------------------
#
#     results.tvalues
#
# Approximately:
#
#     t = coefficient / standard_error
#
#
# Large absolute t:
#
#     stronger evidence that coefficient ≠ 0
#
#
# ------------------------------------------------------------
# p-value
# ------------------------------------------------------------
#
#     results.pvalues
#
# Tests:
#
#     H₀: coefficient = 0
#
# versus:
#
#     H₁: coefficient ≠ 0
#
#
# IMPORTANT:
#
# A small p-value does NOT mean:
#
#     "the model is accurate"
#
# It means roughly:
#
#     "the observed evidence would be relatively unusual
#      if the null hypothesis were true."
#
#
# ============================================================
# 07. CONFIDENCE INTERVALS
# ============================================================
#
#     results.conf_int()
#
# Example conceptual result:
#
#     coefficient = 0.15
#     95% CI = [0.10, 0.20]
#
#
# Visual:
#
#     0.10          0.15          0.20
#       │────────────●─────────────│
#       └──────── 95% CI ─────────┘
#
#
# The interval describes uncertainty around the estimated
# population parameter under the model assumptions.
#
#
# For DA/DS:
#
#     VERY USEFUL
#
# because business questions often involve:
#
#     "How large is the effect?"
#
# rather than simply:
#
#     "Can I predict?"
#
#
# ============================================================
# 08. R² — EXPLAINED VARIATION
# ============================================================
#
#     results.rsquared
#
# Concept:
#
#     R² = proportion of variation in y explained by model
#
#
# Rough visual:
#
#     Total variation
#     ├──────────────────────────────────────┤
#     │████████████████████                  │
#     │       explained by model             │
#     │                         ████████████ │
#     │                         unexplained   │
#     └──────────────────────────────────────┘
#
#
# Example:
#
#     R² = 0.80
#
# often interpreted as:
#
#     "80% of the sample variation in the dependent
#      variable is accounted for by the fitted model."
#
# Do NOT automatically interpret this as:
#
#     "80% prediction accuracy."
#
#
# ============================================================
# 09. ADJUSTED R²
# ============================================================
#
#     results.rsquared_adj
#
# Why does it exist?
#
# Adding variables can increase ordinary R² even when those
# variables are not useful.
#
#
# Therefore:
#
#     R²
#       ↑
#       adding predictors usually cannot decrease it
#
#     adjusted R²
#       ↑
#       penalizes unnecessary predictors
#
#
# Useful for comparing models with different numbers of
# predictors.
#
#
# ============================================================
# 10. F-TEST
# ============================================================
#
#     results.fvalue
#     results.f_pvalue
#
# OLS can test:
#
#     H₀:
#     β₁ = β₂ = β₃ = 0
#
# against:
#
#     At least one coefficient is non-zero.
#
#
# Conceptual difference:
#
#     t-test
#         ↓
#     one coefficient
#
#     F-test
#         ↓
#     multiple coefficients / model-level hypothesis
#
#
# ============================================================
# 11. PREDICTIONS
# ============================================================
#
#     results.predict(X_new)
#
# Example:
#
#     new_X = sm.add_constant(np.array([[2200]]))
#
#     prediction = results.predict(new_X)
#
#
# Concept:
#
#     new observation
#           ↓
#        fitted model
#           ↓
#       predicted y
#
#
# ============================================================
# 12. FORMULA API — VERY IMPORTANT FOR DATA ANALYSIS
# ============================================================
#
# statsmodels provides a formula interface similar to R.
#
#     import statsmodels.formula.api as smf
#
# Example:
#
#     model = smf.ols(
#         "sales ~ advertising",
#         data=df
#     )
#
#     results = model.fit()
#
#
# This is often easier to read than manually constructing X.
#
#
# Formula:
#
#     "y ~ x1 + x2 + x3"
#
# means:
#
#     y = dependent variable
#     x1, x2, x3 = predictors
#
#
# Visual:
#
#     "sales ~ advertising + price"
#              │          │
#              ↓          ↓
#          predictor   predictor
#                 \      /
#                  \    /
#                   ↓  ↓
#                  SALES
#
#
# ============================================================
# 13. FORMULA OPERATORS YOU SHOULD KNOW
# ============================================================
#
#     y ~ x
#
#     y predicted/explained by x
#
#
#     y ~ x1 + x2
#
#     y ~ x1 + x2 + x3
#
# Multiple predictors.
#
#
# ------------------------------------------------------------
# Interaction
# ------------------------------------------------------------
#
#     y ~ x1 * x2
#
# expands conceptually to:
#
#     y ~ x1 + x2 + x1:x2
#
#
# The interaction asks:
#
#     "Does the effect of x1 depend on x2?"
#
#
# Example:
#
#     sales ~ advertising * season
#
#
# ------------------------------------------------------------
# Interaction only
# ------------------------------------------------------------
#
#     y ~ x1:x2
#
# means the interaction term itself.
#
#
# ------------------------------------------------------------
# Categorical variable
# ------------------------------------------------------------
#
#     y ~ C(region)
#
# tells statsmodels to treat region as categorical.
#
#
# Example:
#
#     sales ~ advertising + C(region)
#
#
# Concept:
#
#     Region
#       ├── Tehran
#       ├── Karaj
#       ├── Tabriz
#       └── Shiraz
#
# becomes model variables representing group differences.
#
#
# ============================================================
# 14. MULTIPLE LINEAR REGRESSION
# ============================================================
#
# Most practical DA applications are not:
#
#     y ~ x
#
# but:
#
#     y ~ x1 + x2 + x3 + ...
#
#
# Example:
#
#     results = smf.ols(
#         "salary ~ experience + education + age",
#         data=df
#     ).fit()
#
#
# The important conceptual idea:
#
# Each coefficient estimates the association with y
# while holding the other included predictors constant,
# subject to the model assumptions.
#
#
# This is one of the most important ideas in statistical
# analysis.
#
#
# ============================================================
# 15. CATEGORICAL VARIABLES
# ============================================================
#
# Example:
#
#     salary ~ experience + C(department)
#
#
# Concept:
#
#     Department
#       │
#       ├── Engineering
#       ├── Sales
#       └── Marketing
#
# becomes numerical indicator variables.
#
#
# This is called:
#
#     dummy / indicator coding
#
#
# statsmodels' formula system handles this automatically.
#
#
# ============================================================
# 16. LOG TRANSFORMATIONS
# ============================================================
#
# Formula:
#
#     np.log(y) ~ x
#
# or:
#
#     y ~ np.log(x)
#
#
# Why?
#
# Many real-world variables are highly skewed:
#
#     income
#     price
#     population
#     sales
#     company size
#
#
# Log transformation compresses large values.
#
#
# Example:
#
#     1
#     10
#     100
#     1000
#     10000
#
# after log transformation:
#
#     0
#     2.30
#     4.61
#     6.91
#     9.21
#
#
# Visual:
#
# Original:
#
#     1 ── 10 ───────── 100 ─────────────────── 1000
#
# Log:
#
#     0 ── 2 ── 4 ── 6 ── 8
#
#
# Very useful in DA/DS.
#
#
# ============================================================
# 17. RESIDUALS — THE HEART OF MODEL DIAGNOSTICS
# ============================================================
#
#     results.resid
#
# Residual:
#
#     residual = observed - fitted
#
#
# Concept:
#
#     DATA
#       ↓
#     MODEL
#       ↓
#     FITTED VALUE
#       ↓
#     compare with actual
#       ↓
#     RESIDUAL
#
#
# Example:
#
#     actual    = 100
#     predicted = 92
#
#     residual  = 8
#
#
# A good linear model often produces residuals that look
# approximately random rather than showing systematic patterns.
#
#
# ============================================================
# 18. FITTED VALUES
# ============================================================
#
#     results.fittedvalues
#
# These are:
#
#     ŷ
#
# the values predicted by the model for the observations
# used during fitting.
#
#
# Relationship:
#
#     observed y
#          │
#          ├─────────────┐
#          │             │
#          ↓             ↓
#       fitted y       residual
#
#     residual = y - fitted_y
#
#
# ============================================================
# 19. MODEL DIAGNOSTICS
# ============================================================
#
# This is one of the BIGGEST reasons to learn statsmodels.
#
# A model can produce a beautiful R² and still be statistically
# problematic.
#
#
# You should investigate:
#
#     residual structure
#     heteroskedasticity
#     autocorrelation
#     normality of residuals
#     multicollinearity
#     influential observations
#     outliers
#
#
# ============================================================
# 20. HETEROSKEDASTICITY
# ============================================================
#
# Heteroskedasticity:
#
#     residual variance changes with the predictor/fitted value.
#
#
# Good-looking residual pattern:
#
#     residual
#       ↑
#       │  ●   ● ●   ●
#       │ ● ● ●  ● ●
#       │   ● ● ● ●
#       └────────────────→ fitted
#
#
# Problematic:
#
#     residual
#       ↑
#       │ ●
#       │  ● ●
#       │   ● ● ●
#       │      ● ● ● ●
#       │            ● ● ●
#       └────────────────→ fitted
#
# This is the classic "funnel" shape.
#
#
# Useful statsmodels tools:
#
#     from statsmodels.stats.diagnostic import het_breuschpagan
#     from statsmodels.stats.diagnostic import het_white
#
#
# These test for heteroskedasticity.
#
#
# ============================================================
# 21. BREUSCH-PAGAN TEST
# ============================================================
#
#     het_breuschpagan(residuals, exog)
#
# Concept:
#
#     H₀:
#         constant error variance
#
#     H₁:
#         non-constant error variance
#
#
# If heteroskedasticity exists, one common response is to use
# heteroskedasticity-robust standard errors.
#
#
# ============================================================
# 22. ROBUST STANDARD ERRORS
# ============================================================
#
# Very important in practical DA.
#
# Instead of:
#
#     results = model.fit()
#
# you can use:
#
#     results = model.fit(cov_type="HC3")
#
#
# Common covariance options include:
#
#     HC0
#     HC1
#     HC2
#     HC3
#
#
# Concept:
#
#     ordinary standard errors
#              ↓
#       rely on variance assumptions
#
#     robust standard errors
#              ↓
#       more robust to heteroskedasticity
#
#
# Important:
#
# Robust SEs do NOT magically fix a bad model.
#
#
# ============================================================
# 23. AUTOCORRELATION
# ============================================================
#
# Especially important for:
#
#     time series
#     financial data
#     sensor data
#     repeated measurements
#
#
# Residuals should not contain systematic time dependence when
# the model assumptions require independence.
#
#
# Visual:
#
#     residual
#       ↑
#       │ + + + +       ← cluster
#       │       - - -
#       │             + + +
#       └────────────────────→ time
#
#
# This means:
#
#     error today
#          ↕
#     related to error yesterday
#
#
# ============================================================
# 24. DURBIN-WATSON
# ============================================================
#
#     from statsmodels.stats.stattools import durbin_watson
#
#     dw = durbin_watson(results.resid)
#
#
# Rough interpretation:
#
#     DW ≈ 2
#         → little first-order autocorrelation
#
#     DW < 2
#         → positive autocorrelation tendency
#
#     DW > 2
#         → negative autocorrelation tendency
#
#
# It is a diagnostic, not a universal decision rule.
#
#
# ============================================================
# 25. NORMALITY OF RESIDUALS
# ============================================================
#
# For many classical inference procedures, residual distribution
# matters.
#
# Important tools:
#
#     results.resid
#     statsmodels.stats.stattools
#     statsmodels.stats.diagnostic
#
#
# A Q-Q plot is especially useful.
#
#
# Conceptual Q-Q plot:
#
#     observed
#       ↑
#       │           ●
#       │        ●
#       │      ●
#       │    ●
#       │  ●
#       │ ●
#       └────────────────→ theoretical
#
# If points approximately follow a straight line:
#
#     distribution is approximately compatible with the
#     reference distribution.
#
#
# ============================================================
# 26. MULTICOLLINEARITY
# ============================================================
#
# Suppose:
#
#     income
#     age
#     years_experience
#
# are strongly related.
#
#
# Then:
#
#     X1 ──────────┐
#                  ├── correlated predictors
#     X2 ──────────┤
#                  │
#     X3 ──────────┘
#
#
# This can make coefficient estimates unstable.
#
#
# Common diagnostic:
#
#     Variance Inflation Factor
#     VIF
#
#
#     from statsmodels.stats.outliers_influence import variance_inflation_factor
#
#
# Concept:
#
#     low VIF
#         → predictor has relatively independent information
#
#     high VIF
#         → predictor overlaps strongly with other predictors
#
#
# Important:
#
# Multicollinearity primarily harms interpretation and coefficient
# stability. It does not automatically make predictions useless.
#
#
# ============================================================
# 27. INFLUENTIAL OBSERVATIONS
# ============================================================
#
# Some observations can strongly affect the fitted model.
#
# Important concepts:
#
#     leverage
#     influence
#     Cook's distance
#
#
# Tool:
#
#     influence = results.get_influence()
#
#
# Then:
#
#     influence.cooks_distance
#
#
# Concept:
#
#     normal observations
#
#     ● ● ● ● ● ● ●
#
#              ★
#              ← unusual/influential point
#
#
# An influential point can move the regression line significantly.
#
#
# ============================================================
# 28. OLS SUMMARY
# ============================================================
#
#     print(results.summary())
#
# This is one of the most useful commands in statsmodels.
#
# It brings together:
#
#     coefficients
#     standard errors
#     t-statistics
#     p-values
#     confidence intervals
#     R²
#     adjusted R²
#     F-statistic
#     AIC
#     BIC
#     sample size
#     diagnostic information
#
#
# For learning statistics, study the summary table rather than
# merely copying it into a report.
#
#
# ============================================================
# 29. LOGISTIC REGRESSION
# ============================================================
#
# When y is binary:
#
#     0 / 1
#
# ordinary linear regression is generally not the appropriate
# probabilistic model.
#
# Use:
#
#     sm.Logit
#
# or:
#
#     smf.logit
#
#
# Example:
#
#     model = smf.logit(
#         "failure ~ age + pressure + diameter",
#         data=df
#     )
#
#     results = model.fit()
#
#
# Concept:
#
#     features
#         ↓
#     linear combination
#         ↓
#     logistic function
#         ↓
#     probability
#         ↓
#     P(Y = 1)
#
#
# ============================================================
# 30. LOGISTIC FUNCTION
# ============================================================
#
#     p = 1 / (1 + exp(-z))
#
#
# Shape:
#
#     probability
#       1 ┤                 ─────
#         │             ───
#     0.5 ┤────────────●
#         │        ───
#         │     ───
#       0 ┤─────
#         └────────────────────→ z
#
#
# This guarantees:
#
#     0 < p < 1
#
#
# ============================================================
# 31. LOGIT MODEL COEFFICIENTS
# ============================================================
#
# Logistic regression is commonly interpreted through:
#
#     odds
#     log-odds
#     odds ratios
#
#
# Logistic model:
#
#     log(p / (1-p))
#         =
#     β₀ + β₁x₁ + ...
#
#
# Exponentiating coefficient:
#
#     exp(β)
#
# gives the odds ratio for a one-unit increase in the predictor,
# holding the other variables constant.
#
#
# In statsmodels:
#
#     np.exp(results.params)
#
#
# This is VERY useful in:
#
#     risk analysis
#     medical/statistical studies
#     customer churn
#     failure prediction
#     credit/risk modeling
#
#
# ============================================================
# 32. LOGISTIC PREDICTION
# ============================================================
#
#     results.predict(X)
#
# returns estimated probabilities.
#
#
# Example:
#
#     0.13
#     0.82
#     0.64
#     0.04
#
#
# Then a classification threshold can be applied:
#
#     probability >= 0.5 → class 1
#     probability < 0.5  → class 0
#
#
# But 0.5 is not a universal optimal threshold.
#
#
# ============================================================
# 33. GLM — GENERALIZED LINEAR MODELS
# ============================================================
#
# One of the most important statsmodels areas.
#
#     GLM
#     Generalized Linear Model
#
#
# It generalizes the basic linear model to different response
# distributions.
#
#
# Concept:
#
#     Ordinary linear model
#            ↓
#     Gaussian/normal response
#
#     GLM
#            ↓
#     choose appropriate distribution
#            ↓
#     choose link function
#            ↓
#     model relationship
#
#
# Important families include:
#
#     Gaussian
#     Binomial
#     Poisson
#     Gamma
#     Inverse Gaussian
#
#
# ============================================================
# 34. LINK FUNCTIONS
# ============================================================
#
# GLM separates:
#
#     linear predictor
#             ↓
#        link function
#             ↓
#     expected response
#
#
# Example logistic:
#
#     predictors
#         ↓
#     β₀ + β₁x
#         ↓
#     logistic / inverse-logit
#         ↓
#     probability
#
#
# This is a central statistical modeling concept.
#
#
# ============================================================
# 35. POISSON REGRESSION
# ============================================================
#
# Useful when the target represents counts:
#
#     number of failures
#     number of calls
#     number of accidents
#     number of events
#
#
# Example:
#
#     failures ~ pipe_age + pressure + diameter
#
#
# statsmodels:
#
#     smf.glm(
#         "failures ~ age + pressure",
#         data=df,
#         family=sm.families.Poisson()
#     ).fit()
#
#
# Very useful for:
#
#     reliability
#     event counts
#     demand
#     traffic
#     incident analysis
#
#
# ============================================================
# 36. BINOMIAL GLM
# ============================================================
#
# Useful for:
#
#     success/failure
#     yes/no
#     defective/non-defective
#     conversion/non-conversion
#
#
# Logistic regression is a special case of the binomial GLM.
#
#
# ============================================================
# 37. NEGATIVE BINOMIAL
# ============================================================
#
# Poisson assumes:
#
#     mean ≈ variance
#
# Real count data often have:
#
#     variance > mean
#
# This is called:
#
#     overdispersion
#
#
# Negative Binomial models can be useful when overdispersion
# is present.
#
#
# ============================================================
# 38. GAMMA MODELS
# ============================================================
#
# Useful for positive continuous variables such as:
#
#     waiting time
#     cost
#     insurance claim size
#     duration
#
#
# A Gamma GLM can be useful when the target is:
#
#     > 0
#
# and strongly right-skewed.
#
#
# ============================================================
# 39. GEE — GENERALIZED ESTIMATING EQUATIONS
# ============================================================
#
# GEE is important when observations are correlated.
#
# Example:
#
#     Patient 1 → measurement 1
#              → measurement 2
#              → measurement 3
#
#     Patient 2 → measurement 1
#              → measurement 2
#
#
# These observations are NOT independent.
#
#
# GEE models:
#
#     repeated/clustered observations
#
#
# Concept:
#
#     individual observations
#             ↓
#        grouped/clustered
#             ↓
#       correlation structure
#             ↓
#            GEE
#
#
# Useful in:
#
#     longitudinal data
#     repeated measurements
#     panel-like data
#     clustered observations
#
#
# ============================================================
# 40. MIXED EFFECTS MODELS
# ============================================================
#
# Mixed models combine:
#
#     fixed effects
#     +
#     random effects
#
#
# Example:
#
#     students nested in schools
#
#     School A
#       ├── Student
#       ├── Student
#       └── Student
#
#     School B
#       ├── Student
#       ├── Student
#       └── Student
#
#
# Students from the same school may be more similar.
#
# A mixed model can account for this structure.
#
#
# Important statsmodels class:
#
#     sm.MixedLM
#
#
# Useful for:
#
#     hierarchical data
#     repeated measurements
#     grouped observations
#     longitudinal studies
#
#
# ============================================================
# 41. TIME SERIES — ONE OF STATSMODELS' STRONGEST AREAS
# ============================================================
#
# statsmodels is especially important for classical time-series
# analysis.
#
#
# Time series:
#
#     t₁ → t₂ → t₃ → t₄ → t₅
#      │    │    │    │    │
#      y₁   y₂   y₃   y₄   y₅
#
#
# Unlike ordinary independent observations:
#
#     yₜ may depend on yₜ₋₁
#
#
# ============================================================
# 42. AR — AUTOREGRESSIVE MODEL
# ============================================================
#
# Basic idea:
#
#     yₜ = c + φ₁yₜ₋₁ + error
#
#
# Today's value depends on previous values.
#
#
# Visual:
#
#     yesterday ───────┐
#                     ↓
#     today ← model ← previous values
#
#
# Useful for:
#
#     forecasting
#     sensor data
#     economic data
#     demand
#     environmental data
#
#
# ============================================================
# 43. MA — MOVING AVERAGE MODEL
# ============================================================
#
# MA models use past errors:
#
#     yₜ = c + εₜ + θ₁εₜ₋₁ + ...
#
#
# Important:
#
# "moving average" here is NOT simply the same thing as:
#
#     pandas.Series.rolling().mean()
#
# It is a statistical time-series model based on error terms.
#
#
# ============================================================
# 44. ARIMA
# ============================================================
#
#     ARIMA
#     = AutoRegressive
#       Integrated
#       Moving Average
#
#
# Concept:
#
#     AR
#       ↓
#     previous values
#
#     I
#       ↓
#     differencing
#
#     MA
#       ↓
#     previous errors
#
#
# Important class:
#
#     from statsmodels.tsa.arima.model import ARIMA
#
#
# Example:
#
#     model = ARIMA(
#         series,
#         order=(1, 1, 1)
#     )
#
#     results = model.fit()
#
#
# order = (p, d, q)
#
#     p → AR order
#     d → differencing order
#     q → MA order
#
#
# ============================================================
# 45. DIFFERENCING
# ============================================================
#
# If:
#
#     y₁ = 100
#     y₂ = 105
#     y₃ = 112
#     y₄ = 120
#
# First difference:
#
#     105 - 100 = 5
#     112 - 105 = 7
#     120 - 112 = 8
#
#
# Visual:
#
#     original:
#
#     100 → 105 → 112 → 120
#
#     difference:
#
#       5 → 7 → 8
#
#
# Differencing can help remove trends and make a series more
# stationary.
#
#
# ============================================================
# 46. SARIMAX
# ============================================================
#
# SARIMAX is extremely useful for practical forecasting.
#
#     Seasonal
#     ARIMA
#     with
#     eXogenous variables
#
#
# It can model:
#
#     trend
#     seasonality
#     autoregression
#     moving-average structure
#     external predictors
#
#
# Example:
#
#     electricity_demand
#         ← temperature
#         ← previous demand
#         ← seasonality
#
#
# Important class:
#
#     from statsmodels.tsa.statespace.sarimax import SARIMAX
#
#
# ============================================================
# 47. ACF — AUTOCORRELATION FUNCTION
# ============================================================
#
# ACF asks:
#
#     "How correlated is the series with its own past?"
#
#
# Lag 1:
#
#     yₜ with yₜ₋₁
#
# Lag 2:
#
#     yₜ with yₜ₋₂
#
#
# Visual:
#
#     time:
#
#     ... yₜ₋₃  yₜ₋₂  yₜ₋₁  yₜ
#             ↑     ↑     ↑
#             └─────┴─────┘
#               correlations
#
#
# Useful:
#
#     from statsmodels.graphics.tsaplots import plot_acf
#
#
# ============================================================
# 48. PACF — PARTIAL AUTOCORRELATION
# ============================================================
#
# PACF asks:
#
#     "What is the relationship between yₜ and yₜ₋k
#      after accounting for shorter lags?"
#
#
# Useful for identifying autoregressive structure.
#
#
#     from statsmodels.graphics.tsaplots import plot_pacf
#
#
# ACF and PACF are particularly important when learning ARIMA.
#
#
# ============================================================
# 49. STATIONARITY
# ============================================================
#
# A simplified intuition:
#
# A stationary series has statistical behavior that is relatively
# stable over time.
#
#
# Non-stationary:
#
#     mean changes
#     variance changes
#     trend exists
#     seasonal structure may exist
#
#
# Stationary:
#
#     ──●──●─●──●──●──
#       similar statistical behavior
#
#
# Important test:
#
#     Augmented Dickey-Fuller
#     ADF
#
#
#     from statsmodels.tsa.stattools import adfuller
#
#     result = adfuller(series)
#
#
# Important output:
#
#     test statistic
#     p-value
#     critical values
#
#
# ============================================================
# 50. ADF TEST — IMPORTANT INTERPRETATION
# ============================================================
#
# ADF tests a unit-root-related null hypothesis.
#
# Roughly:
#
#     H₀ → unit root / non-stationary process
#
#     H₁ → stationary process
#
#
# Therefore:
#
#     small p-value
#         → evidence against H₀
#
# But do NOT reduce stationarity analysis to:
#
#     "p < 0.05 = everything is fine."
#
# Use:
#
#     plot
#     domain knowledge
#     ACF/PACF
#     transformations
#     tests
#
# together.
#
#
# ============================================================
# 51. GRANGER CAUSALITY
# ============================================================
#
# statsmodels includes:
#
#     grangercausalitytests
#
#
# IMPORTANT:
#
# Granger causality does NOT establish philosophical or
# experimental causality.
#
# It asks whether past values of X provide predictive information
# about Y beyond information contained in Y's own past.
#
#
# Concept:
#
#     X past ────────────────┐
#                            ↓
#                       predict Y?
#                            ↑
#     Y past ────────────────┘
#
#
# Useful in time-series analysis.
#
#
# ============================================================
# 52. ANOVA
# ============================================================
#
# ANOVA:
#
#     Analysis of Variance
#
# asks whether group means differ.
#
#
# Example:
#
#     Group A → 10, 12, 11
#     Group B → 20, 21, 19
#     Group C → 15, 14, 16
#
#
# Question:
#
#     Are these differences larger than we would expect
#     from within-group variation?
#
#
# Core concept:
#
#     TOTAL VARIATION
#            │
#       ┌────┴────┐
#       ↓         ↓
#   BETWEEN     WITHIN
#   GROUPS      GROUPS
#
#
# ============================================================
# 53. ANOVA WITH OLS
# ============================================================
#
# ANOVA and regression are closely related.
#
# Example:
#
#     model = smf.ols(
#         "score ~ C(group)",
#         data=df
#     ).fit()
#
#
# Then:
#
#     from statsmodels.stats.anova import anova_lm
#
#     table = anova_lm(model)
#
#
# This is a powerful conceptual connection:
#
#     ANOVA
#       ↕
#     regression with categorical predictors
#
#
# ============================================================
# 54. POST-HOC COMPARISONS
# ============================================================
#
# ANOVA may tell you:
#
#     "At least one group differs."
#
# It does NOT automatically tell you:
#
#     "A differs from B"
#     "B differs from C"
#
#
# Post-hoc comparisons investigate pairwise differences.
#
# statsmodels provides multiple-comparison tools under:
#
#     statsmodels.stats.multicomp
#
#
# ============================================================
# 55. T-TESTS
# ============================================================
#
# statsmodels has tools under:
#
#     statsmodels.stats.weightstats
#
# Useful classes/functions include:
#
#     ttest_ind
#     ttest_mean
#     ttest_rel
#
#
# Concept:
#
#     group A mean
#           │
#           │ difference
#           ↓
#     group B mean
#
# Question:
#
#     Is the observed difference sufficiently large relative
#     to its uncertainty?
#
#
# ============================================================
# 56. PROPORTION TESTS
# ============================================================
#
# Very useful in business/data analysis.
#
# Example:
#
#     conversion rate:
#
#     before → 8%
#     after  → 10%
#
#
# Question:
#
#     Is the difference statistically meaningful?
#
#
# statsmodels provides proportion tests such as:
#
#     proportions_ztest
#
#
# Useful for:
#
#     A/B testing
#     conversion rates
#     defect rates
#     success rates
#
#
# ============================================================
# 57. A/B TESTING
# ============================================================
#
# Example:
#
#     Website A
#         1000 visitors
#         80 conversions
#
#     Website B
#         1000 visitors
#         100 conversions
#
#
# Rates:
#
#     A = 8%
#     B = 10%
#
#
# statsmodels can help test whether the observed difference is
# compatible with random sampling variation.
#
#
# Concept:
#
#     observed difference
#            ↓
#     statistical test
#            ↓
#     uncertainty
#            ↓
#     inference
#
#
# VERY USEFUL FOR DA.
#
#
# ============================================================
# 58. CHI-SQUARE TESTS
# ============================================================
#
# Useful for categorical variables.
#
# Example:
#
#     subscription     renewed
#     -------------------------
#     Basic            yes/no
#     Premium          yes/no
#
#
# Question:
#
#     Are the categorical variables associated?
#
#
# statsmodels provides contingency-table functionality.
#
#
# Important area:
#
#     statsmodels.stats.contingency_tables
#
#
# ============================================================
# 59. CONTINGENCY TABLE
# ============================================================
#
# Concept:
#
#                    Renewed
#                  Yes      No
#
#     Basic        80       20
#     Premium      95        5
#
#
# A chi-square test evaluates whether the observed association
# is stronger than expected under independence.
#
#
# ============================================================
# 60. NONPARAMETRIC TESTS
# ============================================================
#
# statsmodels contains useful nonparametric statistical tools,
# although SciPy is often the first library to reach for for
# individual nonparametric tests.
#
#
# Examples of concepts:
#
#     rank tests
#     empirical distribution
#     kernel density
#
#
# Practical rule:
#
#     statsmodels → modeling/inference ecosystem
#     SciPy        → broad scientific/statistical algorithms
#
#
# ============================================================
# 61. SURVIVAL / DURATION ANALYSIS
# ============================================================
#
# statsmodels has duration-model functionality.
#
# This is useful when the target is:
#
#     "time until an event"
#
# Examples:
#
#     time until machine failure
#     time until customer churn
#     time until component failure
#
#
# Concept:
#
#     start
#       │
#       ├───────────────→ event
#       │
#       │
#       └───────────────→ censored observation
#
#
# Censoring is a central survival-analysis concept.
#
#
# For advanced survival analysis, specialized packages such as
# lifelines may often be more convenient.
#
#
# ============================================================
# 62. ROBUST LINEAR MODELS
# ============================================================
#
# Ordinary least squares can be strongly affected by outliers.
#
#
# Example:
#
#     ● ● ● ● ● ● ●
#                     ★
#
# The extreme point may pull the regression line.
#
#
# statsmodels provides:
#
#     RLM
#     Robust Linear Model
#
#
# Concept:
#
#     ordinary regression
#         → sensitive to unusual observations
#
#     robust regression
#         → reduces influence of extreme observations
#
#
# Useful when outliers are genuine observations rather than
# errors that should simply be deleted.
#
#
# ============================================================
# 63. QUANTILE REGRESSION
# ============================================================
#
# Ordinary regression models:
#
#     conditional mean
#
# Quantile regression can model:
#
#     median
#     25th percentile
#     75th percentile
#     etc.
#
#
# Important class:
#
#     smf.quantreg
#
#
# Example:
#
#     model = smf.quantreg(
#         "salary ~ experience",
#         data=df
#     )
#
#     results = model.fit(q=0.5)
#
#
# Here:
#
#     q=0.5
#
# represents the conditional median.
#
#
# Why useful?
#
# The effect of experience might differ for:
#
#     low-income workers
#     middle-income workers
#     high-income workers
#
#
# Ordinary regression can hide this.
#
#
# ============================================================
# 64. WEIGHTED LEAST SQUARES
# ============================================================
#
# OLS:
#
#     every observation has equal weight.
#
# WLS:
#
#     observations receive different weights.
#
#
# Concept:
#
#     observation A → weight 1
#     observation B → weight 2
#     observation C → weight 5
#
#
# Important class:
#
#     sm.WLS
#
#
# Useful when observations have different precision or sampling
# importance under an appropriate statistical design.
#
#
# ============================================================
# 65. GENERALIZED LEAST SQUARES
# ============================================================
#
# Important class:
#
#     sm.GLS
#
# OLS assumes a particular error structure.
#
# GLS allows more general covariance structures.
#
#
# Concept:
#
#     OLS:
#
#     errors
#       ● ● ● ●
#       independent / common variance assumptions
#
#
#     GLS:
#
#     errors
#       ●──●──●
#       │  │  │
#       correlated structure can be modeled
#
#
# This is more advanced.
#
#
# ============================================================
# 66. HETEROSKEDASTICITY-CONSISTENT COVARIANCE
# ============================================================
#
# Important options:
#
#     HC0
#     HC1
#     HC2
#     HC3
#
#
# Practical recommendation:
#
#     learn HC3 first.
#
# Why?
#
# It is commonly useful for regression inference when
# heteroskedasticity is a concern, particularly in finite samples.
#
#
# ============================================================
# 67. CLUSTER-ROBUST STANDARD ERRORS
# ============================================================
#
# Suppose:
#
#     employees belong to companies
#
# Observations within the same company may be correlated.
#
#
# Structure:
#
#     Company A
#       ├── employee
#       ├── employee
#       └── employee
#
#     Company B
#       ├── employee
#       ├── employee
#       └── employee
#
#
# Ordinary SEs may underestimate uncertainty if clustering is
# ignored.
#
# statsmodels supports covariance structures including clustering.
#
#
# ============================================================
# 68. STATISTICAL TESTING TOOLBOX
# ============================================================
#
# Important area:
#
#     statsmodels.stats
#
#
# Think of it as a toolbox:
#
#     statsmodels.stats
#           │
#           ├── hypothesis tests
#           ├── proportions
#           ├── power analysis
#           ├── contingency tables
#           ├── ANOVA
#           ├── multiple comparisons
#           ├── diagnostic tests
#           ├── covariance tools
#           └── meta-analysis
#
#
# ============================================================
# 69. MULTIPLE TESTING
# ============================================================
#
# Suppose you test:
#
#     1 hypothesis
#     2 hypotheses
#     10 hypotheses
#     100 hypotheses
#
#
# The more tests you perform, the more likely you are to obtain
# a small p-value by chance.
#
#
# Therefore:
#
#     multiple testing correction
#
# becomes important.
#
#
# Useful functions:
#
#     multipletests
#
#
# Common corrections include concepts such as:
#
#     Bonferroni
#     Holm
#     FDR / Benjamini-Hochberg
#
#
# VERY IMPORTANT FOR DATA SCIENCE.
#
#
# ============================================================
# 70. FALSE POSITIVE / FALSE DISCOVERY
# ============================================================
#
# Imagine testing 100 independent null hypotheses.
#
# Even if every null is true:
#
#     some p-values may be < 0.05
#
#
# This is why:
#
#     "p < 0.05"
#
# should not be blindly applied to hundreds of simultaneous tests.
#
#
# statsmodels helps control this problem.
#
#
# ============================================================
# 71. POWER ANALYSIS
# ============================================================
#
# Statistical power:
#
#     probability of detecting an effect
#     when that effect really exists.
#
#
# Four important quantities:
#
#     sample size
#     effect size
#     significance level
#     statistical power
#
#
# Concept:
#
#     sample size ↑
#          ↓
#     uncertainty ↓
#          ↓
#     power often ↑
#
#
# statsmodels provides power-analysis classes under:
#
#     statsmodels.stats.power
#
#
# Useful for:
#
#     experiment planning
#     A/B tests
#     research design
#     sample-size planning
#
#
# ============================================================
# 72. EFFECT SIZE VS P-VALUE
# ============================================================
#
# This distinction is CRITICAL.
#
#
# p-value:
#
#     "How strong is the evidence against H₀?"
#
#
# effect size:
#
#     "How large is the effect?"
#
#
# confidence interval:
#
#     "What range of effects is reasonably compatible with
#      the data/model assumptions?"
#
#
# Good statistical analysis:
#
#     effect size
#        +
#     confidence interval
#        +
#     p-value
#        +
#     practical/domain importance
#
#
# NOT:
#
#     p-value alone
#
#
# ============================================================
# 73. DESCRIPTIVE STATISTICS IN STATSMODELS
# ============================================================
#
# statsmodels is NOT the first choice for basic descriptive
# statistics.
#
# Usually:
#
#     pandas
#     statistics
#     NumPy
#
# are more natural for:
#
#     mean
#     median
#     standard deviation
#     quantiles
#     missing-value summaries
#
#
# statsmodels becomes more important when you move from:
#
#     "What does the data look like?"
#
# to:
#
#     "What statistical model explains the data?"
#
#
# ============================================================
# 74. STATSMODELS VS STATISTICS LIBRARY
# ============================================================
#
# Python `statistics`:
#
#     mean
#     median
#     mode
#     variance
#     standard deviation
#     quantiles
#     correlation
#     regression basics
#
#
# statsmodels:
#
#     regression models
#     coefficient inference
#     confidence intervals
#     hypothesis tests
#     ANOVA
#     GLM
#     logistic regression
#     robust regression
#     quantile regression
#     mixed models
#     GEE
#     time series
#     diagnostics
#     A/B testing
#     multiple testing
#     power analysis
#
#
# Mental progression:
#
#     statistics
#          ↓
#     describe data
#
#     statsmodels
#          ↓
#     model data
#          ↓
#     infer
#          ↓
#     diagnose
#
#
# ============================================================
# 75. STATSMODELS VS SCIPY
# ============================================================
#
# SciPy is broader as a scientific-computing library.
#
# SciPy:
#
#     distributions
#     optimization
#     numerical integration
#     signal processing
#     linear algebra
#     statistical tests
#     scientific algorithms
#
#
# statsmodels:
#
#     statistical models
#     regression
#     inference
#     econometrics
#     time series
#     diagnostics
#     statistical testing
#
#
# A useful mental model:
#
#     SciPy
#       ↓
#     "statistical/scientific tool"
#
#     statsmodels
#       ↓
#     "statistical model + inference framework"
#
#
# ============================================================
# 76. STATSMODELS VS SCIKIT-LEARN
# ============================================================
#
# This distinction is extremely important for your DS/ML path.
#
#
# SCIKIT-LEARN:
#
#     preprocessing
#     train/test split
#     cross-validation
#     pipelines
#     classification
#     regression
#     clustering
#     dimensionality reduction
#     hyperparameter tuning
#     predictive evaluation
#
#
# STATSMODELS:
#
#     coefficient interpretation
#     statistical inference
#     p-values
#     confidence intervals
#     hypothesis tests
#     model assumptions
#     diagnostics
#     classical regression
#     econometrics
#     time-series modeling
#
#
# Example:
#
#     Question:
#
#     "Can I predict whether a pipe will fail?"
#
#         → sklearn
#
#
#     Question:
#
#     "Which pipe characteristics are significantly associated
#      with failure, and how large are their effects?"
#
#         → statsmodels
#
#
# ============================================================
# 77. A VERY IMPORTANT ML + STATSMODELS PATTERN
# ============================================================
#
# You can use both.
#
#
#     DATA
#       │
#       ├──────────────→ statsmodels
#       │                    │
#       │                    ├── coefficients
#       │                    ├── p-values
#       │                    ├── CI
#       │                    └── diagnostics
#       │
#       └──────────────→ sklearn
#                            │
#                            ├── prediction
#                            ├── CV
#                            ├── metrics
#                            └── deployment-oriented model
#
#
# This is often much stronger than treating them as competitors.
#
#
# ============================================================
# 78. DATA ANALYSIS — WHAT YOU SHOULD ACTUALLY USE
# ============================================================
#
# For DA, prioritize:
#
#     1. OLS
#     2. formula API
#     3. coefficients
#     4. confidence intervals
#     5. p-values
#     6. R² / adjusted R²
#     7. residuals
#     8. robust standard errors
#     9. ANOVA
#    10. t-tests
#    11. proportion tests
#    12. contingency tables
#    13. multiple testing
#    14. power analysis
#    15. logistic regression
#    16. diagnostics
#
#
# These are HIGH VALUE.
#
#
# ============================================================
# 79. DATA SCIENCE — WHAT YOU SHOULD ACTUALLY USE
# ============================================================
#
# For DS, prioritize:
#
#     1. OLS
#     2. Logistic regression
#     3. GLM
#     4. confidence intervals
#     5. hypothesis testing
#     6. residual analysis
#     7. multicollinearity
#     8. heteroskedasticity
#     9. multiple testing
#    10. time series
#    11. ARIMA/SARIMAX
#    12. ACF/PACF
#    13. stationarity
#    14. robust regression
#    15. quantile regression
#    16. mixed models when data are hierarchical
#
#
# ============================================================
# 80. MACHINE LEARNING — WHAT YOU SHOULD ACTUALLY USE
# ============================================================
#
# For ML, statsmodels is NOT your primary ML library.
#
# Primary:
#
#     NumPy
#     pandas
#     scikit-learn
#     SciPy
#
#
# But statsmodels is highly valuable for:
#
#     statistical understanding
#     feature interpretation
#     baseline models
#     regression analysis
#     feature significance
#     confidence intervals
#     diagnostics
#     experiments
#     time-series forecasting
#
#
# Especially useful if your ML work needs to answer:
#
#     "WHY does this variable matter?"
#
# rather than only:
#
#     "Does the model predict well?"
#
#
# ============================================================
# 81. YOUR DA/DS/ML PRIORITY MAP
# ============================================================
#
#     ██████████████████████████████
#     ESSENTIAL
#
#     OLS
#     Logistic regression
#     Formula API
#     Results object
#     params
#     pvalues
#     bse
#     conf_int()
#     predict()
#     residuals
#     R²
#     adjusted R²
#     robust standard errors
#     diagnostics
#     ANOVA
#     t-tests
#     proportion tests
#     multiple testing
#
#
#     ████████████████████
#     VERY USEFUL
#
#     GLM
#     Poisson
#     Negative Binomial
#     Quantile Regression
#     Robust Regression
#     VIF
#     influence/Cook's distance
#     ACF
#     PACF
#     ADF
#     ARIMA
#     SARIMAX
#
#
#     ████████████
#     ADVANCED / SPECIALIZED
#
#     MixedLM
#     GEE
#     GLS
#     WLS
#     duration models
#     advanced covariance structures
#     state-space models
#     specialized econometric models
#
#
# ============================================================
# 82. THE MOST IMPORTANT STATSMODELS MODULES
# ============================================================
#
# Learn these names:
#
#
#     statsmodels.api
#         ↓
#     main modeling interface
#
#
#     statsmodels.formula.api
#         ↓
#     formula-based modeling
#
#
#     statsmodels.stats
#         ↓
#     statistical tests / inference
#
#
#     statsmodels.stats.diagnostic
#         ↓
#     model diagnostics
#
#
#     statsmodels.stats.outliers_influence
#         ↓
#     influence + VIF
#
#
#     statsmodels.stats.contingency_tables
#         ↓
#     categorical/contingency analysis
#
#
#     statsmodels.stats.multitest
#         ↓
#     multiple-testing correction
#
#
#     statsmodels.stats.power
#         ↓
#     statistical power
#
#
#     statsmodels.tsa
#         ↓
#     time series
#
#
#     statsmodels.graphics.tsaplots
#         ↓
#     ACF/PACF plots
#
#
#     statsmodels.robust
#         ↓
#     robust statistical modeling
#
#
# ============================================================
# 83. A COMPLETE PRACTICAL OLS WORKFLOW
# ============================================================
#
#     import statsmodels.formula.api as smf
#
#     model = smf.ols(
#         "sales ~ advertising + price + C(region)",
#         data=df
#     )
#
#     results = model.fit(cov_type="HC3")
#
#     print(results.summary())
#
#     print(results.params)
#     print(results.bse)
#     print(results.pvalues)
#     print(results.conf_int())
#
#
# Then investigate:
#
#     residuals
#     fitted values
#     influence
#     VIF
#     heteroskedasticity
#
#
# Workflow:
#
#     DATA
#       ↓
#     exploratory analysis
#       ↓
#     choose model
#       ↓
#     fit
#       ↓
#     coefficients
#       ↓
#     uncertainty
#       ↓
#     diagnostics
#       ↓
#     interpretation
#       ↓
#     business/scientific conclusion
#
#
# ============================================================
# 84. A COMPLETE LOGISTIC WORKFLOW
# ============================================================
#
#     import statsmodels.formula.api as smf
#     import numpy as np
#
#     model = smf.logit(
#         "failure ~ age + pressure + diameter",
#         data=df
#     )
#
#     results = model.fit()
#
#     print(results.summary())
#
#     probabilities = results.predict(df)
#
#     odds_ratios = np.exp(results.params)
#
#
# Interpretation:
#
#     coefficient
#         ↓
#     log-odds effect
#         ↓
#     exp(coefficient)
#         ↓
#     odds ratio
#
#
# ============================================================
# 85. A COMPLETE TIME-SERIES WORKFLOW
# ============================================================
#
#     inspect series
#          ↓
#     plot series
#          ↓
#     check trend/seasonality
#          ↓
#     ACF/PACF
#          ↓
#     stationarity analysis
#          ↓
#     transformation/differencing
#          ↓
#     ARIMA/SARIMAX
#          ↓
#     diagnostics
#          ↓
#     forecast
#
#
# Important tools:
#
#     adfuller()
#     plot_acf()
#     plot_pacf()
#     ARIMA()
#     SARIMAX()
#
#
# ============================================================
# 86. MODEL SELECTION — AIC
# ============================================================
#
#     results.aic
#
# AIC:
#
#     Akaike Information Criterion
#
#
# It balances:
#
#     model fit
#     +
#     model complexity
#
#
# Concept:
#
#     fit improves
#         ↓
#     AIC may decrease
#
#     complexity increases
#         ↓
#     AIC penalizes it
#
#
# Generally:
#
#     LOWER AIC → preferred among comparable models
#
#
# Do NOT compare AIC values from completely unrelated datasets
# or incompatible likelihood settings.
#
#
# ============================================================
# 87. BIC
# ============================================================
#
#     results.bic
#
# BIC:
#
#     Bayesian Information Criterion
#
# Similar conceptual purpose:
#
#     fit
#     +
#     complexity penalty
#
# BIC generally imposes a stronger complexity penalty than AIC
# as sample size grows.
#
#
# Lower is generally preferred among comparable models.
#
#
# ============================================================
# 88. AIC/BIC VISUAL
# ============================================================
#
#     Model 1
#       fit:       ███████████████
#       complexity: ███
#
#     Model 2
#       fit:       █████████████████
#       complexity: █████████
#
#
# Model 2 fits better, but is more complicated.
#
# AIC/BIC ask:
#
#     "Is the improvement in fit worth the additional complexity?"
#
#
# ============================================================
# 89. CONFIDENCE INTERVAL VS PREDICTION INTERVAL
# ============================================================
#
# These are NOT the same.
#
#
# Confidence interval:
#
#     uncertainty about a population/model parameter or
#     mean response, depending on the context.
#
#
# Prediction interval:
#
#     uncertainty around a future individual observation.
#
#
# Generally:
#
#     prediction interval
#         wider
#             than
#     confidence interval for mean response
#
#
# Visual:
#
#     narrow:   ───────── mean estimate
#
#     wider:  ═════════════════ individual prediction range
#
#
# ============================================================
# 90. CORRELATION IS NOT CAUSATION
# ============================================================
#
# statsmodels can estimate associations very precisely.
#
# That does NOT automatically establish causality.
#
#
# Example:
#
#     ice cream sales ↑
#     drowning incidents ↑
#
# A third variable:
#
#     temperature
#
# may affect both.
#
#
# Concept:
#
#     temperature
#        ↙     ↘
#     ice cream   drowning
#
#
# Regression can control for measured variables, but causal
# interpretation requires appropriate design/assumptions.
#
#
# ============================================================
# 91. COMMON STATSMODELS MISTAKE:
#     "LOW P-VALUE = IMPORTANT"
# ============================================================
#
# Wrong:
#
#     p < 0.05
#         ↓
#     important variable
#
#
# Better:
#
#     p-value
#       +
#     effect size
#       +
#     confidence interval
#       +
#     sample size
#       +
#     domain significance
#       +
#     model assumptions
#
#
# ============================================================
# 92. COMMON MISTAKE:
#     "HIGH R² = GOOD MODEL"
# ============================================================
#
# Not necessarily.
#
# A high R² can coexist with:
#
#     overfitting
#     nonlinearity
#     heteroskedasticity
#     autocorrelation
#     leakage
#     omitted variables
#     inappropriate functional form
#     influential observations
#
#
# Therefore:
#
#     R² is one diagnostic,
#     not the whole model.
#
#
# ============================================================
# 93. COMMON MISTAKE:
#     "REMOVE ALL OUTLIERS"
# ============================================================
#
# Outlier ≠ error.
#
#
# An unusual observation can be:
#
#     data-entry error
#     measurement error
#     legitimate rare event
#     important population subgroup
#
#
# First ask:
#
#     WHY is it unusual?
#
# Then decide what to do.
#
#
# ============================================================
# 94. COMMON MISTAKE:
#     "STATSmodels IS JUST SKLEARN WITH P-VALUES"
# ============================================================
#
# No.
#
# statsmodels has a different philosophy.
#
#
# sklearn:
#
#     prediction-oriented ML framework
#
# statsmodels:
#
#     statistical modeling + inference framework
#
#
# Their workflows overlap, but their goals differ.
#
#
# ============================================================
# 95. COMMON MISTAKE:
#     CONFUSING MODEL COEFFICIENT WITH CAUSAL EFFECT
# ============================================================
#
# If:
#
#     β₁ = 2
#
# this does not automatically mean:
#
#     "Increasing X causes Y to increase by 2."
#
#
# It means the fitted model estimates an association/effect under
# the model's assumptions and design.
#
#
# Causal interpretation requires stronger assumptions.
#
#
# ============================================================
# 96. THE BIG STATISTICAL PICTURE
# ============================================================
#
# You should mentally organize statistics like this:
#
#
#                    DATA
#                      │
#          ┌───────────┴───────────┐
#          ↓                       ↓
#     DESCRIPTIVE              INFERENTIAL
#          │                       │
#     mean/median/etc.             │
#                                  ↓
#                              MODELING
#                                  │
#                ┌─────────────────┼────────────────┐
#                ↓                 ↓                ↓
#               OLS             LOGIT              GLM
#                │                 │                │
#                └─────────────────┼────────────────┘
#                                  ↓
#                             ESTIMATION
#                                  │
#                ┌─────────────────┼─────────────────┐
#                ↓                 ↓                 ↓
#             effect           uncertainty        prediction
#                │                 │                 │
#                ↓                 ↓                 ↓
#            coefficient          SE             fitted value
#                              CI / p-value
#                                  │
#                                  ↓
#                             DIAGNOSTICS
#                                  │
#             ┌────────────────────┼────────────────────┐
#             ↓                    ↓                    ↓
#        residuals            assumptions           influence
#             │                    │                    │
#             ↓                    ↓                    ↓
#        model quality        validity             robustness
#
#
# ============================================================
# 97. WHERE STATSMODELS FITS IN YOUR PYTHON STACK
# ============================================================
#
# A strong DA/DS/ML stack is:
#
#
#     Python
#       │
#       ├── NumPy
#       │      ↓
#       │   numerical arrays
#       │
#       ├── pandas
#       │      ↓
#       │   tabular data
#       │
#       ├── matplotlib / seaborn
#       │      ↓
#       │   visualization
#       │
#       ├── statistics
#       │      ↓
#       │   basic descriptive statistics
#       │
#       ├── SciPy
#       │      ↓
#       │   scientific/statistical algorithms
#       │
#       ├── statsmodels
#       │      ↓
#       │   statistical modeling + inference
#       │
#       └── scikit-learn
#              ↓
#           predictive ML
#
#
# ============================================================
# 98. WHAT YOU SHOULD MEMORIZE
# ============================================================
#
# DO NOT try to memorize the entire statsmodels API.
#
# Memorize the MODELING PATTERNS.
#
#
# Pattern 1:
#
#     model = smf.ols("y ~ x1 + x2", data=df)
#     results = model.fit()
#
#
# Pattern 2:
#
#     results.params
#     results.bse
#     results.pvalues
#     results.conf_int()
#     results.predict(...)
#
#
# Pattern 3:
#
#     results.resid
#     results.fittedvalues
#
#
# Pattern 4:
#
#     results.summary()
#
#
# Pattern 5:
#
#     model.fit(cov_type="HC3")
#
#
# Pattern 6:
#
#     smf.logit(...)
#
#
# Pattern 7:
#
#     smf.glm(...)
#
#
# Pattern 8:
#
#     ARIMA(...)
#     SARIMAX(...)
#
#
# Pattern 9:
#
#     adfuller(...)
#     plot_acf(...)
#     plot_pacf(...)
#
#
# Pattern 10:
#
#     multipletests(...)
#
#
# ============================================================
# 99. HIGH-VALUE FUNCTIONS / CLASSES TO LEARN FIRST
# ============================================================
#
#                    PRIORITY
#
#     sm.OLS                    ████████████████████
#     smf.ols                   ████████████████████
#     smf.logit                 ████████████████████
#     smf.glm                   ███████████████████
#     results.summary()         ███████████████████
#     results.params           ███████████████████
#     results.pvalues          ███████████████████
#     results.conf_int()       ███████████████████
#     results.predict()        ██████████████████
#     results.resid            ██████████████████
#     results.fittedvalues     █████████████████
#     results.rsquared         ████████████████
#     results.aic              ███████████████
#     results.bic              ███████████████
#
#     anova_lm                 ██████████████
#     proportions_ztest        ██████████████
#     multipletests            ██████████████
#     variance_inflation_factor██████████████
#     het_breuschpagan         █████████████
#     durbin_watson             █████████████
#     get_influence()           █████████████
#
#     ARIMA                    █████████████
#     SARIMAX                  █████████████
#     adfuller                 ████████████
#     plot_acf                 ███████████
#     plot_pacf                ███████████
#
#     MixedLM                  ███████
#     GEE                      ███████
#     GLS                      ██████
#     WLS                      ██████
#     RLM                      ██████
#     QuantReg                 ██████
#
#
# ============================================================
# 100. WHAT IS NOT A PRIORITY FOR YOU
# ============================================================
#
# Given a practical DA → DS → ML learning path, do NOT start by
# trying to master:
#
#     every econometric model
#     every state-space model
#     every covariance estimator
#     every internal class
#     sandbox modules
#     obscure helper functions
#
#
# First master:
#
#     OLS
#     Logistic
#     GLM
#     inference
#     diagnostics
#     hypothesis testing
#     A/B testing
#     time-series fundamentals
#
#
# Then specialize.
#
#
# ============================================================
# 101. YOUR PRACTICAL LEARNING ORDER
# ============================================================
#
# I recommend this order:
#
#
# PHASE 1 — REGRESSION
#
#     OLS
#       ↓
#     simple regression
#       ↓
#     multiple regression
#       ↓
#     formula API
#       ↓
#     categorical variables
#       ↓
#     interactions
#
#
# PHASE 2 — INFERENCE
#
#     coefficients
#       ↓
#     standard errors
#       ↓
#     t-statistics
#       ↓
#     p-values
#       ↓
#     confidence intervals
#       ↓
#     F-tests
#
#
# PHASE 3 — DIAGNOSTICS
#
#     residuals
#       ↓
#     heteroskedasticity
#       ↓
#     robust SE
#       ↓
#     multicollinearity
#       ↓
#     influence
#       ↓
#     autocorrelation
#
#
# PHASE 4 — CATEGORICAL / EXPERIMENTAL
#
#     t-tests
#     ANOVA
#     proportions
#     chi-square
#     A/B testing
#     multiple testing
#     power analysis
#
#
# PHASE 5 — CLASSIFICATION
#
#     Logistic regression
#       ↓
#     probabilities
#       ↓
#     odds
#       ↓
#     odds ratios
#       ↓
#     inference
#
#
# PHASE 6 — GLM
#
#     Gaussian
#     Binomial
#     Poisson
#     Negative Binomial
#     Gamma
#
#
# PHASE 7 — TIME SERIES
#
#     autocorrelation
#     stationarity
#     ACF
#     PACF
#     AR
#     MA
#     ARIMA
#     SARIMAX
#
#
# PHASE 8 — ADVANCED
#
#     robust regression
#     quantile regression
#     mixed effects
#     GEE
#     GLS/WLS
#     survival/duration
#
#
# ============================================================
# 102. THE SINGLE MOST IMPORTANT MENTAL MODEL
# ============================================================
#
# When you use statsmodels, do NOT think:
#
#     "Which function gives me the answer?"
#
#
# Think:
#
#     1. What is my outcome?
#             ↓
#     2. What type of variable is it?
#             ↓
#     3. What is my research/business question?
#             ↓
#     4. What statistical model represents that question?
#             ↓
#     5. What assumptions does that model make?
#             ↓
#     6. Fit the model.
#             ↓
#     7. Examine coefficients.
#             ↓
#     8. Examine uncertainty.
#             ↓
#     9. Test hypotheses where appropriate.
#            ↓
#    10. Diagnose the model.
#            ↓
#    11. Interpret in domain context.
#
#
# ============================================================
# 103. FINAL DA / DS / ML CHEAT SHEET
# ============================================================
#
#     QUESTION                         TOOL
#     -------------------------------------------------------
#     Describe data                   pandas / statistics
#
#     Correlation                     pandas / NumPy / SciPy
#
#     Linear relationship             statsmodels OLS
#
#     Multiple predictors             statsmodels OLS
#
#     Effect significance             statsmodels
#
#     Confidence intervals            statsmodels
#
#     Binary outcome                  statsmodels Logit
#
#     Count outcome                   statsmodels Poisson/GLM
#
#     Overdispersed counts            Negative Binomial
#
#     Group mean comparison           t-test / ANOVA
#
#     Categorical association         contingency tables
#
#     A/B conversion comparison       proportion tests
#
#     Many simultaneous tests         multipletests
#
#     Experiment sample size          power analysis
#
#     Heteroskedasticity              diagnostic tests
#
#     Robust inference                HC3 / robust covariance
#
#     Multicollinearity               VIF
#
#     Influential observations        influence / Cook's distance
#
#     Time dependence                 ACF / PACF
#
#     Stationarity                    ADF
#
#     Classical forecasting           ARIMA / SARIMAX
#
#     Hierarchical data               MixedLM
#
#     Repeated correlated data        GEE
#
#     Outlier-resistant regression    RLM
#
#     Different conditional quantiles QuantReg
#
#     Predictive ML                   scikit-learn
#
#
# ============================================================
# 104. CORE IDEA
# ============================================================
#
# statsmodels is the library you reach for when your question
# changes from:
#
#     "What can I predict?"
#
# to:
#
#     "What does the model tell me about the data?"
#
#
# The central chain is:
#
#
#        DATA
#          ↓
#       MODEL
#          ↓
#      ESTIMATION
#          ↓
#    ┌─────┼─────────────┐
#    ↓     ↓             ↓
# EFFECT  UNCERTAINTY   PREDICTION
#    │     │             │
#    ↓     ↓             ↓
# PARAM   SE/CI         y-hat
#    │     │
#    ↓     ↓
#    └── INFERENCE ─────┘
#             ↓
#        DIAGNOSTICS
#             ↓
#       INTERPRETATION
#
#
# For your DA → DS → ML path, the highest-value part of
# statsmodels is therefore NOT memorizing hundreds of APIs.
#
# It is becoming fluent in:
#
#     OLS
#     Logistic Regression
#     GLM
#     statistical inference
#     confidence intervals
#     hypothesis testing
#     residual diagnostics
#     robust standard errors
#     ANOVA
#     A/B testing
#     multiple testing
#     time-series fundamentals
#
#
# Once these concepts are intuitive, statsmodels becomes much
# easier because the API starts looking like a direct translation
# of statistical ideas into Python.
#
# ============================================================
```
