# MATH LIBRARY 

```python
# ============================================================
# MATH LIBRARY — STATISTICS, NUMERICAL MATH, AND DA/DS/ML
# ============================================================
#
# Python's built-in `math` module is NOT a statistics library.
#
# Its main purpose is:
#
#     numerical mathematics
#          +
#     mathematical functions
#          +
#     floating-point operations
#          +
#     number-theoretic operations
#
# However, many functions are extremely useful for:
#
#     DA  → Data Analysis
#     DS  → Data Science
#     ML  → Machine Learning
#
# especially when you need to understand the mathematical machinery
# underneath statistical and ML algorithms.
#
# ============================================================
# 1. THE BIG PICTURE
# ============================================================
#
# Think of Python's mathematical ecosystem like this:
#
#                         PYTHON MATH
#                              |
#             +----------------+----------------+
#             |                |                |
#             ↓                ↓                ↓
#          statistics        math             NumPy
#             |                |                |
#             ↓                ↓                ↓
#       statistical       mathematical       numerical
#       summaries         functions         arrays
#
#
# `math` is the mathematical toolbox.
#
# It gives you functions for:
#
#     powers
#     roots
#     logarithms
#     exponentials
#     trigonometry
#     rounding
#     floating-point analysis
#     combinatorics
#     number theory
#     special functions
#
# For DA/DS/ML, the most important conceptual areas are:
#
#     arithmetic
#         ↓
#     powers / roots
#         ↓
#     logarithms / exponentials
#         ↓
#     numerical precision
#         ↓
#     distances / geometry
#         ↓
#     probability-related mathematics
#         ↓
#     ML mathematical foundations
#
# ============================================================
# 2. IMPORTING MATH
# ============================================================
#
# The standard approach:
#
# import math
#
# Then:
#
# math.sqrt(25)
# math.log(100)
# math.exp(2)
#
# This makes the origin of the function explicit:
#
# math.sqrt
#    ↑
#    |
#    module
#
# sqrt
#    ↑
#    |
#    function
#
# ============================================================
# 3. WHAT `math` IS REALLY FOR
# ============================================================
#
# A useful mental model:
#
#                  math
#                   |
#       +-----------+-----------+
#       |           |           |
#       ↓           ↓           ↓
#    NUMBERS     FUNCTIONS    SPECIAL
#                 ON          FUNCTIONS
#              NUMBERS
#
# NUMBERS:
#
#     factorial
#     gcd
#     lcm
#     comb
#     perm
#
# FUNCTIONS:
#
#     sqrt
#     log
#     exp
#     sin
#     floor
#     ceil
#
# SPECIAL FUNCTIONS:
#
#     erf
#     gamma
#     lgamma
#
# The module is therefore much broader than "statistics".
#
# ============================================================
# 4. FIRST FOUNDATION — ABSOLUTE VALUE
# ============================================================
#
# math.fabs(x)
#
# Absolute value means:
#
#     distance from zero
#
# Number line:
#
#       -5        0        +5
#        |--------|---------|
#        <-------> <------->
#           5          5
#
# Therefore:
#
# math.fabs(-5)
#     ↓
# 5.0
#
# `fabs()` returns a floating-point result.
#
# Compare:
#
# abs(-5)
#     ↓
# 5
#
# `abs()` is a built-in Python function and is usually more convenient
# for ordinary absolute values.
#
# `math.fabs()` specifically performs floating-point absolute value.
#
# ============================================================
# DA / DS / ML USE
# ============================================================
#
# Absolute differences appear everywhere:
#
#     error = |actual - predicted|
#
#     distance = |x1 - x2|
#
#     deviation = |x - center|
#
# Example:
#
# actual = 100
# predicted = 92
#
# error = math.fabs(actual - predicted)
#
#             ↓
#
#       |100 - 92|
#
#             ↓
#
#            8
#
# This idea leads directly toward:
#
#     MAE
#     Manhattan distance
#     absolute deviation
#     robust statistics
#
# ============================================================
# 5. SQUARE ROOT
# ============================================================
#
# math.sqrt(x)
#
# sqrt asks:
#
#     "What number multiplied by itself gives x?"
#
# Example:
#
# math.sqrt(25)
#
#     ↓
#
# 5.0
#
# because:
#
#     5 × 5 = 25
#
# Visual:
#
#             25
#              ↓
#           sqrt()
#              ↓
#              5
#
# ============================================================
# WHY SQRT IS IMPORTANT
# ============================================================
#
# Square roots appear throughout statistics and ML.
#
# VARIANCE → STANDARD DEVIATION
#
#     variance
#         ↓
#       sqrt()
#         ↓
#     standard deviation
#
# EUCLIDEAN DISTANCE:
#
#     distance = √((x2-x1)² + (y2-y1)²)
#
# NORMAL DISTRIBUTION:
#
#     mathematical formulas contain √(2π)
#
# RMSE:
#
#     mean squared error
#           ↓
#         sqrt()
#           ↓
#         RMSE
#
# ============================================================
# 6. SQUARE ROOT VS ISQRT
# ============================================================
#
# math.isqrt(n)
#
# means:
#
#     integer square root
#
# Example:
#
# math.isqrt(25)
#     → 5
#
# math.isqrt(26)
#     → 5
#
# Why?
#
# sqrt(26) ≈ 5.099
#
# But the integer square root is:
#
#     floor(sqrt(26))
#          ↓
#          5
#
# Therefore:
#
# sqrt()
#     → floating-point square root
#
# isqrt()
#     → integer square root
#
# `isqrt()` is mainly useful in number theory and algorithms rather
# than ordinary DA/DS/ML.
#
# ============================================================
# 7. POW — EXPONENTIATION
# ============================================================
#
# math.pow(x, y)
#
# calculates:
#
#     xʸ
#
# Example:
#
# math.pow(2, 3)
#
#     ↓
#
# 2³
#     ↓
# 8.0
#
# Python also supports:
#
# 2 ** 3
#
# which is usually the clearer choice for ordinary Python code.
#
# ============================================================
# ML CONNECTION
# ============================================================
#
# Powers appear everywhere:
#
#     x²
#     x³
#     distance²
#     variance
#     polynomial features
#     squared error
#
# Example:
#
# error = actual - predicted
#
# squared_error = error ** 2
#
# The mathematical operation is:
#
#     error
#       ↓
#      square
#       ↓
#     error²
#
# ============================================================
# 8. CBRT — CUBE ROOT
# ============================================================
#
# math.cbrt(x)
#
# calculates:
#
#     ∛x
#
# Example:
#
# math.cbrt(27)
#
#     ↓
#
# 3.0
#
# because:
#
#     3 × 3 × 3 = 27
#
# Mathematical relationship:
#
#     ∛x = x^(1/3)
#
# Cube roots are less common than square roots in everyday DA/ML,
# but they are useful when working with transformations and
# mathematical formulas.
#
# ============================================================
# 9. FLOOR
# ============================================================
#
# math.floor(x)
#
# returns the greatest integer less than or equal to x.
#
# Example:
#
# math.floor(4.8)
#     → 4
#
# math.floor(-4.8)
#     → -5
#
# IMPORTANT:
#
# floor does NOT simply mean "remove the decimal part".
#
# Number line:
#
#     -6    -5    -4
#      |-----|-----|
#            ↑
#          -4.8
#
# The greatest integer ≤ -4.8 is -5.
#
# ============================================================
# 10. CEIL
# ============================================================
#
# math.ceil(x)
#
# returns the smallest integer greater than or equal to x.
#
# Example:
#
# math.ceil(4.2)
#     → 5
#
# math.ceil(-4.2)
#     → -4
#
# Visual:
#
#     4.2
#      ↓
#   ceil()
#      ↓
#      5
#
# ============================================================
# DA USE
# ============================================================
#
# floor() and ceil() are useful for:
#
#     → binning
#     → bucketing
#     → pagination
#     → capacity calculations
#     → discrete categories
#     → threshold calculations
#
# Example:
#
# age = 27
# decade = math.floor(age / 10)
#
#     27 / 10
#        ↓
#       2.7
#        ↓
#      floor
#        ↓
#        2
#
# This can help create decade groups such as:
#
#     20–29
#
# ============================================================
# 11. TRUNC
# ============================================================
#
# math.trunc(x)
#
# removes the fractional part.
#
# Example:
#
# math.trunc(4.9)
#     → 4
#
# math.trunc(-4.9)
#     → -4
#
# Compare:
#
#             4.9    -4.9
#
# floor       4       -5
#
# trunc       4       -4
#
# This distinction is important.
#
# ============================================================
# 12. MODULO AND FMOD
# ============================================================
#
# math.fmod(x, y)
#
# calculates the floating-point remainder.
#
# Example:
#
# math.fmod(7, 3)
#     → 1.0
#
# For ordinary Python integer modulo:
#
# 7 % 3
#     → 1
#
# For ordinary DA/DS code, `%` is usually enough.
#
# `fmod()` becomes useful when you specifically need floating-point
# remainder semantics.
#
# ============================================================
# 13. REMAINDER
# ============================================================
#
# math.remainder(x, y)
#
# calculates the IEEE-style floating-point remainder.
#
# This is different from:
#
#     x % y
#
# and can produce negative results depending on the values.
#
# This is primarily a numerical-computing tool.
#
# For beginner DA/DS:
#
#     know what it does
#     ↓
#     do not prioritize memorization
#
# ============================================================
# 14. MODF
# ============================================================
#
# math.modf(x)
#
# separates a number into:
#
#     fractional part
#     integer part
#
# Example:
#
# math.modf(12.75)
#
#     ↓
#
# (0.75, 12.0)
#
# VALUE FLOW:
#
#     12.75
#       |
#       +---- integer part → 12.0
#       |
#       +---- fractional part → 0.75
#
# This is a specialized numerical operation.
#
# ============================================================
# 15. COPYSIGN
# ============================================================
#
# math.copysign(x, y)
#
# takes the magnitude of x and the sign of y.
#
# Example:
#
# math.copysign(5, -1)
#     → -5.0
#
# Conceptually:
#
#     magnitude from x
#            +
#     sign from y
#            ↓
#       new number
#
# This is useful in numerical algorithms but is not a major
# DA/DS/ML function.
#
# ============================================================
# 16. EXPONENTIAL FUNCTION — EXP
# ============================================================
#
# math.exp(x)
#
# calculates:
#
#     eˣ
#
# where:
#
#     e ≈ 2.718281828...
#
# Example:
#
# math.exp(0)
#     → 1.0
#
# math.exp(1)
#     → e
#
# math.exp(2)
#     → e²
#
# ============================================================
# WHY EXP IS EXTREMELY IMPORTANT IN ML
# ============================================================
#
# Exponentials appear in:
#
#     probability
#     normal distributions
#     logistic functions
#     softmax
#     exponential growth
#     likelihood functions
#     optimization
#
# LOGISTIC FUNCTION:
#
#     1
#    ─────────
#    1 + e⁻ˣ
#
# The `e⁻ˣ` part comes directly from exponential mathematics.
#
# SOFTMAX:
#
#            eˣⁱ
#     ─────────────────
#       Σ eˣʲ
#
# Again:
#
#     exp()
#        ↓
#     probability transformation
#
# ============================================================
# 17. EXPM1
# ============================================================
#
# math.expm1(x)
#
# calculates:
#
#     eˣ - 1
#
# It is specifically designed to maintain numerical accuracy when
# x is very close to zero.
#
# Why not simply:
#
# math.exp(x) - 1
#
# Because floating-point subtraction can lose precision when two
# nearly equal numbers are subtracted.
#
# Concept:
#
#     large/close floating-point values
#               ↓
#        subtraction
#               ↓
#       possible precision loss
#
# `expm1()` is designed to reduce this problem.
#
# ============================================================
# 18. LOGARITHM
# ============================================================
#
# math.log(x)
#
# logarithm asks:
#
#     "What exponent produces x?"
#
# If:
#
#     e² = 7.389...
#
# then:
#
#     log(7.389...) ≈ 2
#
# Concept:
#
#     eˣ
#      ↕
#     log(x)
#
# They are inverse operations.
#
# ============================================================
# 19. LOG WITH A BASE
# ============================================================
#
# math.log(x, base)
#
# calculates:
#
#     log_base(x)
#
# Example:
#
# math.log(8, 2)
#     → 3
#
# because:
#
#     2³ = 8
#
# ============================================================
# 20. LOG2
# ============================================================
#
# math.log2(x)
#
# calculates:
#
#     log₂(x)
#
# Example:
#
# math.log2(8)
#     → 3.0
#
# because:
#
#     2³ = 8
#
# ============================================================
# WHY LOG2 MATTERS IN DATA / ML
# ============================================================
#
# Logarithms are fundamental in:
#
#     information theory
#     entropy
#     decision trees
#     information gain
#     binary systems
#     complexity analysis
#
# INFORMATION THEORY:
#
#     information ≈ -log₂(p)
#
# If an event is unlikely:
#
#     p ↓
#
# then:
#
#     -log₂(p) ↑
#
# So rare events carry more information.
#
# ============================================================
# 21. LOG10
# ============================================================
#
# math.log10(x)
#
# calculates:
#
#     log₁₀(x)
#
# Example:
#
# math.log10(1000)
#     → 3.0
#
# because:
#
#     10³ = 1000
#
# Useful in:
#
#     → scientific scales
#     → orders of magnitude
#     → decibel-like transformations
#     → data transformations
#
# ============================================================
# 22. LOG1P
# ============================================================
#
# math.log1p(x)
#
# calculates:
#
#     log(1 + x)
#
# accurately when x is very close to zero.
#
# Example:
#
# math.log1p(0)
#     → 0.0
#
# Like expm1(), this is mainly about NUMERICAL STABILITY.
#
# ============================================================
# 23. LOGARITHMIC TRANSFORMATION IN DATA ANALYSIS
# ============================================================
#
# Suppose data contains:
#
#     [10, 100, 1000, 10000]
#
# The raw scale grows rapidly.
#
# Applying log10:
#
#     10      → 1
#     100     → 2
#     1000    → 3
#     10000   → 4
#
# The transformation changes:
#
#     multiplicative differences
#              ↓
#     additive differences
#
# This is why logarithmic transformations are useful for highly
# skewed data.
#
# IMPORTANT:
#
# In real DS work, you will often use:
#
#     numpy.log()
#     numpy.log1p()
#
# because they operate efficiently on arrays.
#
# ============================================================
# 24. DISTANCE
# ============================================================
#
# math.dist(p, q)
#
# calculates Euclidean distance between two points.
#
# Example:
#
# p = [0, 0]
# q = [3, 4]
#
# math.dist(p, q)
#
#     ↓
#
# √((3-0)² + (4-0)²)
#
#     ↓
#
# √(9 + 16)
#
#     ↓
#
# √25
#
#     ↓
#
# 5.0
#
# ============================================================
# VISUAL
# ============================================================
#
#             q(3,4)
#               *
#              /|
#             / |
#          5 /  | 4
#           /   |
#          /    |
#         *-----*
#       p(0,0)   (3,0)
#             3
#
# This is the Pythagorean theorem.
#
# ============================================================
# WHY DISTANCE MATTERS IN ML
# ============================================================
#
# Distance is fundamental to:
#
#     k-NN
#     k-means
#     clustering
#     nearest-neighbor search
#     anomaly detection
#     geometric algorithms
#
# Conceptually:
#
#     observations
#          ↓
#     calculate distance
#          ↓
#     determine similarity / proximity
#          ↓
#     make decision
#
# ============================================================
# 25. HYPOT
# ============================================================
#
# math.hypot(x, y)
#
# calculates:
#
#     √(x² + y²)
#
# Example:
#
# math.hypot(3, 4)
#     → 5.0
#
# It is essentially the length of a vector:
#
#     vector = (3, 4)
#
#     length = 5
#
# For multiple dimensions, math.hypot() can accept multiple
# coordinates.
#
# Example:
#
# math.hypot(1, 2, 2)
#
#     → √(1² + 2² + 2²)
#     → √9
#     → 3
#
# ============================================================
# 26. FSUM
# ============================================================
#
# math.fsum(iterable)
#
# calculates a floating-point sum with improved numerical accuracy
# compared with a simple repeated floating-point addition.
#
# Example:
#
# values = [0.1, 0.1, 0.1]
#
# math.fsum(values)
#
#     → 0.30000000000000004
#
# Actually, even fsum can display a floating-point representation
# rather than exact decimal arithmetic.
#
# The important idea is:
#
# floating-point arithmetic is approximate.
#
# `fsum()` is designed to reduce accumulated rounding error.
#
# ============================================================
# WHY FSUM MATTERS IN DATA SCIENCE
# ============================================================
#
# Many statistical calculations contain sums:
#
#     mean
#     variance
#     covariance
#     loss functions
#     probabilities
#
# If millions of floating-point values are accumulated, numerical
# error can matter.
#
# `math.fsum()` is therefore useful when accurate floating-point
# summation is important.
#
# In large array-based DS work:
#
#     NumPy
#
# is usually the more natural tool.
#
# ============================================================
# 27. SUMPROD
# ============================================================
#
# math.sumprod(p, q)
#
# calculates:
#
#     p₁q₁ + p₂q₂ + ... + pₙqₙ
#
# Example:
#
# values  = [10, 20, 30]
# weights = [0.2, 0.3, 0.5]
#
# math.sumprod(values, weights)
#
# Conceptually:
#
#     10 × 0.2 = 2
#     20 × 0.3 = 6
#     30 × 0.5 = 15
#
#     2 + 6 + 15
#          ↓
#         23
#
# This operation is extremely important mathematically.
#
# ============================================================
# WHY SUMPRODUCT MATTERS
# ============================================================
#
# The dot product appears in:
#
#     linear algebra
#     weighted averages
#     linear regression
#     neural networks
#     machine learning
#
# A neuron can conceptually calculate:
#
#     x₁w₁ + x₂w₂ + x₃w₃ + ...
#
# which is a dot product.
#
# Therefore:
#
#     sumprod()
#          ↓
#     weighted combination
#          ↓
#     linear model
#          ↓
#     ML foundation
#
# ============================================================
# 28. ISFINITE
# ============================================================
#
# math.isfinite(x)
#
# asks:
#
#     "Is x neither infinity nor NaN?"
#
# Example:
#
# math.isfinite(10)
#     → True
#
# math.isfinite(math.inf)
#     → False
#
# math.isfinite(math.nan)
#     → False
#
# ============================================================
# DATA QUALITY CONNECTION
# ============================================================
#
# Real datasets can contain:
#
#     normal value
#          ↓
#       25.4
#
#     infinity
#          ↓
#       inf
#
#     missing/invalid numerical result
#          ↓
#       nan
#
# Before numerical calculations:
#
#     raw value
#        ↓
#     isfinite()
#        ↓
#     valid?
#      /   \
#    YES    NO
#     ↓      ↓
# process   handle
#
# This is useful for defensive numerical programming.
#
# ============================================================
# 29. ISINF
# ============================================================
#
# math.isinf(x)
#
# checks whether x is positive or negative infinity.
#
# Example:
#
# math.isinf(math.inf)
#     → True
#
# math.isinf(100)
#     → False
#
# ============================================================
# 30. ISNAN
# ============================================================
#
# math.isnan(x)
#
# checks whether x is NaN.
#
# Example:
#
# math.isnan(math.nan)
#     → True
#
# math.isnan(10)
#     → False
#
# IMPORTANT:
#
# NaN is unusual:
#
#     nan == nan
#         ↓
#       False
#
# Therefore:
#
# use math.isnan(x)
#
# rather than:
#
#     x == math.nan
#
# to test for NaN.
#
# ============================================================
# 31. ISCLOSE
# ============================================================
#
# math.isclose(a, b)
#
# asks whether two floating-point values are sufficiently close.
#
# Example:
#
# a = 0.1 + 0.2
# b = 0.3
#
# a == b
#
# may be:
#
# False
#
# because floating-point representation is approximate.
#
# But:
#
# math.isclose(a, b)
#
# → True
#
# ============================================================
# WHY THIS IS IMPORTANT
# ============================================================
#
# Computers represent many decimal fractions approximately.
#
# Conceptually:
#
# decimal number
#       ↓
# binary floating-point representation
#       ↓
# small rounding difference
#
# Therefore:
#
# exact equality
#       ↓
# can be inappropriate
#
# numerical closeness
#       ↓
# often makes more sense
#
# This is VERY important in scientific computing and ML.
#
# ============================================================
# 32. ISclose PARAMETERS
# ============================================================
#
# math.isclose(
#     a,
#     b,
#     rel_tol=1e-09,
#     abs_tol=0.0
# )
#
# `rel_tol`:
#
# relative tolerance.
#
# `abs_tol`:
#
# absolute tolerance.
#
# Near zero, `abs_tol` is particularly important.
#
# Example:
#
# math.isclose(
#     0.0000001,
#     0,
#     abs_tol=0.000001
# )
#
# → True
#
# ============================================================
# 33. NEXTAFTER
# ============================================================
#
# math.nextafter(x, y)
#
# returns the next representable floating-point number after x,
# moving in the direction of y.
#
# This is an advanced floating-point tool.
#
# It is useful for:
#
#     numerical algorithms
#     boundary handling
#     floating-point analysis
#     scientific computing
#
# It is NOT something you need to memorize for ordinary DA/DS.
#
# ============================================================
# 34. ULP
# ============================================================
#
# math.ulp(x)
#
# ULP = Unit in the Last Place.
#
# It measures the spacing between floating-point numbers around x.
#
# Conceptually:
#
#     ... x₁ ... x₂ ... x₃ ...
#              ↑
#          floating-point
#          spacing
#
# `ulp()` helps you understand the precision of floating-point
# representation.
#
# This is useful when learning numerical computing, but rarely a
# day-to-day DA/DS function.
#
# ============================================================
# 35. FREXP
# ============================================================
#
# math.frexp(x)
#
# decomposes a floating-point number into:
#
#     mantissa × 2^exponent
#
# It returns:
#
#     (mantissa, exponent)
#
# such that:
#
#     x = mantissa × 2**exponent
#
# This is mainly useful for numerical analysis and understanding
# floating-point representation.
#
# ============================================================
# 36. LDEXP
# ============================================================
#
# math.ldexp(x, i)
#
# calculates:
#
#     x × 2ⁱ
#
# It is essentially related to the inverse operation of frexp().
#
# Advanced numerical computing:
#
#     frexp()
#        ↓
#     decompose
#        ↓
#     manipulate
#        ↓
#     ldexp()
#        ↓
#     reconstruct
#
# Not a priority for DA/DS/ML.
#
# ============================================================
# 37. GAMMA
# ============================================================
#
# math.gamma(x)
#
# calculates the Gamma function.
#
# For positive integers:
#
#     Γ(n) = (n - 1)!
#
# Therefore:
#
#     Γ(5) = 4!
#          = 24
#
# This extends the factorial idea beyond ordinary positive
# integers.
#
# ============================================================
# WHY GAMMA MATTERS IN STATISTICS
# ============================================================
#
# Gamma functions appear in:
#
#     probability distributions
#     Bayesian statistics
#     continuous probability models
#     likelihood functions
#     mathematical statistics
#
# Examples include:
#
#     Gamma distribution
#     Beta distribution
#     Student-related mathematical expressions
#     Dirichlet-related mathematics
#
# You may not call math.gamma() every day, but understanding the
# concept is useful when moving into advanced statistics.
#
# ============================================================
# 38. LGAMMA
# ============================================================
#
# math.lgamma(x)
#
# calculates:
#
#     log(|Γ(x)|)
#
# Why use log gamma?
#
# Gamma values can become enormous.
#
# Direct computation:
#
#     gamma(x)
#          ↓
#     huge number
#          ↓
#     overflow / numerical difficulty
#
# Logarithmic computation:
#
#     lgamma(x)
#          ↓
#     much more manageable scale
#
# This is a very important numerical-analysis pattern:
#
#     enormous multiplication
#             ↓
#        take logarithm
#             ↓
#       stable computation
#
# ============================================================
# 39. FACTORIAL
# ============================================================
#
# math.factorial(n)
#
# calculates:
#
#     n!
#
# Definition:
#
#     n! = n × (n-1) × ... × 2 × 1
#
# Example:
#
#     5!
#
#     = 5 × 4 × 3 × 2 × 1
#
#     = 120
#
# ============================================================
# WHY FACTORIAL MATTERS
# ============================================================
#
# Factorials appear in:
#
#     permutations
#     combinations
#     probability
#     discrete distributions
#     combinatorics
#
# ============================================================
# 40. COMBINATIONS
# ============================================================
#
# math.comb(n, k)
#
# calculates:
#
#     n choose k
#
# written:
#
#     C(n,k)
#
# or:
#
#     ( n )
#     ( k )
#
# It answers:
#
#     "How many ways can I choose k objects from n objects
#      when order does NOT matter?"
#
# Example:
#
# Choose 2 people from 5:
#
# math.comb(5, 2)
#
#     = 10
#
# ============================================================
# VISUAL
# ============================================================
#
# PEOPLE:
#
# A  B  C  D  E
#
# Choose 2:
#
# AB
# AC
# AD
# AE
# BC
# BD
# BE
# CD
# CE
# DE
#
# TOTAL:
#
#     10
#
# ============================================================
# ML / STATISTICS CONNECTION
# ============================================================
#
# Combinatorics appears in:
#
#     probability
#     sampling
#     binomial distributions
#     feature-selection reasoning
#     search spaces
#
# ============================================================
# 41. PERMUTATIONS
# ============================================================
#
# math.perm(n, k)
#
# calculates the number of ways to choose k objects from n when
# ORDER MATTERS.
#
# Example:
#
# math.perm(5, 2)
#
#     = 5 × 4
#     = 20
#
# Compare:
#
# comb(5, 2) = 10
# perm(5, 2) = 20
#
# Why?
#
# AB and BA are:
#
#     same choice for combinations
#
# but:
#
#     different arrangements for permutations
#
# ============================================================
# 42. GCD
# ============================================================
#
# math.gcd(a, b)
#
# GCD = Greatest Common Divisor.
#
# Example:
#
# math.gcd(12, 18)
#
# Divisors of 12:
#
# 1, 2, 3, 4, 6, 12
#
# Divisors of 18:
#
# 1, 2, 3, 6, 9, 18
#
# Greatest common divisor:
#
#     6
#
# This is mainly number theory rather than DA/DS/ML.
#
# ============================================================
# 43. LCM
# ============================================================
#
# math.lcm(a, b)
#
# LCM = Least Common Multiple.
#
# Example:
#
# math.lcm(4, 6)
#
# Multiples of 4:
#
# 4, 8, 12, 16, ...
#
# Multiples of 6:
#
# 6, 12, 18, ...
#
# First common multiple:
#
#     12
#
# Again, useful mainly in algorithms and number theory.
#
# ============================================================
# 44. SINE
# ============================================================
#
# math.sin(x)
#
# calculates the sine of an angle.
#
# IMPORTANT:
#
# math.sin() expects the angle in RADIANS.
#
# Example:
#
# math.sin(math.pi / 2)
#
#     → 1.0
#
# ============================================================
# 45. COSINE
# ============================================================
#
# math.cos(x)
#
# also expects radians.
#
# Example:
#
# math.cos(0)
#
#     → 1.0
#
# ============================================================
# 46. TANGENT
# ============================================================
#
# math.tan(x)
#
# Example:
#
# math.tan(math.pi / 4)
#
#     → approximately 1.0
#
# ============================================================
# 47. ANGLE CONVERSION
# ============================================================
#
# math.radians(degrees)
#
# converts degrees → radians.
#
# math.degrees(radians)
#
# converts radians → degrees.
#
# Example:
#
# math.radians(180)
#
#     → π
#
# and:
#
# math.degrees(math.pi)
#
#     → 180
#
# ============================================================
# WHY TRIGONOMETRY MATTERS FOR DS/ML
# ============================================================
#
# Trigonometry is not central to ordinary business DA.
#
# But it becomes useful in:
#
#     signal processing
#     computer vision
#     robotics
#     time-series feature engineering
#     spatial data
#     physics-based ML
#     Fourier analysis
#
# ============================================================
# 48. INVERSE TRIGONOMETRY
# ============================================================
#
# math.asin(x)
#     → inverse sine
#
# math.acos(x)
#     → inverse cosine
#
# math.atan(x)
#     → inverse tangent
#
# They return angles in radians.
#
# ============================================================
# 49. ATAN2
# ============================================================
#
# math.atan2(y, x)
#
# calculates the angle of a point/vector while correctly considering
# its quadrant.
#
# Point:
#
#          y
#          ↑
#          |
#          |    *
#          |   /
#          |  /
#          | /
#          +------------→ x
#
# atan2(y, x)
#     ↓
# angle
#
# It is generally more useful than atan(y/x) when you need to know
# the correct quadrant.
#
# Useful in:
#
#     robotics
#     computer vision
#     geometry
#     spatial data
#
# ============================================================
# 50. HYPERBOLIC FUNCTIONS
# ============================================================
#
# math.sinh(x)
# math.cosh(x)
# math.tanh(x)
#
# These are hyperbolic sine, cosine, and tangent.
#
# Of particular importance to ML:
#
#     tanh
#
# because the hyperbolic tangent is historically used as a neural
# network activation function.
#
# Conceptually:
#
#     x
#      ↓
#    tanh()
#      ↓
#  output between -1 and +1
#
# ============================================================
# 51. TANH AND MACHINE LEARNING
# ============================================================
#
# tanh(x) maps values approximately like:
#
#       +1 ┤             ______
#          │          __/
#          │       __/
#        0 ┤------/
#          │   __
#          │__/
#       -1 ┤
#          └────────────────── x
#
# Large positive x:
#
#     tanh(x) → +1
#
# Large negative x:
#
#     tanh(x) → -1
#
# x near zero:
#
#     tanh(x) ≈ x
#
# Modern deep learning usually uses other activations such as ReLU
# variants more often, but tanh remains conceptually important.
#
# ============================================================
# 52. INVERSE HYPERBOLIC FUNCTIONS
# ============================================================
#
# math.asinh(x)
# math.acosh(x)
# math.atanh(x)
#
# These are inverse hyperbolic functions.
#
# They are specialized mathematical tools.
#
# For ordinary DA/DS/ML:
#
#     know they exist
#     ↓
#     understand their mathematical role when needed
#     ↓
#     do not prioritize memorization
#
# ============================================================
# 53. ERF — ERROR FUNCTION
# ============================================================
#
# math.erf(x)
#
# The error function is strongly connected to probability and the
# normal distribution.
#
# It is defined through an integral related to:
#
#     e^(-t²)
#
# You do not need to memorize the integral to use it.
#
# Conceptually:
#
#     normal-distribution mathematics
#              ↓
#            erf()
#              ↓
#     probability calculations
#
# ============================================================
# 54. ERFC
# ============================================================
#
# math.erfc(x)
#
# is the complementary error function:
#
#     erfc(x) = 1 - erf(x)
#
# It is useful in numerical probability calculations.
#
# These functions are more relevant when studying:
#
#     probability theory
#     normal distributions
#     statistical mathematics
#     scientific computing
#
# than ordinary business DA.
#
# ============================================================
# 55. CONSTANTS
# ============================================================
#
# math provides important constants:
#
# math.pi
#     ≈ 3.141592653589793
#
# math.e
#     ≈ 2.718281828459045
#
# math.tau
#     = 2π
#
# math.inf
#     → positive infinity
#
# math.nan
#     → NaN
#
# ============================================================
# PI
# ============================================================
#
# π represents the ratio:
#
#     circumference
#     ─────────────
#       diameter
#
# It appears in:
#
#     circles
#     trigonometry
#     normal distributions
#     probability density functions
#     Fourier mathematics
#     geometry
#
# ============================================================
# 56. TAU
# ============================================================
#
# math.tau = 2π
#
# One full rotation:
#
#     360°
#       ↓
#     2π radians
#       ↓
#     tau
#
# Some mathematical contexts find tau conceptually convenient.
#
# ============================================================
# 57. INF AND NAN
# ============================================================
#
# math.inf
#
# represents positive infinity.
#
# math.nan
#
# represents Not-a-Number.
#
# These values matter greatly in data quality.
#
# A numerical pipeline can conceptually become:
#
#     valid values
#          ↓
#     calculation
#          ↓
#     overflow / invalid operation
#          ↓
#     inf / nan
#
# Therefore:
#
#     isfinite()
#     isinf()
#     isnan()
#
# are useful defensive tools.
#
# ============================================================
# 58. A STATISTICAL EXAMPLE USING MATH
# ============================================================
#
# Suppose we want to understand standard deviation manually.
#
# data = [2, 4, 6]
#
# mean = (2 + 4 + 6) / 3
#
#     = 4
#
# deviations:
#
#     2 - 4 = -2
#     4 - 4 =  0
#     6 - 4 = +2
#
# squared deviations:
#
#     (-2)² = 4
#      0²  = 0
#     (+2)² = 4
#
# population variance:
#
#     (4 + 0 + 4) / 3
#          ↓
#          8 / 3
#
# population standard deviation:
#
#     sqrt(8 / 3)
#
# Therefore:
#
# math.sqrt(8 / 3)
#
# connects the statistical concept to the mathematical operation.
#
# ============================================================
# 59. MATH + STATISTICS RELATIONSHIP
# ============================================================
#
# `statistics` provides:
#
#     statistics.stdev(data)
#
# `math` provides:
#
#     math.sqrt(...)
#
# This illustrates an important distinction:
#
# statistics
#     → statistical meaning
#
# math
#     → mathematical machinery
#
# Example:
#
# variance
#     ↓
# math.sqrt()
#     ↓
# standard deviation
#
# The two modules complement each other.
#
# ============================================================
# 60. EUCLIDEAN DISTANCE AND ML
# ============================================================
#
# Suppose we have two observations:
#
# A = [2, 3]
# B = [5, 7]
#
# Difference:
#
#     Δx = 5 - 2 = 3
#     Δy = 7 - 3 = 4
#
# Distance:
#
#     √(3² + 4²)
#
#     ↓
#
#     √25
#
#     ↓
#
#     5
#
# This is the geometric basis of many algorithms.
#
#     points
#       ↓
#    distances
#       ↓
#    similarity
#       ↓
#    nearest neighbors / clusters
#
# ============================================================
# 61. LOGARITHMS AND LOSS FUNCTIONS
# ============================================================
#
# Logarithms are central to many ML objectives.
#
# Binary cross-entropy contains terms such as:
#
#     -[y log(p) + (1-y) log(1-p)]
#
# Conceptually:
#
# predicted probability
#          ↓
#        log()
#          ↓
#      loss contribution
#          ↓
#       optimization
#
# This is why understanding:
#
#     log()
#     log1p()
#
# is valuable for ML.
#
# In practical ML code, however, you usually use the framework's
# numerical implementation rather than manually calculating every
# term.
#
# ============================================================
# 62. EXPONENTIALS AND SOFTMAX
# ============================================================
#
# Softmax:
#
#             eˣⁱ
#     pᵢ = ───────────
#           Σ eˣʲ
#
# Conceptual flow:
#
# raw scores
#     ↓
# exponentiate
#     ↓
# positive values
#     ↓
# divide by total
#     ↓
# probabilities
#
# `math.exp()` therefore represents one of the fundamental
# mathematical building blocks behind softmax.
#
# ============================================================
# 63. FACTORIAL / COMB / PERM AND PROBABILITY
# ============================================================
#
# Many probability calculations involve counting possible outcomes.
#
#                 POSSIBLE ARRANGEMENTS
#                          |
#             +------------+------------+
#             |                         |
#             ↓                         ↓
#         order matters             order doesn't
#             |                         |
#             ↓                         ↓
#          perm()                    comb()
#
# factorial() is the mathematical foundation beneath both.
#
# This becomes useful when studying:
#
#     binomial probability
#     sampling
#     combinatorics
#     discrete probability
#
# ============================================================
# 64. GAMMA AND ADVANCED STATISTICS
# ============================================================
#
# Factorial:
#
#     1 × 2 × 3 × ... × n
#
# Gamma extends this idea into continuous mathematics.
#
#     factorial
#         ↓
#     Γ(n) = (n-1)!
#         ↓
#     continuous extension
#         ↓
#     probability distributions
#
# This is one reason gamma appears in advanced statistical
# mathematics.
#
# ============================================================
# 65. NUMERICAL STABILITY — A CRITICAL DS CONCEPT
# ============================================================
#
# One of the most important lessons from the math module is:
#
# MATHEMATICAL FORMULA
#          ≠
# COMPUTER IMPLEMENTATION
#
# The mathematical expression may be correct, but a naive numerical
# implementation can lose precision.
#
# Examples:
#
#     exp(x) - 1
#         ↓
#     use expm1(x)
#
#     log(1 + x)
#         ↓
#     use log1p(x)
#
#     large floating-point summation
#         ↓
#     use fsum()
#
#     compare floating-point values
#         ↓
#     use isclose()
#
# This is a major scientific-computing principle.
#
# ============================================================
# 66. FLOATING-POINT MENTAL MODEL
# ============================================================
#
# Do NOT think:
#
#     computer number = exact mathematical real number
#
# Better model:
#
#     mathematical real number
#             ↓
#     finite binary representation
#             ↓
#     approximation
#             ↓
#     floating-point operations
#             ↓
#     possible rounding error
#
# This explains why:
#
#     0.1 + 0.2
#
# may not be represented exactly as:
#
#     0.3
#
# And why:
#
#     math.isclose()
#
# is so useful.
#
# ============================================================
# 67. MATH FUNCTIONS MOST IMPORTANT FOR DA
# ============================================================
#
# For practical Data Analysis, prioritize:
#
# ★★★★★
#
# math.fabs()
# math.sqrt()
# math.floor()
# math.ceil()
# math.log()
# math.log10()
# math.exp()
# math.isfinite()
# math.isnan()
# math.isclose()
#
# ★★★★☆
#
# math.fsum()
# math.log1p()
# math.expm1()
# math.dist()
# math.hypot()
#
# ★★★☆☆
#
# math.factorial()
# math.comb()
# math.perm()
# math.gcd()
# math.lcm()
#
# Trigonometric functions are generally less important for ordinary
# business DA.
#
# ============================================================
# 68. MATH FUNCTIONS MOST IMPORTANT FOR DS
# ============================================================
#
# For Data Science, prioritize:
#
# ★★★★★
#
# sqrt()
# log()
# log2()
# log10()
# exp()
# fsum()
# isfinite()
# isnan()
# isclose()
# dist()
# hypot()
#
# ★★★★☆
#
# factorial()
# comb()
# perm()
# gamma()
# lgamma()
# erf()
# erfc()
#
# Depending on your domain:
#
# sin()
# cos()
# tan()
# atan2()
#
# become important for:
#
#     signals
#     spatial data
#     physics
#     engineering
#     computer vision
#
# ============================================================
# 69. MATH FUNCTIONS MOST IMPORTANT FOR ML
# ============================================================
#
# For Machine Learning, the most conceptually important are:
#
# ★★★★★
#
# exp()
# log()
# sqrt()
# fsum()
# isclose()
# dist()
# hypot()
#
# ★★★★☆
#
# log1p()
# expm1()
# isfinite()
# isnan()
# gamma()
# lgamma()
#
# Domain-dependent:
#
# sin()
# cos()
# tan()
# tanh()
# erf()
# erfc()
#
# Why?
#
#     ML
#      |
#      +-- optimization
#      |      ↓
#      |    exp/log
#      |
#      +-- geometry
#      |      ↓
#      |    distance
#      |
#      +-- probability
#      |      ↓
#      |    exp/log/erf/gamma
#      |
#      +-- numerical computing
#             ↓
#           fsum/isclose
#
# ============================================================
# 70. WHAT YOU SHOULD MASTER
# ============================================================
#
# Given your DA + DS + ML goal, do NOT spend equal time on every
# function in math.
#
# MASTER:
#
#     sqrt()
#     log()
#     log2()
#     log10()
#     exp()
#     fsum()
#     isfinite()
#     isnan()
#     isclose()
#     dist()
#     hypot()
#
# UNDERSTAND WELL:
#
#     floor()
#     ceil()
#     trunc()
#     pow()
#     expm1()
#     log1p()
#     factorial()
#     comb()
#     perm()
#
# KNOW WHEN NEEDED:
#
#     gamma()
#     lgamma()
#     erf()
#     erfc()
#     sin()
#     cos()
#     tan()
#     atan2()
#     tanh()
#
# LOWER PRIORITY FOR ORDINARY DA/DS/ML:
#
#     fmod()
#     remainder()
#     modf()
#     copysign()
#     nextafter()
#     ulp()
#     frexp()
#     ldexp()
#     gcd()
#     lcm()
#     isqrt()
#
# They are still legitimate parts of the library, but they should
# not consume disproportionate learning time.
#
# ============================================================
# 71. COMPLETE MENTAL MAP OF THE MATH MODULE
# ============================================================
#
#                         math
#                          |
#       +------------------+------------------+
#       |                  |                  |
#       ↓                  ↓                  ↓
#    NUMBERS           FUNCTIONS          SPECIAL
#       |                  |             FUNCTIONS
#       |                  |                  |
#       ↓                  ↓                  ↓
# factorial             sqrt()             gamma()
# comb()                cbrt()             lgamma()
# perm()                pow()              erf()
# gcd()                 exp()              erfc()
# lcm()                 log()
# isqrt()               log2()
#                       log10()
#                       sin()
#                       cos()
#                       tan()
#                       tanh()
#                       hypot()
#                       dist()
#
#                          +
#
#                 NUMERICAL PRECISION
#                          |
#             +------------+------------+
#             |            |            |
#           fsum()      isclose()    isfinite()
#                                      |
#                                    isnan()
#                                    isinf()
#
# ============================================================
# 72. A COMPLETE DA/DS/ML CONNECTION MAP
# ============================================================
#
#                         MATH
#                           |
#        +------------------+------------------+
#        |                  |                  |
#        ↓                  ↓                  ↓
#     STATISTICS        GEOMETRY          PROBABILITY
#        |                  |                  |
#        ↓                  ↓                  ↓
#   sqrt / log          dist / hypot       exp / log
#   fsum                sqrt               gamma
#   exp                 pow                erf
#        |                  |                  |
#        +------------------+------------------+
#                           ↓
#                       DATA SCIENCE
#                           ↓
#                    MACHINE LEARNING
#
# ============================================================
# 73. MATH VS STATISTICS
# ============================================================
#
# This distinction is important.
#
# `statistics` asks:
#
#     "What statistical quantity describes my data?"
#
# Example:
#
#     statistics.mean(data)
#
# `math` asks:
#
#     "What mathematical operation do I need?"
#
# Example:
#
#     math.sqrt(x)
#
# Therefore:
#
# statistics.mean()
#       ↓
# statistical concept
#
# math.sqrt()
#       ↓
# mathematical operation
#
# You should learn BOTH, but for different reasons.
#
# ============================================================
# 74. MATH VS NUMPY
# ============================================================
#
# `math` works primarily with individual scalar values.
#
# Example:
#
#     math.sqrt(25)
#
# NumPy is designed for arrays:
#
#     array = [25, 36, 49]
#
#     np.sqrt(array)
#
# Conceptually:
#
# math:
#
#     one value
#        ↓
#     one operation
#
# NumPy:
#
#     many values
#        ↓
#     vectorized operation
#
# This distinction becomes extremely important in DS.
#
# ============================================================
# 75. MATH VS PANDAS
# ============================================================
#
# math:
#
#     mathematical computation
#
# pandas:
#
#     data organization + tabular analysis
#
# Example workflow:
#
#     CSV
#      ↓
#    pandas
#      ↓
#    clean data
#      ↓
#    select numerical column
#      ↓
#    numerical transformation
#      ↓
#    NumPy / math
#
# `math` is therefore usually one small component of a larger
# analytical workflow.
#
# ============================================================
# 76. MATH VS SCIPY
# ============================================================
#
# math:
#
#     core mathematical functions
#
# SciPy:
#
#     advanced scientific computing
#
# SciPy provides much broader functionality for:
#
#     optimization
#     statistics
#     probability distributions
#     integration
#     interpolation
#     signal processing
#     numerical algorithms
#
# Think:
#
# math
#   ↓
# basic mathematical building blocks
#
# SciPy
#   ↓
# advanced scientific machinery
#
# ============================================================
# 77. MATH VS STATSMODELS
# ============================================================
#
# math:
#
#     mathematical operations
#
# statsmodels:
#
#     statistical models + inference
#
# Example:
#
# math.log()
#
# is a mathematical function.
#
# A statistical regression model with:
#
#     coefficients
#     standard errors
#     confidence intervals
#     p-values
#
# belongs to a much higher statistical layer.
#
# ============================================================
# 78. MATH VS SCIKIT-LEARN
# ============================================================
#
# math:
#
#     mathematical building blocks
#
# scikit-learn:
#
#     machine-learning algorithms and workflows
#
# Example:
#
# distance mathematics
#       ↓
# math.dist()
#
#      BUT
#
# nearest-neighbor ML
#       ↓
# scikit-learn
#
# The library hierarchy is therefore:
#
# mathematical operation
#          ↓
#       math
#          ↓
# numerical arrays
#          ↓
#       NumPy
#          ↓
# data structures
#          ↓
#       pandas
#          ↓
# statistical/scientific methods
#          ↓
# SciPy / statsmodels
#          ↓
# machine learning
#          ↓
# scikit-learn
#
# ============================================================
# 79. A PRACTICAL EXAMPLE — ERROR
# ============================================================
#
# actual = 100
# predicted = 92
#
# error = actual - predicted
#
# absolute_error = math.fabs(error)
#
# squared_error = error ** 2
#
# rmse_component = math.sqrt(squared_error)
#
# VALUE FLOW:
#
# actual = 100
# predicted = 92
#       ↓
# error = 8
#       ↓
# absolute error = 8
#       ↓
# squared error = 64
#       ↓
# sqrt(64) = 8
#
# This tiny example connects:
#
#     fabs()
#     exponentiation
#     sqrt()
#
# with fundamental ML error calculations.
#
# ============================================================
# 80. A PRACTICAL EXAMPLE — LOG TRANSFORMATION
# ============================================================
#
# revenue = 10000
#
# transformed = math.log10(revenue)
#
#     10000
#       ↓
#    log10()
#       ↓
#       4
#
# This changes a large multiplicative scale into a much smaller
# logarithmic scale.
#
# In real DS:
#
#     raw feature
#          ↓
#     inspect distribution
#          ↓
#     decide whether log transform is appropriate
#          ↓
#     transform
#          ↓
#     model / analyze
#
# IMPORTANT:
#
# Do not automatically log-transform every variable.
#
# ============================================================
# 81. A PRACTICAL EXAMPLE — NUMERICAL SAFETY
# ============================================================
#
# x = 0.000000001
#
# Instead of:
#
# math.log(1 + x)
#
# use:
#
# math.log1p(x)
#
# because `log1p()` is designed for accuracy when x is close to zero.
#
# Similar pattern:
#
# instead of:
#
# math.exp(x) - 1
#
# use:
#
# math.expm1(x)
#
# General lesson:
#
# mathematically equivalent
#          ≠
# numerically equivalent in accuracy
#
# This is a VERY important concept for scientific computing.
#
# ============================================================
# 82. A PRACTICAL EXAMPLE — FLOATING-POINT COMPARISON
# ============================================================
#
# result = 0.1 + 0.2
#
# expected = 0.3
#
# Do NOT build numerical code around:
#
# result == expected
#
# when exact floating-point equality is not guaranteed.
#
# Prefer:
#
# math.isclose(result, expected)
#
# Concept:
#
#     mathematical equality
#            ↓
#     "these should be the same"
#
# Computer reality:
#
#     approximate floating-point representations
#            ↓
#     tiny numerical difference
#
# Therefore:
#
#     isclose()
#          ↓
#     tolerance-based comparison
#
# ============================================================
# 83. A PRACTICAL EXAMPLE — DISTANCE
# ============================================================
#
# customer_a = [25, 50000]
# customer_b = [30, 55000]
#
# distance = math.dist(customer_a, customer_b)
#
# Mathematically:
#
#     √((30-25)² + (55000-50000)²)
#
#     ↓
#
# huge contribution from income
#
# This reveals an IMPORTANT ML lesson:
#
# raw Euclidean distance can be dominated by features with larger
# numerical scales.
#
# Therefore:
#
#     raw features
#          ↓
#     distance
#          ↓
#     potentially misleading similarity
#
# often requires:
#
#     feature scaling
#          ↓
#     standardized features
#          ↓
#     distance
#
# This is why feature scaling matters for distance-based algorithms.
#
# ============================================================
# 84. MATH AND STANDARDIZATION
# ============================================================
#
# Standardization:
#
#     z = (x - μ) / σ
#
# The mathematical building blocks are:
#
#     subtraction
#     division
#
# and the statistics behind them are:
#
#     mean
#     standard deviation
#
# `math` helps with mathematical operations, while
# `statistics` can calculate the statistical quantities.
#
# This illustrates how Python libraries work together rather than
# existing as isolated subjects.
#
# ============================================================
# 85. MATH AND NORMAL DISTRIBUTION
# ============================================================
#
# The normal probability density function contains:
#
#                1
#     ───────────────────────
#     σ√(2π)
#
# and:
#
#                 (x-μ)²
#               - ───────
#                  2σ²
#     e
#
# This involves:
#
#     math.pi
#     math.sqrt()
#     math.exp()
#
# Therefore:
#
# NORMAL DISTRIBUTION
#        ↓
#     mathematical formula
#        ↓
#     math.pi
#     sqrt()
#     exp()
#
# The `math` module provides the low-level mathematical pieces.
#
# ============================================================
# 86. IMPORTANT: DON'T REIMPLEMENT EVERYTHING
# ============================================================
#
# Knowing the mathematics does NOT mean you should manually
# implement everything in production.
#
# For example:
#
# You should understand:
#
#     standard deviation
#          ↓
#        sqrt()
#
# But for ordinary statistical analysis:
#
#     statistics.stdev(data)
#
# is preferable to manually rebuilding the entire algorithm.
#
# Likewise:
#
# understand:
#
#     Euclidean distance
#
# but use appropriate optimized numerical/ML libraries when
# processing large datasets.
#
# Learning mathematics and writing production code are related but
# different skills.
#
# ============================================================
# 87. FUNCTION PRIORITY TABLE
# ============================================================
#
# +----------------+----------------------+----------------------+
# | Function       | Main concept         | DA/DS/ML importance  |
# +----------------+----------------------+----------------------+
# | sqrt()         | square root          | VERY HIGH            |
# | exp()          | exponential          | VERY HIGH            |
# | log()          | logarithm            | VERY HIGH            |
# | log2()         | base-2 logarithm     | HIGH                 |
# | log10()        | base-10 logarithm    | HIGH                 |
# | fsum()         | accurate sum         | HIGH                 |
# | isclose()      | numerical comparison | HIGH                 |
# | isfinite()     | numeric validation   | HIGH                 |
# | isnan()        | NaN detection        | HIGH                 |
# | dist()         | Euclidean distance   | HIGH for ML         |
# | hypot()        | vector magnitude     | HIGH for ML         |
# | floor()        | lower integer       | MEDIUM              |
# | ceil()         | upper integer       | MEDIUM              |
# | trunc()        | remove fraction      | LOW-MEDIUM          |
# | log1p()        | stable log(1+x)     | HIGH for numerical  |
# | expm1()        | stable exp(x)-1     | HIGH for numerical  |
# | comb()         | combinations        | MEDIUM statistics   |
# | perm()         | permutations        | MEDIUM statistics   |
# | factorial()    | factorial            | MEDIUM statistics   |
# | gamma()        | Gamma function       | ADVANCED statistics |
# | lgamma()       | log Gamma            | ADVANCED statistics |
# | erf()          | error function       | ADVANCED statistics |
# | tanh()         | hyperbolic tangent   | ML / neural nets    |
# +----------------+----------------------+----------------------+
#
# ============================================================
# 88. FUNCTIONS TO KNOW BUT NOT PRIORITIZE
# ============================================================
#
# These are legitimate and useful, but they should NOT consume much
# of your learning time if your immediate goal is DA/DS/ML:
#
#     fmod()
#     remainder()
#     modf()
#     copysign()
#     frexp()
#     ldexp()
#     nextafter()
#     ulp()
#     gcd()
#     lcm()
#     isqrt()
#
# They become more relevant in:
#
#     numerical programming
#     scientific computing
#     algorithms
#     low-level numerical analysis
#
# ============================================================
# 89. THE MOST IMPORTANT CONCEPTUAL GROUPS
# ============================================================
#
# Instead of memorizing ~50 functions independently, remember them
# as groups.
#
# ------------------------------------------------------------
# GROUP 1 — SCALE
# ------------------------------------------------------------
#
# sqrt()
# pow()
# cbrt()
#
#     x
#     ↓
# transformation of magnitude
#
# ------------------------------------------------------------
# GROUP 2 — GROWTH / DECAY
# ------------------------------------------------------------
#
# exp()
# log()
# log2()
# log10()
# expm1()
# log1p()
#
#     exponential
#          ↕
#        logarithm
#
# ------------------------------------------------------------
# GROUP 3 — GEOMETRY
# ------------------------------------------------------------
#
# hypot()
# dist()
#
#     coordinates
#          ↓
#       differences
#          ↓
#       squares
#          ↓
#       sum
#          ↓
#       sqrt
#          ↓
#       distance
#
# ------------------------------------------------------------
# GROUP 4 — NUMERICAL SAFETY
# ------------------------------------------------------------
#
# fsum()
# isclose()
# isfinite()
# isnan()
# isinf()
#
#     calculation
#          ↓
#     numerical reality
#          ↓
#     validate / compare
#
# ------------------------------------------------------------
# GROUP 5 — COUNTING
# ------------------------------------------------------------
#
# factorial()
# comb()
# perm()
#
#     possible arrangements
#          ↓
#     probability / combinatorics
#
# ------------------------------------------------------------
# GROUP 6 — ADVANCED PROBABILITY
# ------------------------------------------------------------
#
# gamma()
# lgamma()
# erf()
# erfc()
#
#     mathematical statistics
#              ↓
#     probability distributions
#
# ============================================================
# 90. GENERALIZABLE PATTERN
# ============================================================
#
# When you encounter a mathematical expression in Data Science,
# ask:
#
#     WHAT mathematical operation is required?
#                     ↓
#     Is it arithmetic?
#                     ↓
#     Is it a power/root?
#                     ↓
#     Is it exponential/logarithmic?
#                     ↓
#     Is it geometric?
#                     ↓
#     Is it probabilistic?
#                     ↓
#     Is numerical precision important?
#
# Then select the appropriate mathematical tool.
#
# ============================================================
# 91. THE DEEPER RELATIONSHIP
# ============================================================
#
# A useful way to organize your learning is:
#
#                  DATA
#                    ↓
#             mathematical operation
#                    ↓
#                   math
#                    ↓
#             statistical meaning
#                    ↓
#                statistics
#                    ↓
#            array computation
#                    ↓
#                  NumPy
#                    ↓
#            tabular computation
#                    ↓
#                 pandas
#                    ↓
#       advanced statistical analysis
#                    ↓
#           SciPy / statsmodels
#                    ↓
#            machine learning
#                    ↓
#             scikit-learn
#
# Do not see these libraries as competitors.
#
# They occupy different layers.
#
# ============================================================
# 92. WHAT YOU SHOULD ACTUALLY LEARN FOR DA
# ============================================================
#
# If your immediate objective is getting strong at Data Analysis,
# focus on:
#
#     math.sqrt()
#     math.log()
#     math.log10()
#     math.exp()
#     math.fsum()
#     math.floor()
#     math.ceil()
#     math.isclose()
#     math.isfinite()
#     math.isnan()
#
# Understand:
#
#     distance
#     logarithmic transformation
#     exponential transformation
#     numerical precision
#
# You do NOT need deep mastery of:
#
#     gamma
#     erf
#     frexp
#     ldexp
#     nextafter
#
# at the beginning.
#
# ============================================================
# 93. WHAT YOU SHOULD ACTUALLY LEARN FOR DS
# ============================================================
#
# For Data Science:
#
# MASTER:
#
#     sqrt
#     log
#     log2
#     log10
#     exp
#     fsum
#     isclose
#     isfinite
#     isnan
#     dist
#     hypot
#
# UNDERSTAND:
#
#     factorial
#     comb
#     perm
#     gamma
#     erf
#
# DEPENDING ON DOMAIN:
#
#     trigonometric functions
#     hyperbolic functions
#
# ============================================================
# 94. WHAT YOU SHOULD ACTUALLY LEARN FOR ML
# ============================================================
#
# For ML, the most important mathematical ideas are:
#
#     powers
#     square roots
#     logarithms
#     exponentials
#     distances
#     numerical precision
#
# Therefore prioritize:
#
#     sqrt()
#     exp()
#     log()
#     log2()
#     log10()
#     fsum()
#     isclose()
#     dist()
#     hypot()
#
# Then understand:
#
#     tanh()
#     gamma()
#     erf()
#     lgamma()
#
# when they arise in specific ML/statistical contexts.
#
# ============================================================
# 95. CORE IDEA
# ============================================================
#
# The `math` module is the LOW-LEVEL MATHEMATICAL TOOLBOX behind
# many higher-level concepts in Data Analysis, Data Science, and
# Machine Learning.
#
# The most important mental model is:
#
#                     MATH
#                       |
#          +------------+------------+
#          |            |            |
#          ↓            ↓            ↓
#        SCALE        GROWTH       GEOMETRY
#          |            |            |
#       sqrt()       exp/log       dist()
#       pow()        log1p()       hypot()
#       cbrt()       expm1()
#          |            |            |
#          +------------+------------+
#                       ↓
#                  STATISTICS
#                       ↓
#                 DATA SCIENCE
#                       ↓
#                 MACHINE LEARNING
#
# The most important lesson is NOT:
#
#     "Memorize every function in math."
#
# It is:
#
#     "Recognize the mathematical operation behind a problem,
#      understand what it means,
#      and know which Python tool performs it."
#
# ============================================================
# 96. FINAL LEARNING PRIORITY FOR YOU
# ============================================================
#
# For a DA → DS → ML path, I would learn `math` in this order:
#
# PHASE 1 — ESSENTIAL
#
#     sqrt()
#     pow() / **
#     floor()
#     ceil()
#     fabs()
#
# PHASE 2 — VERY IMPORTANT
#
#     exp()
#     log()
#     log2()
#     log10()
#     log1p()
#     expm1()
#
# PHASE 3 — NUMERICAL COMPUTING
#
#     fsum()
#     isclose()
#     isfinite()
#     isnan()
#     isinf()
#
# PHASE 4 — ML GEOMETRY
#
#     dist()
#     hypot()
#
# PHASE 5 — PROBABILITY / STATISTICS FOUNDATIONS
#
#     factorial()
#     comb()
#     perm()
#     gamma()
#     lgamma()
#     erf()
#     erfc()
#
# PHASE 6 — DOMAIN-SPECIFIC
#
#     sin()
#     cos()
#     tan()
#     atan2()
#     sinh()
#     cosh()
#     tanh()
#
# PHASE 7 — ADVANCED NUMERICAL DETAILS
#
#     fmod()
#     remainder()
#     modf()
#     copysign()
#     frexp()
#     ldexp()
#     nextafter()
#     ulp()
#
# ============================================================
# FINAL CONNECTION
# ============================================================
#
# You do NOT need to become a "math library expert" to become good
# at DA/DS/ML.
#
# Instead, aim for this mental progression:
#
#     Python syntax
#          ↓
#     mathematical operation
#          ↓
#     statistical meaning
#          ↓
#     data interpretation
#          ↓
#     algorithm
#          ↓
#     model
#
# Example:
#
#     math.sqrt()
#          ↓
#     square root
#          ↓
#     standard deviation / Euclidean distance / RMSE
#          ↓
#     understand spread / similarity / error
#          ↓
#     statistical analysis / ML algorithm
#
# Another:
#
#     math.log()
#          ↓
#     logarithm
#          ↓
#     probability / information / transformation
#          ↓
#     entropy / likelihood / skew reduction
#          ↓
#     statistics / ML
#
# Another:
#
#     math.exp()
#          ↓
#     exponential
#          ↓
#     probability transformation
#          ↓
#     logistic / softmax / probability models
#          ↓
#     machine learning
#
# That is the real purpose of learning the `math` library:
#
#     NOT memorizing functions
#
#                 ↓
#
#     UNDERSTANDING THE MATHEMATICAL BUILDING BLOCKS
#
#                 ↓
#
#     RECOGNIZING THEM INSIDE DA / DS / ML
#
# ============================================================
# END OF MATH LIBRARY — STATISTICS / DA / DS / ML NOTES
# ============================================================
```
